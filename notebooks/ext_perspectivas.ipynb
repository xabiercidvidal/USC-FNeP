# Extensión: Perspectivas

Tarefas e material complementario do tema [Perspectivas](./perspectivas.ipynb).

Aquí van os desenvolvementos que non entran nas transparencias. A notación
(contravariante e covariante, cuadrigradiente, $\hat{p}^\mu = i\partial^\mu$) é a do tema.

## A ecuación de Dirac

<div class="admonition tip">
<p class="admonition-title"><b>[>] Tarefa — Reescribe a ecuación de Dirac</b></p>

Cos operadores $\hat{p}^\mu = i\partial^\mu$, a ecuación de Dirac é
$(\gamma^\mu \hat{p}_\mu - m) \, \Psi = 0$.

1. Desenvolve a contracción $\gamma^\mu \hat{p}_\mu$ na súa parte temporal e espacial, en
   función de $\hat{H}$ e $\hat{\bf p}$. Que signo leva o termo do momento, e de onde sae?
2. Multiplica por $\gamma^0$ pola esquerda e despexa $\hat{H}$. É a forma en que a
   escribiu Dirac en 1928: $\hat{H} = \vec{\alpha} \cdot \hat{\bf p} + \beta \, m$. Que son
   $\vec{\alpha}$ e $\beta$? Que lles esixe que $\hat{H}$ sexa hermítico?
3. Aplica $(\gamma^\nu \hat{p}_\nu + m)$ á ecuación e usa a álxebra
   $\{\gamma^\mu, \gamma^\nu\} = 2 g^{\mu\nu}$. Que ecuación cumpre cada compoñente de
   $\Psi$? Que relación entre $E$ e ${\bf p}$ impón sobre unha onda plana?
4. Substitúe as dúas solucións, $\Psi_s = u_s \, e^{-i p_\mu x^\mu}$ e
   $\Phi_s = v_s \, e^{+i p_\mu x^\mu}$, e obtén as ecuacións dos espinores $u_s$ e $v_s$.
   De onde sae o signo $+m$ da de $v_s$?

</div>

<details>
<summary><b>Solución</b></summary>

1. $\gamma^\mu \hat{p}_\mu = \gamma^0 \hat{p}_0 + \gamma^k \hat{p}_k = \gamma^0 \hat{H} - \vec{\gamma} \cdot \hat{\bf p}$,
   logo

   $$(\gamma^0 \hat{H} - \vec{\gamma} \cdot \hat{\bf p} - m) \, \Psi = 0$$

   O signo menos é o da métrica: $\hat{p}_k = -\hat{p}^k$ (índice abaixo), e o momento
   físico é o de índice arriba.

2. Con $(\gamma^0)^2 = I$:

   $$\hat{H} \, \Psi = (\gamma^0 \vec{\gamma} \cdot \hat{\bf p} + \gamma^0 m) \, \Psi
   \quad \Rightarrow \quad \alpha^k = \gamma^0 \gamma^k, \;\; \beta = \gamma^0$$

   En Pauli-Dirac, $\alpha^k = \begin{pmatrix} 0 & \sigma_k \\ \sigma_k & 0 \end{pmatrix}$ e
   $\beta = \begin{pmatrix} I & 0 \\ 0 & -I \end{pmatrix}$. Que $\hat{H}$ sexa hermítico
   —autovalores reais— esixe $\alpha^k$ e $\beta$ hermíticas, e de aí
   $(\gamma^0)^\dagger = \gamma^0$ e $(\gamma^k)^\dagger = -\gamma^k$.

3. Os termos cruzados en $m$ cancélanse:

   $$(\gamma^\nu \hat{p}_\nu + m)(\gamma^\mu \hat{p}_\mu - m) \, \Psi = (\gamma^\nu \gamma^\mu \, \hat{p}_\nu \hat{p}_\mu - m^2) \, \Psi = 0$$

   Como $\hat{p}_\nu \hat{p}_\mu$ é simétrico en $\mu\nu$, só conta a parte simétrica do
   produto das $\gamma$: $\gamma^\nu \gamma^\mu \, \hat{p}_\nu \hat{p}_\mu =
   \frac{1}{2} \{\gamma^\nu, \gamma^\mu\} \, \hat{p}_\nu \hat{p}_\mu = \hat{p}^\mu \hat{p}_\mu$.
   Con $\hat{p}^\mu \hat{p}_\mu = -\partial^\mu \partial_\mu$:

   $$(\partial_t^2 - \nabla^2 + m^2) \, \Psi = 0$$

   Cada compoñente cumpre a ecuación de **Klein-Gordon**, e sobre unha onda plana,
   $E^2 - {\bf p}^2 = m^2$. A ecuación **lineal** leva dentro a de Einstein: para iso
   pediuse a álxebra das $\gamma$.

4. Sobre $\Psi_s$, $\hat{p}_\mu \to +p_\mu$; sobre $\Phi_s$, $\hat{p}_\mu \to -p_\mu$. As fases
   simplifícanse e quedan

   $$(\gamma^\mu p_\mu - m) \, u_s = 0, \qquad (-\gamma^\mu p_\mu - m) \, v_s = 0
   \;\; \Leftrightarrow \;\; (\gamma^\mu p_\mu + m) \, v_s = 0$$

   O $+m$ é o signo menos da enerxía e o momento da antipartícula. Son as dúas
   ecuacións que se comproban numericamente na tarefa da álxebra de Dirac.

</details>
