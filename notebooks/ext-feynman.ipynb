## Introdución

Esta extensión é o **desenvolvemento matemático** da sección de diagramas de Feynman de
[perspectivas](./perspectivas.ipynb), nos puntos que alí se dan sen demostrar. Responde a catro
preguntas:

1. **De onde sae o propagador $1/(q^2 - m^2)$?** Da ecuación do campo do mediador.
2. **Por que a masa do mediador fixa o alcance?** O potencial de Yukawa, e Coulomb e
   Rutherford como caso sen masa.
3. **Que pasa cando o mediador se pon *on-shell*?** O polo, a anchura e a Breit-Wigner
   relativista: por que un $e^+ + e$ a $\sqrt{s} = m_Z$ produce $Z$ a esgalla.
4. **Como sae a teoría de Fermi do $W$?** O límite de baixa enerxía e os factores de
   $G_F/\sqrt{2} = g_W^2/8m_W^2$.

### Notación

* Unidades naturais, $\hbar = c = 1$; métrica $g_{\mu\nu} = \mathrm{diag}(1, -1, -1, -1)$.
* $q^\mu = (q^0, {\bf q})$ é o cuadrimomento transferido polo mediador; $q^2 = (q^0)^2 - {\bf q}^2$.
* $\Box \equiv \partial_\mu\partial^\mu = \partial_t^2 - \nabla^2$.
* Os factores $i$ e os signos globais das regras de Feynman dependen do convenio; aquí
  seguimos a estrutura, non os factores de fase. [MT5]

## 1. O propagador

### 1.1. Un campo con fonte

Tomamos o mediador máis sinxelo, un campo escalar $\phi$ de masa $m$. Libre, cumpre a ecuación de
Klein-Gordon (a táboa de ecuacións de campo de perspectivas). Se hai algo que o emite —unha
**fonte** $j(x)$, que nun diagrama é o vértice—, a ecuación gaña un termo:

$$
(\Box + m^2)\,\phi(x) = j(x) .
$$

É a mesma estrutura que as ecuacións de Maxwell con cargas: a fonte produce o campo.

**No espazo de momentos.** Con $\phi(x) = \int \frac{\mathrm{d}^4q}{(2\pi)^4}\, e^{-iq\cdot x}\,\tilde\phi(q)$, cada derivada
convértese nun factor, $\partial_\mu \to -iq_\mu$, e $\Box \to -q^2$. A ecuación diferencial pasa a ser
**alxébrica**:

$$
(-q^2 + m^2)\,\tilde\phi(q) = \tilde{j}(q)
\qquad \Rightarrow \qquad
\boxed{
\tilde\phi(q) = -\,\frac{1}{q^2 - m^2}\;\tilde{j}(q)
}
$$

O campo que produce a fonte é a fonte multiplicada por $1/(q^2 - m^2)$: o **propagador** é a
**inversa do operador da ecuación do campo**, escrita en momentos.

### 1.2. Do campo ao diagrama

Nun diagrama de dous vértices, a fonte 1 (un fermión que emite) produce o campo do mediador, e
ese campo actúa sobre a fonte 2 (o fermión que absorbe). A amplitude é

$$
M \;\propto\; \underbrace{g\,\tilde{j}_2}_{\text{vértice 2}}\;\;
\underbrace{\frac{1}{q^2 - m^2}}_{\text{propagador}}\;\;
\underbrace{g\,\tilde{j}_1}_{\text{vértice 1}}
$$

que é exactamente a regra de perspectivas: **dous vértices e un propagador**. O propagador depende
só de $q^2$ e da masa do mediador, porque é unha propiedade do campo, non de quen o emite.
Por conservación en cada vértice, $q$ é o cuadrimomento que perde a fonte 1 e gaña a 2.

**Fóra da capa de masas.** Para unha partícula real, $q^2 = m^2$ e o propagador diverxe. Nun
intercambio, $q^2$ fíxano os fermións externos e en xeral $q^2 \neq m^2$: o mediador é
**virtual** (*off-shell*). En dispersión, $q^2 < 0$, e o denominador nunca se anula.

### 1.3. Mediadores vectoriais

O fotón e os $W^\pm, Z$ teñen espín 1. A súa ecuación con fonte é a de Maxwell (sen masa) ou a de
Proca (con masa):

$$
\partial_\mu F^{\mu\nu} + m^2 A^\nu = j^\nu .
$$

Cunha corrente conservada ($\partial_\nu j^\nu = 0$, sección 4.5 de [ext-dirac](./ext-dirac.ipynb)) pódese
tomar $\partial_\mu A^\mu = 0$, e a ecuación queda $(\Box + m^2)\,A^\nu = j^\nu$: a de antes, compoñente a
compoñente. En momentos,

$$
\tilde{A}^\nu(q) = -\,\frac{1}{q^2 - m^2}\;\tilde{j}^\nu(q) ,
$$

e o campo producido pola corrente 1 acóplase á corrente 2 como $j_2^\mu\, g_{\mu\nu}\, \tilde{A}^\nu$. De aí a
forma de perspectivas:

$$
\frac{g_{\mu\nu}}{q^2 - m^2}, \qquad \text{e para o fotón,} \qquad \frac{g_{\mu\nu}}{q^2} .
$$

O $g_{\mu\nu}$ contrae os índices das dúas correntes. (O propagador completo dun vector con masa
leva ademais un termo $q_\mu q_\nu/m^2$, que se anula ao contraelo con correntes conservadas.)
[MT5.3]

## 2. O alcance: o potencial de Yukawa

### 2.1. Unha fonte en repouso

Unha fonte puntual e **estática**, $j = g\,\delta^3({\bf x})$, non depende do tempo: só intercambia
momento, non enerxía, e $q^0 = 0$, $q^2 = -{\bf q}^2$. A ecuación do campo queda

$$
(-\nabla^2 + m^2)\,\phi({\bf x}) = g\,\delta^3({\bf x})
\qquad \Rightarrow \qquad
\tilde\phi({\bf q}) = \frac{g}{{\bf q}^2 + m^2} ,
$$

o propagador con $q^2 = -{\bf q}^2$. Volvendo ao espazo de posicións, con $\int \mathrm{d}\Omega\,e^{i{\bf q}\cdot{\bf r}} = 4\pi \sin(qr)/qr$:

$$
\phi(r) = g\int \frac{\mathrm{d}^3{\bf q}}{(2\pi)^3}\,\frac{e^{i{\bf q}\cdot{\bf r}}}{{\bf q}^2 + m^2}
= \frac{g}{2\pi^2 r}\int_0^\infty \frac{q\,\sin(qr)}{q^2 + m^2}\,\mathrm{d}q
= \boxed{\frac{g}{4\pi}\,\frac{e^{-mr}}{r}}
$$

(a última integral faise por residuos, co polo en $q = im$).

### 2.2. Que se le

* **O alcance é $1/m$.** O campo cae exponencialmente a distancias maiores que $1/m$ ($\hbar/mc$ no
  SI). É a relación de Yukawa da [introdución](./introduccion.ipynb): o pión, $m_\pi = 140$ MeV, dá
  $1.4$ fm, o tamaño do núcleo; o $W$, $2.5 \cdot 10^{-3}$ fm, unha interacción practicamente puntual.
* **Sen masa, Coulomb.** Con $m = 0$ e a carga como fonte, $\phi = e/4\pi r$, e a enerxía entre dúas
  cargas é $e^2/4\pi r = \alpha/r$: o alcance infinito do electromagnetismo.
* **O propagador é o potencial no espazo de momentos.** O que en mecánica cuántica é a
  aproximación de Born —a amplitude de dispersión é a transformada de Fourier do potencial,
  $\tilde{V}({\bf q})$— aquí é o propagador: $\tilde{V}({\bf q}) \propto g^2/({\bf q}^2 + m^2)$.

### 2.3. Rutherford, outra vez

Para o fotón, $\tilde{V}({\bf q}) \propto e^2/{\bf q}^2$. Nunha dispersión elástica con momento $p$ e ángulo $\theta$,
o momento transferido é

$$
|{\bf q}| = 2p\,\sin\frac{\theta}{2}
$$

(o triángulo isósceles da tarefa de Rutherford), e a sección eficaz vai como o cadrado da
amplitude:

$$
\frac{\mathrm{d}\sigma}{\mathrm{d}\Omega} \;\propto\; \left|\frac{e^2}{{\bf q}^2}\right|^2 = \frac{e^4}{16\,p^4 \sin^4(\theta/2)} .
$$

O $1/\sin^4(\theta/2)$ de Rutherford **é o propagador do fotón ao cadrado**. Cos factores completos
(Born, $m$ reducida), recupérase a fórmula da [introdución](./introduccion.ipynb). Un mediador con
masa cambiaría $1/{\bf q}^4$ por $1/({\bf q}^2 + m^2)^2$, que non diverxe en $\theta \to 0$.

masa na sección eficaz?

## 3. O polo e a anchura

### 3.1. Un mediador real

Nunha aniquilación, o mediador lévase toda a enerxía: $q^2 = s > 0$. Cando $\sqrt{s}$ se achega á
súa masa, o propagador $1/(s - m^2)$ **diverxe**: o mediador deixa de ser virtual e prodúcese de
verdade. Pero unha partícula real e **inestable** non ten unha masa exacta.

**De onde sae a anchura.** Un estado de masa $m$ en repouso evoluciona como $e^{-imt}$; se se
desintegra con anchura $\Gamma$, a súa amplitude decae como $e^{-imt}\,e^{-\Gamma t/2}$ (a probabilidade, como
$e^{-\Gamma t}$). É como se a masa tivese unha parte imaxinaria:

$$
m \;\to\; m - \frac{i\,\Gamma}{2}
\qquad \Rightarrow \qquad
m^2 \;\to\; m^2 - i\,m\,\Gamma \quad (\Gamma \ll m) .
$$

O propagador dun mediador inestable é entón

$$
\boxed{
\frac{1}{s - m^2 + i\,m\,\Gamma}
}
$$

que xa non diverxe: en $s = m^2$ vale $1/(i\,m\Gamma)$, grande pero finito.

### 3.2. A Breit-Wigner relativista

A sección eficaz vai como o propagador ao cadrado:

$$
\sigma(s) \;\propto\; \frac{1}{(s - m^2)^2 + m^2\,\Gamma^2} ,
$$

a **Breit-Wigner relativista**. Preto do pico, $s - m^2 = (\sqrt{s} - m)(\sqrt{s} + m) \simeq 2m\,(\sqrt{s} - m)$, e

$$
\frac{1}{(s - m^2)^2 + m^2\Gamma^2} \simeq \frac{1}{4m^2}\,\frac{1}{(\sqrt{s} - m)^2 + \Gamma^2/4} ,
$$

a Breit-Wigner da [introdución](./introduccion.ipynb), de anchura total a media altura $\Gamma$. A
anchura do pico de OPAL **é** a anchura do propagador do $Z$.

**O pico do $Z$.** O cálculo completo de $e^+ + e \to Z \to f\bar{f}$ no pico dá

$$
\sigma_{pico} = \frac{12\pi}{m_Z^2}\,\frac{\Gamma_{ee}\,\Gamma_{f\bar f}}{\Gamma_Z^2} .
$$

Para hadróns, con $\Gamma_{ee} = 83.9$ MeV, $\Gamma_{had} = 1744$ MeV e $\Gamma_Z = 2495$ MeV:
$12\pi/m_Z^2 = 4.53 \cdot 10^{-3}$ $\mathrm{GeV}^{-2} = 1765$ nb, por $0.0235$, **$41.5$ nb**. OPAL mide uns $30$ nb no
máximo: a diferenza é a radiación dos electróns antes de chocar, que ensancha e rebaixa o
pico.

### 3.3. «$Z$ a esgalla»

A $\sqrt{s} = m_Z$ compiten dous diagramas: o intercambio dun fotón, con $1/s$, e o dun $Z$, con
$1/(s - m_Z^2 + i\,m_Z\Gamma_Z)$. No pico, o cociente de propagadores é

$$
\frac{|1/(i\,m_Z\Gamma_Z)|}{|1/m_Z^2|} = \frac{m_Z}{\Gamma_Z} \simeq 37 ,
$$

e na sección eficaz, que vai co cadrado, máis de mil. Cos acoplamentos incluídos, en
$e^+ + e \to \mu^+ + \mu$:

* só o fotón: $\sigma = 4\pi\alpha^2/3s = 0.0104$ nb;
* no pico do $Z$: $\sigma = 41.5\ \mathrm{nb} \times \Gamma_{\mu\mu}/\Gamma_{had} \simeq 2.0$ nb, **douscentas veces máis**.

É o que resolve a tarefa «Da simetría a un número medible» de perspectivas: a estimación
electromagnética falla por un factor 200 porque o propagador do $Z$ ten o seu polo xusto aí. E
como a anchura total suma todas as canles, incluídas as que non se ven, a forma do pico conta
os neutrinos: $N_\nu = 3$ (tarefa da introdución).

*Cuestións:* canto vale $m_Z/\Gamma_Z$? Que pasaría coa altura do pico se houbese unha cuarta familia
con neutrino lixeiro?

## 4. A teoría de Fermi como límite

### 4.1. O vértice do $W$

A corrente cargada da interacción feble ten estrutura $V - A$ (tema de leptóns), e o vértice
do $W$ cun par de leptóns ou de quarks achega

$$
\frac{g_W}{2\sqrt{2}}\;\gamma^\mu\,(1 - \gamma^5) .
$$

O $\frac{1}{2}(1 - \gamma^5)$ é o proxector de quiralidade $P_L$ (sección 1.7 de [ext-dirac](./ext-dirac.ipynb)), e
o $1/\sqrt{2}$ vén de como se agrupan os campos $W^\pm$ na teoría electrofeble: o acoplamento a cada
corrente cargada é $g_W/\sqrt{2}$, e co proxector, $\frac{g_W}{\sqrt{2}}\cdot\frac{1}{2}(1-\gamma^5) = \frac{g_W}{2\sqrt{2}}(1-\gamma^5)$. [MT11.5]

### 4.2. O límite de baixa enerxía

Na desintegración $\beta$, $q^2 \sim \mathcal{O}(\mathrm{MeV}^2) \ll m_W^2$, e o propagador vólvese unha constante:

$$
\frac{g_{\mu\nu}}{q^2 - m_W^2} \;\longrightarrow\; -\,\frac{g_{\mu\nu}}{m_W^2} .
$$

A amplitude, con dous vértices e o propagador, queda (salvo o signo global)

$$
M = \frac{g_W^2}{8\,m_W^2}\;\left[\bar{u}\,\gamma^\mu(1 - \gamma^5)\,u\right]\left[\bar{u}\,\gamma_\mu(1 - \gamma^5)\,v\right] .
$$

É un acoplamento **puntual** de dúas correntes $V - A$: a teoría de Fermi, que na súa forma moderna se escribe

$$
M = \frac{G_F}{\sqrt{2}}\;\left[\bar{u}\,\gamma^\mu(1 - \gamma^5)\,u\right]\left[\bar{u}\,\gamma_\mu(1 - \gamma^5)\,v\right] .
$$

Comparando:

$$
\boxed{
\frac{G_F}{\sqrt{2}} = \frac{g_W^2}{8\,m_W^2}
}
$$

O $8 = (2\sqrt{2})^2$ son os dous vértices; o $\sqrt{2}$ de $G_F/\sqrt{2}$ é un convenio, que se mantén
por razóns históricas.

### 4.3. Números

Con $G_F = 1.166 \cdot 10^{-5}$ $\mathrm{GeV}^{-2}$ e $m_W = 80.4$ GeV:

$$
g_W = m_W\sqrt{\frac{8\,G_F}{\sqrt{2}}} = 0.653, \qquad \alpha_W = \frac{g_W^2}{4\pi} = 0.034 \simeq \frac{1}{30} .
$$

**O acoplamento feble é maior que o electromagnético** ($\alpha \simeq 1/137$). A febleza a baixa enerxía
pona o propagador: $G_F \propto g_W^2/m_W^2$ é pequena porque $m_W$ é grande. Coa regra de Sargent
(sección 7 de [ext-cinematica](./ext-cinematica.ipynb)), $\Gamma \propto G_F^2 E_0^5$: o muón dá
$\Gamma_\mu = G_F^2\,m_\mu^5/192\pi^3$ e $\tau_\mu = 2.2\ \mu$s.

**Cando deixa de valer.** Cando $q^2$ se achega a $m_W^2$ o propagador xa non é constante, e a teoría de
Fermi —con $G_F$, de dimensións $E^{-2}$— dá seccións eficaces que crecen como $G_F^2\,s$ sen límite. O $W$
**resolve** ese crecemento: por riba de $m_W$ o propagador cae como $1/s$. É o exemplo clásico dunha
**teoría efectiva** que deixa de valer á escala da partícula que «integrou».

*Cuestións:* comproba $g_W = 0.653$. A que $\sqrt{s}$ o termo $q^2/m_W^2$ corrixe nun $10\,\%$ o propagador
constante?
