## Aceleradores.

A historia da Física de Partículas está ligada á dos aceleradores.

Para producir unha partícula necesitamos polo menos unha enerxía igual á súa masa, $E = m c^2$, e para explorala unha lonxitude de onda menor que o seu tamaño, pola relación  $p = h/\lambda$.

Pódense acelerar as partículas estables cargadas: $e, e^+, p, \bar{p}$. 

Podemos crear feixes secundarios ao golpear un feixe de $p$ contra un branco, o que xera partículas secundarias, mesóns ($\pi, K$), que se poden seleccionar nun rango de enerxía e momento. 

Se ademais se deixan desintegrar e se filtran, obtense un feixe de neutrinos $\nu_e, \nu_\mu$.

### Experimentos de branco fixo e de colisión

Existen dous tipos de experimentos principais:

  * De **branco fixo**, *fixed-target*, onde un feixe se fai incidir sobre un branco.
  
  * De **colisión**, *collider*, onde dous feixes se cruzan en determinados puntos.
  
A enerxía dispoñible no CM en ambos os experimentos é moi diferente. 

Para un experimento de branco fixo, $ s= (E, {\bf p}) + (m_t, 0)$, onde $(E, {\bf p})$ é o cuadrimomento da partícula do feixe con masa $m_b$, e $(m_a, 0)$ a masa da partícula do branco (usualmente nucleóns). 

Polo tanto, a enerxía no CM, $\sqrt{s} = \sqrt{(E+m_b)^2 - {\bf p}^2} = \sqrt{m_b^2 + 2 m_b m_a E + m^2_a}$, depende de $\sqrt{E}$ se se poden desprezar as masas. 

Se aumentamos nun factor 100 a enerxía do feixe, a do CM aumenta aproximadamente nun factor 10.

Para un experimento de colisión, $s = (E_a, {\bf p}_a) + (E_b, {\bf p}_b)$, onde $(E_\alpha, p_\alpha)$ son os cuadrimomentos das partículas dos dous feixes. 

No caso de partículas relativistas de igual masa que colisionan de fronte, ${\bf p}_b = - {\bf p}_a$ e $E_a = E_b  = E$, a enerxía no CM é $\sqrt{s} = 2E$. Increméntase linealmente coa enerxía do feixe. 

### Introdución aos aceleradores

Os feixes están compostos de **paquetes** (*bunches*), cunha densidade elevada de partículas; por exemplo, no LHC cada paquete ten da orde de $10^{11}$ protóns.

Existen dous tipos de aceleradores principais:

* Lineais, onde o acelerador é un dispositivo recto, un tubo de feixe en baleiro, que os paquetes percorren unha vez.

* Circulares, onde os paquetes xiran no tubo de baleiro do feixe numerosas voltas, $10^5$.

#### Aceleradores circulares

Os aceleradores cíclicos funcionan de maneira similar, pero o feixe circula por un anel.

As partículas relativistas, ao xirar, perden enerxía por radiación (**radiación sincrotrón**, *synchrotron radiation*) proporcional a $1/m^4$. 

Este efecto é máis dramático para $e$ ca para $p$.

Para 'xirar' as partículas créase un campo magnético dipolar $B$ perpendicular á súa dirección. 

Polo electromagnetismo sabemos que $p = 0.3 B/ \rho$, onde $p$ é o momento, en TeV, $B$ o campo magnético, en T, e $\rho$ o raio, en km.

Para o Tevatron, en Fermilab, Chicago, $\rho = 1$ km, $B = 1.5$ T, o que permite acelerar protóns ata 1 TeV.

Para obter campos máis intensos hai que recorrer a imáns supercondutores.

Por outra banda, as partículas oscilan dentro dos paquetes e, para mantelas xuntas, utilízanse campos magnéticos de cuadrupolos.

Nos **aneis de almacenamento** fanse xirar dous feixes de partículas en sentidos opostos no mesmo anel.

En determinados **puntos de colisión**, mediante un conxunto de imáns, comprímense os paquetes e crúzanse para que teñan lugar as colisións.

Este é o caso do LEP, o colisionador $e^+e^-$ do CERN, de 27 km de circunferencia, que funcionou a finais do século XX.

Tamén é o caso do LHC, que ocupa o túnel onde antes estivo o LEP. O LHC é un colisionador $pp$ cunha enerxía $\sqrt{s} = 7-13$ TeV. 

Os dipolos supercondutores do LHC operan a 8 T. Os paquetes conteñen da orde de $10^{11}$ $p$, que percorren o anel a unha frecuencia de 40 MHz. Os paquetes teñen 40 cm de longo e 1 mm de sección, que se reduce a 10 $\mu$m na zona de colisión.

### Parámetros dun colisionador

Os dous parámetros fundamentais dun colisionador son:
    
 * **Enerxía** no centro de masas, $\sqrt{s}$, que determina a masa das partículas que podemos crear.

 * **Luminosidade**, ou o número de partículas que se cruzan entre elas por unidade de área e tempo. Determina o número de interaccións en cada cruzamento. 

#### Luminosidade

Para dous paquetes con $n_1, n_2$ partículas que se cruzan entre si na dirección $z$ cunha frecuencia $\nu$ e que teñen unha sección transversal $\sigma_T$, a luminosidade é:
$$
\mathcal{L} = \nu \frac{n_1 n_2}{ \sigma_T}
$$
 
Se un tipo de interacción ten unha sección eficaz $\sigma$, esperamos unha frecuencia de interaccións $n = \sigma \mathcal{L}$.

Lembremos que a sección eficaz $\sigma$ se mide en m$^2$ ou en barns (1 barn = $10^{-24}$ cm$^2$).


En xeral $\mathcal{L}(t)$ depende do tempo, da operación do acelerador. A $\mathcal{L}(t)$ chamámoslle **luminosidade instantánea**.

E ao longo dun tempo (tipicamente anos), $\int_t  \mathcal{L}(t) \, \mathrm{d}t$, **luminosidade integrada**, que se mide en (m$^{-2}$ ou b$^{-1}$ -barns-). 
 
As interaccións que se producen no cruzamento dos paquetes coñécense como **suceso**. 

O número de interaccións por suceso é unha variable aleatoria con distribución de Poisson de media $\sigma \mathcal{L}$.

## Interacción das partículas coa materia

As interaccións das partículas poden dividirse en:

   * interaccións das partículas cargadas
   
   * interaccións electromagnéticas de electróns e fotóns
   
   * interaccións fortes de hadróns.


### Interaccións das partículas cargadas

As partículas cargadas relativistas interaccionan electromagneticamente cos electróns dos átomos da materia e perden enerxía por ionización.

Esta perda por distancia percorrida vén dada pola fórmula de Bethe-Bloch (anos 30):

$$
\frac{\mathrm{d}E}{\mathrm{d}x} \simeq - \frac{4\pi \hbar^2 \alpha^2}{m_e} \frac{n_e}{\beta^2} \left[ \ln \left( \frac{2 m_e c^2 \beta^2 \gamma^2}{I_e}\right) - \beta^2\right]
$$

onde $v = \beta c$ é a velocidade, $\gamma$ o factor de Lorentz, $n_e$ a densidade de electróns e $I_e$ o potencial de ionización ($I_e \simeq 10 Z$ eV).

Trátase dun valor medio. A distribución $\mathrm{d}E/\mathrm{d}x$ está relacionada coa flutuación do número de colisións da partícula cos electróns dos átomos.

A densidade de electróns é $n_e = \rho N_A Z/A$, onde $\rho$ é a densidade, $A$ a masa atómica (g/mol), $Z$ o número atómico e $N_A$ o número de Avogadro. 

Se a reescribimos:
$$
\frac{1}{\rho}\frac{\mathrm{d}E}{\mathrm{d}x} \simeq - K \frac{Z}{A \beta^2}\left[ \ln \left( \frac{2  m_e c^2 \beta^2 \gamma^2}{I_e}\right) - \beta^2\right]
$$
Onde $K = \frac{4\pi \hbar^2 \alpha^2 N_A}{m_e} = 0.307$ MeV cm$^2$/mol

Como $Z/A$ é practicamente constante para a maioría dos átomos, a dependencia co material é principalmente proporcional á súa densidade.

O percorrido medio dunha partícula con $\beta$ ata que se detén nun medio denomínase **alcance** de penetración.

A perda de enerxía depende da velocidade da partícula, $\beta$. Distinguimos tres rexións:
 * A perda é máis intensa para $\beta$ baixa (pola dependencia $1/\beta^2$). A ionización é maior ao final da traxectoria, cando a velocidade é moi pequena. Esta rexión coñécese como **pico de Bragg**.
 * No rango de $\beta\gamma$ de $1-10$, a enerxía perdida é mínima; esa rexión denomínase **mip** (*minimum ionizing particle*, partícula de ionización mínima).
     Un muón de 10 GeV en ferro perde en media 13 MeV/cm e o seu alcance é de varios metros.
 * Para $\beta \gamma >100$, a perda aumenta de forma logarítmica. A partir de aquí os efectos de radiación son relevantes.

A fórmula calcula o valor medio. A distribución $\mathrm{d}E/\mathrm{d}x$ está relacionada coa flutuación do número de colisións da partícula cos electróns dos átomos.

### Interaccións electromagnéticas de fotóns e electróns

#### Interaccións dos electróns

As partículas cargadas poden radiar fotóns pola interacción electromagnética cos protóns dos núcleos. Esta radiación chámase **bremsstrahlung**.

$$e^- + (A, Z) \to e^- + \gamma + (A, Z)$$ 

Esta radiación comeza a ser dominante a partir dunha **enerxía crítica** $E_c \sim 800/Z$ MeV; por debaixo domina a ionización.

Os diagramas de Feynman son:

| |
| :-- |
|<img src="./imgs/feynman_bremsstrahlung.png" width = 400 align = "center">|
| Diagramas de Feynman de bremsstrahlung |

Onde o disco representa a corrente do núcleo con $Z$ protóns.

Nótese que a interacción é da orde $Z^2 \alpha^3$


Este proceso pódese calcular en QED (*Quantum Electro-Dynamics*, electrodinámica cuántica) e a súa sección eficaz é: 

$$\sigma_{b} \propto E/m^2.$$

Afecta máis aos electróns ca aos muóns nun factor $(m_e/m_\mu)^2$. 

Os muóns por debaixo de $\mathcal{O}(100)$ GeV perden enerxía principalmente por ionización. 

A perda de enerxía por bremsstrahlung por riba de $E_c$ pode expresarse:

$$
\frac{\mathrm{d}E}{\mathrm{d}x} = - \frac{E}{X_0}, \;\; E(x) = E_0 \, e^{-x/X_0},
$$
onde $X_0$ se denomina **lonxitude de radiación** e $E_0$ é a enerxía inicial do electrón.

A sección eficaz de produción de pares crece rapidamente desde o limiar de produción e pode aproximarse por:

$$
\sigma_\gamma \simeq \frac{7}{9} \frac{1}{n X_0},
$$
onde $n$ é a densidade de núcleos.


A cantidade $\lambda = 1/(n \sigma_\gamma)$ é o **camiño libre medio**, que vale $\lambda \simeq 7/9 X_0$ e que nos indica a cantidade de fotóns que se perden nun feixe monoenerxético de intensidade $I$:

$$
\frac{\mathrm{d}I}{\mathrm{dx}} = - \frac{I}{\lambda}, \;\; I(x) = I_0 e^{-x/\lambda}
$$

Polo tanto, a lonxitude de radiación caracteriza a perda de enerxía dos electróns e a conversión de fotóns en pares para partículas por riba de $\sim 10$ MeV.


Os electróns ou fotóns de alta enerxía, ao atravesaren un medio de $X_0$, producen unha fervenza electromagnética.

### Interaccións fortes dos hadróns

Os hadróns cargados (protóns, pións, kaóns) perden enerxía por ionización. 

Tamén por interaccións fortes cos núcleos da materia.

As interaccións fortes caracterízanse pola **lonxitude de interacción**, $\lambda_I$, que é a distancia media entre interaccións fortes. $\lambda \gt X_0$

|           | Fe    |
|:--        | :--   |
$\lambda_I$ | 17 cm |
$X_0$       | 1.76 cm |

As interaccións de hadróns producen unha fervenza hadrónica.

As fervenzas son máis variables cás electromagnéticas, dado que nelas se poden producir máis tipos de partículas, e tamén $\pi^0$ que se desintegran electromagneticamente, $\pi^0 \to \gamma \gamma$, 

o que dá lugar á súa vez a unha fervenza electromagnética dentro da hadrónica. Isto orixina unha dispersión na enerxía repartida entre os dous tipos de fervenzas.

Unha parte da enerxía tamén se perde en forma de excitacións e roturas nucleares.

## Detectores de partículas

Os detectores de partículas usan como base a interacción das partículas coa materia, principalmente a ionización e a radiación.

En Física de Partículas tratamos de determinar o momento, $p$, a enerxía, $E$, e a identidade das partículas (PID), isto é, se a partícula é un $\mu$, un $e$, un $K$, $\pi$, ...

Os detectores divídense en:
    
   * **Detectores de trazas**: determinan as traxectorias das partículas cargadas.
       Serven xeralmente para medir o momento e determinar os vértices de desintegración.
       Os detectores de trazas están habitualmente inmersos en campos magnéticos, co que o momento se determina a partir da súa curvatura.
    
   * **Calorímetros**: serven para medir a enerxía das partículas, principalmente electróns/fotóns e hadróns.
 
    

### Detectores tipo


Os detectores tipo en colisionadores teñen unha estrutura en capas cilíndricas.

No interior están os detectores de trazas máis precisos (detectores de silicio), mergullados nun campo magnético (solenoidal).

Seguen o calorímetro electromagnético e o hadrónico.

E finalmente os detectores de muóns (que son as partículas máis penetrantes)

### Detectores de trazas

Nos detectores de trazas detéctase a ionización (electróns liberados) producida polo paso da partícula cargada a través do medio para determinar puntos de paso ou *hits*. 

Os detectores están mergullados nun campo magnético, ${\bf B}$ (teslas), que produce unha curvatura da partícula proporcional ao seu momento, ${\bf p}$(GeV), na dirección perpendicular a ${\bf B}$

Se entre ambos hai un ángulo $\theta$:
$$
p \cos \theta = 0.3 B \rho,
$$
onde $\rho$ (m) é o raio de curvatura no plano perpendicular.

Chamamos ao momento no plano perpendicular **momento transverso**, $p_T = p \cos \theta$.

Para CMS, con $B = 4$ T, e un $\pi^\pm$ de $p$ 100 GeV, $\rho \sim 100$ m.  

##### Detectores de silicio

Baséanse en baleirar de portadores libres de carga unha oblea de silicio (aprox. $~300$ $\mu$m de grosor) na que se doparon tiras (*strips*) de tipo $p$, separadas aprox. $~50$ $\mu$m, para crear unións $pn$.

#### Detectores gasosos

Nos detectores gasosos utilízase a ionización ($\sim 30$ eV por ionización) producida polo paso das partículas cargadas en gases nobres, Ar, Xe. 

Os detectores traballan nun rango de voltaxe proporcional (non hai efecto avalancha). 

Os electróns derivan baixo a presenza dun campo eléctrico ${\bf E}$ ata o ánodo, que pode estar formado por fíos (*wires*).

Existen varios tipos de detectores: *wire chambers*, *multiproportial wire chambers (MPWC)*, *time proyection chambers (TPC)*.

##### Cámaras de proxección temporal (TPC)

As TPC adoitan ter forma de barril; poden ser simétricas, con dúas tapas como ánodos e un cátodo central de alta tensión (HV).

Baixo a presenza do campo eléctrico, os $e$ de ionización derivan ata o ánodo.

Son recollidos por un detector segmentado (fíos) en dúas direccións ($R, \phi$) ou ($x, y$), onde se amplifica a súa carga (electrónica frontal).

O tempo de chegada dos electróns, $\Delta t$, serve de estimación da posición en $z$ (o eixe da cámara).  

As TPC son excelentes detectores de trazas

### Calorímetros

#### Calorímetros electromagnéticos

Os calorímetros electromagnéticos están construídos con materiais de alto $Z$, alto $X_0$

Adoitan ter unha estrutura alternada de material pasivo (por exemplo, Pb), onde se desenvolven as fervenzas, e material activo, onde se detecta a ionización.

A resolución en enerxía, $\sigma_E$, está limitada polas flutuacións na produción das partículas (proporcional a $\sqrt{E}$) na fervenza; en xeral:

$$
\frac{\sigma_E}{E} \sim \frac{3 - 10 \mathrm{\%}}{\sqrt{E \;\mathrm{(GeV)}}}
$$


O calorímetro de CMS está formado por cristais de volframato de chumbo, PbWO$_4$, con $X_0 = 0.83$ cm, que é un centelleador inorgánico.

Os **centelleadores** conteñen moléculas centelleadoras que, en vez de ionizarse, se excitan e, ao desexcitarse, emiten luz no visible que pode detectarse con detectores de fotóns (por exemplo, fotomultiplicadores). O número de fotóns é proporcional á enerxía absorbida, aproximadamente 100 eV por fotón de centelleo.

Os centelleadores deben ter unha $X_0$ alta para evitar a conversión dos fotóns.

#### Calorímetros hadrónicos

Os calorímetros hadrónicos están construídos cun material de alta densidade; aínda así, as fervenzas adoitan ocupar moito volume: por exemplo, un hadrón de 100 GeV xera unha fervenza de aproximadamente 2 m de profundidade e 50 cm de anchura.

Están construídos habitualmente en capas de material pasivo e activo. O calorímetro de ATLAS está formado por placas e plástico centelleador.

A resolución en enerxía, $\sigma_E$, está limitada pola produción das partículas, $\sqrt{E}$, pola fracción entre a fervenza electromagnética e a hadrónica, e pola enerxía perdida.

$$
\frac{\sigma_E}{E} \sim \frac{50 \mathrm{\%}}{\sqrt{E \mathrm{(GeV)}}}
$$

### Detectores de luz Cherenkov

O paso de partículas cargadas que se moven a unha velocidade, $v$, maior que a da luz no medio, $v \gt c/n$, onde $n$ é o índice de refracción, xera unha fronte de onda coherente cun determinado ángulo, $\theta$, respecto á dirección ${\bf v}$, no visible, coñecida como luz Cherenkov.

## Sistema de disparo, procesamento e análise de datos

### Sistema de disparo

Os detectores de partículas teñen un **sistema de disparo**, *trigger*, para seleccionar no momento os datos relevantes.

Por exemplo, o LHC ten unha frecuencia de colisións de 40 MHz, 25 ns. O tamaño da información dun suceso en ATLAS é de Mbytes. Actualmente é imposible almacenar todas as colisións que se producen no LHC.

O trigger é un sistema de diversos niveis onde se buscan trazas (principalmente muóns) de alto momento transverso e partículas desprazadas do vértice de interacción (que proveñen posiblemente de partículas de vida curta). Estas son as características principais dos sucesos de interese.

O sistema de disparo de alto nivel, *High Level Trigger*, HLT, selecciona sucesos preprocesados *on-line* nunha granxa de ordenadores in situ.


### Procesamento de datos

Os datos que pasan o trigger **almacénanse e procésanse** posteriormente en centros de computación interconectados, *GRID*. A partir dos sinais de cada sensor (contas de ADC, tempo, etc.) reconstrúense trazas e fervenzas e, de aí, partículas co seu cuadrimomento, posición e identificador. Este procesamento realízase en programas específicos en linguaxes como C++ ou Python.

O produto final son ficheiros de datos procesados (*data summary tape*, DST) que se analizan con **técnicas estatísticas de análise**, moitas veces multivariantes, como as redes neuronais.

O proceso complétase coa produción de **sucesos simulados** por técnicas de Monte Carlo, onde se simulan con gran precisión o detector e a física. A produción de datos simulados realízase tamén en centros de computación e no grid. Os datos simulados serven de referencia e permiten estimar e prever as capacidades dun detector.

### Erros, observación e evidencia

As medidas dos experimentos preséntanse como:

   * a **estimación dun observable**, por exemplo: unha sección eficaz, a vida media, a masa ou unha fracción de desintegración. 
   
   * **un límite no observable**, como a vida media, masa, etc., dunha **busca**.

Os primeiros están suxeitos a unha estimación de erros. Os erros son de dous tipos (e adoitan darse por separado):

  * **estatísticos**: dependen da cantidade de sucesos relevantes dispoñibles.

  * **sistemáticos**: reflicten as nosas incertezas en parámetros que afectan ao observable, que poden ser de diversa índole: de calibración, de estimación de eficiencias de selección, teóricos, etc.
  
A determinación dun límite preséntase cun nivel de confianza e os erros xa están incorporados no propio límite. 


A interpretación dos erros e do nivel de confianza depende de se a análise é **frecuentista** ou **bayesiana**.

Podemos dicir que os frecuentistas parten dunha teoría, $\mu$, e verifican se concorda cos datos, $x$, isto é, calculan a 'probabilidade' $p(x|\mu)$; mentres que os bayesianos tratan de inferir se a teoría é correcta a partir dos datos, $p(\mu | x)$.

Un bayesiano diría: "Atopamos o Higgs!", e un frecuentista: "Os datos casan coa existencia dun Higgs".
