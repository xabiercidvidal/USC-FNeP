### Introdución

O Modelo Estándar define o marco teórico que clasifica a materia e establece as interaccións fundamentais. A súa construción constitúe un dos grandes logros da Física Teórica e está ligada á evolución do noso coñecemento das interaccións fundamentais ao longo do século XX. 

Hai dous piares básicos que sustentan o modelo: que o lagranxiano sexa invariante Lorentz e invariante gauge local. Esta última condición é necesaria para que a teoría sexa *renormalizable* e se poidan calcular observables como a sección eficaz e as anchuras de desintegración sen que aparezan infinitos nos cálculos.

As interaccións electromagnéticas e fortes aparecen no modelo de maneira natural a partir da invariancia gauge local. Porén, a inclusión das interaccións febles require certo encaixe: primeiro, porque estas interaccións non preservan a paridade e separan os campos segundo a quiralidade á esquerda e á dereita; e, segundo, porque tras esta separación, para dotar de masa as partículas cómpre incorporar un mecanismo específico, o mecanismo de Higgs, sen o cal o modelo carece de sentido.

Estas notas presentan unha estrutura matemática do modelo sen pretender ser exhaustivas nin rigorosas. Revisaremos primeiro os lagranxianos dos campos escalar, espinorial e vectorial. Introduciremos despois a invariancia gauge e a derivada covariante nas interaccións electromagnéticas e fortes. Logo discutiremos a interacción electrofeble, veremos como aparecen as correntes cargadas e neutras, e remataremos presentando o mecanismo de Higgs co que dotar de masa os fermións e os bosóns vectoriais $W^\pm$ e $Z^0$. 

## Simetría gauge

Imos introducir o concepto de simetría gauge utilizando para iso o lagranxiano de Dirac e o do campo electromagnético. Veremos que podemos obter a interacción electromagnética como unha consecuencia da posibilidade de rotar o campo fermiónico por unha fase local, isto é, de que o lagranxiano sexa invariante gauge baixo o grupo $U(1)$.

Posteriormente estenderemos o lagranxiano de Dirac para incluír a *cor* dos quarks. Neste caso asignaremos as tres cores dun mesmo quark aos elementos do triplete do grupo $SU(3)_C$, onde $C$ indica cor, e veremos que a invariancia gauge deste lagranxiano baixo transformacións en $SU(3)_C$ é máis rica e complexa que con $U(1)$.

### Lembranza dos diversos lagranxianos 

#### Revisión do lagranxiano

Lembremos que en mecánica clásica definimos o lagranxiano como:

$$
L(q, \dot{q}) = T -V
$$

onde $T$ é a enerxía cinética, $V$ o potencial, $q_i$, con $i=1, \dots, n$, son as coordenadas xeneralizadas e $\dot{q}$ as súas derivadas temporais.

e as ecuacións do movemento veñen dadas polas ecuacións de Euler-Lagrange:

$$
\frac{\mathrm{d}}{\mathrm{dt}} \left( \frac{\partial L}{\partial \dot{q}_i}\right) = \frac{\partial L}{\partial q_i}
$$

que proveñen de esixir que a acción do sistema, $S$, sexa mínima, sendo:

$$
S = \int L(q_i, \dot{q}_i) \, \mathrm{d} t
$$

No caso dunha partícula nun potencial, o lagranxiano é a súa enerxía cinética menos o seu potencial:

$$
L = \frac{1}{2} m \dot{x}^2 - V(x) 
$$

E a súa ecuación do movemento:

$$
m \, \ddot{x} = -\frac{\partial V(x)}{\partial x}
$$

é a ecuación de Newton.

En *teoría clásica de campos* definimos un campo $\phi(x)$ nas coordenadas espazo-temporais $x$ e o lagranxiano do campo como:

$$
L = \int \mathcal{L}(\phi, \partial_\mu \phi) \, \mathrm{d}^3x
$$

onde $\mathcal{L}(\phi, \partial_\mu \phi)$ é a densidade lagranxiana ou, por abuso da linguaxe, o lagranxiano.

As ecuacións do movemento veñen dadas por:

$$
\partial_\mu \left( \frac{\partial \mathcal{L}}{\partial (\partial_\mu \phi)}\right) = \frac{\partial \mathcal{L}}{\partial \phi}
$$

que proveñen de esixir que a acción sexa mínima, sendo a acción:

$$
S = \int \mathcal{L}(\phi, \partial_\mu \phi) \, \mathrm{d}^4x
$$

Nótese que, como a acción non ten dimensións e $\mathrm{d}^4x$ ten dimensións de $[E]^{-4}$, o lagranxiano ten dimensión $[E]^4$. Isto significa que os termos do lagranxiano deben ter dimensión 4 en enerxía.

#### Lagranxiano do campo escalar

Sexa $\phi(x)$ un campo escalar *real* asociado a partículas de masa $m$; o seu lagranxiano vén dado por:

$$
\mathcal{L} = \frac{1}{2} \partial_\mu \phi(x) \, \partial^\mu \phi(x) - \frac{1}{2} m^2 \phi^2(x)
$$

O primeiro termo é a enerxía cinética e o segundo é o potencial asociado á masa da partícula. A masa é entón a enerxía mínima para crear a partícula asociada ao campo.

A ecuación de movemento:

$$
\partial_\mu \partial^\mu \phi(x) + m^2 \phi(x) = 0
$$

é a ecuación de Klein-Gordon

que tamén podemos escribir como:

$$
\square \, \phi(x) + m^2 \phi(x) = 0
$$

onde definimos:

$$
\square \equiv \partial_\mu \partial^\mu
$$

Consideremos agora un campo escalar *complexo*, $\phi$, que está composto por dous campos escalares reais e independentes, $\varphi_1, \varphi_2$:

$$
\phi(x) = \varphi_1(x) + i \varphi_2(x)
$$ 


O lagranxiano do campo complexo corresponde ao dos dous campos reais:

$$
\mathcal{L}  = \partial_\mu \phi^*(x) \, \partial^\mu \phi(x) - m^2 \phi^*(x) \phi(x) = \\
\sum_{i= 1, 2} \frac{1}{2} \partial_\mu \varphi_i(x) \, \partial^\mu \varphi_i(x) - \frac{1}{2} m^2 \varphi_i^2(x)
$$

É doado comprobar que o lagranxiano é invariante respecto a un cambio de fase global

$$
\phi(x) \to \phi'(x) = e^{i\theta} \phi(x)
$$

Por último, nótese tamén que as dimensións do campo, $\phi$, son de enerxía, $[E]$.

#### Lagranxiano de Dirac


O lagranxiano dun campo de Dirac é:

$$
\mathcal{L} = i \bar{\Psi}(x) \gamma^\mu \partial_\mu \Psi(x) - m \bar{\Psi}(x) \Psi(x)
$$

que corresponde a un termo cinético e ao do potencial asociado á súa masa.

Lembremos que $\bar{\Psi}(x)$ é o espinor adxunto definido por:

$$
\bar{\Psi}(x) = \Psi^\dagger(x) \, \gamma^0 
$$

En total hai 8 funcións reais e independentes en $\Psi$, que podemos dar con 4 funcións en $\Psi$ e $\bar{\Psi}$, de maneira similar a considerar como independentes $\phi^*, \phi$ no caso dun campo escalar complexo en vez dos dous campos escalares reais $\varphi_1, \varphi_2$.

Desde o punto de vista das dimensións, o espinor, $\Psi$, ten dimensión $[E]^{3/2}$, de tal forma que cada termo do lagranxiano teña dimensión $[E]^4$.

A ecuación de Dirac obtense a partir da ecuación do movemento derivando respecto ao espinor adxunto, $\bar{\Psi}(x)$:


$$
(i \gamma^\mu \partial_\mu - m)  \, \Psi(x) = 0 
$$


Derivando respecto a $\Psi(x)$, obtemos á súa vez a ecuación de Dirac do espinor adxunto:

$$
i \partial_\mu \bar{\Psi}(x) \, \gamma^\mu +  m \bar{\Psi}(x)  = 0 
$$

Vemos tamén que o lagranxiano de Dirac é invariante baixo un cambio global de fase:

$$
\Psi(x) \to \Psi'(x) = e^{iQ \theta} \Psi(x)
$$

Onde $Q$ é unha constante, que asociaremos á carga eléctrica do fermión.

Isto é, o campo $\Psi(x)$ é invariante baixo unha rotación $U(\theta)$ do grupo $U(1)$

$$
\Psi'(x) = U(\theta) \Psi(x), \;\; U(\theta) = e^{iQ \theta}
$$

Parece *natural* que, se podemos cambiar a fase do campo de Dirac globalmente, isto é, igual en todos os puntos ao mesmo tempo, poidamos cambiala de forma continua nunha rexión ou intervalo temporal, é dicir, cambiala *localmente*. 

Porén, se realizamos un cambio de fase *local*:

$$
\Psi(x) \to \Psi'(x) = e^{iQ \theta(x)} \Psi(x)
$$

onde agora $\theta(x)$ cambia continuamente con $x$.

Isto é, se esiximos que o lagranxiano sexa invariante gauge local respecto a unha rotación, $U(\theta(x))$ do grupo $U(1)$

$$
\Psi'(x) = U(\theta(x)) \Psi(x), \;\;\; U(\theta(x)) =  e^{i Q\theta(x)}
$$

obtemos que o lagranxiano do campo transformado é:

$$
i \bar{\Psi}' \gamma^\mu \partial_\mu \Psi' + m \bar{\Psi}' \Psi' = \\
i \bar{\Psi} \gamma^\mu \partial_\mu \Psi + m \bar{\Psi} \Psi
 - Q \, \bar{\Psi} \gamma^\mu \Psi \, \partial_\mu \theta(x)
$$

No lagranxiano aparece o termo:

$$
- Q \, \bar{\Psi} \gamma^\mu \Psi \, \partial_\mu \theta(x)
$$

que non preserva a invariancia gauge.

Podemos interpretar ese termo como o acoplamento da corrente:

$$
j^\mu(x) \equiv Q \, \bar{\Psi}(x) \gamma^\mu \Psi(x)
$$

cun campo vectorial, $-\partial_\mu \theta(x)$.

O campo é vectorial porque provén dunha derivada covariante e compórtase, polo tanto, como un vector.

#### Lagranxiano do campo electromagnético

Sexa agora $A^\mu(x)$ o campo vectorial do electromagnetismo:

$$
A^\mu(x) = \begin{pmatrix} \phi(x) \\ {\bf A}(x) \end{pmatrix},
$$

onde $\phi(x)$ é o potencial eléctrico e ${\bf A(x)}$ o potencial vector magnético.

Nótese que se trata de 4 funcións reais.

[TODO] Ecuacións de Maxwell

Definimos o tensor de campo de forza, $F^{\mu\nu}$, como:

$$
F^{\mu \nu} = \partial^\mu A^\nu - \partial^\nu A^\mu
$$


O que nos permite dar as ecuacións de Maxwell en ausencia de cargas e correntes como:

$$
\partial_\mu F^{\mu\nu} = 0
$$

Estas ecuacións proveñen do lagranxiano:

$$
\mathcal{L}  = - \frac{1}{4} F_{\mu\nu} F^{\mu\nu}
$$


Nótese que as dimensións do campo $A_\mu$ son de enerxía $[E]$.

Podemos explicitar as ecuacións de Maxwell en función de $A_\mu$:

$$
\partial_\mu \partial^\mu A^\nu - \partial^\nu (\partial_\mu A^\mu) = 0
$$

que podemos comprobar facilmente que é invariante baixo a transformación:

$$
A^\mu(x) \to A^\mu(x) - \partial^\mu \theta(x)
$$

sendo $\theta(x)$ unha transformación continua en $x$.

Isto é, hai un conxunto de campos que representan a mesma física. A diferenza entre eles é a derivada dunha función continua $\theta(x)$.

A esta equivalencia de campos chamámoslle **invariancia gauge**.

Podemos utilizar esta invariancia para escoller unha *gauge* determinada e impoñer, sen perda de xeneralidade, que: 

$$
\partial_\mu A^\mu = 0
$$

A esta gauge chamámoslle de Lorenz (non Lorentz!).

Co que as ecuacións do movemento quedan:

$$
\partial_\mu \partial^\mu A^\nu = 0, \;\;\, \square \, A^\nu = 0
$$

Isto é, as catro compoñentes $A^\mu$ cumpren a ecuación de Klein-Gordon para unha partícula sen masa. De aí que o fotón non teña masa.

As ecuacións do campo e a condición da gauge de Lorenz aínda admiten unha transformación máis do tipo $A^\mu - \partial^\mu \theta$ que as deixa invariantes,  sempre que se cumpra a condición adicional:

$$
\partial_\mu \partial^\mu \theta(x) = 0, \;\;\; \square \, \theta(x) = 0
$$

É doado comprobar que un novo campo con esa condición cumpriría tamén $\partial_\mu A^\mu = 0, \;\; \square A^\mu = 0$.


Así, dos catro graos de liberdade iniciais do fotón, dadas as dúas condicións, só quedan dous graos de liberdade, que corresponden ás dúas polarizacións do fotón.

Sexa agora a corrente electromagnética $j^\mu(x)$

$$
j^\mu(x)_Q = Q \, (\rho(x), \, {\bf j}(x)),
$$

onde $\rho(x)$ é a densidade e ${\bf j}(x)$ a corrente espacial de probabilidade, e $Q$ a carga eléctrica.

As ecuacións de Maxwell, do movemento, son agora:

$$
\partial_\mu F^{\mu\nu} = j^\nu_Q
$$


A corrente é unha corrente conservada. Dado que $F^{\mu\nu}$ é un tensor antisimétrico, cúmprese:

$$
\partial_\nu\partial_\mu F^{\mu\nu} = 0  \Rightarrow \partial_\nu j^\nu_Q = 0.
$$

As ecuacións de Maxwell en presenza de cargas e correntes obtéñense ao incluír no lagranxiano un termo adicional que acopla o fotón coa corrente:

$$
\mathcal{L} = - \frac{1}{4} F_{\mu\nu} F^{\mu\nu} - j^\mu_Q A_\mu
$$

E, finalmente, para representar un fotón masivo, con masa $m$, introducimos un termo de masa no lagranxiano:

$$
\mathcal{L} = - \frac{1}{4}  F_{\mu\nu} F^{\mu\nu} + \frac{1}{2} m^2 A^\mu A_\mu
$$

que se coñece como lagranxiano de Proca.


Que nos daría como ecuación do movemento, en ausencia de cargas:

$$
\partial_\mu \partial^\mu A^\nu - \partial^\nu (\partial_\mu A^\mu) + m^2 A^\nu = 0
$$

Que, se contraemos con $\partial_\nu$, nos queda:

$$
\partial_\mu \partial^\mu (\partial_\nu A^\nu )- \partial_\nu \partial^\nu (\partial_\mu A^\mu) + m^2 \partial_\nu A^\nu = 0, 
$$

Isto é, cúmprese automaticamente, sen necesidade de seleccionala,

$$
\partial_\mu A^\mu = 0
$$

a condición da *gauge de Lorenz*. 

A ecuación do movemento podemos reescribila entón como:

$$
\partial_\mu \partial^\mu A^\nu + m^2 A^\nu = 0, \;\;\; \square \, A^\mu + m^2 A^\mu = 0
$$

Agora as catro compoñentes de $A^\mu$ cumpren a ecuación de Klein-Gordon para unha partícula de masa $m$.

Aínda que a condición da *gauge de Lorenz* se cumpre automaticamente, non obstante aínda podemos impoñer a condición extra:

$$
\square \, \theta = 0
$$

Como neste caso só hai unha condición sobre as catro compoñentes de $A^\mu$, os graos de liberdade do campo vectorial son tres, que corresponden a tres polarizacións.

Non obstante, un campo vectorial con masa non preserva a invariancia gauge, xa que:

$$
\frac{1}{2} m^2 A'^\mu A'_\mu  = \frac{1}{2}m^2 (A^\mu - \partial^\mu \theta) \, (A_\mu - \partial_\mu \theta) \neq \frac{1}{2} m^2 A^\mu A_\mu
$$

por iso dicimos que os termos de masa dos bosóns vectoriais rompen a simetría gauge local.

### Invariancia gauge

#### Electromagnetismo. Derivada covariante.

Vimos, por unha banda, que esixir a invariancia do lagranxiano co cambio de fase local do campo de Dirac implicaba a presenza dun termo de interacción da corrente electromagnética cun campo vectorial; e, por outra banda, que ese mesmo termo pode absorberse no lagranxiano do campo vectorial electromagnético. 

Parece natural que unifiquemos entón ambos nun só lagranxiano que describa as interaccións electromagnéticas dun fermión de carga $Q$ e que, ademais, preserve a invariancia gauge local do fermión:



$$
\mathcal{L}  = i \bar{\Psi} \gamma^\mu \partial_\mu \Psi - m \bar{\Psi}\Psi -  j^\mu_Q  A_\mu - \frac{1}{4} F_{\mu\nu}F^{\mu\nu}
$$

onde a corrente eléctrica vén dada por:

$$
j^\mu_Q \equiv Q \, \bar{\Psi} \gamma^\mu \Psi
$$

$Q$ é a carga eléctrica; por exemplo, para o electrón $Q = -e$.


Que podemos escribir de forma máis compacta se introducimos a **derivada covariante xeneralizada**:

$$
D_\mu = \partial_\mu + i Q A_\mu
$$

Como:

$$
\mathcal{L}  = i \bar{\Psi} \gamma^\mu D_\mu \Psi - m \bar{\Psi}\Psi - \frac{1}{4} F^{\mu\nu}F_{\mu\nu}
$$


Podemos comprobar que este lagranxiano é agora invariante baixo a transformación gauge local de $U(1)$ do campo do fermión, $\Psi(x)$, se o campo vectorial, $A^\mu(x)$, cambia de forma acorde:

$$
\Psi'(x) = e^{i Q\, \theta(x)} \, \Psi(x),
$$
$$
A'_\mu(x) = A_\mu(x) - \partial_\mu \theta(x)
$$


É inmediato verificar que o termo de masa e o cinético do campo vectorial son invariantes baixo esta transformación.

Para que o termo cinético do fermión sexa invariante debe cumprirse:

$$
D'_\mu \, \Psi' = D'_\mu U(\theta )  \,\Psi = U(\theta ) D_\mu \, \Psi
$$

onde:

$$
U(\theta) = e^{iQ\theta(x)}
$$


Comprobémolo:

$$
\left[ \partial_\mu + i Q A'_\mu \right] e^{iQ\theta(x)} \Psi =  \\
\left[ \partial_\mu + i Q (A_\mu - \partial_\mu \theta(x) ) \right]  e^{iQ\theta(x)} \Psi = \\
e^{iQ \theta(x)} \left[\partial_\mu + i Q \partial_\mu \theta(x) + iQA_\mu - i Q \partial_\mu \theta(x) \right] \Psi = \\
e^{iQ \theta(x)} \left[\partial_\mu + iQ A_\mu \right] \Psi
$$

O campo vectorial, de forma xenérica, debe transformarse como:

$$
A'_\mu = U A_\mu U^{-1} + \frac{i}{Q} (\partial_\mu U) \, U^{-1}
$$



Neste caso, con $U(\theta) = e^{iQ\theta(x)}$, temos:


$$
A'_\mu = A_\mu + \frac{i}{Q} (i Q \partial_\mu \theta(x)) = A_\mu -  \partial_\mu \theta(x) 
$$

Podemos concluír entón que, se esiximos que o lagranxiano de Dirac sexa invariante respecto a un cambio de fase *local* do campo $\Psi$, aparece de forma natural a interacción electromagnética entre a carga eléctrica asociada ao fermión e o campo vectorial do fotón. 

Dicimos que o lagranxiano é invariante gauge baixo o grupo $U(1)_Q$, onde $Q$ indica carga.

Tamén vemos que cada fermión pode ter a súa propia carga eléctrica asociada, $Q$.

#### Invariancia gauge baixo cor

Sabemos experimentalmente que existen 6 quarks, $u, d, s, c, b, t$, cada un deles en tres cores: vermello, azul e verde. 


Imos estender o campo de Dirac para incluír a *cor* dos quarks. Farémolo asignando cada cor a un elemento do triplete do grupo $SU(3)_C$, onde $C$ indica cor, e esixindo de novo que o lagranxiano sexa invariante baixo rotacións *locais* no espazo *interno* da cor, en $SU(3)_C$. 

Veremos que este caso é máis complexo ca o do electromagnetismo, dado que os xeradores das rotacións non conmutan, isto é, trátase dun *grupo non abeliano*.

O exemplo do $SU(3)$ pódese estender a outros grupos $SU(N)$, en particular ao $SU(2)$, como veremos máis adiante na sección das interaccións electrofebles.


Sabemos que os quarks veñen en tres cores, *vermello, azul e verde*; podemos asignar o campo de cada cor dun quark, $q^\alpha$, aos elementos dun triplete de $SU(3)$, con $\alpha = r, b, g$. 

así:

$$
q \equiv \begin{pmatrix} q^r \\ q^b \\ q^g \end{pmatrix} 
$$

Estamos a usar unha notación vectorial onde cada elemento corresponde ao espinor de Dirac asociado á cor do quark.

O lagranxiano de Dirac para un quark con tres cores podemos dalo como:

$$
\mathcal{L} = i \bar{q} \gamma^\mu \partial_\mu q - m \bar{q} q 
$$

onde:

$$
\bar{q} = (\bar{q}^r, \bar{q}^b, \bar{q}^g)
$$

Pódese comprobar facilmente que este lagranxiano é invariante baixo a rotación global en $SU(3)_C$.

$$
q \to q' = U({\bf \theta}) \, q, \;\;\; U({\bf \theta}) =  e^{i g_S \theta^a T_a} 
$$

Onde $U({\bf \theta})$ é a matriz de rotación, que é unitaria, $U^\dagger U = I$, e $g_S$ é unha constante, que chamaremos constante de acoplamento forte.

Isto é, rotamos *globalmente* polos ángulos $\theta_a$ sobre cada xerador, $T_a$, de $SU(3)$. 

O grupo $SU(3)$ ten 8 xeradores, $a = 1, \dots, 8$.

Os xeradores podemos expresalos en función das matrices de Gell-Mann:

$$
T_a = \frac{1}{2} \lambda_a
$$

que cumpren as seguintes regras de conmutación:

$$
[\lambda_a, \lambda_b] = 2 i f^{abc} \lambda_c
$$

onde $f^{abc}$ son as constantes de estrutura do grupo, que son reais e totalmente antisimétricas.

O que nos di que os xeradores non conmutan, así que teremos que ter coidado á hora de expandir o expoñente, $e^{i g_S \theta^a T_a}$, para calcular a matriz de rotación, $U({\bf \theta})$.

Nótese que a discusión sería similar se usásemos o grupo $SU(2)$; nese caso o número de xeradores sería tres.

$$
\tau_i = \frac{1}{2} \sigma_i
$$

onde $\sigma_i$ son as matrices de Pauli, con $i = 1, 2, 3$. 

A estrutura do grupo $SU(2)$ é ben coñecida:

$$
[\tau_i, \tau_j] = i \epsilon_{ijk} \tau_j
$$

que se segue inmediatamente das propiedades das matrices de Pauli.

Esiximos agora que o lagranxiano de Dirac do quark con cor sexa invariante baixo rotacións *locais* no espazo interno de $SU(3)_C$, isto é, baixo:

$$
q \to U(\theta(x)) \, q = e^{i g_S \theta^a(x) T_a} q
$$

Para iso, de forma similar a como fixemos no caso do electromagnetismo, introduciremos a derivada covariante:

$$
D_\mu = \partial_\mu + i g_S T_a G^a_\mu(x)
$$

Onde $g_S$ será a constante forte, e $G^a_\mu(x)$ os 8 campos vectoriais asociados a cada xerador. Nótese que o índice $a$ está contraído, logo significa un sumatorio nos 8 termos. Fisicamente, cada campo vectorial $G_\mu^a(x)$ corresponde a cada un dos 8 tipos de gluóns da cromodinámica cuántica.

Do mesmo xeito que fixemos no electromagnetismo, as transformacións dos campos $G^a_\mu(x)$ serán tales que a derivada covariante se transforme igual ca $q$, isto é:

$$
D'_\mu q' = D'_\mu e^{i g_S \theta^a(x) T_a} q = e^{i g_S \theta^a(x) T_a} D_\mu q
$$

para preservar a invariancia gauge local.

As condicións sobre a derivada covariante e sobre o campo transformado son iguais ca no caso do electromagnetismo.

O campo transformado será:

$$
G'^a_\mu T_a = U G^a_\mu T_a U^{-1} + \frac{i}{g_S} (\partial_\mu U)  U^{-1}
$$

Consideremos unha transformación infinitesimal:

$$
e^{i g_S \theta^a T_a} \simeq  I + i g_S \theta^a T_a
$$

Polo tanto, o termo:

$$
\frac{i}{g_s} (\partial_\mu U) U^{-1} \simeq - \partial_\mu \theta^a T_a
$$

Mentres que o termo:

$$
G^a_\mu U T_a U^{-1} \simeq G^a_\mu \, (I + i g_s \theta^b T_b) \, T_a \, (I - i g_S \theta^c T_c) \\\simeq G^a_\mu \left(T_a + i g_s \theta^b [T_a, T_b] \right) = G^a_\mu (T_a - g_S f_{abc} \theta^b T_c)  = \\ G^a_\mu T_a - g_S f_{abc} T_a \theta^b G^c_\mu 
$$



Logo, compoñente a compoñente:

$$
G'^a_\mu  = G^a_\mu  -  \partial_\mu \theta^a - g_S f_{abc} \theta^b G^c_\mu
$$


Vemos, pois, que a transformación dos campos de gluóns é complexa e involucra todos os gluóns. 

Tamén vemos que nos aparece a constante $g_S$ na transformación. Isto fai que o acoplamento forte de todos os quarks sexa o mesmo, independente do seu sabor, mentres que a cada fermión podiámoslle asociar unha carga eléctrica propia.


[EXT] comentario sobre a parte cinética do lagranxiano

Finalmente, se consideramos os seis sabores de quarks, $f = u, d, c, s, b, t$, o lagranxiano de QCD é:

$$
\sum_{f} i \bar{q}_f \gamma^\mu D_\mu q - m \bar{q}q - \frac{1}{4} G^a_{\mu\nu} G^{\mu\nu}_a
$$


## Unificación electrofeble

As interaccións febles, en concreto as cargadas, violan a paridade, e iso implica que tratan de forma diferente as compoñentes de quiralidade á esquerda e á dereita dun campo fermiónico.

Sabemos tamén que nas correntes cargadas, mediadas polos bosóns vectoriais $W^\pm$, só intervén a quiralidade á esquerda dos fermións. Porén, nas correntes neutras, mediadas polo bosón vectorial $Z^0$, interveñen ambas as quiralidades e con pesos diferentes.

Parece natural entón que, para describir as interaccións febles, debamos tratar as dúas quiralidades por separado.


Sabemos que $\Psi$ se divide nas súas dúas compoñentes de quiralidade, vía os proxectores á esquerda e á dereita, $P_L, \, P_R$:

$$
\Psi = \Psi_L + 
\Psi_R, \;\; \Psi_R = P_R \, \Psi, \;\; \Psi_L = P_L \, \Psi
$$

Tamén sabemos que a corrente, a forma vectorial, se descompón en dúas correntes, unha á esquerda e outra á dereita, pero que o termo escalar, o asociado á masa, mestura as dúas quiralidades, isto é:

$$
\bar{\Psi} \gamma^\mu \Psi = \bar{\Psi}_R \gamma^\mu \Psi_R + \bar{\Psi}_L \gamma^\mu \Psi_L,
$$
$$
\bar{\Psi} \Psi = \bar{\Psi}_R \Psi_L + \bar{\Psi}_L \Psi_R
$$

Vimos tamén que a quiralidade se intercambiaba baixo a inversión de paridade:

$$
P : \;\; \Psi_L \leftrightarrow \Psi_R
$$

Para construír o lagranxiano das interaccións febles, imos separar entón as dúas compoñentes de quiralidade.

O lagranxiano de Dirac quédanos:

$$
i \bar{\Psi} \gamma^\mu \partial_\mu \Psi - m \bar{\Psi} \Psi = \\ i \bar{\Psi}_R \gamma^\mu \partial_\mu \Psi_R + i \bar{\Psi}_L \gamma^\mu \partial_\mu \Psi_L  - m \bar{\Psi}_R \Psi_L - m \bar{\Psi}_L \Psi_R 
$$



Pero, para separar completamente as dúas quiralidades, estamos obrigados a que o fermión non teña masa!

Admitamos polo de agora fermións sen masa e ignoremos polo momento (ata máis adiante) o feito de que os fermións si teñen masa.

Nese caso, temos dous campos independentes:

$$
i \bar{\Psi}_R \gamma^\mu \partial_\mu \Psi_R + i \bar{\Psi}_L \gamma^\mu \partial_\mu \Psi_L 
$$



Sabemos tamén que nas desintegracións febles cargadas sempre aparecen os fermións en parellas, por exemplo o leptón cargado, o electrón, co neutrino do seu sabor, o $\nu_e$. 

Para acomodar este feito, podemos situar os campos á esquerda do leptón cargado e do seu neutrino nun duplete que chamaremos de quiralidade á esquerda:

$$
L_e \equiv \begin{pmatrix} \nu_{eL} \\ e_L \end{pmatrix} 
$$

onde, para facilitar a lectura, imos introducir a notación:

$$
\nu_{eL} \equiv u^{(\nu_e)}_L(p), \;\; e_L \equiv u^{(e)}_L(p), 
$$

isto é, o espinor á esquerda do neutrino é $\nu_{eL}$ e o do electrón, $e_L$


Estamos, pois, a asociar os espinores á esquerda das parellas nun duplete que chamamos de isospín feble, do grupo $SU(2)_L$, onde $L$ indica a quiralidade á esquerda.

Neste caso, o $\nu_{eL}$ ten terceira compoñente de isospín feble $+1/2$ e o $e_L$ ten $-1/2$.


Ao colocarmos os campos das parellas no duplete, podemos entender a acción dun campo $W^\pm$ como subir ou baixar os elementos do duplete.

Nótese que esta agrupación é similar a cando introducimos a notación $q$ como un triplete en cuxos elementos situamos os campos asociados ás cores de cada sabor de quark.

Sabemos, non obstante, que nas correntes neutras participan os espinores á dereita, logo cómpre incluílos.

A compoñente á dereita situámola nun singlete de isospín feble.

$$
e_R
$$

onde $e_R$ é o espinor á dereita do electrón.

Neste caso, o Modelo Estándar predí que o neutrino non ten compoñente á dereita: non existe $\nu_R$.

A primeira familia de leptóns é un duplete á esquerda e un singlete á dereita:

$$
L_e \equiv \begin{pmatrix} \nu_{eL} \\ e_L \end{pmatrix}, \;\;\; e_R 
$$

E de forma similar agrupariamos as outras dúas familias leptónicas

Para os quarks a situación é similar: os espinores á esquerda forman un duplete e os da dereita un singlete de isospín. 

O espinor do quark con maior carga do duplete sitúase no elemento de arriba do duplete, isto é, ten terceira compoñente de isospín feble $I_{W3} =1/2$, mentres que o de menor carga aparece abaixo no duplete, con $I_{W3} = -1/2$. A diferenza de carga entre os dous é 1.

Finalmente, os dous quarks da parella teñen ambos singletes á dereita.

Así pois, os fermións agrúpanse en dupletes de quiralidade á esquerda e singletes á dereita.

Para a primeira familia de fermións:

$$
L_e \equiv \begin{pmatrix} \nu_{eL} \\ e_L \end{pmatrix}, \;\; e_R
$$

$$
Q_u \equiv \begin{pmatrix} u_L \\ d_L \end{pmatrix}, \;\; u_R, \; d_R
$$

*Nota:* Non confundir o espinor $u(p)$ coa notación $u_L$; esta última significa o espinor á esquerda do quark $u$.

Tamén hai que ter coidado cos espinores conxugados: $\bar{\nu}_L$ corresponde ao conxugado do espinor á esquerda do neutrino, e non á compoñente á esquerda do antineutrino (que non existe!).


A seguinte táboa mostra o isospín feble e a súa terceira compoñente para os leptóns da primeira familia. 

Na primeira familia, a electrónica:

| | | | |
| :--  | :-- | :-- | :-- |
|  | $\nu_{eL}$ | $e_L$ | $e_R$ |
| $I_{W}$| +1/2 | +1/2 | 0 |
| $I_{W3}$| +1/2 | -1/2 | 0 |

A táboa para o resto de familias leptónicas sería similar.


E a seguinte táboa mostra o isospín e a terceira compoñente para a primeira familia de quarks:

| | | | | | 
| :--  | :-- | :-- | :-- | :-- |
|  | $u_L$ | $d_L$ | $u_R$ | $d_R$ |
| $I_{W}$| +1/2 | 1/2 | 0 | 0 |
| $I_{W3}$| +1/2 | -1/2 | 0 | 0 |

E de forma similar obteriamos a táboa para o resto de familias de quarks.

### Simetría gauge local $SU(2)$

Separemos agora o lagranxiano de Dirac nas súas compoñentes do duplete á esquerda e do singlete á dereita.

Indicamos aquí soamente o da primeira familia de leptóns, pero sería idéntico para os quarks e para o resto de familias de fermións.


$$
i \bar{L}_e \gamma^\mu \partial_\mu L_e + i \bar{e}_R \gamma^\mu \partial_\mu e_R  = 0
$$

Podemos comprobar facilmente que o lagranxiano é invariante baixo unha rotación global do duplete no espazo de isospín feble.

$$
L_e \to L'_e = U(\theta) L_e, \;\;\; U(\theta) = e^{i g_W \theta^i \tau_i}
$$

Neste caso, o espazo interno $SU(2)_L$ ten tres xeradores, $\tau_i$, que están relacionados coas matrices de Pauli, 

$$
\tau_i = \frac{\sigma_i}{2}
$$

con $i = 1, 2, 3$.

Nótese que a rotación ten tres fases, $\theta_i$, asociadas a cada xerador.


E $g_W$ é unha constante real, a constante de acoplamento feble.


Esiximos agora que o lagranxiano sexa invariante baixo unha transformación gauge *local* do espazo interno do isospín feble, $SU(2)_L$, onde $L$ indica que só afecta aos estados de quiralidade á esquerda.

O lagranxiano sería invariante baixo a rotación continua local:

$$
L_e \to L'_e = U(\theta(x)) \, L_e
$$

Isto é, se modificamos as fases do espazo do isospín de forma continua no espazo-tempo.

$$
L'_e  = e^{i g_W \theta^i(x) \tau_i} \, L_e
$$

O lagranxiano quedará invariante.

O que á súa vez implica que para cada xerador $\tau_i$ existe un campo vectorial $W^i_\mu(x)$ ao que se acoplan as correntes dos membros do duplete.

Introducimos a derivada covariante:

$$
D_\mu = \partial_\mu + i g_W W^i_\mu \tau_i
$$

Podemos dar explicitamente a expresión:

$$
W^i_\mu \tau_i = \frac{1}{2} \begin{pmatrix} W^3_\mu & W^1_\mu - i W^2_\mu \\ W^1_\mu + i W^2_\mu & -W^3_\mu \end{pmatrix}
$$

simplemente facendo uso das matrices de Pauli.



Podemos tamén reescribir:

$$
\frac{g_W}{2}(W^1_\mu - i W^2_\mu) \sigma_+ +
\frac{g_W}{2}(W^1_\mu + i W^2_\mu) \sigma_- +
\frac{g_W}{2} W^3_\mu \sigma_3 
$$

onde, lembremos:

$$
\sigma_\pm = \frac{1}{2}(\sigma_1 \pm i \sigma_2)
$$


Se definimos os campos:

$$
W^\pm_\mu = \frac{1}{\sqrt{2}} (W^1_\mu \mp i W^2_\mu)
$$

nótese que:

$$
(W^\pm_\mu)^\dagger = W^\mp_\mu
$$

Temos entón que:

$$
g_W W^i_\mu \tau_i = \frac{g_W}{\sqrt{2}} W^+_\mu \sigma_+ + \frac{g_W}{\sqrt{2}} W^-_\mu\sigma_- + \frac{g_W}{2} W^3_\mu \sigma_3 
$$

Os termos asociados que aparecen no lagranxiano son:

$$
\frac{g_W}{\sqrt{2}} \bar{L} W^+_\mu \sigma_+ L + \frac{g_W}{\sqrt{2}} \bar{L} W^-_\mu\sigma_- L + \frac{g_W}{2} \bar{L}W^3_\mu L = \\
j^\mu_+ W^+_\mu + j^\mu_-W^-_\mu + j^\mu_3 W^3_\mu
$$

Onde definimos as correntes cargadas como:

$$
j^\mu_\pm = \frac{g_W}{\sqrt{2}} \bar{L} \gamma^\mu \sigma_\pm L
$$

cunha constante de acoplamento $g_W/\sqrt{2}$


Se desenvolvemos explicitamente a corrente, podemos comprobar que efectivamente obtemos a corrente esperada:

$$
 j^\mu_+ = \frac{g_W}{\sqrt{2}} \bar{L} \sigma^+ \gamma^\mu L= 
$$
$$
 \frac{g_W}{\sqrt{2}}
 \begin{pmatrix} \bar{\nu}_{eL}, \bar{e}_L \end{pmatrix} \, 
\begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}  \gamma^\mu
\begin{pmatrix} \nu_{eL} \\ e_L \end{pmatrix} =
\frac{g_W}{\sqrt{2}} \bar{\nu}_{eL} \, \gamma^\mu \, e_L 
$$

Así pois, conseguimos acomodar as correntes cargadas febles e apareceron os campos vectoriais $W^{\pm}$

Porén, o termo de interacción:

$$
j^\mu_3 W^3_\mu = \frac{g_W}{2} (\bar{\nu}_{eL} \gamma^\mu \nu_{eL} - \bar{e}_L \gamma^\mu e_L)
$$

non corresponde ás correntes neutras!, porque nas correntes neutras interveñen tamén os espinores á dereita do electrón, $e_R$.

Para implementar as interaccións neutras, imos estender a simetría do lagranxiano esixindo que sexa *tamén* invariante gauge local respecto a unha fase, isto é, respecto ao grupo $U(1)_Y$, onde $Y$ indica hipercarga. 

Así, introducimos a derivada covariante estendida como:

$$
D_\mu = \partial_\mu + i g_Y B_\mu + i g_W W^i_\mu \tau_i 
$$

sendo $B_\mu$ o campo vectorial asociado e $g_Y$ a constante de acoplamento da hipercarga.

Dado que se trata dunha simetría de tipo $U(1)$, cada un dos campos do lagranxiano, neste caso $L, e_R$, ten asociado un valor de hipercarga, $Y, Y_R$, respectivamente. 

Nótese que, se considerásemos que as partículas teñen masa, estes campos xa non serían independentes e non poderiamos asociarlles un valor distinto de hipercarga. 

Nótese tamén que a hipercarga non pode ser a carga eléctrica, porque sabemos que os elementos do duplete teñen un mesmo valor de hipercarga $Y$, mentres que o neutrino non ten carga eléctrica e o electrón ten carga -1, e ambos comparten duplete. 

Imos tentar asociar os termos que nos aparecen no lagranxiano:

$$
j^\mu_3 W^3_\mu + j^\mu_Y B_\mu
$$

coa interacción feble neutra e coa interacción electromagnética:

$$
j^\mu_Z Z_\mu + j^\mu_Q A_\mu
$$

Onde $Z_\mu, A_\mu$ son os campos asociados ao bosón $Z^0$ e ao fotón; e $j^\mu_Z$ e $j^\mu_{Q}$ as correntes neutras e electromagnéticas, respectivamente.

Pola expresión anterior, podemos considerar que os campos *físicos* $A_\mu, \, Z_\mu$ son unha combinación lineal dos campos subxacentes, $W^3_\mu, \, B_\mu$.

A relación entre ambos viría dada pola matriz de mestura:

$$
\begin{pmatrix} A_\mu \\ Z_\mu \end{pmatrix} = 
\begin{pmatrix} \cos \theta_W  &  \sin \theta_W  \\
-\sin \theta_W  &  \cos \theta_W  \end{pmatrix}
\begin{pmatrix} B_\mu \\  W^3_\mu \end{pmatrix}
$$

onde $\theta_W$ se denomina ángulo de Weinberg.

Para facilitar a lectura definimos: 

$$
s_W \equiv \sin \theta_W, \;\; c_W \equiv \cos \theta_W.
$$




Se reescribimos as interaccións fundamentais en función dos campos físicos:

$$
j^\mu_Y B_\mu + j^\mu_{3} W^3_\mu =  \\
 j^\mu_Y \left( c_W A_\mu - s_W Z_\mu \right) + j^\mu_{3} \left(s_W A_\mu + c_W Z_\mu\right)  = \\
(s_W j^\mu_{3} +  c_W j^\mu_Y) \, A_\mu + 
(c_W j^\mu_{3} - s_W j^\mu_Y) \, Z_\mu 
$$

No primeiro sumando, o termo que se acopla ao fotón, $A_\mu$, debe corresponderse coa corrente electromagnética:

$$
j^\mu_Q = s_W j^\mu_{3} +  c_W j^\mu_Y
$$

E o segundo, coas correntes neutras:

$$
j^\mu_Z = c_W j^\mu_{3} -  s_W j^\mu_Y
$$

#### Correntes electromagnéticas

Se desenvolvemos a expresión da corrente electromagnética para o caso da primeira familia dos leptóns, obtemos:

$$
j^\mu_Q = s_W \frac{g_W}{2} \, \bar{L} \gamma^\mu \sigma_3 L +  c_W g_Y Y \, \bar{L} \gamma^\mu L  + c_W g_Y Y_R \, \bar{e}_R \gamma^\mu e_R 
$$

Isto é:

$$
j^\mu_{Q} = 
\left(s_W \frac{g_W}{2} + c_W g_Y Y \right) \, \bar{\nu}_{eL} \gamma^\mu \nu_{eL} + 
\left(-s_W \frac{g_W}{2} + c_W g_Y Y \right) \bar{e}_L \gamma^\mu e_L + c_W g_Y Y_R \, \bar{e}_R \gamma^\mu e_R
 $$

O neutrino non ten corrente eléctrica, logo debe cumprirse que:

$$
s_W g_W  = - 2 c_W g_Y Y
$$

mentres que para o electrón as correntes á esquerda e á dereita deben ter un mesmo factor, que é a súa carga eléctrica $Q = -e$.

$$
-s_W \frac{g_W}{2} + c_W g_Y Y = - e \\
c_W g_Y Y_R = -e 
$$

Ambas as condicións se cumpren se:

$$
e = g_W s_W = g_Y c_W
$$

isto é:

$$
g_Y = g_W \tan \theta_W
$$

O ángulo de Weinberg pode expresarse como:

$$
s_W = \frac{g_Y}{\sqrt{g^2_W+g^2_Y}}, \;\;\; c_W = \frac{g_W}{\sqrt{g^2_W+g^2_Y}},

$$

e ademais:

$$
Y_R = -1, Y = -1/2
$$


En xeral, vemos que para cada campo fermiónico á dereita se cumprirá: 

$$
Y_R = Q,
$$

No caso do $\nu_e$, de forma natural $Y_R = 0$.

Mentres que a hipercarga ten un valor tal que nos permite obter a carga dos elementos do duplete:

$$
Q = Y + I_{W3}
$$

onde $I_{W3}$ é a terceira compoñente de isospín feble.

Isto é, o elemento de arriba do duplete ten sempre unha carga eléctrica +1 respecto á do de abaixo.

A seguinte táboa mostra os valores de hipercarga para os leptóns e os quarks á esquerda:

| |  $L_e$ | $Q_u$ | 
| :--  | :--  | :--  |
| $Y$| -1/2 | 1/6 | 

E para as compoñentes á dereita:

| |  $e_R$ | $u_R$ | $d_R$ |
| :--   | :--  | :--  | :-- |
| $Y_R$ | -1 | 2/3 | -1/3 |





Recuperamos, pois, a descrición das interaccións electromagnéticas. 


#### Correntes neutras


Se regresamos ás correntes febles neutras, atopámonos con:

$$
j^\mu_Z = c_W j^\mu_3 - s_W j^\mu_Y =  \\ 
c_W j^\mu_3 - \frac{s_W}{c_W} \left( j^\mu _Q - s_W j^\mu_3 \right)= \\
\frac{1}{c_W} \left( j^\mu_3 - s_W j^\mu_Q \right)
$$


Se explicitamos a corrente para o caso da primeira familia de leptóns:

$$
j^\mu_Z = \frac{g_W}{2c_W} \, \bar{L} \gamma^\mu \sigma_3 L  + \frac{s_W}{c_W} e \, \bar{e}_R \gamma^\mu e_R + \frac{s_W}{c_W} e \, \bar{e}_L \gamma^\mu e_L
$$

Introducindo o valor de $e = s_W g_W$:

$$
j^\mu_Z = \frac{g_W}{c_W} \left( \frac{1}{2}\bar{\nu}_{eL} \gamma^\mu \nu_{eL}\right) + 
\frac{g_W}{c_W} \left[ s^2_W \bar{e}_R \gamma^\mu e_R + (-\frac{1}{2} + s^2_W) \bar{e}_L \gamma^\mu e_L\right]
$$



Se definimos:

$$
g_Z \equiv \frac{g_W}{c_W}
$$

Obtemos:

$$
j^\mu_Z = g_Z \left( \frac{1}{2}\bar{\nu}_{eL} \gamma^\mu \nu_{eL}\right) + 
g_Z \left[ s^2_W \bar{e}_R \gamma^\mu e_R + (-\frac{1}{2} + s^2_W) \bar{e}_L \gamma^\mu e_L\right]
$$

O acoplamento coas compoñentes á esquerda e á dereita do neutrino é:

$$
c^{(\nu_e)}_L= \frac{1}{2}, \;\; c^{(\nu_e)}_R = 0
$$

Mentres que para o electrón o acoplamento á esquerda e á dereita vale:

$$
c^{(e)}_L = -\frac{1}{2} + s^2_W, \;\; c^{(e)}_R = s^2_W
$$

De forma xenérica, as correntes neutras dun fermión con espinor $u$ son: 

$$
j^\mu_Z = -Q_f s^2_W g_Z \, \bar{u}_R \gamma^\mu u_R + \left( I_{W3} - Q_f s^2_W \right) g_Z \, \bar{u}_L \gamma^\mu u_L
$$

onde $Q_f$ é a carga eléctrica do fermión en unidades da carga do electrón (por exemplo, para o $e$, $Q_f = -1$) e $I_{W3}$ é a terceira compoñente de isospín feble.

Que podemos dar de forma compacta:

$$
j^\mu_Z = g_Z c_R \, (\bar{u}_R \gamma^\mu u_R) + g_Z c_L \, (\bar{u}_L \gamma^\mu u_L)
$$

sendo $c_R, \, c_L$ os factores á dereita e á esquerda:

$$
c_R = -Q_f s^2_W, \;\;\; c_L = I_{W3} - Q_f s^2_W
$$ 


A seguinte táboa dá os distintos acoplamentos á esquerda e á dereita da corrente neutra para a primeira familia de fermións:

| | | | 
| :-- | :-- | :-- |
| | ----- $c_R$ ---- | ----- $c_L$ ---- |
| $\nu_e$ | 0 | $\frac{1}{2}$| 
| $e$ | $s^2_W$ | $-\frac{1}{2} + s^2_W$| 
| $u$ | $-\frac{2}{3}s^2_W$ | $\frac{1}{2} - \frac{2}{3}s^2_W$| 
| $d$ | $\frac{1}{3}s^2_W$ | $-\frac{1}{2} + \frac{1}{3} s^2_W$| 

E de forma similar obteriámola para o resto de familias.

#### Resumo

Construímos un lagranxiano que é invariante gauge local respecto aos grupos $SU(2)_L\times U(1)_Y$,
de tal forma que nos acomoda as correntes neutras, febles e electromagnéticas. 

Agrupamos os campos dos fermións en dupletes á esquerda e singletes á dereita, o que impón que os fermións non teñan masa. 

Obtivemos catro bosóns vectoriais: dous deles son os bosóns $W^\pm$,
que median as correntes cargadas, e os outros dous son os bosóns $A, Z^0$, que á súa vez median as correntes 
electromagnéticas e neutras, respectivamente, e que son combinacións lineais dos bosóns subxacentes $B, W^3$
a través dun ángulo de mestura, $s_W$.

O modelo ten dous parámetros, $g_Y, g_W$, que se corresponden cos parámetros físicos $e, s_W$, e que fixan os acoplamentos das tres correntes: $e, g_W, g_Z$. 

Porén, pagamos un prezo elevado: os fermións e os bosóns vectoriais non teñen masa. 

Na seguinte sección imos tratar como resolver este problema.

## O mecanismo de Higgs

Brout, Englert e Higgs desenvolveron nos anos 60 un mecanismo que permite dotar de masa os bosóns vectoriais e tamén os fermións, e que preserva a invariancia gauge do lagranxiano, fundamental para que a teoría sexa *renormalizable*.

Este mecanismo introduce un campo escalar cuxa partícula asociada é o bosón de Higgs. Pero non foi ata a segunda década do século XXI cando se descubriu o bosón de Higgs nos experimentos ATLAS e CMS. 

Inspeccionemos primeiro o termo do lagranxiano asociado á  masa dos fermións

$$
m \bar{\Psi} \Psi = m \bar{\Psi}_R \Psi_L + m \bar{\Psi}_L \Psi_R
$$

Imos tentar substituír $m$ polo acoplamento a un campo escalar, $\phi$, que a substitúa. Lembremos que $m$ e $\phi$ teñen ambos dimensión de enerxía, $[E]$.


Consideremos unha interacción dos fermións cun campo escalar complexo, $\phi$, cunha constante de acoplamento $\lambda$, que denominamos acoplamento de Yukawa. Para iso introduciremos no lagranxiano un termo:

$$
\lambda \Psi \Psi \phi
$$


E consideremos ademais que o campo $\phi$ ten un valor non nulo no estado de mínima enerxía, isto é, no baleiro, $\phi_0$:

$$
\phi(x) = \phi_0 + H(x)
$$

O termo que introducimos dá lugar a dous, que corresponden a un termo de masa e a outro de interacción cun campo $H(x)$

$$
\lambda \Psi \Psi \phi =  \lambda \phi_0 \, \bar{\Psi} \Psi + \lambda \bar{\Psi} \Psi H
$$

Onde podemos asociar a masa a:

$$
m \equiv  \lambda \phi_0
$$

Pero como obtén un campo escalar ese valor no baleiro, $\phi_0$?

Examinemos primeiro o lagranxiano dun campo escalar complexo, $\phi(x)$:

$$
\partial_\mu \phi^* \partial^\mu \phi - m^2 \phi^* \phi
$$

O termo de masa podemos entendelo como un potencial parabólico en $\phi^* \phi$.

$$
V(\phi) = m^2 \phi^* \phi
$$

Consideremos a primeira extensión dese potencial:

$$
V(\phi) = \mu^2 \, \phi^* \phi + \lambda \, (\phi^* \phi)^2
$$

O potencial debe estar acoutado inferiormente para que a súa enerxía mínima teña un límite inferior; isto implica que $\lambda \gt 0$, xa que o termo cuadrático é o dominante.

#### Bosón de Goldstone

Reescribimos agora o noso campo complexo, $\phi(x)$, con dous campos reais, $\varphi_1(x), \varphi_2(x)$, respecto ao valor do campo que corresponde ao mínimo do potencial que ten só compoñente real. 

$$
\phi(x) = \frac{1}{\sqrt{2}} \left( v + \varphi_1(x) + i \varphi_2(x) \right)
$$

Isto é, seleccionamos un valor do campo específico, de entre todos os posibles mínimos, respecto ao que desenvolver o campo. Esta elección rompe a simetría dos valores do campo onde o potencial é mínimo. De aí que chamemos a esta elección *rotura espontánea de simetría*.

Certamente, a física non pode quedar afectada pola elección deste mínimo.

Así temos:

$$
\phi^*\phi = \frac{1}{2}\left[(v+\varphi_1)^2 + \varphi^2_2 \right]=  \frac{1}{2} \left[ v^2 + 2v \varphi_1 + (\varphi^2_1 + \varphi^2_2)\right] \\
$$

e

$$
(\phi^*\phi)^2 = \frac{1}{4} \left[ (v^2+2v\varphi_1)^2 + 2 (v^2+2v\varphi_1) (\varphi^2_1 + \varphi^2_2) + 
(\varphi^2_1 + \varphi^2_2)^2 \right] \\
\frac{1}{4} \left[ (v^4+ 4v^3 \varphi_1+ 4 v^2 \varphi^2_1)+ 2 v^2(\varphi^2_1 + \varphi^2_2) + 
4v\varphi_1 (\varphi^2_1 + \varphi^2_2) + 
(\varphi^2_1 + \varphi^2_2)^2 \right]
$$

O potencial, ao tomar $\mu^2 = -\lambda v$, é:

$$
V(\phi) = - \lambda v (\phi^*\phi) + \lambda (\phi^*\phi)^2
$$

Nótese que no desenvolvemento do potencial os termos só en $\varphi_1$ e só en $(\varphi^2_1 + \varphi^2_2)$ se anulan.

O potencial queda:

$$
V(\phi) = -\frac{\lambda v^4}{4} + \lambda v^2 \varphi^2_1 + 4 v \lambda \varphi_1 (\varphi^2_1+\varphi^2_2) + \frac{1}{4} \lambda (\varphi^2_1+\varphi^2_2)^2 
$$


onde o primeiro termo é o valor mínimo do potencial $V(\phi)$ e o segundo é o termo de masa do campo $\varphi_1$, con masa:

$$
\frac{1}{2} m^2_{\varphi_1} = \lambda v^2 = - \mu^2, \;\;\; (\mu^2 <0)
$$

e os dous últimos termos corresponden á interacción entre os campos $\varphi_1, \varphi_2$.


Logo, dos dous campos reais, un, o asociado á compoñente real, adquire masa, isto é, require un mínimo de enerxía para crear a súa partícula asociada; esta masa correspóndese coa curvatura arredor da base do sombreiro. O segundo campo, o asociado á compoñente imaxinaria, non ten masa: o seu potencial é nulo, o que se corresponde co feito de que o potencial non cambia ao desprazarse ao longo da base do sombreiro.

A este segundo campo coñéceselle como bosón de Goldstone, e reflicte o feito de que o mínimo do potencial ten unha simetría.

a partir da súa compoñente real $\varphi_1(x)$ e dunha rotación $\theta(x)$.

Onde o primeiro termo é o termo cinético do campo $H$.

E os outros dous termos son a interacción entre o campo $H$ e o bosón $A_\mu$, con vértices de tres e catro liñas, respectivamente.

Ás veces dise na literatura que o campo $A_\mu$ *comeu* o bosón de Goldstone $\varphi_2$.

### Masa dos bosóns vectoriais

O caso anterior é un caso académico, onde usamos a simetría gauge U(1) do lagranxiano do campo escalar para dotar de masa un hipotético bosón vectorial.

Pero o problema real é como dotar de masa os bosóns vectoriais febles, $W^\pm$ e $Z^0$.

Neste caso necesitaremos tres bosóns de Goldstone para xerar a masa dos tres bosóns vectoriais, o que nos leva a que o campo de Higgs debe ter polo menos 4 compoñentes reais. 


Consideremos que o campo de Higgs é un duplete de campos escalares complexos:

$$
\Phi(x) \equiv \begin{pmatrix} \phi^+(x) \\ \phi^0(x) \end{pmatrix} 
$$

onde $\phi^+(x)$ é un campo escalar complexo cunha carga eléctrica asociada $+1$, mentres que $\phi^0(x)$ é tamén un campo escalar complexo con carga eléctrica nula.

O porqué de que sexa un duplete quedará máis claro cando, na seguinte sección, dotemos de masa os fermións.

Consideremos que o lagranxiano deste campo é:

$$
\mathcal{L} = (D_\mu \Phi)^\dagger D^\mu \Phi - V(\Phi)
$$

onde:

$$
V(\Phi) = \mu^2 (\Phi^\dagger \Phi) + \lambda (\Phi^\dagger \Phi)^2
$$
$$
D_\mu = \partial_\mu + i g_W W^i_\mu \tau_i + i g_Y B_\mu
$$

con $\mu^2<0, \lambda \gt0$, sendo $g_W, \, g_Y$ as intensidades dos acoplamentos feble e de hipercarga, respectivamente, e $\tau_i = \frac{\sigma_i}{2}$ os xeradores do grupo $SU(2)$. 



Este lagranxiano é, polo tanto, invariante gauge baixo o grupo $SU(2)_L \times U(1)_Y$, e o campo está sometido a un potencial do tipo sombreiro mexicano.

A hipercarga do duplete é $Y = 1/2$, de tal forma que a carga eléctrica do elemento de abaixo do duplete de Higgs, $\Phi$, sexa nula:

$$
Q = I_{W3} + Y
$$

O mínimo do potencial é de novo $v/\sqrt{2}$

$$
|\Phi_0| = \sqrt{-\frac{\mu^2}{2\lambda}} \equiv \frac{v}{\sqrt{2}}
$$

Rompemos a simetría do valor do campo no mínimo do potencial e desenvolvemos o campo de forma xenérica a partir do valor no mínimo que ten só compoñente real no elemento de abaixo do duplete: 

$$
\Phi(x) = e^{i \tau_i \theta^i(x)} \frac{1}{\sqrt{2}}\begin{pmatrix} 0 \\ v+ H(x)\end{pmatrix}
$$

onde $H(x)$ é un campo escalar real, e $\theta^i(x)$ son 3 funcións reais.

Podemos usar a invariancia gauge para seleccionar unha gauge específica onde $\theta^i(x) = 0$, que chamamos unitaria.

Nótese que esta gauge nos asegura que o campo que sobrevive estea asociado co campo *físico* que chamamos de Higgs.


A derivada covariante:

$$
D_\mu \Phi = \left(\partial_\mu + i g_W \frac{\sigma_i}{2} W^i_\mu + i Y g_Y B_\mu\right) \Phi
$$

onde $Y= \frac{1}{2}$.

Se a expresamos co campo desenvolvido a partir do valor no baleiro e na gauge unitaria:

$$
D_\mu \Phi =  \\ \frac{1}{\sqrt{2}} \begin{pmatrix} 0 \\ \partial_\mu H \end{pmatrix} + 
 \frac{i}{2} \begin{pmatrix} g_W W^3_\mu + g_Y B_\mu & g_W(W^1_\mu - i W^2_\mu) \\ g_W(W^1_\mu + i W^2_\mu) & -g_W W^3_\mu + g_Y B_\mu\end{pmatrix} \frac{1}{\sqrt{2}} \begin{pmatrix} 0 \\ v+H \end{pmatrix} 
$$
$$
 \frac{1}{\sqrt{2}} \begin{pmatrix} 0 \\ \partial_\mu H \end{pmatrix} +
 \frac{i}{2\sqrt{2}} \begin{pmatrix} \sqrt{2} g_W W^+_\mu \\ - g_W W^3_\mu + g_Y B_\mu \end{pmatrix} (v+H)
$$


Podemos reescribir o termo onde aparecen os campos $W^3_\mu, B_\mu$ cos dos bosóns físicos, $A_\mu, Z_\mu$:

$$
-g_W W^3_\mu + g_Y B_\mu = -g_W (s_W A_\mu + c_W Z_\mu) + g_Y (c_WA_\mu - s_W Z_\mu) = \\
(-g_W s_W + g_Y c_W) A_\mu - (g_W c_W + g_Y s_W) Z = -\frac{g_W}{c_W} Z_\mu = - g_Z Z_\mu
$$

onde usamos que $g_Y = g_W \tan \theta_W, g_Z = g_W/c_W$.

Ou, equivalentemente, como o fotón non ten masa, sabemos que para que se anule o seu factor na expresión anterior debe cumprirse: 

$$
g_W s_W = g_Y c_W
$$

isto é, debe cumprirse a seguinte relación entre as intensidades de acoplamento:

$$
g_Y = g_W \tan \theta_W
$$

que é a relación que nos aparecera antes, ao desenvolver a unificación electrofeble.

Así:

$$
D_\mu \Phi =
 \frac{1}{\sqrt{2}} \begin{pmatrix} 0 \\ \partial_\mu H \end{pmatrix} +
 \frac{i}{2\sqrt{2}} \begin{pmatrix} \sqrt{2} g_W W^+_\mu \\ -g_Z Z_\mu \end{pmatrix} (v+H)
$$


Tendo en conta que $(W^+)^\dagger = W^-$, o termo cinético do lagranxiano na gauge unitaria queda:

$$
(D_\mu \Phi)^\dagger D^\mu \Phi = \\ \frac{1}{2} \partial_\mu H \partial^\mu H + (v+H)^2 \left[ \frac{1}{4} g^2_W W^-_\mu W^{+\mu} + \frac{g^2_Z}{8} Z_\mu Z^\mu \right]
$$

Logo, obtivemos un termo de masa para os bosóns vectoriais $W^\pm, Z^0$, que é:

$$
M_W = \frac{1}{2} v g_W, \;\;\; M_Z = \frac{1}{2} v g_Z 
$$

O que establece a mesma relación entre as masas ca entre os acoplamentos:

$$
\frac{M_W}{M_Z} = \frac{g_W}{g_Z} = c_W 
$$

Nótese que as constantes de acoplamento feble, $g_W, g_Y$, que fixan as interaccións febles cos fermións, xunto con $v$, fixan tamén a masa dos bosóns vectoriais, $M_W, M_Z$. 

Isto é, o modelo de Higgs non introduce parámetros adicionais asociados á masa dos bosóns $W^\pm, Z^0$, e esta está fixada polos parámetros $g_Y, g_W, v$.

Pero nótese tamén que, ademais dos termos de masa dos bosóns, apareceron no lagranxiano unha serie de termos que corresponden ás interaccións do Higgs cos bosóns vectoriais $W^\pm, Z^0$. 

A física que se deduce destas interaccións é moi rica e complexa, pero non a trataremos nestas notas. 

Si facemos notar, porén, que a intensidade da interacción dos campos vectoriais co Higgs:

$$
g_W = 2 \frac{M_W}{v}, \;\;\; g_Z = 2 \frac{M_Z}{v}
$$

é proporcional á masa dos bosóns.

Lembremos agora que nos diagramas de Feynman das correntes febles nos aparece o factor (ao cadrado) asociado aos vértices e ao propagador, que, se o cuadrimomento transferido é pequeno comparado coa masa dos bosóns, podemos aproximar por:

$$
\frac{g^2_W}{2} \frac{4}{M^2_W} = \frac{1}{2v^2} 
$$

E sabemos que este factor está asociado á constante de Fermi:

$$
G_F = \frac{1}{2v^2} \Rightarrow v = \left(\sqrt{2} G_F\right)^{-1/2} = 246 \;\; \mathrm{GeV}
$$

Isto é, o valor esperado no baleiro, $v$, está asociado á constante de Fermi, $G_F$.


Así pois, a unificación electrofeble ten catro parámetros físicos: dous asociados ao bosón de Higgs, $G_F, m_H$, a constante de acoplamento de Fermi e a masa do Higgs; e $e, s_W$, a carga eléctrica do electrón e o seno do ángulo de Weinberg. 

Desde o punto de vista teórico, os parámetros son $v, \lambda, g_Y, g_W$.


### Masas dos fermións

Para dotar de masa os fermións debemos recuperar o termo do lagranxiano que acople o campo á esquerda co da dereita a través do campo de Higgs.

$$
m \bar{\Psi} \Psi = m \bar{\Psi}_R\Psi_L + m \bar{\Psi}_L \Psi_R 
$$


Ao situarmos os campos á esquerda nun duplete e os da dereita nun singlete de isospín, debemos establecer un acoplamento co campo de Higgs de tal forma que sexa un escalar e invariante Lorentz. Por iso, introducimos os termos:

$$
\lambda_e \, \bar{L} \Phi \, e_R + \lambda_e \, (\bar{L} \Phi e_R)^\dagger
$$

onde $\lambda_e$ é unha constante de acoplamento de Yukawa. O segundo termo é simplemente o hermítico conxugado, $\mathrm{h.c.}$ 

Tras a rotura espontánea de simetría, o termo queda:

$$
\lambda \begin{pmatrix} \bar{\nu}_L, & \bar{e}_L \end{pmatrix} \frac{1}{\sqrt{2}} \begin{pmatrix} 0 \\ v+H \end{pmatrix} e_R + \mathrm{h.c.}
$$

Isto é:

$$
\frac{1}{\sqrt{2}} \lambda \bar{e}_L e_R (v+H)+ \mathrm{h.c.}
$$

Aparécenos un termo de masa do electrón:

$$
\frac{1}{\sqrt{2}} \lambda_e v (\bar{e}_L e_R + \bar{e}_R e_L) = m_e \bar{e} e
$$

e un termo de interacción do electrón co campo de Higgs:

$$
\frac{1}{\sqrt{2}} \lambda_e H (\bar{e}_L e_R + \bar{e}_R e_L) = \frac{\lambda_e}{\sqrt{2}} H \, \bar{e} e
$$




A masa do electrón é entón:

$$
m_e = \frac{1}{\sqrt{2}} \lambda_e v
$$

E a interacción do electrón co Higgs é proporcional a:

$$
\lambda_e = \sqrt{2} \frac{m_e}{v}
$$

a masa do electrón.

E procederiamos de forma similar para o resto de familias leptónicas.

Neste caso, ao non existir o campo á dereita do neutrino, non podemos dotalo de masa.

Se consideramos agora a primeira familia dos quarks, uns termos similares aos anteriores dotarían de masa o quark $d$, pero para dotar de masa o quark $u$ necesitamos introducir termos adicionais.

Para iso, definimos primeiro o conxugado do duplete do campo de Higgs:

$$
\Phi^c = i \sigma_2 \Phi^*
$$

que, tras a rotura espontánea de simetría e a elección da gauge unitaria, queda:

$$
\Phi^c = \begin{pmatrix} 0 & 1 \\ -1 & 1 \end{pmatrix} \frac{1}{\sqrt{2}} \begin{pmatrix} 0 \\ v+H \end{pmatrix} =
\frac{1}{\sqrt{2}} \begin{pmatrix} v+H \\  0\end{pmatrix}
$$

Se introducimos agora os termos de interacción dos quarks co campo de Higgs como:

$$
\lambda_u \bar{Q}_u \Phi u_R + \lambda_d \bar{Q}_u \Phi^c  d_R + \mathrm{h.c.}
$$

onde $\lambda_u, \lambda_d$ son as constantes de acoplamento de Yukawa dos quarks $u, d$, respectivamente.

Nótese que ambos os termos son escalares respecto ao isospín feble, e acóplannos os campos á esquerda e á dereita de cada quark.

Tras a rotura espontánea de simetría e na gauge unitaria, obtemos:

$$
\lambda_u  (\bar{u}_L u_R + \bar{u}_R u_L) \frac{1}{\sqrt{2}} (v+H) + \lambda_d (\bar{d}_L d_R + \bar{d}_R d_L) \frac{1}{\sqrt{2}} (v+H)
$$

Isto é, os termos de masa para os quarks $u, d$:

$$
m_u = \frac{1}{\sqrt{2}} \lambda_u v, \;\;\; m_d = \frac{1}{\sqrt{2}} \lambda_d v,
$$

E os da interacción dos quarks co bosón de Higgs, cuxa constante de acoplamento de Yukawa:

$$
\lambda_u = \sqrt{2} \frac{m_u}{v}, \;\;\; \lambda_d = \sqrt{2} \frac{m_d}{v}
$$

é proporcional á masa de cada quark.

Procederiamos da mesma forma para o resto de familias de quarks.

Nótese, polo tanto, que para cada fermión cargado introducimos unha constante de acoplamento de Yukawa. Isto é, hai tantos parámetros libres adicionais como masas dos quarks.

E tamén que o acoplamento co campo do bosón de Higgs é proporcional á masa de cada fermión. O bosón de Higgs acóplase con máis intensidade aos fermións máis pesados.

En resumo, o Higgs acóplase aos fermións e aos bosóns vectoriais, $W^\pm, Z^0$, proporcionalmente á súa masa.


Recapitulemos: para dotar de masa a primeira familia de fermións introducimos os seguintes termos de interacción:

$$
\lambda_e \bar{L}_e \Phi e_R + \lambda_u \bar{Q}_u \Phi^c u_R + \lambda_d \bar{Q}_u \Phi d_R + \mathrm{h.c.}
$$

que, ademais de dotar de masa o electrón e os quarks $u, d$, fixan as interaccións destes co bosón de Higgs.

### Parámetros do modelo

Recapitulemos e resumamos os parámetros que aparecen no Modelo Estándar.

O campo de Higgs introduce dous parámetros, $v, \lambda$, que experimentalmente están asociados a $G_F$ e á masa do Higgs $m_H$

A interacción electrofeble introduce dous parámetros, $g_W, g_Y$, que experimentalmente están asociados á carga do electrón, $e$ (ou á constante de estrutura fina $\alpha$), e ao seno do ángulo de Weinberg $s_W$. E que, xunto con $v$ (ou $G_F$), establecen a masa dos bosóns vectoriais $M_W, M_Z$.

A interacción forte introduce unha constante de acoplamento $g_S$ ou, experimentalmente, a constante $\alpha_S$.

Finalmente, temos tantos acoplamentos de Yukawa como fermións cargados, isto é, 3 por familia. Para a primeira familia, $\lambda_e, \lambda_u, \lambda_d$ ou, experimentalmente, as masas $m_e, m_u, m_d$.

Non consideramos na nosa discusión a mestura de sabores dos quarks, que nos introduciría 4 parámetros adicionais, asociados á matriz CKM: tres ángulos de mestura e unha fase complexa.

Se consideramos a masa dos neutrinos e a súa mestura, o modelo ten 3 máis 4 (ou 6) parámetros adicionais: 3 das masas e 4 da matriz de mestura PMNS, e dúas posibles fases se os neutrinos son de Majorana.

V_{\text{1-loop}}(\phi) \;=\; \sum_i \frac{n_i\, m_i^4(\phi)}{64\pi^2}\left[\ln\!\bigg(\frac{m_i^2(\phi)}{\mu^2}\bigg)-c_i\right],
$$
onde:
 * \(i\) percorre os graos de liberdade que se acoplan a \(\phi\) (top, W, Z, Higgs, Goldstones, \dots),
 * \(n_i\) conta graos de liberdade con signo (p. ex. \(n_t=-12\) para o quark top: - para fermións, 3 cores e factor 4 de Dirac),
 * \(m_i(\phi)\) son as masas dependentes do campo (ver máis abaixo),
 * \(\mu\) é a escala de renormalización,
 *  \(c_i\) depende do esquema: en \(\overline{\mathrm{MS}}\) tipicamente \(c_{\,\text{escalares/fermións}}=\tfrac{3}{2}\), \(c_{\,\text{gauge}}=\tfrac{5}{6}\).

% 2a) masas dependentes do campo (exemplos dominantes no SM)
$$
m_t(\phi)=\frac{y_t}{\sqrt{2}}\phi,\qquad
m_W(\phi)=\frac{g}{2}\phi,\qquad
m_Z(\phi)=\frac{\sqrt{g^2+g'^2}}{2}\phi,\qquad
m_h^2(\phi)= -m^2 + 3\lambda\phi^2.
$$

% 3) Potencial efectivo mellorado polo GR (apropiado para \phi \gg v)
$$
V_{\text{eff}}(\phi)\simeq \frac{1}{4}\,\lambda_{\text{eff}}(\mu=\phi)\;\phi^4,
$$
onde \(\lambda_{\text{eff}}(\mu)\) é a constante de acoplamento \(\lambda\) avaliada coas ecuacións do grupo de renormalización (RG) ata a escala \(\mu\sim\phi\). A estabilidade do baleiro depende esencialmente do signo de \(\lambda_{\text{eff}}(\mu)\) a escalas altas.

% 4) Función beta (esquema \overline{MS}, primeira orde — termo dominante polo top)
$$
\beta_\lambda \equiv \mu\frac{d\lambda}{d\mu}
= \frac{1}{16\pi^2}\Big( 12\lambda^2 + 6\lambda y_t^2 - 3 y_t^4 - 3\lambda (3g^2+g'^2) + \tfrac{3}{8}(2g^4+(g^2+g'^2)^2) \Big) + \cdots
$$
(a contribución principal que empurra \(\lambda\) cara a valores negativos vén do termo \(-\tfrac{3}{16\pi^2}y_t^4\) cando \(y_t\) é grande).

% 5) Condición de metaestabilidade
$$
\exists\; \mu_* \quad\text{tal que}\quad \lambda_{\text{eff}}(\mu_*) < 0
\quad\Longrightarrow\quad \text{o baleiro electrofeble é metaestable (falso mínimo).}
$$

% 6) Taxa de desintegración por efecto túnel (aproximación semiclásica)
$$
\Gamma/V \;\simeq\; A\, e^{-S_E[\phi_{\text{bounce}}]/\hbar},
$$
coa estimación de orde de magnitude (para un potencial dominado por \phi^4 negativa)
$$
S_E \;\approx\; \frac{8\pi^2}{3\,|\lambda_{\text{eff}}(\mu_B)|},
$$
onde \(\mu_B\) é a escala característica do \emph{bounce} (aproximadamente a escala onde \(\lambda_{\text{eff}}\) alcanza o seu valor mínimo negativo). O prefactor \(A\) é tipicamente \(\sim \mu_B^4\) ata factores adimensionais.
