## O lagranxiano do SM

### Da Física Clásica á Teoría Cuántica de Campos

A dinámica dun sistema en Física Clásica queda determinada polo lagranxiano: 

$$
L(q, \dot{q}) = T -V
$$

que depende das coordenadas xeneralizadas $q_i$, con $i=1, \dots, n$ e das súas derivadas temporais $\dot{q}$

as ecuacións do movemento veñen dadas polas ecuacións de Euler-Lagrange:

$$
\frac{\mathrm{d}}{\mathrm{dt}} \left( \frac{\partial L}{\partial \dot{q}_i}\right) = \frac{\partial L}{\partial q_i}
$$

No caso dunha partícula nun potencial, o lagranxiano é a súa enerxía cinética menos o seu potencial:

$$
L = \frac{1}{2} m \dot{x}^2 - V(x) 
$$

as ecuacións do movemento dan:
$$
m \ddot{x} = -\frac{\partial V(x)}{\partial x}
$$

que é a ecuación de Newton.

En teoría cuántica de campos, TCC, introducimos un campo $\phi(x)$ definido no espazo $x$ cun lagranxiano:

$$
L = \int \mathcal{L}(\phi, \partial_\mu \phi) \, \mathrm{d}^3x 
$$

onde $\mathcal{L}$ é a densidade lagranxiana ou, por abuso da linguaxe, o lagranxiano.

As ecuacións do movemento son:

$$
\partial_\mu \left( \frac{\partial \mathcal{L}}{\partial (\partial_\mu \phi)}\right) = \frac{\partial \mathcal{L}}{\partial \phi}
$$

En TCC entendemos as partículas como excitacións do campo $\phi$.

### Lagranxiano do campo escalar

Sexa $\phi(x)$ un campo escalar asociado a partículas de masa $m$; o seu lagranxiano vén dado por:

$$
\mathcal{L}(\phi) = \frac{1}{2} (\partial_\mu \phi) \, (\partial^\mu \phi) - \frac{1}{2} m^2 \phi^2
$$

E a súa ecuación de movemento:

$$
\partial_\mu \partial^\mu \phi + m^2 \phi = 0
$$

É a ecuación de Klein-Gordon

### Lagranxiano dos fermións (espín 1/2)

Sexa $\Psi$ un campo espinorial de Dirac; o seu lagranxiano é:

$$
\mathcal{L}(\Psi) = i \bar{\Psi} \gamma^\mu \partial_\mu \Psi - m \bar{\Psi} \Psi
$$

Onde:
$$
\Psi = \begin{pmatrix} \psi_1 \\ \psi_2 \\ \psi_3 \\ \psi_4\end{pmatrix},
$$

cada $\psi_i$ con $i=1, \dots, 4$ é un campo complexo. E as $\gamma$ son matrices complexas $4 \times 4$, con regras de conmutación determinadas (ver [apendice-Dirac])


A ecuación de Euler-Lagrange con respecto a $\bar{\Psi}$ dá:

$$
(i \gamma^\mu \partial_\mu - m)  \, \Psi = 0 
$$

que é a ecuación de Dirac

### Lagranxiano do campo vectorial

Sexa o campo vectorial $A^\mu(x)$, por exemplo o electromagnético, $A^\mu = (\phi, {\bf A})$, onde $\Phi$ é o potencial eléctrico e ${\bf A}$ o potencial vector magnético.

O tensor de forza é:

$$
F^{\mu \nu} = \partial^\mu A^\nu - \partial^\nu A^\mu
$$

e a corrente $j^\mu = (\rho, {\bf j})$, onde $\rho$ é a densidade eléctrica e ${\bf j}$ a corrente eléctrica.

A ecuación do movemento:

$$
\partial_\mu F^{\mu\nu} = j^\nu
$$

dá as ecuacións de Maxwell.

A ecuación obtense do lagranxiano:

$$
\mathcal{L}(A) = - \frac{1}{4} F_{\mu\nu} F^{\mu\nu} - j^\mu A_\mu
$$

onde o termo $j^\mu A_\mu$ representa a interacción dunha corrente eléctrica co fotón.

Para o fotón illado:

$$
\mathcal{L}(A) = - \frac{1}{4} F_{\mu\nu} F^{\mu\nu}
$$

E para un fotón masivo, con masa $m_A$:

$$
\mathcal{L}(A) = - \frac{1}{4}  F_{\mu\nu} F^{\mu\nu} + \frac{1}{4} m_A^2 A^\mu A_\mu
$$

que se coñece como lagranxiano de Proca.


## Simetría gauge

### QED

O lagranxiano dun fermión é invariante respecto a unha transformación global de U(1), isto é, respecto a unha fase global, $q\theta$:

$$
\Psi(x) \to \Psi'(x) = e^{iq\theta} \Psi(x)
$$

onde $q$ é o valor da carga conservada asociada ao campo e á simetría conservada.

o cal é inmediato de comprobar, dado que $\bar{\Psi'} = \bar{\Psi} \, e^{-i\theta}$



Falamos de **invariancia gauge local** se o lagranxiano é invariante baixo unha transformación local dun grupo de simetría.

No caso do campo do fermión, sexa unha transformación local baixo o grupo $U(1)$

$$
\Psi(x) \to \Psi'(x) = e^{i q \theta(x)} \Psi
$$

onde $\theta(x)$ depende agora do punto local $x$, e $q$, no caso do electromagnetismo, é o acoplamento electromagnético, neste caso a carga eléctrica do fermión, por exemplo $-e$ para o electrón.

O lagranxiano baixo esta transformación queda:

$$
\mathcal{L'}(\Psi') = i \bar{\Psi} \gamma^\mu \partial_\mu \Psi - m \bar{\Psi} \Psi - q \bar{\Psi} \gamma^\mu \Psi (\partial_\mu \theta)
$$


O seguinte termo rompe a invariancia:

$$
- q \bar{\Psi} \gamma^\mu \Psi \, (\partial_\mu \theta(x))
$$

que interpretamos como o acoplamento dunha corrente $j^\mu = q \bar{\Psi} \gamma^\mu \Psi$, a electromagnética, cun campo vectorial $A_\mu = \partial_\mu \theta(x)$:

$$
 - j^\mu \, A_{\mu}
$$

Se consideramos o lagranxiano ao que lle engadimos o termo do campo vectorial:

$$
\mathcal{L}_{EL}(\Psi, A) = i \bar{\Psi} \gamma^\mu \partial_\mu \Psi - m \bar{\Psi}\Psi -  j^\mu  A_\mu - \frac{1}{4} F^{\mu\nu}F_{\mu\nu}
$$

Este é o lagranxiano do electromagnetismo se $q$ é a carga eléctrica.


O lagranxiano do EM é agora invariante baixo a transformación gauge local de U(1) do campo do fermión, $\Psi(x)$, se o campo vectorial, $A^\mu(x)$, cambia de forma acorde:

$$
\Psi'(x) = e^{iq \theta(x)} \Psi(x), \;\;\; A'_\mu(x) = A_\mu(x) - \partial_\mu \theta(x)
$$

o que se pode expresar máis elegantemente substituíndo:

$$
\partial_\mu \to D_\mu = \partial_\mu + i q A_\mu
$$

$D_\mu$ denomínase **derivada covariante xeneralizada**.

O lagranxiano do electromagnetismo, que describe o fermión, o fotón e a súa interacción, queda:

$$
\mathcal{L}_{EM}(\Psi, A) = i \bar{\Psi} \gamma^\mu D_\mu \Psi - m \bar{\Psi}\Psi - \frac{1}{4} F^{\mu\nu}F_{\mu\nu}
$$


#### Non invariancia do termo de masas do bosón

Non obstante, o termo de masas do bosón vectorial $A$ non é invariante gauge:

$$
\frac{1}{4} m_A^2 A'^\mu A'_\mu  = \frac{1}{4}m_A^2 (A^\mu - \partial^\mu \theta) \, (A_\mu - \partial_\mu \theta)
$$

por iso dicimos que os termos de masa dos bosóns vectoriais rompen a simetría gauge.

### Clase acelerada sobre o Higgs no SM:

| | | | |
| :--: |:--: | :--: | :--: |
| ----- Tipo ----- | --------------- Escalar ---------------- | ---------- Espinor ------------ | ------ Bosón vectorial ---------- |
| | Klein-Gordon | Dirac | Maxwell |
|Ecuación | $(\partial^\mu\partial_\mu + m^2) \, \phi = 0$ | $(i\gamma^\mu \partial_\mu - m) \, \Psi = 0$ | $\partial_\mu F^{\mu\nu} = j^\nu$ |
|Lagranxiano| $\frac{1}{2} (D_\mu \phi)^\dagger (D^\mu \phi) - \frac{1}{2} m^2 \phi^\dagger \phi$ | $i\bar{\Psi}\gamma^\mu D_\mu \Psi - m \bar{\Psi}\Psi$|  $-\frac{1}{4}F^{\mu\nu}F_{\mu\nu} \left[- \frac{1}{4} m^2 A_\mu A^\mu \right]$|

Onde:

$$
D_\mu = \partial_\mu + i g A_\mu, \;\;\; F^{\mu\nu} = \partial^\mu A^\nu - \partial^\nu A^\mu
$$

sendo $g$ unha constante de acoplamento, por exemplo $g=e$; e se $A_\mu$ é o fotón, obtemos o electromagnetismo.

O lagranxiano de espinores e bosóns vectoriais:

$$
i \bar{\Psi} \gamma^\mu D_\mu \Psi - m \bar{\Psi} \Psi - \frac{1}{4} F^{\mu\nu}F_{\mu\nu}
$$

É invariante baixo a simetría gauge local, dada por $\theta(x)$:
    
$$
\Psi'(x) = e^{i g \theta(x)} \, \Psi(x), \;\;\; A'_\mu(x) = A_\mu(x) -i\partial_\mu \theta(x)
$$ 

Pero non o termo de masas do bosón vectorial:

$$
\frac{1}{2} m_A^2 A'^\mu A'_\mu  = \frac{1}{2}m_A^2 (A^\mu - \partial^\mu \theta) \, (A_\mu - \partial_\mu \theta)
$$

Nótese que no lagranxiano aparece o termo:
    
$$
i \bar{\Psi} \gamma^\mu D_\mu \Psi = \dots + (i g \bar{\Psi} \gamma^\mu \Psi) \, A_\mu  
$$

que é o acoplamento dunha corrente fermiónica, cun acoplamento $g = e$, ao fotón, que interpretamos como o vértice:

$$
e \bar{\Psi} \gamma^\mu \Psi \, A_\mu = j^\mu_{EL} \, A_\mu
$$

### Como lles dá o Higgs masa a fermións e bosóns?

$$
\phi(x) = \frac{1}{\sqrt{2}}\left(v + h(x)\right) 
$$

A posible interacción no lagranxiano entre os fermións e este campo, controlada por un acoplamento (chamado de **Yukawa**), $\lambda$

$$
-\lambda \bar{\Psi} \Psi \, \phi = -\frac{\lambda v}{\sqrt{2}} \bar{\Psi} \Psi - \frac{\lambda}{\sqrt{2}} \bar{\Psi} \, \Psi h(x)
$$

O primeiro dá a masa do fermión:

$$
m= \frac{\lambda v}{\sqrt{2}}
$$

E o segundo a interacción (a desintegración, por exemplo) do Higgs nun par fermión-antifermión

$$

$$

m_A = \frac{g v}{\sqrt{2}}

### Como aparece o campo de Higgs?

$$

$$
\phi(x) = \frac{1}{\sqrt{v}} \left( v + h(x)\right)

## A teoría electrofeble en detalle

O Modelo Estándar desenvolveuse nos anos 60 por A. Salam, S. Weinberg e S. Glashow.

Establece a unificación das forzas feble e electromagnética na forza electrofeble.

O modelo inclúe, para ser coherente, as forzas electrofeble e forte.

Está baseado en teorías gauge locais.

O SM establece a invariancia respecto a U(1)$_Y$ SU(2)$_L$ SU(3)$_C$

Onde $Y$ é a hipercarga feble, $L$ é a quiralidade dos fermións e $C$ a cor. O SM introduce para o grupo SU(2)$_L$ o isospín feble, $I_W$. 

Consideraremos aquí só a teoría electrofeble, sen a cor, pero o tratamento para esta é similar. 

Cada fermión terá asociada unha carga de hipercarga, $Y$, e de isospín feble, $I_w$.

### Hipercarga e isospín feble.

O SM organiza as partículas dependendo da súa quiralidade.

Lembremos que todo espinor de Dirac se pode descompoñer en partes de quiralidade á esquerda e á dereita.

Cada xeración agrúpase nun doblete á esquerda para leptóns, isto é $I_W = 1/2$, e quarks, e singletes á dereita, $I_W = 0$:

$$
\begin{pmatrix} \nu_{eL} \\ e_L \end{pmatrix}, \, e_R; \;\;\;  
\begin{pmatrix} u_L \\ d'_L \end{pmatrix}, \, u_R, d'_r. 
$$

Do mesmo xeito para as outras dúas xeracións.

Nótese que non existe o singlete á dereita para os neutrinos!

Os dobletes teñen terceira compoñente de isospín feble, $I^3_W$:

$$
I^3_W(\nu_{eL}) = 1/2, \;\; I^3_W(e_L) = -1/2; \\
I^3_W(u_L) = 1/2, \;\; I^3_W(d'_L) = -1/2 
$$

Por exemplo, para o $\nu_e$:

$$
\frac{1}{2}\sigma ^2 \begin{pmatrix} \nu_{eL} \\ 0 \end{pmatrix} = \frac{1}{2} 
\begin{pmatrix} 1 & 0 \\ 0  & -1 \end{pmatrix}
\begin{pmatrix} \nu_{eL} \\ 0 \end{pmatrix} = 
+ \frac{1}{2} \begin{pmatrix} \nu_{eL} \\ 0 \end{pmatrix}
$$

Mentres que os singletes non teñen terceira compoñente de isospín feble:

$$
I^3_W (e_R) = I^3_W(u_R) = I^3_W(d'_R) = 0
$$

Definimos a hipercarga feble, $Y_W$, como:
    
$$
Y_W \equiv 2 (Q - I^3_W)
$$

O que nos dá

$$
Y_W(\nu_{eL}) = -1, \;\; Y_W(e_L) = -1; \\
Y_W(u_L) = 1/2, \;\; I^3_W(d'_L) = -1/2 
$$


### Interaccións febles e electromagnéticas


O campo do fermión debe ser invariante baixo transformacións locais

   * unha fase local, $\theta(x)$, do grupo U(1), relacionada coa hipercarga feble $Y$, cuxo acoplamento denotamos por $g_Y$,
   
   * unha tripla fase local, $\alpha(x)_i$ con $i=1, 2, 3$, do grupo SU(2), asociada ao isospín feble SU(2)$_L$, cuxo acoplamento é $g_W$.

$$
\Psi'(x) = e^{ig' \frac{\theta(x)}{2}} \, e^{ig \bf{T} \cdot {\bf \alpha}(x)} \,\, \Psi(x)
$$

onde ${\bf T}$ son os xeradores de SU(2), isto é, as matrices de Pauli, $T_i = \frac{\sigma_i}{2}$ con $i=1, 2, 3$. 

O factor $1/2$ no primeiro expoñente é por conveniencia.


Impoñer invariancia gauge neste grupo equivale a estender a derivada covariante xeneralizada:

$$
D_\mu \equiv \partial_\mu - \frac{g'}{2} B_\mu - g \frac{\sigma}{2} \cdot {\bf W}_\mu
$$

Onde:

  * $B_\mu$ é o campo vectorial asociado ao grupo U(1)

  * ${\bf W}_\mu$ son tres campos vectoriais, $W^i_\mu$, con $i=1, 2, 3$, asociados ao grupo SU(2)

### Inciso

A seguinte táboa mostra os xeradores, os campos e o acoplamento dos grupos de simetría do SM. 



|      |   hipercarga feble   | ---- isospín feble ---- |   ----- cor  ----- |
| :--: | :--:                 | :--:           | :--:     |
| grupo| U$_Y$(1) | SU(2)$_L$ | SU(3) |
| xeradores | $\frac{1}{2}$ | $T^i = \frac{1}{2} \sigma^i$ | $\frac{1}{2}\lambda^j$ |
| mediadores | $B_\mu$ | $W^i_\mu$| $a^j_\mu$ |
| acoplamentos    | $g'$ | $g$| $g_S$ |

onde $\sigma^i$ son as tres matrices de Pauli e $\lambda^j$ as 8 matrices de Gell-Mann.

$g_s$ é o acoplamento forte e $a^j_\mu$ os campos vectoriais dos oito gluóns, $j = 1, \dots, 8$.


A derivada covariante xeneralizada dará lugar á interacción de catro correntes que se acoplan cos catro campos vectoriais $B_\mu, {\bf W}_\mu$:


$$
(j^\mu_{1W} W^1_\mu + j^\mu_{2W} W^2_\mu + j^\mu_{3W} W^3_\mu) +  j^\mu_Y B_\mu
$$

Onde (con $i=1, 2, 3$):

$$
j^\mu_{iW} = g_W \bar{\Psi} \frac{\sigma_i}{2} \gamma^\mu \Psi, \;\;\; j^\mu_Y = \frac{g_Y}{2} \bar{\Psi} \gamma^\mu \Psi
$$

Imos relacionar estas correntes coas correntes físicas: a do electromagnetismo e as das correntes febles.




Se reescribimos os campos e as correntes:

$$
W^\pm_\mu = \frac{1}{\sqrt{2}}(W^1_\mu \mp i W^2_\mu), \;\; j^\mu_{\pm} = \frac{g_W}{\sqrt{2}} (j^\mu_1 \pm i j^\mu_2)
$$

Obtemos:

$$
 \left(j^\mu_+ W^+_\mu + j^\mu_- W^-_\mu \right) + j^\mu_3 W^3_\mu +  j^\mu_Y B_\mu
$$


Onde os dous primeiros sumandos corresponden á interacción das correntes febles cargadas cos bosóns $W^\pm$.


Podemos reescribir explicitamente as correntes febles cargadas:

$$
j^\mu_{\pm} = \frac{g_W}{\sqrt{2}} \bar{\Psi} \sigma^\pm \Psi
$$
con:
$$
\sigma^\pm = \frac{\sigma_1}{2} \pm i \frac{\sigma_2}{2}, \;\;
\sigma^+ = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}, \;\;
\sigma^- = \begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}
$$

Para maior claridade escribimos expresamente a corrente positiva, a que se acopla a $W^+$, para os leptóns da primeira familia:

$$
 j^\mu_+ = \begin{pmatrix} \bar{u}(\nu_e)_{L}, \bar{u}(e)_L \end{pmatrix} \, \gamma^\mu \frac{1}{2}\left(\sigma^1 + i \sigma^2 \right)  \, \begin{pmatrix} \bar{u}(\nu_e)_{L} \\ \bar{u}(e)_L \end{pmatrix} = \\
\begin{pmatrix} \bar{u}(\nu_e)_{L}, \bar{u}(e)_L \end{pmatrix} \, \gamma^\mu \sigma^+  \, \begin{pmatrix} \bar{u}(\nu_e)_{L} \\ \bar{u}(e)_L \end{pmatrix} = 
 \begin{pmatrix} \bar{u}(\nu_e)_L, \bar{u}(e)_L \end{pmatrix} \, \gamma^\mu \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix} \begin{pmatrix} \bar{u}(\nu_e)_L \\ \bar{e}_L \end{pmatrix} = \\
 \bar{u}(\nu_e)_L \, \gamma^\mu \, u(e)_L = 
 \bar{u}(\nu_e) \, \gamma^\mu \frac{1}{2} (I - \gamma^5) \, u(e)
$$

onde $u(\nu_e), u(e)$ son os espinores de Dirac do $\nu_e$ e do $e$ respectivamente.

E do mesmo xeito procederiamos para $j^\mu_-$

Tamén por claridade podemos expresar a corrente $j^\mu_3$:
    
$$
j^\mu_3 =  \begin{pmatrix} \bar{u}(\nu_e)_{L}, \bar{u}(e)_L \end{pmatrix} \, \gamma^\mu \frac{1}{2} \sigma^2  \, \begin{pmatrix} u(\nu_e)_{L} \\ u(e)_L \end{pmatrix} = \\
\begin{pmatrix} \bar{u}(\nu_e)_L, \bar{u}(e)_L \end{pmatrix} \, \gamma^\mu \frac{1}{2}\begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix} \begin{pmatrix} u(\nu_e)_L \\ u(e)_L) \end{pmatrix} = \\
\frac{1}{2} \bar{u}(\nu_e)_L \gamma^\mu u(\nu_e)_L - \frac{1}{2} \bar{u}(e)_L \gamma^\mu u(e)_L = \\
I^3_W(\nu_{eL}) \,\,  \bar{u}(\nu_e)_L \gamma^\mu u(\nu_e)_L + I^3_W(e_L) \,\, \bar{u}(e)_L \gamma^\mu u(e)_L
$$
    
Nótese que os factores en cada sumando corresponden ao valor da súa terceira compoñente de isospín feble, isto é: $I^3_W(\nu_{eL}) = 1/2, \; I^3_W(e_L) = -1/2$.

#### As correntes neutras e electromagnéticas

Para recuperar a interacción das correntes electromagnéticas co fotón, a teoría electrofeble establece que existe unha rotación entre os campos $B_\mu, W^3_\mu$ e os campos físicos $A_\mu, Z_\mu$, onde $A$ é o fotón e $Z$ o bosón $Z$ neutro.

$$
A_\mu = B_\mu \cos \theta_W  + W^3_\mu \sin \theta_W +\\
Z_\mu = -B_\mu \sin \theta_W + W^3_\mu \cos \theta_W
$$

onde $\theta_W$ se denomina ángulo de Weinberg.



Se reescribimos os dous sumandos anteriores

$$
j^\mu_3 W^3_\mu + j^\mu_Y B_\mu =  \\
( \sin \theta_W j^\mu_3 +  \cos \theta_W j^\mu_Y) \, A_\mu + 
(\cos \theta_W j^\mu_3 - \sin \theta_W j^\mu_Y) \, Z_\mu 
$$

Neste caso, o primeiro sumando corresponde ás interaccións electromagnéticas, onde a intensidade do acoplamento é $e$ e $Q(f)$ é a carga do fermión en unidades de $e$, por exemplo $Q(e) = -1$ para o electrón.

$$
j^\mu_{EL} = Q(f) \, e \, \Psi \gamma^\mu \Psi
$$


No caso dos neutrinos, que non interaccionan electromagneticamente, a súa corrente EL é nula:

$$
\left[ I^3_W(\nu_{eL}) \, g_W \sin \theta_W + Y_W(\nu_{eL}) \, \frac{g_Y}{2} \cos \theta_W  \right] \, \bar{u}(\nu_{e})_L \gamma^\mu u(\nu_e)_L = 0
$$

onde $I^3_W(\nu_{eL})$ é a terceira compoñente de isospín e $Y_W(\nu_{eL})$ a hipercarga feble do $\nu_{eL}$. Ambas son as cargas do $\nu_{eL}$ respecto á terceira compoñente de isospín e á hipercarga feble.

Como: 
$$
I^3_W(\nu_{eL}) = 1/2, \;\;\; Y_W(\nu_{eL}) = 2 \, (Q(\nu_{eL}) - I^3_W(\nu_{eL})) = 2 \, (0 -1/2) = -1
$$

dado que a carga do neutrino é nula, $Q(\nu_{eL}) = 0$, obtemos a relación:

$$
-g_W \sin \theta_W + g_Y \cos \theta_W = 0, \;\;\; \tan \theta_W = \frac{g_Y}{g_W}
$$


As correntes que se acoplan ao fotón e ao $Z$ simplifícanse:
    
$$
\sin \theta_W (j^\mu_3 + \frac{1}{2} j^\mu_Y) \, A_\mu + \frac{1}{\cos \theta_W} (j^\mu_3 \cos^2 \theta_W- \sin^2 \theta_W j^\mu_Y) \, Z_\mu
$$

Dada a definición que fixemos de hipercarga, $Y = 2(Q-I^3_W)$, vemos que efectivamente o primeiro sumando, para un fermión f que ten carga eléctrica Q, é:

$$
Q  \, g_W \sin \theta_W \,\, \bar{\Psi} \gamma^\mu \Psi  = Q \, e \, \bar{\Psi} \, \gamma^\mu \Psi
$$

Cúmprese a relación:

$$
g_W \sin \theta_W = e = \sqrt{4 \pi \alpha}
$$

onde $e$ é a carga eléctrica e $\alpha$ a constante de estrutura fina.

Se atendemos á interacción co $Z$ e substituímos $j^\mu_Y = 2 (j^\mu_Q - j^\mu_3)$, obtemos:

$$
\frac{g_W}{\cos \theta_W} (j^\mu_3 - Q \, \sin^2 \theta_W j^\mu_Q) \, Z_\mu
$$

Sexa, por exemplo, o neutrino: como a súa carga é nula, $Q(\nu_{eL}) = 0$, e a súa terceira compoñente de isospín feble é $I^3_W(\nu_{eL}) = 1/2$, obtemos:

$$
\frac{g_W}{\cos \theta_W} \frac{1}{2} \; \bar{u}(\nu_e)_L \gamma^\mu u(\nu_e)_L \, Z_\mu 
$$

Podemos identificar a constante de acoplamento $g_Z$:

$$
g_Z = \frac{g_W}{ \cos \theta_W}
$$


De forma xeral, para un fermión obteremos para as súas compoñentes á esquerda e á dereita:

$$
g_Z \left[ \left( I^3_W - Q \sin^2 \theta_W \right) \bar{u}_L \gamma^\mu u_L - Q \sin^2 \theta_W \bar{u}_R \gamma^\mu u_R \right] 
$$

Onde $I^3_W$ é a terceira compoñente de isospín feble do fermión á esquerda e $Q$ a carga do fermión en unidades de $e$.


Que podemos reescribir:

$$
g_Z \left[ \left( I^3_W - Q \sin^2 \theta_W \right) \bar{u} \gamma^\mu \frac{1}{2} (I - \gamma^5) u - Q \sin^2 \theta_W \bar{u} \gamma^\mu \frac{1}{2}(I + \gamma^5) u \right] = \\
g_Z \; \bar{u} \gamma^\mu \left[ \left( I^3_W - 2 Q(f) \sin^2 \theta_W \right) I - I^3_W \gamma^5 \right] u
$$

Onde identificamos os factores vectorial e axial das interaccións neutras dun fermión:

$$
c_V = I^3_W - 2 Q \sin^2 \theta_W, \;\;\; c_A = I^3_W
$$

A seguinte táboa indica o valor de $I^3_W, Q, c_V, c_A$ dos fermións da primeira xeración:

| fermión | --- $I^3_W$ --- | --- $Q$ --- | --------- $c_V$ --------- | --- $c_A$ --- |
| :--:    | :--:    | :--: | :--:  | :--: |
| $\nu_e$ | 1/2     | 0   | 1/2     | 1/2  |
| $e$     | -1/2    | -1  | $-1/2 + s^2$ | -1/2|
| $u$     | 1/2     | +2/3  | $1/2 - (2/3) s^2$ | 1/2 |
| $d$     | -1/2    | -1/3 | $-1/2 + (1/3)s^2$ | -1/2|


onde $s^2_W = \sin^2 \theta_W$

e para os antifermións:

| antifermión | --- $I^3_W$ --- | --- $Q$ --- | ---------- $c_V$ ---------- | --- $c_A$ --- |
| :--:    | :--:    | :--: | :--:  | :--: |
| $e^+$ | 1/2     | +1   | $1/2 - s^2$      | 1/2  |
| $\bar{\nu}_e$  | -1/2  | 0  | -1/2 | -1/2|
| $\bar{d}$     | 1/2     | +1/3  | $1/2 - (1/3) s^2$ | 1/2 |
| $\bar{u}$     | -1/2    | -2/3 | $-1/2 + (2/3) s^2$ | -1/2|

### As correntes físicas: electromagnéticas, e febles cargadas e neutras

Lembremos pois as **relacións entre as constantes de acoplamento**:
    
$$
e = g_W \sin \theta_W = g_Z \sin \theta_W \cos \theta_W
$$
    
Isto é, $g_W, g_Z$ veñen dadas por $e$ e o ángulo de Weinberg $\theta_W$.

O valor $\theta_W \simeq 33^o$ determinouse de numerosas maneiras (ver despois)

$$
\sin^2 \theta_W = 0.23146 \pm 0.000012
$$

A parte electrofeble do SM, en particular a interacción co $Z$, queda determinada por tres parámetros:
    
$$
e, \sin \theta_W, M_W
$$
    
e experimentalmente por:
    
$$
\alpha, G_F, M_Z
$$

a constante de estrutura fina, $\alpha$, a constante de Fermi, $G_F$, e a masa do bosón, $M_Z$

As correntes electromagnéticas non cambian a carga nin o sabor, e conservan a paridade.

#### As correntes cargadas e o vértice do $W$

As correntes cargadas cambian o sabor, a carga nunha unidade e a terceira compoñente de isospín.

Nas correntes mediadas por $W^+$ $I^3_W$ sobe unha unidade, e nas de $W^-$ baixa.

Só interveñen os estados de quiralidade á esquerda dos fermións.

| |
|:--:|

No vértice cun $W^+$ introducimos a constante de acoplamento $g_W/\sqrt{2}$ entre os dous fermións, o de arriba, $f_\uparrow$, e o de abaixo, $f_\downarrow$, do doblete de quiralidade á esquerda:

$$
W^+: \;\; \frac{g_W}{\sqrt{2}} \, \bar{u}(f_\uparrow)_L \gamma^\mu u(f_\downarrow)_L = \frac{g_W}{\sqrt{2}} \bar{u}(f_\uparrow) \gamma^\mu \frac{1}{2}(I-\gamma^5) u(f_\downarrow), \\
W^-: \;\; \frac{g_W}{\sqrt{2}} \, \bar{u}(f_\downarrow)_L \gamma^\mu u(f_\uparrow)_L = \frac{g_W}{\sqrt{2}} \bar{u}(f_\downarrow) \gamma^\mu \frac{1}{2}(I-\gamma^5) u(f_\uparrow),
$$


Dito doutro xeito, do vértice das correntes cargadas podemos dicir que:
    
  * ten a constante de acoplamento $g_W/\sqrt{2}$
  
  * só afecta aos espinores á esquerda dos fermións ou, equivalentemente, aparece un factor $\frac{1}{2} (I-\gamma^5)$
  
  * sobe ou baixa a terceira compoñente, ou aparece como factor a matriz $\sigma^\pm = \frac{1}{2}(\sigma^1 \pm i \sigma^2)$, que media entre os estados dos dobletes

Vexámolo de forma práctica para o vértice dun $W^+$ cun $e$ e un $\nu_e$:

$$
\frac{g_W}{\sqrt{2}} \begin{pmatrix}\bar{u}(\nu_e), & 0 \end{pmatrix} \gamma^\mu \frac{1}{2} (I - \gamma^5) \sigma^+ 
\begin{pmatrix} 0 \\ u(e)\end{pmatrix} = \frac{g_W}{\sqrt{2}} \bar{u}(\nu_e)_L \gamma^\mu u(e)_L 
$$

#### O vértice do Z

As interaccións co $Z$ non cambian o sabor da partícula, nin tampouco a súa carga eléctrica, nin a súa terceira compoñente de isospín feble.

Pero cambian de forma non trivial a paridade.

O bosón $Z$ actúa de xeito diferente para a quiralidade á esquerda e á dereita, dependendo do fermión.

A corrente neutra co bosón $Z$ ten a forma:

$$
Z: \;\;\; g_Z \left( c_L \, \bar{u}_L \gamma^\mu \bar{u}_L + c_R \, \bar{u}_R \gamma^\mu \bar{u}_R \right) = g_Z \,\bar{u} \frac{1}{2}\left(c_V I - c_A \gamma^5\right) u 
$$

onde $g_Z$ é a constante de acoplamento e $c_L, c_R$ os factores asociados á parte de quiralidade.

Ou, equivalentemente, dous factores $c_V, c_A$, vectorial e axial.

No vértice de Feynman dunha interacción co bosón $Z$ intervirán os factores:
    
   * a intensidade: $g_Z$
   
   * a parte de quiralidade á esquerda: $c_L = I^3_W - Q \sin^2 \theta_W$, onde $Q$ é a carga eléctrica do fermión en unidades de $e$ e $I^3_W$ a terceira compoñente de isospín feble
   
   * a parte de quiralidade á dereita: $c_R = -Q \sin^2 \theta_W$

Por exemplo, para o neutrino

Non ten parte á dereita, nin tampouco carga, $Q(\nu)=0$; só $I^3_W(\nu) = 1/2$

A súa compoñente $c_L = 1/2$ e $c_R = 0$

a corrente neutra co $Z$ será:
    
$$
\frac{g_Z}{2} \bar{u}(\nu_e)_L \gamma^\mu u(\nu_e)_L 
$$

Mentres que para o electrón

Ten carga $Q(e)=-1$ e a súa parte á esquerda $I^3_W(e_L) = -1/2$. 

Polo tanto $c_L = -\frac{1}{2} + \sin^2 \theta_W$ e $c_R = \sin^2 \theta_W$.

Así, a corrente neutra co $Z$ é:
    
$$
g_Z \left(  \left(-\frac{1}{2} + \sin^2 \theta_W\right) \, \bar{u}_L \gamma^\mu u_L  + \sin^2 \theta_W \, \bar{u}_R \gamma^\mu u_R \right)
$$

De forma equivalente, podemos dar as compoñentes vectorial e axial das correntes neutras:
    
$$
c_V = I^3_W - Q \sin^2 \theta_W, \;\; c_A = I^3_W 
$$

## Anchura de desintegración do $Z$

O colisionador LEP $e+e^+$, que operou durante os anos 90, verificou con gran detalle as predicións do SM, especialmente a física do Z.

Durante un período o LEP operou a $\sqrt{s} = m_Z = 91$ GeV producindo millóns de $Z$.

Un dos estudos máis importantes do LEP é a medida da anchura de desintegración do $Z$.

### Anchura de desintegración parcial

Sexa a desintegración $ Z \to f + \bar{f}$ nun par fermión-antifermión.

A anchura de desintegración vén dada por:

$$
\Gamma = \frac{p^*}{8 \pi s} \langle |M_{fi}|^2 \rangle
$$

onde neste caso $\sqrt{s} = m_Z$ e $p^* = m_z/2$, se desprezamos a masa do fermión en comparación coa do $m_Z$

O cálculo de $M_{fi}$, aínda que non é complicado, escapa ao nivel deste curso,

aínda que xa sabemos que debe involucrar a corrente neutra:

$$
g_z \left( c_L \bar{u}_L \gamma^\mu v_L + c_R \bar{u}_R \gamma^\mu v_R \right)
$$

O seu valor é:

$$
\langle |M_{fi}|^2 \rangle = \frac{2}{3} (c^2_L + c^2_R) g^2_Z m^2_Z
$$



A anchura parcial de desintegración $Z \to f + \bar{f}$ é:

$$
\Gamma(Z \to f + \bar{f}) = \frac{m_Z}{16 \pi m^2_Z} \frac{2}{3} (c^2_L + c^2_R) g^2_Z m^2_Z = \frac{g^2_Z m_Z}{24 \pi} (c^2_L + c^2_R)
$$

Por exemplo, para $\nu_e$, como $c^2_L = 1/4$

$$
\Gamma (Z \to \nu_e + \bar{\nu}_e) = \frac{g^2_Z m_Z}{96 \pi} = 166 \; \mathrm{MeV}
$$

*Cuestión*: Calcula as anchuras de desintegración parciais e total do $Z$ e as súas fraccións de desintegración.

```python
GF  = units.value("Fermi coupling constant")
s2t = units.value("weak mixing angle")
MW  = 80.34 # GeV 
MZ  = 91.19 # GeV
g2w = 8 * MW**2 * GF /np.sqrt(2)
g2z = g2w/(1-s2t) 
gamma_nu = g2z*MZ/(96*np.pi)
print('gw    {:5.4f}'.format(np.sqrt(g2w)))
print('gz    {:5.4f}'.format(np.sqrt(g2z)))
print('Gamma nue {:5.4f} GeV'.format(gamma_nu))
```

```python
nc   = 3 # 3 cores para os quarks
pars = {'e':(-0.5+s2t, s2t), 'nue': (0.5, 0), 'u' : (0.5-2*s2t/3, -2*s2t/3), 'd': (-0.5+s2t/3, s2t/3)}
gamma = lambda cl, cr: g2z*MZ/(24*np.pi)*(cl**2 + cr**2)
gammas = {}
for key in pars.keys():
    cl, cr = pars[key]
    gammai = gamma(cl, cr)
    gammai = nc * gammai if key in ('u', 'd') else gammai
    comment = 'Gamma ' + key + ' {:5.4f} GeV'.format(gammai) 
    print(comment)
    gammas[key] = gammai
```

```python
gamma_total = 3 * gammas['nue'] + 3 * gammas['e'] + 2 * gammas['u'] + 3 * gammas['d']
print('Gamma total {:5.4f} GeV '.format(gamma_total))
for key in gammas.keys():
    #n  = 2 if key == 'u' else 3
    br = 100 * gammas[key]/gamma_total
    print('BR ' + key + ' {:3.1f} %'.format(br))
```

### Sección eficaz

Aínda que non a calcularemos, imos presentar agora os distintos elementos que se necesitan na sección eficaz de $e+e^+ \to f + \bar{f}$ no $Z$, onde, lembremos, a contribución do $\gamma$ é pequena.

Sabemos que a sección eficaz vén dada por:

$$
\sigma = \frac{1}{64 \pi^2 s} \frac{p^*_f}{p^*_i} \int_\Omega \langle |M_{fi}|^2 \rangle \, \mathrm{d}\Omega
$$

Onde neste caso $p^*_i = p^*_f = m_Z/2$ se desprezamos as masas dos fermións en comparación con $m_Z$.

Ao estarmos tan próximos ao polo do $Z$ debemos incluír no propagador a parte asociada á súa anchura $\Gamma_Z$, que habitualmente desprezamos:

$$
\frac{1}{q^2 - m^2_Z} \to \frac{1}{q^2 - (m_Z - i \Gamma_Z/2)^2} \simeq \frac{1}{q^2 - m^2_Z +i m_z\Gamma_Z}
$$

que ao elevala ao cadrado nos dá a expresión:

$$
P_Z(s) \equiv \left|\frac{1}{q^2 - m^2_Z +i m_z\Gamma_Z} \right|^2 = \frac{1}{(s-m^2_Z)^2 + m^2_Z \Gamma^2_Z} 
$$

#### Inciso: sobre a anchura e a masa

O cálculo correcto faise en TCC; aquí só damos unha aproximación:

Se unha partícula con masa $m$ é estable, a súa función de onda en repouso é:

$$
\Psi \propto e^{-imt}.
$$

Se consideramos que é inestable cunha anchura $\Gamma = 1/\tau$, modificamos a función de onda:

$$
\Psi \propto e^{-imt} e ^{-\Gamma t/2}
$$

O que é equivalente a substituír:

$$
m \to m - i \Gamma/2
$$

o proceso $e+e^+\to \mu+\mu^+$, mediado só polo fotón, xa o estudamos (ver hadróns).

As distintas combinacións de helicidade son as mesmas:

| |

| As posibles combinacións de helicidade de $e^++e \to \mu^+ + \mu$ no CM no réxime relativista |

No elemento de matriz, onde antes nos apareceu $e^2$ agora aparecerá $g^2_Z$ e os acoplamentos $c_L, c_R$ correspondentes.

Os $M_{fi}$ conteñen os acoplamentos das correntes e as distribucións angulares seguintes:

|      | | | | 
| :--: | :--: | :--: | :--: |
| $LR \to LR$ | $LR \to RL$ | $RL \to LR$ | $RL \to RL$ |
| $g^2_Z s (1+\cos \theta)$| $g^2_z s (1-\cos \theta)$ | $g^2_Z s (1-\cos\theta)$ | $g^2_Z s (1+\cos\theta)$ | 
| $c_L(e) \, c_L(\mu)$ |$c_L(e) \, c_R(\mu)$ | $c_R(e) \, c_L(\mu)$| $c_R(e) \, c_R(\mu)$ |


Para obter a sección eficaz calculamos a media dos catro elementos de matriz e integramos no ángulo sólido.

Non facemos o cálculo, pero damos o resultado. 

A sección eficaz é:  

$$
\sigma (e+e^+\to Z \to \mu+\mu^+)= \frac{1}{64\pi^2 s} \frac{1}{4}\frac{g^4_z s^2}{(s-m^2_Z)^2 + m^2_Z\Gamma^2_Z} 
\frac{16\pi}{3} [c^2_L(e) + c^2_R(e)] [c^2_L(\mu) + c^2_R(\mu)] = \\
 \frac{1}{48\pi} \frac{g^4_z s}{(s-m^2_Z)^2 + m^2_Z\Gamma^2_Z}  [c^2_L(e) + c^2_R(e)] [c^2_L(\mu) + c^2_R(\mu)]
$$

Na segunda igualdade poden identificarse os distintos factores que interveñen.

Que podemos reescribir introducindo as anchuras de desintegración parciais:

$$
\Gamma_{ee} = \frac{g^2_z m_z}{24} [c^2_L(e) + c^2_R(e)] \\
\Gamma_{\mu\mu} = \frac{g^2_z m_z}{24} [c^2_L(\mu) + c^2_R(\mu)] \\
$$

Polo tanto:

$$
\sigma (e+e^+\to Z \to \mu+\mu^+)= \frac{12 \pi s}{m^2_Z} \frac{\Gamma_{ee} \Gamma_{\mu\mu}}{(s-m^2_Z)^2 + m^2_Z\Gamma^2_Z} 
$$


Observamos:
    
  * Trátase dunha función de tipo Breit-Wigner
  
  * No polo, $\sqrt{s} = m_z$, obtemos o máximo da sección eficaz:
  
  $$
  \sigma_{\max} = \frac{12\pi}{m^2_Z} \frac{\Gamma_{ee}\Gamma_{\mu\mu}}{\Gamma^2_Z} 
  $$
  
  * Nos valores $\sqrt{s} = m_Z \pm \Gamma_Z/2$ a sección eficaz:
  
  $$
  \sigma \left(\sqrt{s} = m_Z \pm \Gamma_Z/2 \right) = \frac{\sigma_{\max}}{2}
  $$
  
  Ou, o que é o mesmo, a anchura a media altura, FWHM, é $\Gamma_Z$

*Cuestión*: Calcula o máximo da sección eficaz para $e+e^+\to \mu+\mu^+$ e $e+e^+\to \mathrm{hadróns}$, e a curva da sección eficaz para hadróns no pico do $Z$.

```python
Gee = 0.0838 # GeV 
Gmm = 0.0838 # GeV
Guu = 0.2894 # GeV
Gdd = 0.3713 # GeV
Ghad = 2 * Guu + 3 * Gdd
mz  = 91.19  # GeV
Gz  =  2.45  # GeV
sigma_mumu = (12 * np.pi)/(mz**2) * (Gee * Gmm)/Gz**2
sigma_had  = (12 * np.pi)/(mz**2) * (Gee * Ghad)/Gz**2
hbarc  = 0.197 * units.femto # GeV m
barn =  1e-28 # m^2
sigma_to_barn = (hbarc**2) / barn
print('sigma max (ee->mumu) {:3.1f} nbarns '.format(sigma_to_barn * sigma_mumu / units.nano))
print('sigma max (ee->had) {:3.1f} nbarns '.format(sigma_to_barn * sigma_had /units.nano))
```

## O bosón de Higgs

A teoría electrofeble necesita do mecanismo de Higgs para ser coherente.

O modelo de Higgs dá resposta (quizais a única posible) aos seguintes problemas graves:

  * Os bosóns febles son masivos; porén, nas teorías gauge os bosóns non teñen masa. Como dotar de masa os bosóns $W^\pm, Z$?


  * A simetría electrofeble, U(1)$_Y$ SU(2)$_L$, separa os fermións á dereita e á esquerda, $f_L, f_R$, e esixe que non teñan masa.


  * outras seccións eficaces, $W^+ + W^- \to W^+ + W^-$, crecen indefinidamente a non ser que un escalar cos acoplamentos apropiados as controle.
  

O mecanismo de Higgs permite resolver os tres problemas dunha forma económica.

É imposible discutir sobre o mecanismo de Higgs sen recorrer á TCC e ao lagranxiano do SM.

Imos dar só unhas ideas que nos permitan entender o seu comportamento e descubrimento.

### O bosón de Higgs e as súas implicacións

A forma máis económica do mecanismo de Higgs é considerar a existencia dun campo escalar complexo, $\phi$, que interacciona consigo mesmo cun potencial $V(\phi)$.

O campo é un doblete de U(1)$_Y$SU(2)$_L$:

$$
\phi = \begin{pmatrix} \phi^+ \\ \phi^o \end{pmatrix}
$$
onde $\phi^0, \phi^+$ son campos complexos

O potencial $V(\Phi)$ coñécese como o potencial do sombreiro mexicano

$$
V(\phi) = \frac{\mu^2}{2} (\phi^\dagger \phi) + \frac{\lambda}{4} (\phi^\dagger \phi)^2
$$

A elección dun dos seus mínimos denomínase **rotura espontánea de simetría**.

Expresamos o campo de Higgs, $h(x)$, como unha perturbación respecto ao valor no baleiro: 

$$
\langle \phi(x) \rangle = \frac{1}{\sqrt{2}}\begin{pmatrix}  0 \\ v + h(x) \end{pmatrix}
$$

Esta elección coñécese como **gauge unitario**.

A elección non cambia a física subxacente, pero dá lugar aos campos físicos coñecidos.

*Nota adicional:* Matematicamente, dos 4 graos de liberdade iniciais, $\phi$, queda un, o bosón de Higgs, $h(x)$; os outros tres trasládanse ás compoñentes lonxitudinais dos bosóns $W^\pm, Z$.

Ao establecer a invariancia gauge, o campo de Higgs dá lugar ás interaccións:

 * entre os bosóns febles, $W^\pm, Z$, e o valor $v$ do baleiro, que interpretamos como a **masa dos bosóns $W^\pm, Z$**.
 
 * entre as dúas compoñentes de quiralidade dos fermións e o *vev*, o que dá lugar ás **masas dos fermións**.
 
 * entre o Higgs, $h(x)$, e os bosóns febles, $W, Z$, que dan lugar a **desintegracións do Higgs nos bosóns vectoriais**
 
 * entre o Higgs, $h(x)$, e os fermións, que á súa vez dan lugar a **desintegracións en pares fermión-antifermión**
 
*Nota adicional:* Ademais, o Higgs acóplase consigo mesmo en vértices de tres e catro ramas, e cos bosóns vectoriais febles en vértices de catro ramas.

### As masas

A fricción dos bosóns e dos fermións co *vev* dá lugar ás masas:

Na táboa móstrase como se relacionan as masas co *vev* e as constantes de acoplamento:

|  ---- fermión ---- |  ---- bosóns $W^\pm$  -----  | ------ bosón $Z$ ------ | ----- Higgs ----- |
| :--: | :--: | :--: | :--: |
| $m_f = \frac{1}{\sqrt{2}}\lambda_f v $ | $m_W = \frac{1}{2} g_W v$| $ m_Z = \frac{1}{2} \frac{g_W}{\cos \theta_W} v$ | $m_H = 2 \lambda v^2$|

Observamos:
    
   * Hai un parámetro do modelo, $\lambda_f$, que se chama **acoplamento de Yukawa**, para dar a masa de cada fermión, ademais do vev $v$. Hai tantos acoplamentos de Yukawa como fermións cargados!
    
   * Que a masa dos bosóns vectoriais, $m_W, m_Z$, está fixada por $v$ e polos acoplamentos, $g_W, g_Z$.
   
   * Que a masa do Higgs depende de $v^2$ e de $\lambda$. O parámetro $\lambda$ só intervén na masa do Higgs.
    
O mecanismo de Higgs establece unha relación entre as masas dos bosóns vectoriais:

$$
 \cos \theta_W = \frac{m_W}{m_Z}
$$

A partir dos datos experimentais, $M_z, \sin^2 \theta_W$, o SM predí:

$$
m_W \simeq 80.34 \;\; \mathrm{GeV}
$$

Que se aproxima moito ao valor experimental.

Esta relación deulle credibilidade ao mecanismo de Higgs antes de que se descubrise o Higgs.

*Nota adicional*: Na determinación da masa debemos considerar tamén diagramas de orde superior; as correccións que introducen son de tipo logarítmico, e a súa dependencia principal está relacionada coa masa do $t$.

Estas correccións predixeron unha masa do $t$ de ~175 GeV. O top descubriuse no Tevatron de Fermilab en 1995.

```python
s2t = units.value("weak mixing angle")
MW  = 80.378 # GeV 
MZ  = 91.1876 # GeV
cw = np.sqrt(1 - s2t)
mw_pred = MZ * cw
print(' Masa do W esperada a primeira orde {:4.3f} GeV, medida {:4.3f} GeV'.format(mw_pred, MW))
```

Finalmente, o valor do $vev$ pódese determinar a partir de $g_W, m_W$, ou simplemente de $G_F$:

$$
v = \frac{2m_W}{g_W} = \sqrt{\frac{1}{\sqrt{2} G_F}} = 246 \; \mathrm{GeV}
$$

#### Fermións

Os **fermións** son as partículas fundamentais, teñen espín 1/2 e son os constituíntes básicos da materia. 

Divídense en:

  * **quarks**: se interaccionan fortemente
  
  * **leptóns**: se non interaccionan fortemente
    
| forte | electromagnética | feble | 
|---      |:--       | :--          |
| quarks | quarks e leptóns cargados  | quarks e leptóns |

Os neutrinos só senten a forza feble.

#### Antipartículas

Os fermións rexense pola ecuación de Dirac. Son espinores de Dirac. Por cada fermión existe un antifermión.

Unha **antipartícula** ten as cargas e os números cuánticos opostos aos da súa partícula (a excepción do espín e da masa, que non cambian)

Se $f$ é un fermión, denotamos $\bar{f}$ o seu antifermión,
excepto para os leptóns cargados, onde indicamos o seu antileptón coa carga positiva: $e^+, \mu^+, \tau^+$

#### As tres xeracións de fermións

Os fermións (tanto leptóns como quarks) agrúpanse en **tres xeracións** ou familias. 

As tres xeracións de leptóns compórtanse con **universalidade** (da mesma maneira) fronte ás interaccións se se ten en conta que as súas masas son diferentes. 


En cada xeración, tanto leptóns como quarks agrúpanse en **dobletes** con posicións **arriba** e **abaixo**.

A súa posición está asociada á carga feble que se denomina **isospín feble**.

#### Creación de partículas

Para crear unha partícula necesitamos polo menos unha enerxía equivalente á masa da nova partícula. $E = m c^2$.

Un par fermión-antifermión pódese aniquilar (mediante o intercambio de bosóns neutros) para producir outro par fermión-antifermión. $(e^- + e^- \to \mu^- + \mu^+)$.

Sempre que producimos un antifermión producimos tamén o seu fermión. Sempre que creamos antimateria creamos materia.

Existen dous tipos principais de experimentos de creación de partículas:

  * colisionador: feixes de $e, e^+, p, \bar{p}$ colisionan entre si, por exemplo o [LEP](https://es.wikipedia.org/wiki/Large_Electron-Positron_collider) ($e^+e^-$)

  * branco fixo (*fixed target*): feixes de $e, p, \mu, \nu$ contra un material (átomos, núcleos). Experimentos de neutrinos, por exemplo o experimento [Gargamelle](https://en.wikipedia.org/wiki/Gargamelle).

 

#### Colisionadores

Colisionadores recentes:

| colisionador | Laboratorio | Tipo       | Período    | Enerxía, $\sqrt{s}$, (GeV) | Luminosidade, $\mathcal{L}$, ($cm^{-2}s^{-1})$ |
| :--      | :--        | :--        | :--       | :--              | :-- |
| PEP-II   | SLAC       | $e^+e^-$   | 1999-2009 | 10.5             | $1.2 \, 10^{34}$|
| LEP      | CERN       | $e^+e^-$   | 1989-2000 | 90-209           |  $10^{32}$ |
| Tevatron | Fermilab   | $p\bar{p}$ | 1987-2012 | 1960             | $4 \, 10^{32}$ |
| LHC      | CERN       | $p p$      | 2009-     | 14000            | $10^{34}$| 

Os dous parámetros fundamentais dun colisionador son:
    
 * **Enerxía** no centro de masas, $\sqrt{s}$, que limita a masa das partículas que podemos crear.

 * **Luminosidade**, ou o número de partículas que lanzamos unhas contra outras por unidade de área e tempo.
Delimita o número de interaccións interesantes en cada cruzamento. 




Para dous paquetes con $n_1, n_2$ partículas que se cruzan entre si na dirección $z$ cunha frecuencia $f$ e que teñen unha sección transversal gaussiana $\sigma_x, \sigma_y$, a luminosidade é:
 
 $$
 \mathcal{L} = f \frac{n_1 n_2}{ 4 \pi \sigma_x \sigma_y}
 $$
 
 Se un tipo de interacción ten unha sección eficaz $\sigma$, esperamos $n = \sigma \mathcal{L}$ interaccións por cruzamento. Lembrade que a sección eficaz $\sigma$ se mide en $m^2$ ou, en unidades naturais, en barns (1 barn = $10^{-24} cm^2$.

 E ao longo dun tempo (tipicamente anos), $N = \int_t \sigma \mathcal{L}(t) \, \mathrm{d}t$. 
 
 Onde $\mathcal{L}(t)$ é a luminosidade instantánea. Á cantidade $\int_t \mathcal{L}(t) \, \mathrm{d}t$ chamámola luminosidade integrada.

 Á interacción por cruzamento coñécese como **suceso**.

##### Propiedades dos portadores

A interacción entre dous fermións descríbese mediante o intercambio doutra partícula (virtual), que é o portador.

Cada portador intervén segundo a carga (eléctrica, isospín feble, cor) do fermión ao que se acopla.

Os portadores son bosóns vectoriais (espín 1); son os transmisores da forza.


A intensidade aproximada para dúas partículas a unha distancia de $10^{-15}$ m. 

| forza            | intensidade | bosón            | Q (e) | masa (GeV) |
| :--               | :--        | :--              | :--   | :--     | 
| forte            | 1          | $g$ (gluón)      | 0     | 0       | 
| electromagnetismo | $10^{-3}$  | $\gamma$ (fotón) | 0     | 0       | 
| feble             | $10^{-8}$  | $W^{\pm}$        | $\pm1$  | 80.4 | 
|                   |           | $Z^0$              | 0  | 91.2 |
| gravidade          | $10^{-37}$| 

A masa do bosón determina o alcance (en distancia) da interacción. O alcance das interaccións febles é de $10^{-15}$ m. 

O rango de $100$ GeV, aproximadamente a masa do bosón $Z$, coñécese como **escala electrofeble.**

Os bosóns $W^\pm$ están cargados. As súas interaccións coñécense como **correntes cargadas** (CC).

As interaccións do bosón $Z^0$ denomínanse tamén **correntes neutras** (NC). 

Existen 8 tipos de gluóns, que cambian a cor dos quarks, son electricamente neutros e de masa nula. 

Os quarks e os gluóns exhiben a propiedade de confinamento e non poden observarse illados. É por iso que, a pesar da súa masa nula, o alcance da forza forte é tamén de $10^{-15}$ m.

#### Elementos da interacción.

As ideas fundamentais sobre a interacción entre partículas:

  * prodúcese mediante un portador ou mediador (Yukawa).
  
  * ten lugar entre dúas correntes de fermións (Fermi).
  
  * Regra de Ouro de Fermi. A probabilidade da interacción é proporcional a $|M_{if}|^2$, onde $M_{if}$ é a matriz de transición entre os estados iniciais e finais.

#### Diagramas de árbore e de lazo

Nun diagrama de Feynman cada vértice introduce unha constante de acoplamento, $g$. 

A matriz de transición, $M_{if}$, dun diagrama de árbore será proporcional a $g^2$, e a probabilidade dunha interacción, a $|M_{if}|^2 \propto g^4$.

Nalgunhas ocasións utilízase a constante adimensional, $\alpha \propto g^2$, que caracteriza a intensidade dun diagrama. Para o electromagnetismo asociamos $\alpha$ á constante de estrutura fina:
$$
\alpha = \frac{e^2}{4 \pi \epsilon_0 \hbar c}
$$
