## Sobre a ecuación de Dirac

Dirac propuxo en 1928 unha ecuación que describe a dinámica dos fermións libres.

Construíu a súa ecuación de tal forma que fose invariante Lorentz, que aparecesen linealmente os operadores de enerxía e momento da mecánica cuántica,  $\hat{E} = i \partial_0, \; \hat{p} = -i \nabla$, e que contivese implicitamente a ecuación de Einstein, $E^2 = m^2 + p^2$.

A solución da ecuación de Dirac, $\Psi(x)$, é un novo tipo de obxecto físico, un **espinor**, que describe os fermións libres, partículas cun momento angular intrínseco ou espín de $1/2$. $\Psi(x)$ non é só unha función de ondas, senón catro, postas en forma de columna. Este novo obxecto transformarase cunhas regras específicas baixo transformacións Lorentz que, como veremos, se derivan da invariancia Lorentz da propia ecuación de Dirac. En particular, un espinor baixo rotacións regresa á súa forma orixinal tras xirar dúas veces, $4\pi$, isto é, un xiro aféctalle "a metade", de aí que o seu *espín* sexa 1/2. 

Dúas das solucións de Dirac tiñan enerxía negativa. O propio Dirac propuxo unha interpretación moi iluminadora do seu significado, indicando que as solucións de enerxía negativa se correspondían con antipartículas. A antipartícula sería a *ausencia* de partícula e, polo tanto, tería as cargas opostas. O descubrimento do positrón supuxo o respaldo definitivo da existencia das antipartículas e da ecuación de Dirac. 

As catro solucións independentes de Dirac corresponden a un fermión e ás súas dúas compoñentes de espín, por exemplo o electrón con carga $-e$, e ao antifermión, por exemplo o positrón, que ten a carga oposta, $+e$, e ás súas dúas compoñentes de espín. A ecuación de Dirac introduce de forma natural o **espín**, o momento angular intrínseco dunha partícula, e as **antipartículas**. Todo un logro!

Na ecuación aparecen ademais catro matrices, $\gamma^\mu$ con $\mu = 0, 1, 2, 3$, iguais en todos os sistemas inerciais, que, aínda que admiten diversas representacións, cumpren unhas regras de conmutación que permiten, como unha engrenaxe, ao asocialas ao momento e á enerxía da partícula, recuperar de novo a relación $E^2 = m^2 + p^2$. Podemos dicir que as matrices $\gamma$ asociadas cos operadores enerxía e momento están codificando, "encriptando", linealmente a famosa ecuación de Einstein.

A ecuación de Dirac é un dos grandes logros da Física Teórica: matematicamente é dunha simplicidade sorprendente e, ao mesmo tempo, dunha riqueza física desbordante.

No que segue desvelaremos algúns dos seus segredos e intentaremos asocialos aos elementos fundamentais das interaccións.

### Notación

* Unidades naturais, $\hbar = c = 1$. Métrica $g_{\mu\nu} = \mathrm{diag}(1, -1, -1, -1)$.
* Índices gregos $\mu, \nu = 0, 1, 2, 3$; latinos $i, j, k = 1, 2, 3$. Un índice repetido súmase.
* $x^\mu = (t, {\bf x})$, $p^\mu = (E, {\bf p})$, $p_\mu = (E, -{\bf p})$ e
  $p_\mu x^\mu = Et - {\bf p} \cdot {\bf x}$. As compoñentes físicas do momento son $p^k$.
* ${\bf p} = \mathrm{p} \, \hat{\bf v}$: $\mathrm{p} = |{\bf p}|$ é o módulo e $\hat{\bf v}$ a dirección.
* Operadores da mecánica cuántica: $\hat{H} = i\partial_t$, $\hat{\bf p} = -i\nabla$, isto é,
  $\hat{p}^\mu = i\partial^\mu$.
* Con ${\bf p}$ en $z$: $p^\mu = (E, 0, 0, \mathrm{p})$, e os dous números que aparecerán en todas partes,

$$
\kappa \equiv \frac{\mathrm{p}}{E+m}, \qquad N \equiv \sqrt{E+m}.
$$

* $I$ é a identidade ($2\times2$ ou $4\times4$, segundo o contexto); $\sigma_k$, as matrices de Pauli.
* Espinores: $u_s(p)$ (fermión) e $v_s(p)$ (antifermión), $s = 1, 2$; os seus bloques $2\times1$
  de arriba e de abaixo levan superíndice $a$, $b$. Biespinores base: $\chi_s$ e $\xi_s$. Con
  ${\bf p}$ en $z$ tamén $u_\pm$, $v_\pm$ (helicidade).
* Espín: ${\bf S} = \frac{1}{2}{\bf \Sigma}$. Transformación de Lorentz dun espinor: a matriz
  $S(\Lambda)$ (non confundir co espín).
* Rapidez dun *boost*: $\eta$. Ángulo dunha rotación arredor de $z$: $\phi$.

## 1. Como se constrúe a ecuación de Dirac?

**Punto de partida.** En mecánica cuántica a enerxía e o momento son operadores:

$$
\hat{H} = i\partial_t, \qquad \hat{\bf p} = -i\nabla.
$$

Se se substitúen en $E^2 = {\bf p}^2 + m^2$ obtense a ecuación de Klein-Gordon,
$-\partial^2_t \Psi = (-\nabla^2 + m^2)\Psi$. É de **segunda orde** no tempo, e a súa densidade
de probabilidade pode ser negativa. [MT4.1]

**A idea de Dirac:**

* primeira orde en $\partial_t$, como a ecuación de Schrödinger;
* a relatividade trata $t$ e ${\bf x}$ por igual: primeira orde tamén en $\nabla$;
* a ecuación de Einstein, $E^2 = {\bf p}^2 + m^2$, debe seguir cumpríndose.

A ecuación máis xeral **lineal** nas derivadas é:

$$
(i\gamma^0 \partial_t + i\gamma^k \partial_k - m) \, \Psi = 0,
$$

con catro coeficientes $\gamma^\mu$ que non coñecemos (o de $m$ tómase igual á unidade sen
perda de xeneralidade).

**Paso 1: de derivadas a enerxía e momento.** Cos operadores, $i\partial_t \to E$ e
$i\partial_k \to -p^k$:

$$
(\gamma^0 E - \gamma^k p^k) \, \Psi = m \, \Psi,
$$

que en forma covariante é $(\gamma^\mu p_\mu - m)\Psi = 0$.

**Paso 2: elevar ao cadrado.** Aplicando dúas veces o operador, $(\gamma^0 E - \gamma^k p^k)^2 \Psi = m^2 \Psi$.
Desenvolvemos sen cambiar a orde das $\gamma$, porque non sabemos se conmutan:

$$
(\gamma^0 E - \gamma^k p^k)^2 = (\gamma^0)^2 E^2 + \sum_k (\gamma^k)^2 (p^k)^2
+ \sum_{i<k} \{\gamma^i, \gamma^k\} \, p^i p^k - \sum_k \{\gamma^0, \gamma^k\} \, E \, p^k
$$

**Paso 3: esixir a ecuación de Einstein.** Para que isto sexa $E^2 - {\bf p}^2$ para calquera
$E$ e ${\bf p}$, termo a termo:

$$
(\gamma^0)^2 = I, \qquad (\gamma^k)^2 = -I, \qquad \{\gamma^0, \gamma^k\} = 0, \qquad
\{\gamma^i, \gamma^k\} = 0 \;\; (i \neq k),
$$

que se resume en:

$$
\boxed{
\{\gamma^\mu, \gamma^\nu\} \equiv \gamma^\mu \gamma^\nu + \gamma^\nu \gamma^\mu = 2 \, g^{\mu\nu} \, I
}
$$

As $\gamma$ **codifican linealmente** a ecuación de Einstein.

**Paso 4: as $\gamma^\mu$ son matrices e $\Psi$ ten catro compoñentes.** Dous números non nulos
non poden anticonmutar: as $\gamma^\mu$ son matrices, e $\Psi$ é unha columna sobre a que actúan.
De que dimensión?

* Fan falta **catro** matrices que anticonmuten dúas a dúas. Con $2\times2$ só hai tres, as
  de Pauli.
* As $\gamma^\mu$ teñen traza nula e autovalores $\pm 1$ (ou $\pm i$), logo a súa dimensión é par.

A dimensión mínima é $4$: $\Psi$ ten catro compoñentes e é un **espinor de Dirac**.

**Paso 5: hermiticidade.** Multiplicando o paso 1 por $\gamma^0$ pola esquerda obtense a
forma hamiltoniana, $E \, \Psi = H \, \Psi$, con

$$
H = \gamma^0 \gamma^k \, p^k + \gamma^0 \, m.
$$

$H$ debe ser hermítico (os seus autovalores, as enerxías, son reais). Logo
$(\gamma^0)^\dagger = \gamma^0$ e $(\gamma^0\gamma^k)^\dagger = (\gamma^k)^\dagger \gamma^0 = \gamma^0\gamma^k = -\gamma^k\gamma^0$, é dicir:

$$
\boxed{
(\gamma^0)^\dagger = \gamma^0, \qquad (\gamma^k)^\dagger = -\gamma^k
}
$$

**Paso 6: unha representación.** Só a álxebra está fixada; calquera conxunto de matrices que
a cumpra serve, e a física non depende da elección. A habitual, e a que usaremos, é a
**representación de Pauli-Dirac**, en bloques $2\times2$:

$$
\boxed{
\gamma^0 = \begin{pmatrix} I & 0 \\ 0 & -I \end{pmatrix}, \qquad
\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ -\sigma_k & 0 \end{pmatrix}
}
$$

coas matrices de Pauli

$$
\sigma_1 = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \;\;
\sigma_2 = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \;\;
\sigma_3 = \begin{pmatrix} 1 & 0 \\ 0 & -1\end{pmatrix}, \qquad
\{\sigma_i, \sigma_j\} = 2 \, \delta_{ij} \, I, \;\; [\sigma_i, \sigma_j] = 2i \, \epsilon_{ijk} \sigma_k.
$$

**Resultado.** Volvendo ás derivadas:

$$
\boxed{
(i \gamma^\mu \partial_\mu - m) \, \Psi(x) = 0
}
$$

onde $\partial_\mu \equiv \partial/\partial x^\mu = (\partial_t, \nabla)$ é o **cuadrigradiente**.
*Nota:* non confundir coa *derivada covariante* $D_\mu = \partial_\mu + i q A_\mu$, que aparece ao
impor a simetría *gauge*.

O prezo da linealidade —matrices e catro compoñentes— é o que nos vai dar o **espín** e
as **antipartículas**. [MT4.2]

*Cuestión:* comproba que as $\gamma^\mu$ de Pauli-Dirac cumpren a álxebra e a hermiticidade.

### Que son as matrices $\gamma$?

**Como o número $i$.** Con números reais, $x^2 + y^2$ non se pode escribir como produto de
dous factores lineais. Fai falta un número novo, $i$, con $i^2 = -1$:

$$
x^2 + y^2 = (x + iy)(x - iy).
$$

Dirac necesita o mesmo con $E^2 - {\bf p}^2$, unha suma de catro cadrados con signos, e ningún
número serve. Fan falta catro obxectos novos que:

* ao elevalos ao cadrado den o signo da métrica: $(\gamma^0)^2 = I$, $(\gamma^k)^2 = -I$;
* **anticonmuten** entre si, para que no cadrado se cancelen os termos cruzados
  ($E\,p^k$, $p^i p^k$).

Eses obxectos son as $\gamma^\mu$, e resultan ser matrices. En tres dimensións xa o coñecemos
polo espín en mecánica cuántica: $(\vec{\sigma} \cdot {\bf p})^2 = {\bf p}^2$, é dicir,
$\vec{\sigma} \cdot {\bf p}$ é a «raíz cadrada» de ${\bf p}^2$. Dirac fai o mesmo no
espazo-tempo, con catro matrices en lugar de tres:

$$
(\gamma^\mu p_\mu)^2 = p_\mu p^\mu = E^2 - {\bf p}^2 = m^2 .
$$

**A métrica escrita con matrices.** A álxebra

$$
\{\gamma^\mu, \gamma^\nu\} = 2 \, g^{\mu\nu} \, I
$$

di que as $\gamma^\mu$ se comportan como catro **vectores unitarios do espazo-tempo**:
o temporal de cadrado $+1$, os espaciais de cadrado $-1$, e «ortogonais» entre si (o
anticonmutador de dúas distintas é cero). Son a métrica de Minkowski, $g^{\mu\nu}$, escrita con
matrices. Por iso son as mesmas en todos os sistemas inerciais, e por iso $\gamma^\mu p_\mu$,
o cuadrimomento contraído con elas, ten como cadrado o invariante $m^2$.

## 2. Cales son as solucións? Fermión de masa $m$ que se move en $z$

Por simplicidade tomamos o momento lineal na dirección $z$, ${\bf p} = \mathrm{p}\,\hat{z}$: das
tres $\gamma^k$ só sobrevive $\gamma^3$, e por iso en todo o que segue aparece $\sigma_3$. O
caso xeral obtense cambiando $\sigma_3$ por $(\hat{\bf v}\cdot\vec{\sigma})$ (sección 5).

**Ansatz: onda plana.** Probamos unha columna de catro constantes, $w$, por unha onda plana:

$$
\Psi(x) = w \, e^{-i p_\mu x^\mu} = w \, e^{-i(Et - \mathrm{p} z)}.
$$

Como $i\partial_\mu \Psi = p_\mu \Psi$, a ecuación de Dirac vólvese **alxébrica**:

$$
(\gamma^\mu p_\mu - m) \, w = 0, \qquad \gamma^\mu p_\mu = \gamma^0 E - \gamma^3 \mathrm{p}.
$$

En bloques $2\times2$ (representación de Pauli-Dirac), cada termo é:

$$
\gamma^0 E = \begin{pmatrix} E \, I & 0 \\ 0 & -E \, I \end{pmatrix}, \qquad
-\gamma^3 \mathrm{p} = \begin{pmatrix} 0 & -\mathrm{p} \, \sigma_3 \\ \mathrm{p} \, \sigma_3 & 0 \end{pmatrix}, \qquad
-m = \begin{pmatrix} -m \, I & 0 \\ 0 & -m \, I \end{pmatrix},
$$

e sumando, con $w = (w^a, w^b)$ separado nos seus dous bloques de dúas compoñentes:

$$
\begin{pmatrix} (E-m) \, I & -\mathrm{p} \, \sigma_3 \\ \mathrm{p} \, \sigma_3 & -(E+m) \, I \end{pmatrix}
\begin{pmatrix} w^a \\ w^b \end{pmatrix} = 0 .
$$

Cada fila de bloques é unha ecuación; quedan **dúas ecuacións acopladas** entre $w^a$ e $w^b$:

$$
(E - m) \, w^a - \mathrm{p} \, \sigma_3 \, w^b = 0, \qquad
\mathrm{p} \, \sigma_3 \, w^a - (E + m) \, w^b = 0 .
$$

**Os dous signos da enerxía.** $(\gamma^\mu p_\mu - m) \, w = 0$ é un sistema lineal
**homoxéneo** de catro ecuacións. Se a matriz ten determinante non nulo, é invertible e a
única solución é $w = 0$: non hai onda. Para que exista unha solución $w \neq 0$ o determinante
ten que **anularse**.

Como $\sigma_3 = \mathrm{diag}(1, -1)$, o sistema sepárase en dous sistemas $2\times2$
independentes, un para as compoñentes $(w_1, w_3)$ e outro para $(w_2, w_4)$:

$$
\begin{pmatrix} E-m & -\mathrm{p} \\ \mathrm{p} & -(E+m) \end{pmatrix}
\begin{pmatrix} w_1 \\ w_3 \end{pmatrix} = 0, \qquad
\begin{pmatrix} E-m & +\mathrm{p} \\ -\mathrm{p} & -(E+m) \end{pmatrix}
\begin{pmatrix} w_2 \\ w_4 \end{pmatrix} = 0 .
$$

Os dous teñen o mesmo determinante,

$$
-(E-m)(E+m) + \mathrm{p}^2 = -(E^2 - \mathrm{p}^2 - m^2),
$$

e o da matriz $4\times4$ é o seu produto:

$$
\det(\gamma^\mu p_\mu - m) = (E^2 - \mathrm{p}^2 - m^2)^2 = 0
\quad \Rightarrow \quad E^2 = \mathrm{p}^2 + m^2 .
$$

**É a ecuación de Einstein!** A onda plana só é solución se $E$ e $\mathrm{p}$ son os dunha
partícula de masa $m$. Pero a ecuación fixa $E^2$, non $E$:

$$
E = \pm \sqrt{\mathrm{p}^2 + m^2},
$$

e cada raíz aparece dúas veces (unha por cada sistema $2\times2$, que serán os dous estados de
espín): **catro solucións**, dúas de enerxía positiva e dúas de enerxía negativa.

Vese tamén directamente nas ecuacións acopladas: despexando
$w^b = \frac{\mathrm{p}}{E+m} \sigma_3 w^a$ da segunda e substituíndo na primeira,

$$
\left(E - m - \frac{\mathrm{p}^2}{E+m}\right) w^a = \frac{E^2 - \mathrm{p}^2 - m^2}{E+m} \, w^a = 0,
$$

que con $w^a \neq 0$ esixe de novo $E^2 = \mathrm{p}^2 + m^2$.

As de enerxía negativa escríbense cambiando $E \to -E$ e $\mathrm{p} \to -\mathrm{p}$ na fase,
de forma que nas fórmulas $E$ sexa sempre positiva:

$$
\boxed{
\Psi_s(x) = u_s(p) \, e^{-i p_\mu x^\mu}, \qquad \Phi_s(x) = v_s(p) \, e^{+i p_\mu x^\mu}, \qquad
E = +\sqrt{\mathrm{p}^2 + m^2}, \;\; s = 1, 2
}
$$

Os espinores cumpren:

$$
\boxed{
(\gamma^\mu p_\mu - m) \, u_s(p) = 0, \qquad (\gamma^\mu p_\mu + m) \, v_s(p) = 0
}
$$

Cos operadores, $\Psi_s$ dá enerxía $+E$ e momento $+\mathrm{p}$; $\Phi_s$ dá $-E$ e $-\mathrm{p}$.
$\Phi_s$ é a **antipartícula** (ver *Que son as solucións $\Phi_s$?*, máis abaixo). [MT4.6]

**Espinores $u_s$.** Con $w = u$ e $E > 0$, as dúas ecuacións acopladas son:

$$
(E - m) \, u^a = \mathrm{p} \, \sigma_3 \, u^b, \qquad (E + m) \, u^b = \mathrm{p} \, \sigma_3 \, u^a.
$$

Da segunda, $u^b = \kappa \, \sigma_3 \, u^a$. A primeira non engade nada, porque
$\kappa \, \mathrm{p} = \mathrm{p}^2/(E+m) = E - m$.

O bloque de arriba é libre: tomamos os dous independentes,

$$
\chi_1 = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \qquad \chi_2 = \begin{pmatrix} 0 \\ 1 \end{pmatrix},
$$

e obtemos:

$$
u_s(p) = N \begin{pmatrix} \chi_s \\ \kappa \, \sigma_3 \, \chi_s \end{pmatrix}, \qquad
u_1(p) = N \begin{pmatrix} 1 \\ 0 \\ \kappa \\ 0 \end{pmatrix}, \;\;
u_2(p) = N \begin{pmatrix} 0 \\ 1 \\ 0 \\ -\kappa \end{pmatrix}.
$$

**Espinores $v_s$.** Con $+m$ a matriz é
$\begin{pmatrix} (E+m) \, I & -\mathrm{p} \, \sigma_3 \\ \mathrm{p} \, \sigma_3 & -(E-m) \, I \end{pmatrix}$, e
agora convén despexar o bloque de arriba da primeira fila: $v^a = \kappa \, \sigma_3 \, v^b$.

O bloque libre é o de abaixo. Tomamos

$$
\xi_1 = \begin{pmatrix} 0 \\ 1 \end{pmatrix}, \qquad \xi_2 = \begin{pmatrix} 1 \\ 0 \end{pmatrix},
$$

en orde cruzada, para que $s = 1$ sexa espín arriba tamén na antipartícula (ver sección 3):

$$
v_s(p) = N \begin{pmatrix} \kappa \, \sigma_3 \, \xi_s \\ \xi_s \end{pmatrix}, \qquad
v_1(p) = N \begin{pmatrix} 0 \\ -\kappa \\ 0 \\ 1 \end{pmatrix}, \;\;
v_2(p) = N \begin{pmatrix} \kappa \\ 0 \\ 1 \\ 0 \end{pmatrix}.
$$

**Normalización.** Tómase $u^\dagger_s u_s = v^\dagger_s v_s = 2E$ ($2E$ partículas por unidade de
volume, un convenio que se mantén baixo *boosts*):

$$
u^\dagger_s u_s = N^2 (1 + \kappa^2) = N^2 \, \frac{2E}{E+m} = 2E \quad \Rightarrow \quad N = \sqrt{E+m}.
$$

Na sección 4 veremos que este $N$ é xusto o que dá o *boost* do espinor en repouso.

**En repouso** ($\mathrm{p} = 0$, $\kappa = 0$, $N = \sqrt{2m}$) só sobreviven as compoñentes de
arriba en $u$ e as de abaixo en $v$:

$$
u_s(m) = \sqrt{2m} \begin{pmatrix} \chi_s \\ 0 \end{pmatrix}, \qquad
v_s(m) = \sqrt{2m} \begin{pmatrix} 0 \\ \xi_s \end{pmatrix}.
$$

Todo o movemento está en $\kappa$:

| caso | $\kappa$ | que queda |
|:--|:--:|:--|
| en repouso, $\mathrm{p} = 0$ | $0$ | un bloque: o espinor de Pauli de sempre |
| en movemento, $m > 0$ | $0 < \kappa < 1$ | o *boost* "acende" o outro bloque |
| sen masa, ou $E \gg m$ | $\to 1$ | os dous bloques pesan igual |

### Que son as solucións $\Phi_s$?

Cos operadores, $\Phi_s$ ten enerxía e momento **negativos**:

$$
i \partial_0 \, \Phi_s = -E \, \Phi_s, \qquad -i \partial_k \, \Phi_s = -p^k \, \Phi_s .
$$

Hai dúas lecturas clásicas.

#### O mar de Dirac

Dirac propuxo unha idea enxeñosa e iluminadora para explicar as solucións de enerxía negativa, baseada en que no baleiro todos os estados negativos están ocupados. Esta idea tamén se coñece como o mar de Dirac.

De tal forma que a enerxía mínima para crear un electrón, para facelo saltar da enerxía negativa, é de $2m_e$. Ao facer saltar un electrón do mar, neste queda un oco, que se comporta como unha partícula de carga positiva: o positrón.

#### Feynman e Stückelberg

Stückelberg (1941) e Feynman (1949) propuxeron que as antipartículas son "partículas que van
cara atrás no tempo".

A idea: na fase só aparecen os produtos $Et$ e ${\bf p}\cdot{\bf x}$, e cambiar o signo de
$E$ é indistinguible de cambiar o de $t$,

$$
e^{-i(-E)\,t} = e^{+iEt} = e^{-iE\,(-t)} .
$$

É unha igualdade, non unha transformación: a fase por si soa non distingue unha enerxía negativa que
avanza no tempo dun reloxo que vai cara atrás. A fase da partícula xira en sentido
horario; a de $\Phi_s$, en sentido antihorario, como o reloxo da partícula marchando cara atrás.

| |
|:--:|
|<img src="./imgs/dirac_reloj_fase.png" width = 420 align="center">|
| a fase no plano complexo: partícula (horario) e antipartícula (antihorario) |

Coa fase completa, $\Phi_s$ é unha partícula co que dan os operadores, enerxía $-E$ e momento
$-{\bf p}$:

$$
e^{+i(Et - {\bf p}\cdot{\bf x})} = e^{-i\left((-E)\,t \, - \, (-{\bf p})\cdot{\bf x}\right)} .
$$

A proposta de Feynman e Stückelberg: esa partícula de cuadrimomento $(-E, -{\bf p})$ propágase
**cara atrás no tempo**. Inverter o sentido de propagación inverte o fluxo de todo o
cuadrimomento, e o que se ve cara adiante no tempo é unha **antipartícula** de enerxía $+E$ e
momento $+{\bf p}$:

$$
(-E, -{\bf p}) \text{ cara atrás no tempo} \quad \equiv \quad (+E, +{\bf p}) \text{ cara adiante}.
$$

O que converte esta equivalencia formal en física é o **proceso**: emitir unha partícula de
enerxía $-E$ e momento $-{\bf p}$ cara adiante no tempo equivale a absorber unha antipartícula
de enerxía $+E$ e momento $+{\bf p}$, e viceversa. Por iso nos diagramas de Feynman a frecha da
liña dun antifermión apunta cara atrás no tempo. [MT4.7.1]

*Regra práctica:* nesta visión, as cantidades físicas da antipartícula lense en $\Phi_s$ invertendo o signo do que dan os operadores. Equivale a usar sobre $\Phi_s$ os operadores:

$$
\hat{H}^{(v)} = - i \partial_0, \;\; \hat{\bf p}^{(v)} = + i \nabla, \;\; \hat{\bf S}^{(v)} = - \hat{\bf S}
$$

Invértense a enerxía e o momento (todo o cuadrimomento, $p^\mu \to -p^\mu$), o momento angular (espín, orbital e total) e as cargas. A helicidade, ${\bf S} \cdot \hat{\bf p}$, non cambia, porque se inverten os seus dous factores; a quiralidade si, e de aí que para os antifermións se defina $v_R = P_L \, v$.

*Ollo:* non é a regra de «o que cambia ao inverter o tempo»: a enerxía e a carga non cambian baixo inversión temporal e, porén, aquí invértense. [MT4.7.3]

A interpretación das antipartículas da ecuación de Dirac é *controvertida*, e o seu sentido aparece de forma natural cando se cuantifica o campo de Dirac, isto é, en teoría cuántica de campos.

Tanto na teoría cuántica de campos como experimentalmente, a antipartícula é igual que a súa partícula en espín, masa e vida media, pero coas cargas opostas.


## 3. Que espín e que helicidade teñen as solucións? (${\bf p}$ en $z$)

O operador **espín** é

$$
\boxed{
{\bf S} \equiv \frac{1}{2} {\bf \Sigma}, \qquad \Sigma_k \equiv \begin{pmatrix} \sigma_k & 0 \\ 0 & \sigma_k \end{pmatrix}
}
$$

Con ${\bf p}$ en $z$, $S_3$ conmuta con $H$ e os catro espinores son autoestados de $S_3$.
Como $\Sigma_3 = \mathrm{diag}(1, -1, 1, -1)$:

$$
S_3 \, u_1 = +\tfrac{1}{2} \, u_1, \qquad S_3 \, u_2 = -\tfrac{1}{2} \, u_2, \qquad
S_3 \, v_1 = -\tfrac{1}{2} \, v_1, \qquad S_3 \, v_2 = +\tfrac{1}{2} \, v_2 .
$$

Na antipartícula o operador dá o valor **cambiado de signo**, igual que con $E$ e ${\bf p}$
(regra práctica de Feynman-Stückelberg): o espín físico de $v_1$ é $+\frac{1}{2}$ e o de $v_2$,
$-\frac{1}{2}$. De aí a orde cruzada dos $\xi_s$.

A **helicidade** é a proxección do espín na dirección do momento:

$$
\boxed{
h \equiv {\bf S} \cdot \hat{\bf v} = \frac{1}{2}
\begin{pmatrix} \hat{\bf v} \cdot \vec{\sigma} & 0 \\ 0 & \hat{\bf v} \cdot \vec{\sigma} \end{pmatrix}
}
$$

Con ${\bf p}$ en $z$, $h = S_3$. Na antipartícula invértense á vez o espín e o momento, e a
helicidade non cambia de signo. Así, os catro espinores son tamén autoestados de helicidade:

$$
u_+ = u_1, \quad u_- = u_2, \quad v_+ = v_1, \quad v_- = v_2 .
$$

$u_{+}(p), v_{+}(p)$ son espinores de **helicidade positiva** (o espín ten o mesmo sentido que o momento).

$u_{-}(p), v_{-}(p)$ teñen **helicidade negativa** (o espín e o momento teñen sentidos opostos).

## 4. Como se rota un espinor e como se lle fai un *boost*? (en $z$)

**A regra.** Baixo unha transformación de Lorentz $\Lambda$ (rotación ou *boost*), un cuadrivector
transfórmase con $\Lambda$, $p^\mu \to \Lambda^\mu{}_\nu \, p^\nu$. Un espinor faino cunha
matriz complexa $4\times4$, $S(\Lambda)$:

$$
\Psi'(x) = S(\Lambda) \, \Psi(\Lambda^{-1} x), \qquad
u(p) \;\to\; S(\Lambda) \, u(p) \;\; \text{(un espinor de momento } \Lambda p \text{)}.
$$

$S(\Lambda)$ queda fixada ao esixir que a ecuación de Dirac teña a mesma forma en todos os
sistemas inerciais, $S^{-1}\gamma^\mu S = \Lambda^\mu{}_\nu \, \gamma^\nu$ (a dedución está en
[ext-trans-Lorentz](./ext-trans-Lorentz.ipynb)). En $z$:

* **rotación** de ángulo $\phi$ arredor de $z$ (activa, en sentido antihorario). O xerador é o
  espín:

$$
\boxed{
S(\phi \, \hat{z}) = e^{-i \phi \, S_3} = \cos\frac{\phi}{2} \, I - i \sin\frac{\phi}{2} \, \Sigma_3
}
$$

* ***boost*** de rapidez $\eta$ en $z$:

$$
\boxed{
S(\eta \, \hat{z}) = e^{\frac{\eta}{2} \gamma^0 \gamma^3} = \cosh\frac{\eta}{2} \, I + \sinh\frac{\eta}{2} \, \gamma^0 \gamma^3
}
$$

A serie da exponencial súmase porque $(\Sigma_3)^2 = I$ e $(\gamma^0\gamma^3)^2 = I$. O
importante: **os ángulos aparecen divididos por dous**, $\phi/2$ e $\eta/2$.

Na representación de Pauli-Dirac ($\gamma^0\gamma^3 = \begin{pmatrix} 0 & \sigma_3 \\ \sigma_3 & 0 \end{pmatrix}$):

$$
S(\phi \, \hat{z}) = \begin{pmatrix} e^{-i\frac{\phi}{2}\sigma_3} & 0 \\ 0 & e^{-i\frac{\phi}{2}\sigma_3} \end{pmatrix}, \qquad
S(\eta \, \hat{z}) = \begin{pmatrix} \cosh\frac{\eta}{2} \, I & \sinh\frac{\eta}{2} \, \sigma_3 \\ \sinh\frac{\eta}{2} \, \sigma_3 & \cosh\frac{\eta}{2} \, I \end{pmatrix}
$$

A rotación non mestura os bloques de arriba e de abaixo; o *boost* si: é o *boost* o que
"acende" o segundo bloque.

### 4.1. Rotación arredor de $z$ dos espinores $u_s$ con ${\bf p}$ en $z$

Unha rotación arredor de $z$ non cambia ${\bf p} = \mathrm{p}\,\hat{z}$: o espinor rotado debe ser
unha solución co mesmo momento. Como $\sigma_3$ é diagonal,
$e^{-i\frac{\phi}{2}\sigma_3} = \mathrm{diag}(e^{-i\phi/2}, e^{+i\phi/2})$ e

$$
S(\phi \, \hat{z}) = \mathrm{diag}\left(e^{-i\phi/2}, \, e^{+i\phi/2}, \, e^{-i\phi/2}, \, e^{+i\phi/2}\right).
$$

Sobre os espinores:

$$
S(\phi \, \hat{z}) \, u_1(p) = N \begin{pmatrix} e^{-i\phi/2} \\ 0 \\ \kappa \, e^{-i\phi/2} \\ 0 \end{pmatrix} = e^{-i\phi/2} \, u_1(p), \qquad
S(\phi \, \hat{z}) \, u_2(p) = e^{+i\phi/2} \, u_2(p),
$$

isto é,

$$
\boxed{
S(\phi \, \hat{z}) \, u_s(p) = e^{-i m_s \phi} \, u_s(p), \qquad m_s = \pm\tfrac{1}{2}
}
$$

**Que se le:**

* Rotar arredor do eixe do espín non cambia o estado: só lle engade unha fase, como a calquera
  autoestado de $J_z$ en mecánica cuántica, $e^{-i\phi J_z}|j, m\rangle = e^{-im\phi}|j, m\rangle$.
* A fase vai con **medio ángulo**: unha volta, $\phi = 2\pi$, dá $S = -I$ e $u_s \to -u_s$. Fan
  falta **dúas voltas**, $4\pi$, para volver ao espinor de partida. Unha fase global non se ve, pero
  o signo si se mediu, facendo interferir dous feixes de neutróns dos que só un se
  rota. [Rauch et al., Phys. Lett. A 54 (1975) 425; Werner et al., PRL 35 (1975) 1053]
* Na antipartícula a fase vai ao revés: $S(\phi\,\hat{z}) \, v_1 = e^{+i\phi/2} \, v_1$, coherente
  co seu espín físico $+\frac{1}{2}$ (o operador dá o signo cambiado).

### 4.2. *Boost* en $z$ do espinor en repouso

Partimos de $u_s(m) = \sqrt{2m} \, (\chi_s, 0)$ e facémoslle un *boost* en $z$ de rapidez $\eta$,
a que leva a partícula do repouso ao momento $\mathrm{p}$:

$$
\cosh\eta = \frac{E}{m}, \qquad \sinh\eta = \frac{\mathrm{p}}{m}.
$$

Coas fórmulas do ángulo metade:

$$
\cosh\frac{\eta}{2} = \sqrt{\frac{\cosh\eta + 1}{2}} = \sqrt{\frac{E+m}{2m}}, \qquad
\sinh\frac{\eta}{2} = \sqrt{\frac{\cosh\eta - 1}{2}} = \sqrt{\frac{E-m}{2m}}, \qquad
\tanh\frac{\eta}{2} = \sqrt{\frac{E-m}{E+m}} = \frac{\mathrm{p}}{E+m} = \kappa .
$$

Aplicando $S(\eta\,\hat{z})$:

$$
S(\eta \, \hat{z}) \, u_s(m) = \sqrt{2m} \begin{pmatrix} \cosh\frac{\eta}{2} \, \chi_s \\ \sinh\frac{\eta}{2} \, \sigma_3 \, \chi_s \end{pmatrix}
= \sqrt{2m} \, \cosh\frac{\eta}{2} \begin{pmatrix} \chi_s \\ \tanh\frac{\eta}{2} \, \sigma_3 \, \chi_s \end{pmatrix}
= \sqrt{E+m} \begin{pmatrix} \chi_s \\ \kappa \, \sigma_3 \, \chi_s \end{pmatrix}
$$

$$
\boxed{
S(\eta \, \hat{z}) \, u_s(m) = u_s(p)
}
$$

É o espinor da sección 2, obtido agora **sen resolver a ecuación**.

**Que se le:**

* $\kappa = \tanh(\eta/2)$: é a tanxente da **metade da rapidez**. Vai de $0$ (repouso) a $1$
  ($\eta \to \infty$), e por iso mide canto de relativista é o fermión.
* $N = \sqrt{E+m}$ non é arbitrario: é o que dá o *boost* partindo de $\sqrt{2m}$ en repouso.
* O *boost* é ao longo do eixe do espín e non cambia $S_3$: $u_1(m)$, espín arriba, pasa a ser
  $u_1(p) = u_+(p)$, de helicidade positiva.
* O *boost* **non é unitario**: $u^\dagger u$ pasa de $2m$ a $2E$ (a densidade crece como $E/m$,
  porque o volume se contrae). O que non cambia é $\bar{u}u = u^\dagger\gamma^0 u = N^2(1-\kappa^2) = 2m$.
  Por iso fai falta o espinor adxunto, $\bar{u} = u^\dagger \gamma^0$ (ver *O espinor adxunto*).
* O mesmo para a antipartícula: $S(\eta\,\hat{z}) \, v_s(m) = \sqrt{2m} \, (\sinh\frac{\eta}{2}\,\sigma_3\,\xi_s, \; \cosh\frac{\eta}{2}\,\xi_s) = v_s(p)$.

2. Aplica $S(-\eta\,\hat{z})$ a $u_s(p)$. Que obtés?

## 5. Rotacións e *boosts* nunha dirección calquera

As fórmulas da sección 4 xeneralízanse cambiando o eixe $z$ por un eixe unitario calquera, $\hat{\bf n}$:

* **rotación** de ángulo $\alpha$ arredor de $\hat{\bf n}$:

$$
\boxed{
S(\alpha \, \hat{\bf n}) = e^{-i \alpha \, \hat{\bf n} \cdot {\bf S}}
= \cos\frac{\alpha}{2} \, I - i \sin\frac{\alpha}{2} \, (\hat{\bf n} \cdot {\bf \Sigma})
}
$$

* ***boost*** de rapidez $\eta$ na dirección $\hat{\bf n}$:

$$
\boxed{
S(\eta \, \hat{\bf n}) = e^{\frac{\eta}{2} \gamma^0 (\hat{\bf n} \cdot \vec{\gamma})}
= \cosh\frac{\eta}{2} \, I + \sinh\frac{\eta}{2} \, \gamma^0 (\hat{\bf n} \cdot \vec{\gamma})
}
$$

porque de novo $(\hat{\bf n} \cdot {\bf \Sigma})^2 = I$ e $[\gamma^0 (\hat{\bf n} \cdot \vec{\gamma})]^2 = I$.

### 5.1. Rotación nunha dirección calquera

**Rotar o espín.** Levamos o eixe $z$ a unha dirección calquera
$\hat{\bf v} = (\sin\theta\cos\phi, \, \sin\theta\sin\phi, \, \cos\theta)$ con dúas rotacións:
primeiro un ángulo $\theta$ arredor de $y$ e despois un ángulo $\phi$ arredor de $z$,

$$
S(R) \equiv S(\phi\,\hat{z}) \, S(\theta\,\hat{y}), \qquad
S(\theta\,\hat{y}) = \begin{pmatrix} e^{-i\frac{\theta}{2}\sigma_2} & 0 \\ 0 & e^{-i\frac{\theta}{2}\sigma_2} \end{pmatrix}, \qquad
e^{-i\frac{\theta}{2}\sigma_2} = \begin{pmatrix} \cos\frac{\theta}{2} & -\sin\frac{\theta}{2} \\ \sin\frac{\theta}{2} & \cos\frac{\theta}{2} \end{pmatrix}.
$$

Como a rotación non mestura bloques, abonda con ver que lles fai a $\chi_1$ e $\chi_2$:

$$
\chi_1 \xrightarrow{\;\theta\,\hat{y}\;} \begin{pmatrix} \cos\frac{\theta}{2} \\ \sin\frac{\theta}{2} \end{pmatrix}
\xrightarrow{\;\phi\,\hat{z}\;} e^{-i\phi/2} \begin{pmatrix} \cos\frac{\theta}{2} \\ e^{i\phi}\sin\frac{\theta}{2} \end{pmatrix} = e^{-i\phi/2} \, \chi_+(\hat{\bf v}),
$$

$$
\chi_2 \xrightarrow{\;\theta\,\hat{y}\;} \begin{pmatrix} -\sin\frac{\theta}{2} \\ \cos\frac{\theta}{2} \end{pmatrix}
\xrightarrow{\;\phi\,\hat{z}\;} e^{-i\phi/2} \begin{pmatrix} -\sin\frac{\theta}{2} \\ e^{i\phi}\cos\frac{\theta}{2} \end{pmatrix} = e^{-i\phi/2} \, \chi_-(\hat{\bf v}).
$$

Saen, **sen resolver ningunha ecuación de autovalores**, os biespinores de helicidade da
sección *Helicidade nunha dirección calquera*, cos mesmos ángulos. A fase común
$e^{-i\phi/2}$ é global e omítese (alí elixiuse real a compoñente de arriba). En repouso,
$S(R) \, u_1(m) \propto \sqrt{2m} \, (\chi_+(\hat{\bf v}), 0)$: o espín apunta agora en $\hat{\bf v}$,
$(\hat{\bf v}\cdot\vec{\sigma}) \, \chi_\pm = \pm \chi_\pm$.

**Rotar o fermión en movemento.** Aplicamos a mesma rotación a $u_\pm(p)$ con ${\bf p}$ en $z$.
Agora xira tamén o momento, $\mathrm{p}\,\hat{z} \to \mathrm{p}\,\hat{\bf v}$, e $\sigma_3$ convértese na
compoñente de $\vec\sigma$ ao longo do eixe rotado, $R \, \sigma_3 \, R^{-1} = \hat{\bf v}\cdot\vec{\sigma}$
(con $R = e^{-i\frac{\phi}{2}\sigma_3} e^{-i\frac{\theta}{2}\sigma_2}$). Resultado:

$$
S(R) \, u_\pm(\mathrm{p}\,\hat{z}) \propto N \begin{pmatrix} \chi_\pm(\hat{\bf v}) \\ \pm\kappa \, \chi_\pm(\hat{\bf v}) \end{pmatrix} = u_\pm(\mathrm{p}\,\hat{\bf v}),
$$

que son os $u_r(p)$ de helicidade $r = \pm$. **A helicidade non cambia baixo rotacións**: xiran á
vez o espín e o momento.

Para a antipartícula, $\xi_1 = \chi_2$ e $\xi_2 = \chi_1$, logo $\xi_1 \to \chi_-(\hat{\bf v}) = \xi_+(\hat{\bf v})$ e
$\xi_2 \to \xi_-(\hat{\bf v})$: obtéñense igualmente os $v_\pm(\mathrm{p}\,\hat{\bf v})$.

### 5.2. *Boost* nunha dirección calquera

Como $\gamma^0\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ \sigma_k & 0 \end{pmatrix}$,
o *boost* de rapidez $\eta$ en $\hat{\bf v}$ é o de $z$ cambiando $\sigma_3$ por $\hat{\bf v}\cdot\vec{\sigma}$:

$$
S(\eta\,\hat{\bf v}) = \begin{pmatrix} \cosh\frac{\eta}{2} \, I & \sinh\frac{\eta}{2} \, (\hat{\bf v}\cdot\vec{\sigma}) \\ \sinh\frac{\eta}{2} \, (\hat{\bf v}\cdot\vec{\sigma}) & \cosh\frac{\eta}{2} \, I \end{pmatrix}
= \sqrt{\frac{E+m}{2m}} \begin{pmatrix} I & \kappa \, (\hat{\bf v}\cdot\vec{\sigma}) \\ \kappa \, (\hat{\bf v}\cdot\vec{\sigma}) & I \end{pmatrix}.
$$

Sobre o espinor en repouso, con calquera $\chi$:

$$
\boxed{
S(\eta\,\hat{\bf v}) \, \sqrt{2m} \begin{pmatrix} \chi \\ 0 \end{pmatrix} = N \begin{pmatrix} \chi \\ \kappa \, (\hat{\bf v}\cdot\vec{\sigma}) \, \chi \end{pmatrix}
}
$$

que é a solución xeral $u_s(p)$ con ${\bf p} = \mathrm{p}\,\hat{\bf v}$. O *boost* non toca o bloque de
arriba: o espín da solución é o que tiña en repouso.

* Se en repouso o espín está en $z$ ($\chi_1$) e o *boost* vai en $\hat{\bf v} \neq \hat{z}$, o resultado
  **non** é autoestado de helicidade: espín en $z$, momento en $\hat{\bf v}$.
* Se en repouso o espín xa apunta en $\hat{\bf v}$ ($\chi_\pm(\hat{\bf v})$), o resultado é $u_\pm(\mathrm{p}\,\hat{\bf v})$.

### 5.3. Dous camiños, o mesmo espinor

Rotar e logo acelerar, ou acelerar e logo rotar:

$$
u_1(m) \xrightarrow{\;S(\eta\,\hat{z})\;} u_+(\mathrm{p}\,\hat{z}) \xrightarrow{\;S(R)\;} u_+(\mathrm{p}\,\hat{\bf v}),
\qquad
u_1(m) \xrightarrow{\;S(R)\;} \sqrt{2m}\begin{pmatrix}\chi_+(\hat{\bf v}) \\ 0\end{pmatrix} \xrightarrow{\;S(\eta\,\hat{\bf v})\;} u_+(\mathrm{p}\,\hat{\bf v}),
$$

porque $S(\eta\,\hat{\bf v}) \, S(R) = S(R) \, S(\eta\,\hat{z})$: un *boost* en $\hat{\bf v}$ é o *boost* en
$z$ visto desde o sistema rotado. (Agás a fase global $e^{-i\phi/2}$.)

## O espinor adxunto, a norma e a corrente

En mecánica cuántica a nosa función de ondas está normalizada, e toda transformación física preserva a norma. Ademais existe unha corrente de probabilidade que cumpre a ecuación de continuidade.

Imos identificar agora a norma e a corrente para as solucións da ec. de Dirac, pero para iso necesitamos introducir antes o espinor adxunto.


### O espinor adxunto

Definimos o espinor adxunto como:

$$
\bar{\Psi}(x) \equiv \Psi^\dagger(x) \gamma^0
$$

Na representación de Pauli-Dirac:

$$
\bar{\Psi}(x) = (\psi^*_1(x), \psi^*_2(x), -\psi^*_3(x), -\psi^*_4(x)),
$$

que ten forma de vector fila.

### A ecuación de Dirac para o espinor adxunto


A ec. de Dirac para o espinor adxunto obtense tras aplicar o transposto conxugado á ec. de Dirac:

$$
- i \partial_\mu \Psi^\dagger \, (\gamma^\mu)^\dagger - m \Psi^\dagger = 0 \\
$$

se primeiro usamos $(\gamma^0)^\dagger = \gamma^0, \, (\gamma^k)^\dagger = - \gamma^k$, logo multiplicamos por $\gamma^0$ pola dereita, e aplicando $\gamma^k\gamma^0 = - \gamma^0\gamma^k$, obtemos:

$$
-i\partial_0 \Psi^\dagger \gamma^0 + i \partial_k \Psi^\dagger \gamma^k - m \Psi^\dagger = 0 \\
-i\partial_0 \Psi^\dagger \gamma^0 \gamma^0 + i \partial_k \Psi^\dagger \gamma^k  \gamma^0 - m \Psi^\dagger \gamma^0 = 0 \\
-i\partial_0 \Psi^\dagger \gamma^0 \gamma^0 - i \partial_k \Psi^\dagger \gamma^0  \gamma^k - m \Psi^\dagger \gamma^0 = 0 \\
-i\partial_\mu \bar{\Psi} \gamma^\mu - m \bar{\Psi}= 0 \\
$$

Finalmente:

$$
i\partial_\mu \bar{\Psi} \, \gamma^\mu + m \bar{\Psi}= 0
$$

que é a ecuación de Dirac para o espinor adxunto.


### Norma

A norma asociada ás solucións da ecuación de Dirac é:

$$
\bar{\Psi}(x)\Psi(x)
$$

que é un escalar Lorentz, preserva a unitariedade (ver [ext-trans-Lorentz](./ext-trans-Lorentz.ipynb)), e cuxo valor (por convenio) se toma como $2m$ para os fermións, $\bar{u}u = 2m$, e $-2m$ para os antifermións, $\bar{v}v = -2m$.


### Corrente fermiónica de probabilidade

A corrente de probabilidade fermiónica é:

$$
\bar{\Psi} \gamma^\mu \Psi = (\rho, {\bf j})
$$

onde $\rho$ é a densidade e ${\bf j}$ será a corrente de probabilidade do espinor de Dirac.

Nótese que os catro elementos da corrente corresponden aos valores do índice $\mu$.

Na extensión [ext-trans-Lorentz] demóstrase que efectivamente a corrente se transforma como un vector baixo transformacións de Lorentz. Vexamos aquí que cumpre a ecuación de conservación:

$$
\frac{\partial \rho}{\partial t} + \nabla \cdot {\bf j} = 0.
$$

Ou en forma covariante:

$$
\partial_\mu j^\mu = 0
$$


A densidade é:

$$
\rho = \bar{\Psi} \gamma^0 \Psi = \Psi^\dagger \Psi = \sum_{i = 1}^4 |\psi_i|^2
$$

Podemos comprobar que é certo:

$$
\partial_\mu \left(\bar{\Psi} \gamma^\mu \Psi\right) = (\partial_\mu \bar{\Psi}) \gamma^\mu \Psi + \bar{\Psi} \gamma^\mu (\partial_\mu\Psi)
$$

Das ecuacións de Dirac do espinor e do espinor adxunto, temos que:

$$
\partial_\mu \bar{\Psi} \gamma^\mu = -i m \bar{\Psi}, \;\;\; \gamma^\mu \partial_\mu \Psi = +i m \Psi
$$

Se substituímos:

$$
(\partial_\mu \bar{\Psi} \gamma^\mu) \Psi + \bar{\Psi} (\gamma^\mu \partial_\mu\Psi) = (-i m \bar{\Psi}) \Psi + \bar{\Psi} (i m \Psi) = 0.
$$

Logo:

$$
\partial_\mu \left(\bar{\Psi} \gamma^\mu \Psi\right) = 0, \;\; \partial_\mu j^\mu = 0.
$$

Concluímos que $j^\mu = \bar{\Psi} \gamma^\mu \Psi$ é unha corrente conservada, que chamaremos corrente fermiónica.


## Helicidade nunha dirección calquera

Para un momento nunha dirección calquera, ${\bf p} = \mathrm{p} \, \hat{\bf v}$, todo o da
sección 2 vale cambiando $\sigma_3$ por $(\hat{\bf v} \cdot \vec{\sigma})$:

$$
u_s(p) = N \begin{pmatrix} \chi_s \\ \kappa \, (\hat{\bf v} \cdot \vec{\sigma}) \, \chi_s \end{pmatrix}, \qquad
v_s(p) = N \begin{pmatrix} \kappa \, (\hat{\bf v} \cdot \vec{\sigma}) \, \xi_s \\ \xi_s \end{pmatrix},
$$

con

$$
(\hat{\bf v} \cdot \vec{\sigma}) = \begin{pmatrix} v_z & v_x - i v_y \\ v_x+iv_y & -v_z \end{pmatrix} =
\begin{pmatrix} \cos\theta & \sin\theta \, e^{-i\phi} \\ \sin\theta \, e^{i\phi} & -\cos\theta \end{pmatrix},
\qquad (\hat{\bf v} \cdot \vec{\sigma})^2 = I,
$$

onde $\theta, \phi$ son os ángulos polar e azimutal de $\hat{\bf v}$.

Pero agora $\chi_1, \chi_2$ xa non son autoestados de helicidade: teñen o espín en $z$ e o momento
en $\hat{\bf v}$.

### Solucións autoestados de helicidade

Podemos preguntarnos agora cales son as solucións da ec. de Dirac que son á súa vez autoestados de helicidade, isto é, que teñen o seu espín aliñado co momento.

Os autovalores de helicidade, como os da terceira compoñente de espín, son $\pm 1/2$. 

$$
h \, u_s(p) = \pm \frac{1}{2} u_s(p)
$$

O que se traduce en:

$$
\frac{1}{2} \begin{pmatrix} (\vec{\sigma} \cdot \hat{\bf v}) & 0 \\ 0  & (\vec{\sigma} \cdot \hat{\bf v}) \end{pmatrix} \, 
\begin{pmatrix} \chi_r \\ \kappa (\vec{\sigma} \cdot \hat{\bf v}) \chi_r  \end{pmatrix} = \pm \frac{1}{2} \, \begin{pmatrix} \chi_r \\  \kappa (\vec{\sigma} \cdot \hat{\bf v}) \chi_r\end{pmatrix}
$$

Isto tradúcese en dúas ecuacións, pero a segunda é unha identidade se se cumpre a primeira:

$$
(\sigma \cdot \hat{\bf v}) \, \chi_r(\hat{\bf v}) =  \pm \, \chi_r(\hat{\bf v}) \\
\kappa \, (\sigma \cdot \hat{\bf v}) \left[ (\sigma \cdot \hat{\bf v}) \, \chi_r(\hat{\bf v}) \right]=  \pm \kappa \, (\sigma \cdot \hat{\bf v}) \, \chi_r(\hat{\bf v})
$$



Imos atopar, pois, o biespinor, $\chi_r(\hat{\bf v})$, que cumpre

$$
(\sigma \cdot \hat{\bf v}) \, \chi_r(\hat{\bf v}) =  \pm \, \chi_r(\hat{\bf v})
$$

que vemos que depende agora de $\hat{\bf v}$. 

Se o escribimos explicitamente:

$$
(\vec{\sigma} \cdot \hat{\bf v}) = \begin{pmatrix} v_z & v_x - i v_y \\ v_x+iv_y & -v_z \end{pmatrix} = 
\begin{pmatrix} \cos\theta & \sin\theta e^{-i\phi} \\ \sin\theta e^{i\phi} & -\cos\theta \end{pmatrix}
$$

A condición das dúas compoñentes $a, b$ de $\chi_r(\hat{\bf v})$ é:

$$
\begin{pmatrix} \cos\theta & \sin\theta e^{-i\phi} \\ \sin\theta e^{i\phi} & -\cos\theta \end{pmatrix}
\begin{pmatrix} a \\ b \end{pmatrix} = 
\pm \begin{pmatrix} a \\ b \end{pmatrix}
$$



Isto é, con $\lambda = \pm 1$:

$$
a \cos \theta + b \sin \theta e^{-i\phi}  = \lambda \, a, \\
b \sin \theta e^{-i\phi}  = (\lambda -  \cos\theta) \, a \\
$$

Polo tanto, as compoñentes $a, b$ cumpren:

$$
\frac{b}{a} = \frac{\lambda -\cos \theta}{\sin\theta} e^{i\phi}
$$


Para o caso de helicidade positiva, $\lambda = +1$, obtemos:

$$
\frac{b}{a} = \frac{\lambda -\cos \theta}{\sin\theta} e^{i\phi} = e^{i\phi} \tan \frac{\theta}{2}
$$

Resolvendo tamén para $\lambda = -1$, impoñendo que teñan norma unidade, e facendo uso de que sen perda de xeneralidade podemos tomar a compoñente de arriba, $a$, como real, os biespinores quedan:

$$
\chi_+(\hat{\bf v}) = \begin{pmatrix} \cos \frac{\theta}{2} \\ e^{i\phi}\sin \frac{\theta}{2}\end{pmatrix}, \;\;\;
\chi_-(\hat{\bf v}) = \begin{pmatrix} -\sin \frac{\theta}{2} \\ e^{i\phi}\cos \frac{\theta}{2}\end{pmatrix}, \;\;\;
$$



Nótese que o biespinor definido antes, $\chi_s$,  é simplemente $\chi_r(\hat{k})$, que corresponde a seleccionar como dirección $\hat{\bf v} = \hat{k}$.

De forma similar, os biespinores dos antifermións que son autoestados de helicidade quedan:

$$
\xi_+(\hat{\bf v}) = \begin{pmatrix} -\sin \frac{\theta}{2} \\ e^{i\phi}\cos \frac{\theta}{2}\end{pmatrix}, \;\;\;
\xi_-(\hat{\bf v}) = \begin{pmatrix} \cos \frac{\theta}{2} \\ e^{i\phi}\sin \frac{\theta}{2}\end{pmatrix}, \;\;\;
$$

Así pois, os espinores autoestados de helicidade que son solucións da ec. de Dirac son entón:

$$
u_r(p) = N \begin{pmatrix} \chi_r(\hat{\bf v}) \\ \lambda_r \kappa \, \chi_r(\hat{\bf v}) \end{pmatrix}, \;\;\;
v_r(p) = N \begin{pmatrix} -\lambda_r \kappa \, \xi_r(\hat{\bf v}) \\ \xi_r(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

onde $r = \pm, \, \lambda_\pm = \pm$.


Se reescribimos explicitamente os espinores de helicidade na dirección $z$ ($\theta = \phi = 0$), volvemos obter:

$$
\boxed{
u_{+}(p) = N \begin{pmatrix} 1 \\ 0 \\ \kappa  \\ 0 \end{pmatrix}, \;
u_{-}(p) = N \begin{pmatrix} 0 \\ 1 \\ 0 \\ -\kappa \end{pmatrix}, \;
v_{+}(p) = N \begin{pmatrix} 0  \\ -\kappa \\ 0 \\ 1 \end{pmatrix}, \;
v_{-}(p) = N \begin{pmatrix} \kappa  \\ 0 \\ 1 \\ 0 \end{pmatrix}. 
}
$$

co factor de normalización $N = \sqrt{E+m}$

*Nota*: Podemos reinterpretar o espinor $u_+(p)$ como aquel que se obtén ao rotar un espinor inicialmente en $u_1(p)$, co seu espín aliñado na dirección cara arriba en $z$, ata situar o seu espín aliñado con $\hat{\bf v}$, coa dirección e o sentido do momento, ${\bf p}$.

É o que se fai na sección 5.1, sen resolver a ecuación de autovalores: rotar arredor de $y$ un ángulo $\theta$ e despois arredor de $z$ un ángulo $\phi$.

## Quiralidade

A violación de paridade das interaccións febles está relacionada cunha propiedade do espinor de Dirac chamada quiralidade.

Dicimos que un espinor ten quiralidade á dereita ou á esquerda se é autoestado de $\gamma^5$ con autovalores +1 e -1, respectivamente. Isto é:

$$
\boxed{
\gamma^5 \, u_R \equiv + u_R, \;\;\; \gamma^5 \, u_L \equiv - u_L
}
$$

A quiralidade é unha propiedade *matemática* que non ten un equivalente físico simple, ao contrario que a helicidade, que si ten un equivalente físico: a proxección do espín na dirección do momento.

Veremos que un espinor de Dirac pode descompoñerse nun espinor de quiralidade á esquerda e noutro á dereita, tales que a inversión de paridade intercambia a quiralidade, a dereita pola esquerda e viceversa.

Isto fai que unha interacción que sexa invariante por paridade non faga distinción entre as dúas quiralidades, pero que unha que non o sexa favoreza ou mesmo suprima unha quiralidade.

#### A matriz $\gamma^5$

Imos introducir unha quinta matriz, $\gamma^5$, que vai estar moi relacionada cunha propiedade que chamaremos *quiralidade* e que xoga un papel fundamental nas interaccións febles.

A matriz $\gamma^5$ defínese:

$$
\gamma^5 \equiv i \gamma^0 \gamma^1 \gamma^2 \gamma^3,
$$

E ten as seguintes propiedades:

$$
(\gamma^5)^2 = I, \;\; \gamma^{5\dagger} = \gamma^5, \;\; \gamma^5\gamma^\mu = - \gamma^\mu \gamma^5
$$

que se derivan das propiedades das matrices $\gamma^\mu$.

*Cuestión:* Verifica as propiedades de $\gamma^5$

Na representación de Pauli-Dirac:

$$
\gamma^5 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix}
$$

Os proxectores de quiralidade son:

$$
\boxed{
P_L = \frac{1}{2} (1-\gamma^5), \;\;\; P_R = \frac{1}{2} (1+\gamma^5)
}
$$

De tal forma que $\Psi(x)$ se descompón en dúas partes, á esquerda e á dereita.

$$
\Psi_L(x) = P_L \, \Psi(x), \;\;\, \Psi_R(x) = P_R \, \Psi(x)
$$

Os proxectores son en realidade dúas matrices que actúan sobre a parte espinorial de $\Psi(x)$.


Vemos que efectivamente un espinor de quiralidade á dereita ou á esquerda cumpre:

$$
\gamma^5 u_R = + u_R \Rightarrow P_R \,u_R = \frac{1}{2}(1+\gamma^5) \, u_R = u_R, 
$$
$$
\gamma^5 u_L = - u_L \Rightarrow P_L \,u_L = \frac{1}{2}(1-\gamma^5) \, u_L = u_L. 
$$


Isto é, para os espinores dos fermións, actúan sobre $u(p)$:

$$
u_L(p) = P_L \, u(p), \;\;\, u_R(p) = P_R \, u(p)
$$

onde eliminamos o subíndice $s$ dos espinores $u_s(p)$ por simplicidade.

Imos comprobar que se trata de proxectores.

Estes deben cumprir:

$$
P_R + P_L = I, \;\; P^2_R = P_R, \;\;, P^2_L = P_L, \;\; P_L P_R = P_R P_L = 0
$$

A primeira é inmediata.

Comprobemos que $P^2_R = P_R$

$$
\left[\frac{1}{2} (1+\gamma^5) \right]^2 =
\frac{1}{4}(I + 2\gamma^5 + (\gamma^5)^2) = \frac{1}{4}(I + 2\gamma^5 + I) = \frac{1}{2}(1+\gamma^5)
$$

dado que $(\gamma^5)^2 = I$.

Do mesmo xeito procederiamos con $P^2_L = P_L$

E finalmente, $P_L \, P_R = P_R \, P_L = 0$:

$$
\frac{1}{2}(1-\gamma^5) \frac{1}{2}(1+\gamma^5) = \frac{1}{4} (I-(\gamma^5)^2) = 0
$$


Imos aplicar paridade sobre $u_L(p), \; u_R(p)$; lembremos que a representación espinorial da inversión por paridade é $S(P) = \gamma^0$

$$
\frac{1}{2}(1\pm\gamma^5) \, u(p) \to S(P) \, \frac{1}{2}(1 \pm \gamma^5) \, u(E, -{\bf p}) = \\
\gamma^0 \, \frac{1}{2}(1 \pm \gamma^5) \, u(E, -{\bf p}) = \\ \frac{1}{2} (1 \mp \gamma^5) (\gamma^0 \, u(E, -{\bf p})) = \frac{1}{2}(1 \mp \gamma^5) u(p)
$$

Logo a inversión por paridade intercambia os espinores de quiralidade á dereita polos de quiralidade á esquerda e viceversa:

$$
u_L(p) \leftrightarrow u_R(p)
$$


Para os espinores dos antifermións definimos a quiralidade de forma complementaria á dos fermións, isto é:

$$
v_R(p) =  P_L\, v(p), \;\;\; v_L(p) = P_R \, v(p), \;\;\; 
$$

Igualmente, a inversión por paridade intercambia as súas quiralidades

$$
v_R(p) \leftrightarrow v_L(p) 
$$

### Os espinores na súa expresión quiral

Aplicámolos sobre os espinores de Dirac que son autoestados de helicidade. Lembremos que son:

$$
u_r(p) = N \begin{pmatrix} \chi_r(\hat{\bf v}) \\ \lambda_r \kappa \, \chi_r(\hat{\bf v}) \end{pmatrix}, \;\;\;
v_r(p) = N \begin{pmatrix} -\lambda_r \kappa \, \xi_r(\hat{\bf v}) \\ \xi_r(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

onde $r = \pm, \, \lambda_\pm = \pm 1$.



Obtemos:

$$
P_R u_\pm(p) = \frac{1}{2} (1 \pm \kappa) N \begin{pmatrix} \chi_\pm(\hat{\bf v}) \\ \chi_\pm(\hat{\bf v}) \end{pmatrix}, \;\;\;
P_L u_\pm(p) = \frac{1}{2} (1 \mp \kappa) N \begin{pmatrix} \chi_\pm(\hat{\bf v}) \\ -\chi_\pm(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

Que podemos explicitar na seguinte táboa:

| helicidade $\downarrow$ \ quiralidade  $\rightarrow$| --- R --- | --- L --- |
| :--: | :--: | :--: | 
| + | $\frac{1}{2}(1+\kappa)$ | $\frac{1}{2}(1-\kappa)$ |
| - | $\frac{1}{2}(1-\kappa)$ | $\frac{1}{2}(1+\kappa)$ |



Consideremos o caso en que a partícula é ultrarrelativista, $E \gg m$, ou sen masa, $m=0$, isto é, cando $\kappa = 1$.

Nese caso o espinor de helicidade negativa só ten compoñente de quiralidade á esquerda, e o espinor de helicidade positiva só de quiralidade á dereita.


$$
u_R(p) = u_+(p) = N \begin{pmatrix} \chi_+(\hat{\bf v}) \\ \chi_+(\hat{\bf v}) \end{pmatrix}, \;\;\;
u_L(p) = u_-(p) = N \begin{pmatrix} \chi_-(\hat{\bf v}) \\ -\chi_-(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

só cando $\kappa = 1$.

Que explicitamos para o caso $\hat{\bf v} = \hat{k}$, isto é, eliximos o eixe $z$ como a dirección de movemento da partícula,

e no caso ultrarrelativista, 

$$
u_R(p) =  u_+(p) = \sqrt{E} \begin{pmatrix} 1 \\ 0 \\ 1  \\ 0 \end{pmatrix}, \;\;\;
u_L(p) =  u_-(p) = \sqrt{E} \begin{pmatrix} 0 \\ 1 \\ 0  \\  -1 \end{pmatrix}, \;\;\;
$$

onde, lembremos, $N = \sqrt{E + m } = \sqrt{E}$


Para os espinores dos antifermións:

$$
v_R(p) = P_L v_\pm(p) = \frac{1}{2} (1 \pm \kappa)  N \begin{pmatrix} - \xi_\pm(\hat{\bf v}) \\ \xi_\pm(\hat{\bf v}) \end{pmatrix}, \;\;\;
v_L(p) = P_R v_\pm(p) = \frac{1}{2} (1 \mp \kappa)  N \begin{pmatrix}  \xi_\pm(\hat{\bf v}) \\ \xi_\pm(\hat{\bf v}) \end{pmatrix}, \;\;\;
$$

que explicitamos para o caso $\hat{\bf v} = \hat{k}$ e no caso ultrarrelativista, 

$$
v_R(p) =  P_L v_+(p) = \sqrt{E} \begin{pmatrix} 0 \\ -1 \\ 0  \\ 1 \end{pmatrix}, \;\;\;
v_L(p) =  P_R v_-(p) = \sqrt{E} \begin{pmatrix} 1 \\ 0 \\ 1  \\  0 \end{pmatrix}, \;\;\;
$$

Podemos especificar os proxectores de quiralidade na representación de Pauli-Dirac:

$$
P_R = \frac{1}{2}\begin{pmatrix} I & I \\ I & I \end{pmatrix}, \;\;\;
P_L = \frac{1}{2}\begin{pmatrix} I & -I \\ -I & I \end{pmatrix},
$$


### A representación de Weyl ou quiral

Imos considerar a representación de Weyl (ou quiral) das matrices $\gamma^\mu$:

$$
\gamma^0 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix}, \;\;
\gamma^k = \begin{pmatrix} 0 & \sigma^k \\ -\sigma^k & 0 \end{pmatrix}, \;\;
\gamma^5 = \begin{pmatrix} -I & 0 \\ 0 & I \end{pmatrix}
$$

Co que:

$$
P_L = \begin{pmatrix} I & 0 \\ 0 & 0 \end{pmatrix}, \;
P_R = \begin{pmatrix} 0 & 0 \\ 0 & I \end{pmatrix} \;
$$

Isto é, o espinor de Dirac está composto de dous biespinores, $u_L, u_R$, chamados de Weyl, que se corresponden coa parte de quiralidade á esquerda e á dereita do espinor de Dirac.

$$
u(p) = \begin{pmatrix} u_L(p) \\ u_R(p)\end{pmatrix}
$$

Se agora consideramos o caso revelador dun fermión sen masa, $m=0$, a ec. de Dirac queda:

$$
\gamma^\mu p_\mu \, u(p) = 0
$$

que se descompón en dúas ecuacións independentes:

$$
(E - {\bf p} \cdot \vec{\sigma}) \, u_R(p) = 0, \\
(E + {\bf p} \cdot \vec{\sigma}) \, u_L(p) = 0.
$$

Isto é, o espinor de Dirac componse de dous espinores de Weyl independentes. 

Ou, noutras palabras, é o termo de masa na ec. de Dirac o que fai que os dous espinores de Weyl estean conectados.


dado que $E = \text{p}$, onde ${\bf p} = \text{p} \hat{\bf v}$, estas ecuacións son equivalentes a:

$$
(\hat{\bf v} \cdot \vec{\sigma}) \; u_R(p) = + u_R(p), \\ (\hat{\bf v} \cdot \vec{\sigma}) \; u_L(p) = - u_L(p).
$$

Isto é, hai unha conexión un a un entre helicidade e quiralidade. O espinor de Weyl á esquerda ten helicidade negativa e o da dereita, positiva.

Neste caso das partículas sen masa, a quiralidade e a helicidade son ambas propiedades estacionarias e invariantes Lorentz.

No Modelo Estándar este caso correspondía ao neutrino.


### A norma e a corrente baixo quiralidade

Como $(\gamma^5)^\dagger = \gamma^5$, os proxectores non se ven afectados pola operación $\dagger$:

$$
\frac{1}{2}(1\pm \gamma^5)^\dagger = \frac{1}{2}(1\pm \gamma^5), \;\; P^\dagger_{L/R} = P_{L/R}
$$

Mentres que o intercambio con $\gamma^0$ intercambia os proxectores:

$$
\gamma^0 \, P_{L/R} = P_{R/L} \, \gamma^0
$$

Isto separa en dúas as formas invariantes Lorentz.

Para a escalar $\bar{\Psi}(x) \Psi(x)$, se a escribimos só en función dos seus espinores $\bar{u}(p) u(p)$:

$$
\bar{u}(p) \, u_L(p) = \bar{u}(p) \, P_L u_L(p) = (u^\dagger(p) \gamma^0) P_L \, u_L(p) = \\ 
(u^\dagger(p)  P_R \gamma^0) \, u_L(p) = 
(P_R u(p))^\dagger \gamma^0 u_L(p) = u^\dagger_R(p) \gamma^0 u_L(p) = \bar{u}_R(p) u_L(p)
$$

E de forma recíproca:

$$
\bar{u}(p) \, u_R(p) = \bar{u}_L \, u_R(p)
$$

Logo a forma escalar:

$$
\bar{u}(p) \, u(p) = \bar{u}_R \, u_L(p) + \bar{u}_L \, u_R(p)
$$

componse de dúas partes nas que as dúas quiralidades se acoplan unha á outra. 

Isto implica que un termo escalar no lagranxiano (o que estará asociado á masa) esixe a existencia das dúas quiralidades.

Lembrando que $\gamma^5\gamma^\mu = - \gamma^\mu\gamma^5$, vemos que:

$$
\gamma^\mu \, P_{L/R} = P_{R/L} \gamma^\mu
$$

é dicir, cambiar de lado $\gamma^\mu$ intercambia os proxectores.

Vexamos agora como queda a forma vectorial, $\bar{\Psi}(x)\gamma^\mu\Psi(x)$, escrita nas súas compoñentes de quiralidade. Desenvolvemos novamente a expresión cos espinores, $\bar{u}(p) \gamma^\mu u(p)$.

$$
\bar{u}(p) \gamma^\mu u_L(p) = \bar{u}(p) \gamma^\mu P_L u_L(p) =  (u^\dagger(p) \gamma^0) \gamma^\mu P_L \, u_L(p) = \\ 
u^\dagger(p) \gamma^0 P_R \, \gamma^\mu \, u_L(p) = (u^\dagger(p) P_L) \gamma^0 \gamma^\mu \, u_L(p) = \\
(P_L u(p))^\dagger \gamma^0 \gamma^\mu u_L(p) = (u^\dagger_L(p) \gamma^0) \gamma^\mu u_L(p) = \bar{u}_L(p) \gamma^\mu u_L(p)
$$

De forma similar:

$$
\bar{u}(p) \gamma^\mu u_R(p) = \bar{u}_R(p) \gamma^\mu u_R(p)
$$

Así, a forma vectorial queda:

$$
\bar{u}(p) \gamma^\mu u(p) = \bar{u}_R(p) \gamma^\mu u_R(p) + \bar{u}_L(p) \gamma^\mu u_L(p)
$$

Isto é, a corrente divídese en dúas: unha corrente de quiralidade á esquerda e outra á dereita.

En resumo:

$$
\bar{\Psi}\Psi =  \bar{\Psi}_R \Psi_L + \bar{\Psi}_L \Psi_R, \\
\bar{\Psi} \gamma^\mu \Psi = \bar{\Psi}_R \gamma^\mu \Psi_R + \bar{\Psi}_L \gamma^\mu \Psi_L
$$
