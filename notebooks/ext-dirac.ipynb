### Notación

* Unidades naturais, $\hbar = c = 1$. Métrica $g_{\mu\nu} = \mathrm{diag}(1, -1, -1, -1)$.
* Índices gregos $\mu, \nu = 0, 1, 2, 3$; latinos $i, j, k = 1, 2, 3$. Un índice repetido súmase.
* $x^\mu = (t, {\bf x})$, $p^\mu = (E, {\bf p})$, $p_\mu = (E, -{\bf p})$, $p_\mu x^\mu = Et - {\bf p}\cdot{\bf x}$.
  As compoñentes físicas do momento son $p^k$.
* ${\bf p} = \mathrm{p}\,\hat{\bf v}$: $\mathrm{p} = |{\bf p}|$ é o módulo e $\hat{\bf v}$ a dirección, de ángulos
  polar e azimutal $\theta, \phi$.
* Cuadrigradiente: $\partial_\mu = \partial/\partial x^\mu = (\partial_t, \nabla)$.
* O acento circunflexo indica unha **dirección** sobre un vector unitario ($\hat{\bf v}$, $\hat z$, $\hat{\bf n}$) e un
  **operador** sobre o momento: $\hat{p}^k = -i\partial_k$ (seccións 3.3, 3.7 e 5.3).
* $\not{p} \equiv \gamma^\mu p_\mu$ (lese «p barrada»).
* Dous números que aparecen en todas partes:

$$
\kappa \equiv \frac{\mathrm{p}}{E+m}, \qquad N \equiv \sqrt{E+m} .
$$

* $I$ é a identidade ($2\times2$ ou $4\times4$, segundo o contexto); $\sigma_k$, as matrices de Pauli.
* Espín: ${\bf S} = \frac{1}{2}{\bf \Sigma}$. A transformación de Lorentz dun espinor é a matriz
  $S(\Lambda)$: **non confundir** unha coa outra.
* Rapidez dun *boost*: $\eta$, con $\cosh\eta = E/m$ e $\sinh\eta = \mathrm{p}/m$.
* Na clase os espinores con ${\bf p}$ en $z$ chámanse pola súa helicidade: $u_+ \equiv u_1$, $u_- \equiv u_2$
  (sección 3.7).

## 1. A ecuación de Dirac

### 1.1. Punto de partida

En mecánica cuántica a enerxía e o momento son operadores:

$$
\hat{H} = i\partial_t, \qquad \hat{\bf p} = -i\nabla .
$$

Se se substitúen en $E^2 = {\bf p}^2 + m^2$ obtense a ecuación de **Klein-Gordon**,
$-\partial_t^2\Psi = (-\nabla^2 + m^2)\Psi$. É de **segunda orde** no tempo, e a súa densidade de
probabilidade pode ser negativa. [MT4.1]

### 1.2. A idea de Dirac (1928)

* primeira orde en $\partial_t$, como a ecuación de Schrödinger;
* a relatividade trata $t$ e ${\bf x}$ por igual: primeira orde tamén en $\nabla$;
* a ecuación de Einstein debe seguir cumpríndose.

A ecuación máis xeral **lineal** nas derivadas é

$$
(i\gamma^0 \partial_t + i\gamma^k \partial_k - m) \, \Psi = 0 ,
$$

con catro coeficientes $\gamma^\mu$ que non coñecemos (o de $m$ tómase igual á unidade).

### 1.3. Ao cadrado, a ecuación de Einstein

**Cos operadores**, $i\partial_t \to E$ e $i\partial_k \to -p^k$, a ecuación lese

$$
(\gamma^0 E - \gamma^k p^k) \, \Psi = m \, \Psi ,
\qquad \text{é dicir,} \qquad (\not{p} - m)\,\Psi = 0 .
$$

**Aplicándoa dúas veces**, $(\gamma^0 E - \gamma^k p^k)^2\,\Psi = m^2\,\Psi$. Desenvolvemos sen cambiar a
orde das $\gamma$, porque non sabemos se conmutan:

$$
(\gamma^0 E - \gamma^k p^k)^2 = (\gamma^0)^2 E^2 + \sum_k (\gamma^k)^2 (p^k)^2
+ \sum_{i<k} \{\gamma^i, \gamma^k\}\, p^i p^k - \sum_k \{\gamma^0, \gamma^k\}\, E\, p^k .
$$

**Esiximos a ecuación de Einstein**: para que isto sexa $E^2 - {\bf p}^2$ para calquera $E$ e ${\bf p}$,

$$
(\gamma^0)^2 = I, \qquad (\gamma^k)^2 = -I, \qquad \{\gamma^0, \gamma^k\} = 0, \qquad
\{\gamma^i, \gamma^k\} = 0 \;\; (i \neq k),
$$

que se resume na **álxebra das $\gamma$**:

$$
\boxed{
\{\gamma^\mu, \gamma^\nu\} = \gamma^\mu\gamma^\nu + \gamma^\nu\gamma^\mu = 2\, g^{\mu\nu}\, I
}
$$

### 1.4. As $\gamma$ son matrices, e $\Psi$ ten catro compoñentes

Dous números non nulos non poden anticonmutar: as $\gamma^\mu$ son **matrices**, e $\Psi$ é unha
columna sobre a que actúan. Fan falta catro matrices que anticonmuten dúas a dúas: con $2\times2$
só hai tres (as de Pauli); teñen traza nula e autovalores $\pm1$ (ou $\pm i$), logo a súa dimensión
é par. A mínima é $4$: $\Psi$ ten **catro compoñentes**, é un **espinor de Dirac**.

**Hermiticidade.** Multiplicando por $\gamma^0$ obtense a forma hamiltoniana,
$E\,\Psi = H\,\Psi$ con $H = \gamma^0\gamma^k p^k + \gamma^0 m$. Como $H$ debe ser hermítico (enerxías
reais):

$$
(\gamma^0)^\dagger = \gamma^0, \qquad (\gamma^k)^\dagger = -\gamma^k .
$$

### 1.5. A representación de Pauli-Dirac

Só a álxebra está fixada; a física non depende da elección. A habitual, en bloques $2\times2$:

$$
\boxed{
\gamma^0 = \begin{pmatrix} I & 0 \\ 0 & -I \end{pmatrix}, \qquad
\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ -\sigma_k & 0 \end{pmatrix}
}
\qquad
\sigma_1 = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \;
\sigma_2 = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \;
\sigma_3 = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}
$$

con $\{\sigma_i, \sigma_j\} = 2\delta_{ij} I$ e $[\sigma_i, \sigma_j] = 2i\epsilon_{ijk}\sigma_k$. Volvendo ás derivadas:

$$
\boxed{
(i\gamma^\mu \partial_\mu - m)\, \Psi(x) = 0
}
$$

[MT4.2]

### 1.6. Que son as matrices $\gamma$?

**Como o número $i$.** Con números reais, $x^2 + y^2$ non se pode escribir como produto de dous
factores lineais; fai falta un número novo con $i^2 = -1$:

$$
x^2 + y^2 = (x + iy)(x - iy) .
$$

Dirac necesita o mesmo con $E^2 - {\bf p}^2$, e ningún número serve: fan falta obxectos cuxo
cadrado dea o signo da métrica e que **anticonmuten**, para cancelar os termos cruzados. En
tres dimensións xa o coñecemos: $(\vec\sigma\cdot{\bf p})^2 = {\bf p}^2$, así que $\vec\sigma\cdot{\bf p}$ é a
«raíz cadrada» de ${\bf p}^2$. Dirac fai o mesmo no espazo-tempo:

$$
\not{p}^{\,2} = (\gamma^\mu p_\mu)^2 = p_\mu p^\mu = m^2 .
$$

**A métrica escrita con matrices.** A álxebra $\{\gamma^\mu, \gamma^\nu\} = 2g^{\mu\nu}I$ di que as
$\gamma^\mu$ se comportan como catro vectores unitarios do espazo-tempo: o temporal de cadrado
$+1$, os espaciais de cadrado $-1$, e «ortogonais» entre si.

### 1.7. A quinta matriz, $\gamma^5$

Coas catro $\gamma^\mu$ constrúese unha quinta:

$$
\boxed{
\gamma^5 \equiv i\,\gamma^0\gamma^1\gamma^2\gamma^3
}
$$

As súas propiedades saen só da álxebra $\{\gamma^\mu, \gamma^\nu\} = 2g^{\mu\nu}I$:

* **Anticonmuta coas catro $\gamma^\mu$**, $\{\gamma^5, \gamma^\mu\} = 0$. Para pasar $\gamma^\mu$ dun lado ao outro
  do produto hai que cruzala coas tres $\gamma$ distintas (un signo menos cada unha) e con ela
  mesma (sen signo): $(-1)^3 = -1$.
* **O seu cadrado é a identidade**, $(\gamma^5)^2 = I$. Reordenando $\gamma^0\gamma^1\gamma^2\gamma^3\,\gamma^0\gamma^1\gamma^2\gamma^3$
  ata xuntar cada $\gamma$ coa súa parella saen $(-1)^{3+2+1} = +1$ e
  $(\gamma^0)^2(\gamma^1)^2(\gamma^2)^2(\gamma^3)^2 = (+1)(-1)^3 = -1$; co $i^2 = -1$ de diante, $+I$.
* **É hermítica**, $(\gamma^5)^\dagger = \gamma^5$. Con $(\gamma^0)^\dagger = \gamma^0$ e $(\gamma^k)^\dagger = -\gamma^k$:
  $(\gamma^5)^\dagger = -i\,(\gamma^3)^\dagger(\gamma^2)^\dagger(\gamma^1)^\dagger(\gamma^0)^\dagger = i\,\gamma^3\gamma^2\gamma^1\gamma^0$, e darlle a volta
  ao produto de catro custa seis permutacións: $(-1)^6 = +1$.

**En Pauli-Dirac.** Con $\gamma^1\gamma^2 = \mathrm{diag}(-i\sigma_3, -i\sigma_3)$ (porque $\sigma_1\sigma_2 = i\sigma_3$) e multiplicando
por $\gamma^3$ e $\gamma^0$:

$$
\gamma^5 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix} :
$$

intercambia o bloque de arriba co de abaixo.

**Non cambia con Lorentz**: $S^{-1}\gamma^5 S = \gamma^5$ para calquera transformación $S(\Lambda)$ dos
espinores. Demóstrase na sección 3.2, cos xeradores.

Con ela defínense os proxectores de **quiralidade**, $P_{L,R} = \frac{1}{2}(I \mp \gamma^5)$, e a corrente
axial (sección 4.4). A quiralidade está desenvolvida en [ext-quiral](./ext-quiral.ipynb). [MT4.2, MT6.4]

## 2. As solucións: os espinores

### 2.1. Ondas planas e as ecuacións dos espinores

Buscamos solucións de onda plana: unha columna de catro constantes por unha fase,

$$
\Psi_s(x) = u_s(p)\, e^{-i p_\mu x^\mu}, \qquad \Phi_s(x) = v_s(p)\, e^{+i p_\mu x^\mu}, \qquad
E = +\sqrt{{\bf p}^2 + m^2} .
$$

Toda a dependencia en $(t, {\bf x})$ está na fase; o **espinor** depende só de $p$. As derivadas
actúan só sobre a fase ($i\partial_\mu \to \pm p_\mu$), e a ecuación de Dirac convértese nunha
ecuación **alxébrica** para os espinores:

$$
\boxed{
(\not{p} - m)\, u_s(p) = 0, \qquad (\not{p} + m)\, v_s(p) = 0
}
$$

Son as ecuacións fundamentais dos espinores. Resolvémolas en tres pasos: en repouso, co
momento en $z$ e co momento nunha dirección calquera.

### 2.2. Einstein outra vez, e os dous signos da enerxía

A ecuación $(\not{p} - m)\,w = 0$ é un sistema lineal **homoxéneo**: só ten solución $w \neq 0$ se o
determinante se anula. Multiplicando por $(\not{p} + m)$ vese nunha liña:

$$
(\not{p} + m)(\not{p} - m)\, w = (\not{p}^{\,2} - m^2)\, w = (p^2 - m^2)\, w = 0
\quad \Rightarrow \quad E^2 = {\bf p}^2 + m^2 .
$$

A onda plana só é solución se $E$ e ${\bf p}$ son os dunha partícula de masa $m$. Pero a ecuación
fixa $E^2$, non $E$: hai solucións con $E = \pm\sqrt{{\bf p}^2 + m^2}$. As de enerxía negativa
escríbense cambiando o signo de toda a fase, de modo que nas fórmulas $E > 0$: son as $\Phi_s$,
con $v_s$.

**Lectura limpa:** como $\not{p}^{\,2} = m^2$, a matriz $\not{p}$ só ten autovalores $\pm m$, cada un dúas
veces (ten traza nula):

$$
\not{p}\, u = +m\, u, \qquad \not{p}\, v = -m\, v .
$$

Partícula e antipartícula son os **dous subespazos** da mesma matriz.

### 2.3. En repouso, ${\bf p} = 0$

Con ${\bf p} = 0$, $E = m$ e $\not{p} = m\,\gamma^0$. As dúas ecuacións son diagonais:

$$
\not{p} - m = m\,(\gamma^0 - I) = \begin{pmatrix} 0 & 0 \\ 0 & -2m\, I \end{pmatrix}, \qquad
\not{p} + m = m\,(\gamma^0 + I) = \begin{pmatrix} 2m\, I & 0 \\ 0 & 0 \end{pmatrix} .
$$

* $(\not{p} - m)\,u = 0$ só esixe que o **bloque de abaixo** sexa cero: o de arriba, $\chi$, é libre.
* $(\not{p} + m)\,v = 0$ esixe que o **bloque de arriba** sexa cero: o de abaixo, $\xi$, é libre.

Dúas compoñentes libres en cada caso: os **dous estados de espín**, co espinor de Pauli de sempre.
Coa normalización $u^\dagger u = 2E = 2m$ (sección 2.4), $N = \sqrt{2m}$:

$$
u_1(m) = \sqrt{2m} \begin{pmatrix} 1 \\ 0 \\ 0 \\ 0 \end{pmatrix}, \;\;
u_2(m) = \sqrt{2m} \begin{pmatrix} 0 \\ 1 \\ 0 \\ 0 \end{pmatrix}, \;\;
v_1(m) = \sqrt{2m} \begin{pmatrix} 0 \\ 0 \\ 0 \\ 1 \end{pmatrix}, \;\;
v_2(m) = \sqrt{2m} \begin{pmatrix} 0 \\ 0 \\ 1 \\ 0 \end{pmatrix}
$$

A partícula vive arriba e a antipartícula abaixo: son os autoespazos de $\gamma^0$ con autovalor
$+1$ e $-1$. En $v$ a orde das compoñentes está cruzada para que $s = 1$ sexa espín físico cara arriba
tamén na antipartícula (sección 2.6).

### 2.4. Co momento en $z$

Tomamos ${\bf p} = (0, 0, \mathrm{p})$. Con $\not{p} = \gamma^0 E - \gamma^3 \mathrm{p}$ e $u = (u^a, u^b)$ en bloques:

$$
\not{p} - m =
\begin{pmatrix} (E - m)\, I & -\mathrm{p}\,\sigma_3 \\ \mathrm{p}\,\sigma_3 & -(E + m)\, I \end{pmatrix} .
$$

* $\gamma^0$ (a enerxía) é diagonal: **non mestura** arriba con abaixo.
* $\gamma^3$ (o momento) está fóra da diagonal: **liga** os dous bloques, con peso $\mathrm{p}$.

Cada fila de bloques é unha ecuación:

$$
(E - m)\, u^a = \mathrm{p}\,\sigma_3\, u^b, \qquad (E + m)\, u^b = \mathrm{p}\,\sigma_3\, u^a .
$$

Da segunda, o bloque de abaixo queda **fixado** polo de arriba, $u^b = \kappa\,\sigma_3\,u^a$, con
$\kappa = \mathrm{p}/(E+m)$. A primeira non engade nada: con $\sigma_3^2 = I$ só esixe $E^2 = \mathrm{p}^2 + m^2$.
Quedan **dúas compoñentes libres**, as de $u^a$, como en repouso. Para $v$ (con $+m$) o bloque libre é
o de abaixo, e $v^a = \kappa\,\sigma_3\,v^b$.

**Normalización.** Tómase $u^\dagger u = v^\dagger v = 2E$. A densidade $\Psi^\dagger\Psi$ non é invariante:
cun *boost* o volume contráese e a densidade crece como $E$. A única normalización válida en
todos os sistemas é proporcional a $E$; o 2 é un convenio, o mesmo que para os bosóns. Así,
$N^2(1 + \kappa^2) = 2E$ dá $N = \sqrt{E+m}$, e os catro espinores son

$$
\boxed{
u_1 = N \begin{pmatrix} 1 \\ 0 \\ \kappa \\ 0 \end{pmatrix}, \;\;
u_2 = N \begin{pmatrix} 0 \\ 1 \\ 0 \\ -\kappa \end{pmatrix}, \;\;
v_1 = N \begin{pmatrix} 0 \\ -\kappa \\ 0 \\ 1 \end{pmatrix}, \;\;
v_2 = N \begin{pmatrix} \kappa \\ 0 \\ 1 \\ 0 \end{pmatrix}
}
\qquad \kappa = \frac{\mathrm{p}}{E+m}, \;\; N = \sqrt{E+m}
$$

Como cambian con $\kappa$ (en repouso, en movemento, ultrarrelativista): ver [perspectivas](./perspectivas.ipynb), *Que relativista é o espinor?*

### 2.5. Co momento nunha dirección calquera

O único cambio é $\mathrm{p}\,\sigma_3 \to \vec\sigma\cdot{\bf p}$ nos bloques de fóra da diagonal:

$$
\not{p} - m =
\begin{pmatrix} (E - m)\, I & -\vec\sigma\cdot{\bf p} \\ \vec\sigma\cdot{\bf p} & -(E + m)\, I \end{pmatrix},
\qquad
u^b = \frac{\vec\sigma\cdot{\bf p}}{E+m}\, u^a .
$$

A primeira ecuación volve non engadir nada porque $(\vec\sigma\cdot{\bf p})^2 = {\bf p}^2$, que fai o papel de
$\sigma_3^2 = I$. Cos bloques libres $\chi_1 = (1, 0)$, $\chi_2 = (0, 1)$ e, en orde cruzada,
$\xi_1 = (0, 1)$, $\xi_2 = (1, 0)$:

$$
\boxed{
u_s(p) = N \begin{pmatrix} \chi_s \\ \dfrac{\vec\sigma\cdot{\bf p}}{E+m}\,\chi_s \end{pmatrix}, \qquad
v_s(p) = N \begin{pmatrix} \dfrac{\vec\sigma\cdot{\bf p}}{E+m}\,\xi_s \\ \xi_s \end{pmatrix}, \qquad N = \sqrt{E+m}
}
$$

A normalización é a mesma: $(\vec\sigma\cdot{\bf p})^2 = {\bf p}^2$ dá outra vez $u^\dagger u = 2E$. Con
${\bf p} = \mathrm{p}\,\hat{z}$ recupéranse os espinores da sección 2.4, e con ${\bf p} = 0$ os da 2.3.

### 2.6. Enerxía, momento e antipartícula

Sobre a fase, $i\partial_t\,\Phi_s = -E\,\Phi_s$ e $-i\nabla\,\Phi_s = -{\bf p}\,\Phi_s$: en $\Phi_s$ os operadores dan o
signo cambiado, e a regra práctica é invertelo en todo (enerxía, momento, espín e cargas) para ler a
antipartícula. A interpretación, en [perspectivas](./perspectivas.ipynb) (*Antipartículas* e *Feynman e Stückelberg*).

## 3. Transformacións de Lorentz: espín e helicidade

### 3.1. A condición sobre $S(\Lambda)$

Baixo unha transformación de Lorentz $\Lambda$, $x' = \Lambda x$, un cuadrivector transfórmase con $\Lambda$ e
o cuadrigradiente, que é covariante, coa inversa: $\partial'_\mu = (\Lambda^{-1})^\nu{}_\mu\,\partial_\nu$. Un
espinor transfórmase cunha matriz complexa $4\times4$, $S(\Lambda)$:

$$
\Psi'(x') = S(\Lambda)\, \Psi(x) .
$$

**O que esiximos:** a ecuación de Dirac ten a **mesma forma** en todos os sistemas inerciais, coas
**mesmas** $\gamma^\mu$. Se $(i\gamma^\nu\partial_\nu - m)\,\Psi = 0$, debe cumprirse

$$
(i\gamma^\mu\partial'_\mu - m)\, S\,\Psi
= \left[ i\,\gamma^\mu (\Lambda^{-1})^\nu{}_\mu\, \partial_\nu - m \right] S\,\Psi = 0 .
$$

Multiplicando por $S^{-1}$ pola esquerda, isto coincide coa ecuación orixinal se
$S^{-1}\gamma^\mu S\,(\Lambda^{-1})^\nu{}_\mu = \gamma^\nu$, é dicir:

$$
\boxed{
S^{-1}(\Lambda)\, \gamma^\mu\, S(\Lambda) = \Lambda^\mu{}_\nu\, \gamma^\nu
}
$$

É a condición que define a **representación espinorial** da transformación de Lorentz: rotar as
$\gamma$ con $\Lambda$ é o mesmo que conxugalas con $S$.

**Por que as $\gamma$ poden ser as mesmas en todos os sistemas?** Esíxeo o principio de
relatividade (a mesma ecuación para todos), e é posible porque o único que se lles pide ás $\gamma$ é a súa
álxebra, $\{\gamma^\mu, \gamma^\nu\} = 2g^{\mu\nu}I$, e a métrica é a mesma en todos os sistemas
($\Lambda^\mu{}_\alpha\Lambda^\nu{}_\beta\, g^{\alpha\beta} = g^{\mu\nu}$). Un conxunto rotado, $\Lambda^\mu{}_\nu\gamma^\nu$, cumpre
a mesma álxebra, e polo **teorema fundamental de Pauli** dous conxuntos que cumpren a álxebra están
relacionados por unha semellanza: existe $S$ con $\Lambda^\mu{}_\nu\gamma^\nu = S^{-1}\gamma^\mu S$. Rotar as $\gamma$
equivale a un cambio de base no espazo dos espinores, que se lle pasa a $\Psi$.

**Que significa para $\not{p}$.** As $\gamma^\mu$ son matrices fixas, non as compoñentes dun cuadrivector.
$\gamma^\mu v_\mu$ ten o índice contraído, pero non é un número invariante: da condición,

$$
\gamma^\mu v'_\mu = S\, (\gamma^\nu v_\nu)\, S^{-1},
$$

transfórmase **por semellanza**, á vez que o espinor. Iso é a **covarianza** da ecuación de
Dirac. Os números invariantes aparecen ao pechar co adxunto, $\bar\Psi = \Psi^\dagger\gamma^0$: vémolo
na sección 4.

Para a dedución detallada dos xeradores en todas as direccións, ver
[ext-trans-Lorentz](./ext-trans-Lorentz.ipynb).

### 3.2. Transformacións infinitesimais: os xeradores

Buscamos $S = I + \omega$, con $\omega$ pequena, que cumpra a condición para unha transformación
infinitesimal. Como $S^{-1} = I - \omega$, a condición queda

$$
S^{-1}\gamma^\mu S \simeq \gamma^\mu + [\gamma^\mu, \omega] = \Lambda^\mu{}_\nu\,\gamma^\nu .
$$

**Rotación $\delta\phi$ arredor de $z$** (antihoraria): $\gamma^1 \to \gamma^1 - \delta\phi\,\gamma^2$,
$\gamma^2 \to \gamma^2 + \delta\phi\,\gamma^1$, e $\gamma^0, \gamma^3$ non cambian. $\omega$ debe conmutar con
$\gamma^0$ e $\gamma^3$: probamos $\omega = b\,\gamma^1\gamma^2$. Con $[\gamma^1, \gamma^1\gamma^2] = -2\gamma^2$ sae
$b = \delta\phi/2$, e como $\gamma^1\gamma^2 = -i\,\Sigma_3$:

$$
S = I - i\,\frac{\delta\phi}{2}\,\Sigma_3 = I - i\,\delta\phi\, S_3 .
$$

**O xerador das rotacións é o espín**, $S_3 = \frac{1}{2}\Sigma_3$.

***Boost* $\delta\eta$ en $z$**: $\gamma^0 \to \gamma^0 + \delta\eta\,\gamma^3$, $\gamma^3 \to \gamma^3 + \delta\eta\,\gamma^0$, e
$\gamma^1, \gamma^2$ non cambian. Probamos $\omega = a\,\gamma^0\gamma^3$. Con $[\gamma^0, \gamma^0\gamma^3] = 2\gamma^3$ sae
$a = \delta\eta/2$:

$$
S = I + \frac{\delta\eta}{2}\,\gamma^0\gamma^3 .
$$

**Transformación finita.** Encadéanse moitas infinitesimais: $S$ é a exponencial do xerador.
Os ángulos aparecen **divididos por dous**.

**$\gamma^5$ non cambia con Lorentz.** $\gamma^5$ conmuta cos dous xeradores: con $\Sigma_k$ (diagonal por
bloques, co mesmo bloque arriba e abaixo) e con $\gamma^0\gamma^k$ (anticonmuta con cada factor). Logo
conmuta con calquera $S(\Lambda)$, que é a súa exponencial:

$$
S^{-1}\,\gamma^5\, S = \gamma^5 .
$$

Usarémolo na sección 4.4.


### 3.3. Rotación en $z$: o espín

**En $z$.** Como $\Sigma_3^2 = I$, a serie da exponencial súmase:

$$
\boxed{
S(\phi\,\hat{z}) = e^{-i\phi\,S_3} = \cos\frac{\phi}{2}\, I - i \sin\frac{\phi}{2}\, \Sigma_3
= \mathrm{diag}\left(e^{-i\phi/2},\, e^{+i\phi/2},\, e^{-i\phi/2},\, e^{+i\phi/2}\right)
}
$$

A rotación non mestura os bloques de arriba e de abaixo. Sobre os espinores con ${\bf p}$ en $z$ (unha
rotación arredor de $z$ non cambia ${\bf p}$) só engade unha **fase**:

$$
S(\phi\,\hat{z})\, u_1 = e^{-i\phi/2}\, u_1, \qquad S(\phi\,\hat{z})\, u_2 = e^{+i\phi/2}\, u_2,
\qquad \text{é dicir,} \qquad S(\phi\,\hat{z})\, u_s = e^{-i m_s \phi}\, u_s ,
$$

con $m_s = \pm\frac{1}{2}$ os autovalores de $S_3$: $S_3 u_1 = +\frac12 u_1$, $S_3 u_2 = -\frac12 u_2$. Con
$\phi = 2\pi$, $S = -I$: o espinor volve **cambiado de signo**; fan falta dúas voltas, $4\pi$.

**Por que é un espín, un momento angular intrínseco?**

1. **Contar estados.** As rotacións actúan sobre as dúas compoñentes libres da sección 2.3; un
   momento angular $s$ ten $2s + 1$ estados, e $2s + 1 = 2$ dá $s = \frac{1}{2}$.
2. **Quen xera as rotacións.** O xerador sobre o espinor é ${\bf S} = \frac{1}{2}{\bf \Sigma}$, que
   cumpre a álxebra do momento angular, $[S_i, S_j] = i\epsilon_{ijk}S_k$. É **intrínseco**: actúa sobre
   as compoñentes, non depende de ${\bf x}$. O medio ángulo, $e^{-i m_s\phi}$, é o $\frac{1}{2}$.
3. **A conservación esíxeo.** O hamiltoniano de Dirac non conmuta co momento angular orbital;
   só co total: $[H, {\bf L}] \neq 0$, $[H, {\bf J}] = 0$, ${\bf J} = {\bf L} + {\bf S}$ (demostrámolo
   a continuación). [MT4.4]

Na antipartícula o operador dá o signo cambiado (sección 2.6): $S_3 v_1 = -\frac{1}{2}v_1$, pero o
espín físico de $v_1$ é $+\frac{1}{2}$.

**$[H, {\bf L}] \neq 0$, pero $[H, {\bf J}] = 0$.** O hamiltoniano da sección 1.4, co momento como
operador, $\hat{p}^k = -i\partial_k$, é

$$
H = \alpha_k\, \hat{p}^k + \gamma^0 m, \qquad
\alpha_k \equiv \gamma^0\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ \sigma_k & 0 \end{pmatrix} .
$$

Usamos $[x^j, \hat{p}^k] = i\,\delta_{jk}$, $L_i = \epsilon_{ijk}\, x^j \hat{p}^k$ e $[\sigma_i, \sigma_j] = 2i\,\epsilon_{ijk}\,\sigma_k$.

* **O orbital.** $\gamma^0 m$ e as $\alpha_k$ son matrices constantes: conmutan con $x^j$ e $\hat{p}^k$. Só
  conta o conmutador de $\hat{p}^l$ con $x^j$:

  $$
  [H, L_i] = \alpha_l\, \epsilon_{ijk}\, [\hat{p}^l, x^j]\, \hat{p}^k = -i\, \epsilon_{ijk}\, \alpha_j\, \hat{p}^k
  = -i\, (\vec\alpha \times \hat{\bf p})_i \neq 0 .
  $$

* **O espín.** $S_i = \frac{1}{2}\Sigma_i$ actúa só sobre as compoñentes: conmuta con $\hat{p}^k$ e con
  $\gamma^0$ (os dous son diagonais por bloques, co mesmo bloque arriba e abaixo). Coas $\alpha_k$,
  bloque a bloque, $[\alpha_k, \Sigma_i] = 2i\,\epsilon_{kil}\,\alpha_l$, e

  $$
  [H, S_i] = \frac{1}{2}\, \hat{p}^k\, [\alpha_k, \Sigma_i] = i\, \epsilon_{ilk}\, \alpha_l\, \hat{p}^k
  = +i\, (\vec\alpha \times \hat{\bf p})_i .
  $$

Os dous cancélanse na suma:

$$
\boxed{
[H, {\bf J}] = [H, {\bf L}] + [H, {\bf S}] = 0, \qquad {\bf J} = {\bf L} + {\bf S}
}
$$

**Que se le:**

* Nin ${\bf L}$ nin ${\bf S}$ se conservan por separado; só a suma. O espín non se postula: fai falta para
  que se conserve o momento angular.
* **O $\frac{1}{2}$ está fixado.** Con ${\bf S} = \lambda\,{\bf \Sigma}$ sairía $[H, {\bf S}] = 2\lambda\, i\,(\vec\alpha\times\hat{\bf p})$, que
  só cancela $[H, {\bf L}]$ se $\lambda = \frac{1}{2}$: a ecuación fixa o valor do espín.
* **Con ${\bf p}$ en $z$**, $(\vec\alpha\times{\bf p})_3 = \alpha_1 p^y - \alpha_2 p^x = 0$: $[H, S_3] = 0$, e $S_3$ é un bo
  número cuántico das solucións da sección 2.4.
* Noutra dirección $S_3$ non se conserva, pero a **helicidade** si: sección 3.7.

### 3.4. Rotación nunha dirección calquera (só indicado)

Un ángulo $\alpha$ arredor dun eixe $\hat{\bf n}$. Como $(\hat{\bf n}\cdot{\bf \Sigma})^2 = I$, igual ca en $z$:

$$
\boxed{
S(\alpha\,\hat{\bf n}) = e^{-i\alpha\,\hat{\bf n}\cdot{\bf S}} = \cos\frac{\alpha}{2}\, I - i \sin\frac{\alpha}{2}\, (\hat{\bf n}\cdot{\bf \Sigma})
}
$$

**Levar o espín a unha dirección $\hat{\bf v}$.** Primeiro un ángulo $\theta$ arredor de $y$, despois $\phi$
arredor de $z$: $S(R) = S(\phi\,\hat{z})\,S(\theta\,\hat{y})$, con
$e^{-i\frac{\theta}{2}\sigma_2} = \begin{pmatrix} \cos\frac{\theta}{2} & -\sin\frac{\theta}{2} \\ \sin\frac{\theta}{2} & \cos\frac{\theta}{2} \end{pmatrix}$.
Sobre os biespinores (sen demostración):

$$
\chi_1 \;\to\; e^{-i\phi/2} \begin{pmatrix} \cos\frac{\theta}{2} \\ e^{i\phi}\sin\frac{\theta}{2} \end{pmatrix} \equiv e^{-i\phi/2}\,\chi_+(\hat{\bf v}),
\qquad
\chi_2 \;\to\; e^{-i\phi/2} \begin{pmatrix} -\sin\frac{\theta}{2} \\ e^{i\phi}\cos\frac{\theta}{2} \end{pmatrix} \equiv e^{-i\phi/2}\,\chi_-(\hat{\bf v}) ,
$$

que cumpren $(\hat{\bf v}\cdot\vec\sigma)\,\chi_\pm = \pm\chi_\pm$: o espín apunta agora en $\hat{\bf v}$.

**O fermión en movemento.** Aplicada a $u_\pm$ con ${\bf p}$ en $z$, a rotación xira tamén o momento,
$\mathrm{p}\,\hat{z} \to \mathrm{p}\,\hat{\bf v}$, e $\sigma_3 \to \hat{\bf v}\cdot\vec\sigma$:

$$
S(R)\, u_\pm(\mathrm{p}\,\hat{z}) \propto N \begin{pmatrix} \chi_\pm(\hat{\bf v}) \\ \pm\kappa\,\chi_\pm(\hat{\bf v}) \end{pmatrix} = u_\pm(\mathrm{p}\,\hat{\bf v}) :
$$

espín e momento xiran xuntos, e a **helicidade** (o espín na dirección do momento) non cambia.

### 3.5. *Boost* en $z$

Como $(\gamma^0\gamma^3)^2 = I$ e $\gamma^0\gamma^3 = \begin{pmatrix} 0 & \sigma_3 \\ \sigma_3 & 0 \end{pmatrix}$:

$$
\boxed{
S(\eta\,\hat{z}) = e^{\frac{\eta}{2}\,\gamma^0\gamma^3} = \cosh\frac{\eta}{2}\, I + \sinh\frac{\eta}{2}\,\gamma^0\gamma^3
= \begin{pmatrix} \cosh\frac{\eta}{2}\, I & \sinh\frac{\eta}{2}\,\sigma_3 \\ \sinh\frac{\eta}{2}\,\sigma_3 & \cosh\frac{\eta}{2}\, I \end{pmatrix}
}
$$

A diferenza da rotación, o *boost* **mestura** os bloques. Sobre o espinor en repouso,
$u_s(m) = \sqrt{2m}\,(\chi_s, 0)$, coas fórmulas do ángulo metade
$\sqrt{2m}\cosh\frac{\eta}{2} = \sqrt{E+m}$ e $\tanh\frac{\eta}{2} = \frac{\mathrm{p}}{E+m}$:

$$
S(\eta\,\hat{z})\, u_s(m) = \sqrt{2m} \begin{pmatrix} \cosh\frac{\eta}{2}\,\chi_s \\ \sinh\frac{\eta}{2}\,\sigma_3\,\chi_s \end{pmatrix}
= \sqrt{E+m} \begin{pmatrix} \chi_s \\ \kappa\,\sigma_3\,\chi_s \end{pmatrix} = u_s(p) .
$$

* Recuperamos, **sen resolver a ecuación**, o espinor da sección 2.4. E $\kappa = \tanh\frac{\eta}{2}$: a
  tanxente da metade da rapidez.
* $N = \sqrt{E+m}$ é o que dá o *boost* partindo de $\sqrt{2m}$ en repouso.
* O *boost* vai ao longo do eixe do espín e non cambia $S_3$: $u_1(m)$ pasa a $u_+(p)$.
* **Non é unitario**: $u^\dagger u$ pasa de $2m$ a $2E$, mentres que $\bar u u = N^2(1 - \kappa^2) = 2m$ non cambia.
* A fase non cambia: depende de $p_\mu x^\mu$, invariante Lorentz; tras o *boost* escríbese
  $p'_\mu x'^\mu$, pero vale o mesmo.

### 3.6. *Boost* nunha dirección calquera (só indicado)

Como $\gamma^0\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ \sigma_k & 0 \end{pmatrix}$, abonda con cambiar
$\sigma_3$ por $\hat{\bf n}\cdot\vec\sigma$:

$$
\boxed{
S(\eta\,\hat{\bf n}) = e^{\frac{\eta}{2}\,\gamma^0(\hat{\bf n}\cdot\vec\gamma)}
= \begin{pmatrix} \cosh\frac{\eta}{2}\, I & \sinh\frac{\eta}{2}\,(\hat{\bf n}\cdot\vec\sigma) \\ \sinh\frac{\eta}{2}\,(\hat{\bf n}\cdot\vec\sigma) & \cosh\frac{\eta}{2}\, I \end{pmatrix}
}
$$

Sobre o espinor en repouso, con calquera $\chi$, dá a solución xeral da sección 2.5 con
${\bf p} = \mathrm{p}\,\hat{\bf n}$:

$$
S(\eta\,\hat{\bf n})\, \sqrt{2m} \begin{pmatrix} \chi \\ 0 \end{pmatrix}
= N \begin{pmatrix} \chi \\ \kappa\,(\hat{\bf n}\cdot\vec\sigma)\,\chi \end{pmatrix}
= N \begin{pmatrix} \chi \\ \dfrac{\vec\sigma\cdot{\bf p}}{E+m}\,\chi \end{pmatrix} .
$$

O *boost* non toca o bloque de arriba: o espín é o que tiña en repouso. Se en repouso apunta en
$z$ e o *boost* vai noutra dirección, o resultado non é autoestado de helicidade; se xa apunta en
$\hat{\bf n}$, $\chi = \chi_\pm(\hat{\bf n})$, o resultado é $u_\pm(\mathrm{p}\,\hat{\bf n})$.

### 3.7. Helicidade

A **helicidade** é a proxección do espín sobre a dirección do momento, $\hat{\bf v} = {\bf p}/\mathrm{p}$:

$$
\boxed{
h \equiv {\bf S}\cdot\hat{\bf v} = \frac{1}{2}
\begin{pmatrix} \hat{\bf v}\cdot\vec\sigma & 0 \\ 0 & \hat{\bf v}\cdot\vec\sigma \end{pmatrix}
}
$$

Sen un eixe externo, o único eixe que ten unha partícula libre é o do seu propio momento.

**Consérvase.** Co momento como operador, e como $\gamma^0$ conmuta con ${\bf \Sigma}$,

$$
[H, {\bf \Sigma}\cdot\hat{\bf p}] = \hat{p}^k \hat{p}^i\, [\alpha_k, \Sigma_i]
= 2i\,\epsilon_{kil}\,\hat{p}^k\hat{p}^i\,\alpha_l = 0
$$

(simétrico por antisimétrico): $h$ conmuta con $H$ co momento en **calquera** dirección, mentres
que $S_3$ só o fai con ${\bf p}$ en $z$ (sección 3.3).

**Os seus autoestados.** Con ${\bf p}$ en $z$, $h = S_3$: $h\,u_1 = +\frac{1}{2}\,u_1$ e $h\,u_2 = -\frac{1}{2}\,u_2$, e por iso
$u_+ \equiv u_1$, $u_- \equiv u_2$. Nunha dirección calquera son os $u_\pm(\mathrm{p}\,\hat{\bf v})$ da sección 3.4:
os dous bloques, $\chi_\pm(\hat{\bf v})$ e $\pm\kappa\,\chi_\pm(\hat{\bf v})$, son autoestados de $\hat{\bf v}\cdot\vec\sigma$ co mesmo
autovalor, $\pm 1$.

**A antipartícula.** O operador dá cambiados de signo o espín e o momento (sección 2.6), así que a
helicidade física **non** cambia: $v_+ = v_1$, $v_- = v_2$.

**Non é invariante Lorentz se $m \neq 0$.** As rotacións e os *boosts* ao longo de ${\bf p}$ consérvana
(seccións 3.4 e 3.5). Pero un observador que se mova na mesma dirección máis rápido ca a
partícula ve o momento invertido e o espín igual: a helicidade cambia de signo. Só se $m = 0$ non
existe ese observador, e a helicidade é unha propiedade da partícula. A cantidade invariante é a
**quiralidade** ([ext-quiral](./ext-quiral.ipynb)), que coincide coa helicidade no límite $m \to 0$ (ext-quiral, sección 7). [MT4.8]

## 4. As formas bilineais e a corrente

$M_{fi}$ é un invariante Lorentz, e constrúese con espinores. Os espinores non o son:
transfórmanse con $S(\Lambda)$. Esta sección demostra que $\bar\Psi\Psi$ é un **escalar** e
$\bar\Psi\gamma^\mu\Psi$ un **cuadrivector** (e o mesmo con $\gamma^5$), que a corrente **se conserva**, e como se le como
**corrente eléctrica**.

### 4.1. O adxunto e como se transforma

$\Psi^\dagger\Psi$ non serve: $\Psi'^\dagger\Psi' = \Psi^\dagger S^\dagger S\,\Psi$, e $S$ non é unitaria para os *boosts*
(en repouso $u^\dagger u = 2m$; en movemento, $2E$). O que si se cumpre, para calquera $S(\Lambda)$, é

$$
\boxed{
\gamma^0\, S^\dagger\, \gamma^0 = S^{-1}
}
$$

* **Rotación.** $S(\phi\,\hat{z}) = \cos\frac{\phi}{2}\,I - i\sin\frac{\phi}{2}\,\Sigma_3$. $\Sigma_3$ é hermítica e diagonal
  por bloques, así que conmuta con $\gamma^0$ e $S^\dagger = \cos\frac{\phi}{2}\,I + i\sin\frac{\phi}{2}\,\Sigma_3 = S^{-1}$
  (é unitaria). Entón $\gamma^0 S^\dagger \gamma^0 = S^\dagger (\gamma^0)^2 = S^{-1}$.
* ***Boost*.** $S(\eta\,\hat{z}) = \cosh\frac{\eta}{2}\,I + \sinh\frac{\eta}{2}\,\gamma^0\gamma^3$. Como
  $(\gamma^0\gamma^3)^\dagger = \gamma^{3\dagger}\gamma^{0\dagger} = -\gamma^3\gamma^0 = \gamma^0\gamma^3$, é hermítica: $S^\dagger = S$ (**non** é
  unitaria). Pero $\gamma^0$ anticonmuta con $\gamma^0\gamma^3$, e
  $\gamma^0 S^\dagger \gamma^0 = \cosh\frac{\eta}{2}\,I - \sinh\frac{\eta}{2}\,\gamma^0\gamma^3 = S(-\eta\,\hat{z}) = S^{-1}$.

Noutra dirección é igual ($\Sigma_k$ e $\gamma^0\gamma^k$ teñen as mesmas propiedades), e calquera
transformación é un produto de rotacións e *boosts*. Con isto, o **adxunto**
$\bar\Psi \equiv \Psi^\dagger\gamma^0$ transfórmase coa inversa:

$$
\bar\Psi' = \Psi^\dagger S^\dagger \gamma^0 = \Psi^\dagger\gamma^0\,(\gamma^0 S^\dagger \gamma^0) = \bar\Psi\, S^{-1} .
$$

(Metemos $(\gamma^0)^2 = I$ entre $\Psi^\dagger$ e $S^\dagger$.)

### 4.2. $\bar\Psi\Psi$ é un escalar

$$
\bar\Psi'\Psi' = \bar\Psi\, S^{-1} S\, \Psi = \bar\Psi\Psi
\qquad \Rightarrow \qquad
\boxed{\bar\Psi\Psi \;\text{ é invariante Lorentz}}
$$

A fase cancélase ($e^{+ip_\mu x^\mu}e^{-ip_\mu x^\mu} = 1$), así que para as ondas planas abonda cos espinores.
**Comprobación** con $u_1$ en $z$: $\bar u_1 = u_1^\dagger\gamma^0 = N\,(1, 0, -\kappa, 0)$, e

$$
\bar u_1 u_1 = N^2(1 - \kappa^2) = (E+m) - \frac{\mathrm{p}^2}{E+m} = \frac{(E+m)^2 - \mathrm{p}^2}{E+m} = 2m ,
$$

o mesmo que en repouso, $\bar u_1(m)\, u_1(m) = 2m$. En cambio $u_1^\dagger u_1 = N^2(1 + \kappa^2) = 2E$. A
diferenza é o signo menos do bloque de abaixo, que pon $\gamma^0$. Cun espinor xenérico,
$\bar u_s u_s = N^2\left(1 - \frac{{\bf p}^2}{(E+m)^2}\right) = 2m$, e para a antipartícula $\bar v_s v_s = -2m$.

### 4.3. $\bar\Psi\gamma^\mu\Psi$ é un cuadrivector

Coa condición da sección 3.1, $S^{-1}\gamma^\mu S = \Lambda^\mu{}_\nu\gamma^\nu$:

$$
\bar\Psi'\gamma^\mu\Psi' = \bar\Psi\, S^{-1}\gamma^\mu S\, \Psi = \Lambda^\mu{}_\nu\, \bar\Psi\gamma^\nu\Psi
\qquad \Rightarrow \qquad
\boxed{j^\mu \equiv \bar\Psi\gamma^\mu\Psi \;\text{ é un cuadrivector}}
$$

As $\gamma^\mu$ non cambian; o que se transforma son os espinores, e o resultado é o dun
cuadrivector.

**Comprobación** co *boost* en $z$. En repouso, $\bar u\gamma^0 u = u^\dagger u = 2m$ e
$\bar u\gamma^k u = u^\dagger\gamma^0\gamma^k u = 0$ ($\gamma^0\gamma^k$ só ten bloques fóra da diagonal, e o de
abaixo é cero): $j^\mu = (2m, 0, 0, 0)$. O *boost* de rapidez $\eta$ lévaa a

$$
\Lambda^\mu{}_\nu\, j^\nu = (2m\cosh\eta, 0, 0, 2m\sinh\eta) = (2E, 0, 0, 2\mathrm{p}) = 2p^\mu ,
$$

que é o que dá o cálculo directo con $u_s(p)$ (tarefa de perspectivas): $j^0 = N^2(1 + \kappa^2) = 2E$ e
$j^3 = 2N^2\kappa = 2\mathrm{p}$. Cun momento calquera, $\bar u_s\gamma^\mu u_s = 2p^\mu$.

### 4.4. O pseudoescalar e o axial

Como $\gamma^5$ conmuta con $S$ (sección 3.2), metendo $S S^{-1} = I$ onde faga falta:

$$
\bar\Psi'\gamma^5\Psi' = \bar\Psi\, S^{-1}\gamma^5 S\, \Psi = \bar\Psi\gamma^5\Psi,
\qquad
\bar\Psi'\gamma^\mu\gamma^5\Psi' = \bar\Psi\, (S^{-1}\gamma^\mu S)(S^{-1}\gamma^5 S)\, \Psi = \Lambda^\mu{}_\nu\, \bar\Psi\gamma^\nu\gamma^5\Psi .
$$

$\bar\Psi\gamma^5\Psi$ é invariante e $\bar\Psi\gamma^\mu\gamma^5\Psi$ un cuadrivector, como S e V. O que os distingue
deles é a **paridade** (o *pseudo* e o *axial*), que se ve no tema de leptóns.

**Valores para $u_\pm$ con ${\bf p}$ en $z$** (o caso $f = i$). En Pauli-Dirac,
$\gamma^0\gamma^5 = \begin{pmatrix} 0 & I \\ -I & 0 \end{pmatrix}$ e $\gamma^0\gamma^k\gamma^5 = \Sigma_k$. Con $u = (u^a, u^b)$:

* **Pseudoescalar**: $\bar u\gamma^5 u = u^{a\dagger}u^b - u^{b\dagger}u^a = 0$, porque $\kappa$ é real. Anúlase
  para unha partícula libre; só sobrevive nunha transición, $\bar u_f\gamma^5 u_i$.
* **Axial, parte temporal**: $\bar u\gamma^0\gamma^5 u = u^\dagger\gamma^5 u = 2\,\mathrm{Re}(u^{a\dagger}u^b) = \pm 2N^2\kappa = \pm 2\mathrm{p}$.
* **Axial, parte espacial**: $\bar u\gamma^k\gamma^5 u = u^\dagger\Sigma_k u$. Só $\Sigma_3$ sobrevive, e
  $u_\pm^\dagger\Sigma_3 u_\pm = \pm N^2(1 + \kappa^2) = \pm 2E$.

$$
\bar u_\pm\gamma^\mu\gamma^5 u_\pm = \pm 2\,(\mathrm{p}, 0, 0, E) = \pm 2m\, s^\mu,
\qquad s^\mu = \left(\frac{\mathrm{p}}{m}, 0, 0, \frac{E}{m}\right) .
$$

$s^\mu$ é o **cuadrivector de espín**: o *boost* de $(0, 0, 0, 1)$, o espín en $z$ en repouso, porque
$\Lambda\,(0, 0, 0, 1) = (\sinh\eta, 0, 0, \cosh\eta)$. Cumpre $s_\mu p^\mu = 0$ e $s_\mu s^\mu = -1$. A corrente
axial leva o espín, e por iso cambia de signo entre $u_+$ e $u_-$; a vectorial, $2p^\mu$, non.

### 4.5. A corrente consérvase

**A ecuación do adxunto.** Tomando o hermítico conxugado de $i\gamma^\mu\partial_\mu\Psi - m\Psi = 0$:

$$
-i\,\partial_\mu\Psi^\dagger\,\gamma^{\mu\dagger} - m\,\Psi^\dagger = 0 .
$$

Multiplicando por $\gamma^0$ pola dereita e usando $\gamma^{\mu\dagger}\gamma^0 = \gamma^0\gamma^\mu$ (é dicir,
$\gamma^0\gamma^{\mu\dagger}\gamma^0 = \gamma^\mu$, que se comproba coas matrices de Pauli-Dirac):

$$
\boxed{
i\,\partial_\mu\bar\Psi\,\gamma^\mu + m\,\bar\Psi = 0
}
$$

**A conservación.** Derivando o produto e usando as dúas ecuacións,
$\gamma^\mu\partial_\mu\Psi = -im\,\Psi$ e $\partial_\mu\bar\Psi\,\gamma^\mu = +im\,\bar\Psi$:

$$
\partial_\mu j^\mu = (\partial_\mu\bar\Psi)\,\gamma^\mu\Psi + \bar\Psi\gamma^\mu(\partial_\mu\Psi)
= im\,\bar\Psi\Psi - im\,\bar\Psi\Psi = 0 .
$$

Con $j^\mu = (\rho, {\bf j})$, é a **ecuación de continuidade**:

$$
\partial_t\rho + \nabla\cdot{\bf j} = 0, \qquad \rho = j^0 = \Psi^\dagger\Psi \geq 0 .
$$

Integrando nun volume, o que diminúe dentro é o que sae pola superficie, e
$Q = \int \rho\,\mathrm{d}^3x$ consérvase. A diferenza de Klein-Gordon (sección 1.1), a densidade é
**sempre positiva**: era un dos obxectivos de Dirac.

### 4.6. A corrente eléctrica

Multiplicada pola carga $q$ do fermión ($q = -e$ para o electrón), a corrente conservada é a
**corrente eléctrica**, e a súa conservación é a da carga:

$$
j^\mu_{em} = q\,\bar\Psi\gamma^\mu\Psi = (q\rho,\, q\,{\bf j}) .
$$

Para a onda plana, $j^\mu = 2p^\mu$:

$$
\rho = 2E, \qquad {\bf j} = 2{\bf p} = \rho\,\frac{{\bf p}}{E} = \rho\,{\bf v} ,
$$

densidade por velocidade, como unha corrente clásica. Hai $2E$ partículas por unidade de volume (a
normalización) movéndose con ${\bf v} = {\bf p}/E$.

**A antipartícula.** $\bar v_s\gamma^\mu v_s = 2p^\mu$ tamén, pero a súa carga é a oposta: a súa corrente
eléctrica é $-q\,2p^\mu$. Na teoría dunha partícula iso dáo a regra de Feynman-Stückelberg
(inverter as cargas en $\Phi_s$); en QFT sae directamente.

É a esta corrente á que se acopla o fotón, como **corrente por campo**, $q\,j^\mu A_\mu$: a
simetría *gauge* de [perspectivas](./perspectivas.ipynb). [MT4.3, MT4.5.1]

*Cuestións sobre a sección 4:*

## 5. A simetría *gauge* local e o fotón

En [perspectivas](./perspectivas.ipynb) afírmase que a interacción do fermión co fotón **sae**
de esixir que a física non cambie cunha fase local. Aquí está a demostración.

### 5.1. Fase global e fase local

**Global.** Cunha fase constante, $\Psi' = e^{iq\theta}\,\Psi$, a ecuación de Dirac non cambia: a fase sae
da derivada e cancélase. Tampouco cambian os bilineais, $\bar\Psi' = \bar\Psi\,e^{-iq\theta}$, e
$\bar\Psi'\Gamma\Psi' = \bar\Psi\Gamma\Psi$. A fase global de $\Psi$ non é observable.

**Local.** Cunha fase que depende do punto, $\theta = \theta(x)$,

$$
\Psi'(x) = e^{iq\theta(x)}\,\Psi(x),
$$

a derivada actúa tamén sobre a fase, $\partial_\mu\Psi' = e^{iq\theta}\left(\partial_\mu\Psi + iq\,(\partial_\mu\theta)\,\Psi\right)$, e
sobra un termo:

$$
(i\gamma^\mu\partial_\mu - m)\,\Psi' = e^{iq\theta}\left[(i\gamma^\mu\partial_\mu - m)\,\Psi - q\,\gamma^\mu(\partial_\mu\theta)\,\Psi\right]
= -q\,e^{iq\theta}\,\gamma^\mu(\partial_\mu\theta)\,\Psi \neq 0 .
$$

Se $\Psi$ é solución, $\Psi'$ xa non o é: a ecuación libre **non** é invariante baixo unha fase local.

### 5.2. A derivada covariante

Para absorber o termo que sobra fai falta un campo vectorial, $A_\mu(x)$, que se transforme á vez
que $\Psi$ e compense $\partial_\mu\theta$:

$$
\boxed{
D_\mu \equiv \partial_\mu + iq\,A_\mu(x), \qquad
\Psi' = e^{iq\theta}\,\Psi, \qquad A'_\mu = A_\mu - \partial_\mu\theta
}
$$

**A demostración.** Cos campos transformados:

$$
D'_\mu\Psi' = \left(\partial_\mu + iq\,A'_\mu\right) e^{iq\theta}\,\Psi
= e^{iq\theta}\left(\partial_\mu\Psi + iq\,(\partial_\mu\theta)\,\Psi + iq\,A_\mu\Psi - iq\,(\partial_\mu\theta)\,\Psi\right)
= e^{iq\theta}\,D_\mu\Psi .
$$

$D_\mu\Psi$ transfórmase **como $\Psi$**, coa mesma fase: de aí o nome de *covariante*. Entón

$$
(i\gamma^\mu D'_\mu - m)\,\Psi' = e^{iq\theta}\,(i\gamma^\mu D_\mu - m)\,\Psi = 0 ,
$$

e a ecuación $(i\gamma^\mu D_\mu - m)\,\Psi = 0$ é **invariante** baixo a fase local. Desenvolvendo $D_\mu$:

$$
(i\gamma^\mu\partial_\mu - m)\,\Psi = q\,\gamma^\mu A_\mu\,\Psi :
$$

o fermión libre á esquerda e, á dereita, o termo **novo**: a interacción con $A_\mu$, con
intensidade $q$. A mesma $A_\mu$ serve para todos os fermións, cada un coa súa carga: por iso o
acoplamento é **universal**.

### 5.3. O acoplamento mínimo

Con $i\partial_\mu \leftrightarrow p_\mu$, a derivada covariante é $iD_\mu = i\partial_\mu - qA_\mu \leftrightarrow p_\mu - qA_\mu$. Con
$A^\mu = (\phi, {\bf A})$, é dicir, $A_\mu = (\phi, -{\bf A})$:

$$
E \;\to\; E - q\phi, \qquad {\bf p} \;\to\; {\bf p} - q{\bf A} ,
$$

a **substitución mínima** do electromagnetismo clásico. En forma hamiltoniana (multiplicando por $\gamma^0$,
como na sección 1.4, con $\alpha_k = \gamma^0\gamma^k$):

$$
i\partial_t\Psi = \left[\vec\alpha\cdot(\hat{\bf p} - q{\bf A}) + \gamma^0 m + q\phi\right]\Psi ,
$$

o análogo relativista de $H = ({\bf p} - q{\bf A})^2/2m + q\phi$. A simetría local non engadiu nada que non
soubésemos do electromagnetismo: **dedúxoo**.

### 5.4. O campo do fotón

$A_\mu$ é un campo coa súa propia ecuación, que tamén debe ser invariante baixo
$A_\mu \to A_\mu - \partial_\mu\theta$.

**O tensor de campo é invariante.** Con $F_{\mu\nu} \equiv \partial_\mu A_\nu - \partial_\nu A_\mu$ (as súas compoñentes son
${\bf E}$ e ${\bf B}$):

$$
F'_{\mu\nu} = F_{\mu\nu} - \partial_\mu\partial_\nu\theta + \partial_\nu\partial_\mu\theta = F_{\mu\nu} .
$$

Os campos eléctrico e magnético non cambian: o que cambia, $A_\mu$, non é observable.

**As ecuacións de Maxwell**, coa corrente do fermión como fonte,

$$
\boxed{
\partial_\mu F^{\mu\nu} = q\,\bar\Psi\gamma^\nu\Psi
}
$$

son invariantes: $F^{\mu\nu}$ non cambia, e a corrente tampouco ($\bar\Psi'\gamma^\nu\Psi' = \bar\Psi\gamma^\nu\Psi$). Sen fonte, é
a ecuación do fotón libre da táboa de perspectivas.

**A conservación é obrigatoria.** Derivando outra vez, $\partial_\nu\partial_\mu F^{\mu\nu} = 0$, porque $F^{\mu\nu}$ é
antisimétrico e $\partial_\nu\partial_\mu$ simétrico. Logo $\partial_\nu\,(q\,\bar\Psi\gamma^\nu\Psi) = 0$: Maxwell só é consistente
se a corrente se conserva, e a de Dirac consérvase (sección 4.5). A carga que se conserva é a
da simetría.

**O fotón non ten masa.** Unha masa engadiría un termo $m_\gamma^2 A^\nu$ á ecuación,
$\partial_\mu F^{\mu\nu} + m_\gamma^2 A^\nu = q\,\bar\Psi\gamma^\nu\Psi$. Baixo a transformación cambia en
$-m_\gamma^2\,\partial^\nu\theta$, que non se compensa con nada: a simetría local esixe $m_\gamma = 0$. [MT10.1]
