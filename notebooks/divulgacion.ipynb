### Partículas a alta enerxía

Pero durante os primeiros segundos do Universo, as partículas tiñan enerxías altas. Os momentos iniciais foron unha 'sopa' de partículas de alta enerxía.

O Universo comezou cun Big Bang, pero agora vólvese 'zen' e tranquilo co paso do tempo.


Cando o Universo arrefriou lixeiramente, as partículas formaron primeiro protóns e neutróns, despois núcleos, despois átomos e, finalmente, estrelas e galaxias.

Agora sabemos que algunhas partículas se desintegran e se converten noutras, por exemplo:


$$
n \to p + e + \bar{\nu}_e
$$

E temos moita, moita sorte!!

Despois da súa creación, o neutrón vive 10 minutos, tempo dabondo para crear os primeiros núcleos estables de helio do Universo, o que levou 6 minutos!

### Os catro fabulosos

Na desintegración do neutrón aparecen as catro partículas principais que compoñen o noso Universo hoxe:

$$
n \to p + e + \bar{\nu}_e
$$

Isto chámase: *desintegración beta*!

## Introdución ao Modelo Estándar

O Modelo Estándar da física de partículas é un dos maiores logros da física. Por desgraza, o seu nome é moi pobre!

Clasifica todas as partículas elementais e as interaccións entre elas.

Todo o que sabemos, con todos os detalles, está aquí: *Particle Data Group* ([PDG](https://pdg.lbl.gov/))



Volvamos á *desintegración beta*.

Esta desintegración pode ocorrer en certos núcleos: un neutrón convértese nun protón, e escapan un electrón e un neutrino.

| :--: |

Durante un momento, todo era sinxelo: neutrón, protón, electrón, neutrino.

Pero ao observar as interaccións dos raios cósmicos (partículas extraterrestres que entran na atmosfera), apareceu unha nova partícula: o muón!

O muón era coma o electrón, pero con maior masa. O muón é o irmán do electrón!

Para que necesitamos o muón? O noso universo era tan sinxelo, con só catro partículas!

Non só iso, este home:

Esta ecuación é tan importante que deberías poñela como imán no teu frigorífico!

Por que é tan importante?

Dirac unificou as dúas grandes revolucións, a relatividade e a mecánica cuántica, para obter a súa ecuación.

Pero nos anos 60 chegou a paz. Os físicos puxeron orde no caos!

Hai dous tipos de partículas:

* **hadróns**, como o neutrón e o protón, que están feitos de **quarks**.

* **leptóns**, como o electrón e o neutrino, e o muón, o neutrino muónico, e o tau e o neutrino tauónico!

Só os quarks $(u, d)$ que forman o protón e o neutrón e o (e, $\nu$) son estables!



Os quarks teñen carga eléctrica: u (2/3) e d (-1/3), e sempre aparecen en grupos, nunca sós! Forman **hadróns**.
