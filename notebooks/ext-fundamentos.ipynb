## 1. Normalización en mecánica cuántica relativista

En mecánica cuántica normalizamos a función de ondas, $\psi$, nun cubo de lados $L$, con volume $V = L^3$, á unidade:

$$
\int_V \psi^* \psi \, \mathrm{d}^3x = 1 
$$

É dicir, a partícula está contida en $V$.

O número de estados dispoñibles nese volume é:

$$
\frac{V \, \mathrm{d}^3{\bf p}}{(2\pi)^3},
$$

que vén de expresar $\frac{\mathrm{d}^3 {\bf x} \, \mathrm{d}^3 {\bf p}}{h^3}$ en unidades naturais, onde $h = 2\pi$.

O volume, porén, non é un invariante Lorentz.

Un cubo de lado $L$ visto desde un sistema no que se move con factor de Lorentz $\gamma$
contráese na dirección do movemento:

$$
V' = \frac{L^3}{\gamma} = \frac{V}{\gamma},
$$

e a densidade de partículas, $1/V$, **aumenta** nun factor $\gamma$.

Se nese cubo estaba contida a partícula con enerxía $E_0 $, no sistema de referencia con factor $\gamma$ a súa enerxía é:

$$
E' = \gamma E_0
$$


Vemos que a densidade, $1/V$, e a enerxía, $E$, se transforman da mesma forma, cun factor $\gamma$.

En mecánica cuántica relativista, normalizamos a función de ondas, $\Psi$, nun volume $V$ a $2E$:

$$
\int_V \Psi^* \Psi \, \mathrm{d}^3 x = 2E,
$$

O factor $2$ é por conveniencia.

É dicir, $2E$ partículas por unidade de volume. Esta normalización si é invariante porque a densidade, $1/V$, e a enerxía, $E$, cambian co mesmo factor ao cambiar o sistema de referencia.

## 2. Elemento de matriz relativista

Como consecuencia da normalización relativista da función de ondas $\Psi$, o elemento de matriz relativista $M_{fi} = \langle \Psi_f| H_{int}| \Psi_i \rangle$ relaciónase co non relativista $T_{fi} = \langle \psi_f | H_{int}| \psi_i \rangle$ mediante un factor que provén da normalización. 

$$
M_{fi} = \langle \Psi_f | H_{int}| \Psi_i \rangle = \left(\prod_i \sqrt{2E_i} \right) \langle  \psi_f | H_{int}| \psi_i \rangle =  \left(\prod_i \sqrt{2E_i} \right)  T_{fi}
$$
onde o índice $i$ percorre as $n$ partículas iniciais e finais.

As expresións da taxa de transición (a anchura) ou da sección eficaz no tratamento relativista incorporan polo tanto o seguinte factor:

$$
\frac{1}{\prod_i (2E_i)},
$$

debido á normalización relativista.

## 3. As fórmulas xerais: anchura e sección eficaz

A **regra de ouro de Fermi** dá a taxa de transición como o elemento de matriz ao cadrado pola
densidade de estados finais. Aplicada a unha partícula que se desintegra dá a **anchura**,
$\Gamma$; aplicada a dúas que chocan, dividida polo fluxo, a **sección eficaz**, $\sigma$.
Escribímolas primeiro na súa forma non relativista e despois na relativista, cos factores
$1/(2E)$ da normalización.

### Anchura de desintegración

A expresión non relativista da anchura de desintegración dunha partícula de enerxía $E_a$ a $m$ partículas finais é:


$$
\Gamma  = (2\pi)^4 \int |T_{fi}|^2 \delta^4 \left(\sum_k p_k - p_a \right) \, \prod_i \frac{\mathrm{d}^3{\bf p}_i}{(2\pi)^3}
$$

onde $p_a$ é o cuadrimomento da partícula inicial e $p_i$ (ou $p_k$) os das finais, $i = 1, \dots, m$. Lembremos que as deltas de Dirac impoñen a conservación da enerxía e do momento entre a partícula inicial e as finais. 

E a relativista:

$$
\Gamma = \frac{(2\pi)^4}{2 E_a} \int |M_{fi}|^2 \delta^4 \left(\sum_k p_k - p_a \right) \, \prod_i \frac{\mathrm{d}^3{\bf p}_i}{(2\pi)^3 (2E_i)}
$$

### Sección eficaz da interacción de dúas partículas $a + b$

Sexa a interacción entre dúas partículas $a, b$ que dá lugar a $m$ partículas finais.

Sexa un fluxo de partículas $a$, cunha densidade de $N_a$ partículas nun volume $V$, que se move contra $N_b$ partículas branco nun volume $V$ que á súa vez se moven en sentido oposto con $v_b$.

Asignámoslles unha sección eficaz $\sigma$ de interacción ás partículas branco $b$. A frecuencia de interaccións corresponderá ao número de partículas $a$ que atravesan, por unidade de tempo, un cilindro de base $\sigma$ e altura $(v_a + v_b)\,\mathrm{d}t$ para cada partícula $b$.

| |
|:--|
|<img src="./imgs/fun_drawing_xsection.png" width = 400 align="center">|
| (a) unha partícula $a$ percorre en $\mathrm{d}t$ un cilindro de lonxitude $(v_a+v_b)\,\mathrm{d}t$ [MT3.4] |
| (b) visión transversal do cilindro, as partículas $b$ nel co disco da sección eficaz|


É dicir, o número de interaccións nun volume $V$ e nun tempo $\mathrm{d}t$ é:

$$
\mathrm{d}N = (v_a + v_b) \sigma \, n_a N_b \, \mathrm{d}t
$$

O factor $(v_a + v_b) n_a$ é o fluxo de partículas $a$, $\phi_a$, con velocidade $v_a$ que atravesan unha superficie perpendicular á súa velocidade, cando a partícula $b$ se move contra elas con velocidade $v_b$.

Logo, a frecuencia de interaccións, $R$, nun volume, V, onde hai $N_b$ partículas branco, ao que chega un fluxo $\phi_a$ de partículas $a$ é:

$$
R = \phi_a \, \sigma \, N_b
$$



En mecánica cuántica adóptase a normalización dunha partícula nunha unidade de volume, $V = 1, n_a = 1, n_b = 1$; nese caso:

$$
R = (v_a + v_b) \, \sigma, \;\;\; \sigma = \frac{R}{(v_a+v_b)}
$$

Con esa normalización a unha partícula por unidade de volume, a expresión non relativista para a sección eficaz de $a+b$ para producir $m$ partículas é:

$$
\sigma = \frac{(2\pi)^4}{|v_a + v_b|} \int |T_{fi}|^2 \delta^4 \left(\sum_k p_k - p_a - p_b \right) \, \prod_i \frac{\mathrm{d}^3{\bf p}_i}{(2\pi)^3},
$$

onde $v_a, v_b$ son os módulos das velocidades, considerando que as partículas colisionan nunha mesma dirección e teñen sentidos opostos. As deltas de Dirac impoñen a condición de conservación da enerxía e do momento entre os estados inicial e final.

A expresión relativista, co factor $1/(2E)$ de cada partícula, é:

$$
\sigma = \frac{(2\pi)^4}{(2 E_a) (2 E_b) (\beta_a + \beta_b)} \int |M_{fi}|^2 \delta^4 \left(\sum_k p_k - p_a -p_b\right) \, \prod_i \frac{\mathrm{d}^3{\bf p}_i}{(2\pi)^3 (2E_i)}
$$

Na versión relativista, cada un dos factores da sección eficaz é invariante Lorentz:

   * o elemento de matriz ao cadrado: $|M_{fi}|^2 = | \langle \Psi_f| H_{int} | \Psi_i\rangle |^2$
   
   * a densidade de estados: $\frac{\mathrm{d}^3 {\bf p}}{(2\pi)^3 (2E)}$
   
   * o termo asociado ao fluxo: $4 E_a E_b (\beta_a + \beta_b)$ (para feixes colineais)
   
Nótese que, a partir da propiedade da función delta de Dirac, $\int_{E>0} \delta(E^2-{\bf p}^2 - m^2) \, \mathrm{d}E = \frac{1}{2E}$, podemos reescribir:

$$
\frac{\mathrm{d}^3{\bf p}}{(2\pi)^3 (2E)} = \int \delta(p^2-m^2) \, \theta(p^0) \frac{\mathrm{d}^4p}{(2\pi)^3}
$$

*Cuestión*: Demostrar que $4 E_a E_b (\beta_a + \beta_b) = 4 \sqrt{(p_a \cdot p_b)^2 - m^2_a m^2_b}$, que no CM vale $4 \, p^*_i \sqrt{s}$

## 4. Cinemática a dous corpos: invariantes de Mandelstam

| | 
| :-- |
| <img src="./imgs/feynman_stchannels.png" width = 400 align="center">|
| Diagramas asociados aos invariantes de Mandelstam: (esquerda) dispersión, (dereita) aniquilación|


Os cuadrimomentos transferidos, $q^2$, entre as correntes, dos diagramas da figura, denótanse con:

$$
t = (p_c - p_a)^2 = (p_d - p_b)^2, \;\;\; s = (p_a +p_b)^2 = (p_c + p_d)^2,
$$

onde $p_\alpha$, $\alpha = a, b, c, d$ son os cuadrimomentos das partículas, 

e denomínanse **invariantes de Mandelstam**, que corresponden ás **canles** $t$ de dispersión e $s$ de aniquilación, respectivamente.


### O invariante $s$: enerxía e momento no CM

A cantidade $\sqrt{s}$ é a enerxía no centro de masas (CM) dunha aniquilación

$$
s = (p_a + p_b)^2 = (E^*_a + E^*_b)^2 - ({\bf p}^*_a +{\bf p}^*_b)^2 = (E^*_a + E^*_b)^2,
$$

dado que ${\bf p}^*_i = {\bf p}^*_a = -{\bf p}^*_b$ (véxase a figura da dispersión de dous corpos arriba).

Na literatura denótase $\sqrt{s}$ para indicar a **enerxía no centro de masas**.

Podemos calcular $E^*_a, E^*_b$ a partir de $s$ e das masas $m_a, m_b$.

Se calculamos $E^{*2}_a$ tendo en conta que $E^*_a = (\sqrt{s} - E^*_b)$, e $p^*_a = p^*_b = p^*_i$, obtemos:

$$
E^{*2}_a = m^2_a + p_i^{*2} = E^{*2}_b - 2 \sqrt{s}E^*_b + s = m^2_b + p_i^{*2} - 2 \sqrt{s}E^*_b + s \\
m^2_a = m^2_b + s - 2 \sqrt{s} E^*_b
$$


e como o tratamento para $a$ é idéntico ao de $b$, obtemos:

$$
E^*_a = \frac{s + m^2_a - m^2_b}{2 \sqrt{s}}, \;\; E^*_b = \frac{s + m^2_b - m^2_a}{2 \sqrt{s}}.
$$



O momento será:
    
$$
p^{*}_i = \sqrt{E^{*2}_a - m^2_a} = \sqrt{E^{*2}_b - m^2_b}, \qquad 2 p^{*2}_i = E^{*2}_a + E^{*2}_b - m^2_a - m^2_b
$$

polo tanto

$$
p^{*2}_i = \frac{E^{*2}_a + E^{*2}_b - m^2_a - m^2_b}{2} = \\ 
\frac{\left[s + (m^2_a - m^2_b) \right]^2 + \left[s - (m^2_a - m^2_b)\right]^2 - 4 s (m^2_a + m^2_b)}{8 s} = \\
\frac{s^2 + (m^2_a - m^2_b)^2 - 2 s (m^2_a + m^2_b)}{4s}
$$



como: 

$$
m^2_a - m^2_b = (m_a + m_b) (m_a - m_b), \\ 2(m^2_a + m^2_b) = (m_a+m_b)^2 + (m_a-m_b)^2
$$

temos:

$$
p^{*2}_i = \frac{s^2 + (m_a+m_b)^2(m_a-m_b)^2 - s \left[ (m_a+m_b)^2 + (m_a - m_b)^2 \right] }{4s} = \\
       = \frac{\left[s- (m_a+m_b)^2\right] \, \left[s - (m_a-m_b)^2\right]}{4s}
$$

polo que o momento inicial é:

$$
p^*_i = \frac{\sqrt{\left[s- (m_a+m_b)^2\right] \, \left[s - (m_a-m_b)^2\right]}}{2\sqrt{s}}
$$


Como a situación é idéntica para as partículas finais, temos:

$$
E^*_c = \frac{s + m^2_c - m^2_d}{2 \sqrt{s}}, \;\; E^*_d = \frac{s + m^2_d - m^2_c}{2 \sqrt{s}} \\
p^*_f = \frac{\sqrt{\left[s- (m_c+m_d)^2\right] \, \left[s - (m_c-m_d)^2\right]}}{2 \sqrt{s}}
$$

No caso da desintegración a dous corpos, $a \to c + d$, a enerxía no CM é $ \sqrt{s} = m_a$.

Os momentos e enerxías finais son nese caso:

$$
E^*_c = \frac{m^2_a + m^2_c - m^2_d}{2 m_a}, \;\; E^*_d = \frac{m^2_a + m^2_d - m^2_c}{2 m_a} \\
p^*_f = \frac{1}{2m_a}\sqrt{\left[m^2_a- (m_c+m_d)^2\right] \, \left[m^2_a - (m_c-m_d)^2\right]}
$$

## 5. Anchura de desintegración a dous corpos

Consideremos a desintegración dunha partícula $a$ en dúas, $b, c$, con masas $m_a, m_b, m_c$.

| |
|:--:|
|<img src="./imgs/intro_drawing_decay2.png" width = 400 align="center">|
| Esquema da desintegración da partícula $a$ en dúas, $b, c$, no CM|

O sistema máis conveniente para tratar o proceso é o CM, onde a partícula $a$ está en repouso e ten $E = m_a$.

As partículas $b, c$ saen con momentos opostos ${\bf p}^* = {\bf p}_b = - {\bf p}_c$. 

O módulo do momento das partículas saíntes, $p^*$, é (véxase a sección 4): 

$$
p^*= \frac{1}{2m_a} \sqrt{[m^2_a - (m_b+m_c)^2] \, [m^2_a - (m_b - m_c)^2] }
$$

A anchura de desintegración, $\Gamma$, é a taxa de transición da regra de ouro de Fermi.

Integramos usando $\delta({\bf p}_b + {\bf p}_c)$ en  $\mathrm{d}^3p_{c}$ directamente.

$$
\Gamma = \frac{1}{(2 E) (2\pi)^2} \int |M_{fi}|^2 \delta \left(E - E_b - E_c \right) \frac{p^2_b }{4 E_b E_c}\,  \mathrm{d}p_b \mathrm{d}\Omega^*,
$$

onde $E_b = \sqrt{m^2_b + p^2_b}, \; E_c = \sqrt{m^2_c + p^2_b}$ e $\mathrm{d}\Omega^*$ é o diferencial de ángulo sólido.


Se aplicamos a seguinte propiedade da delta de Dirac: $\int \delta(f(p))\,  \mathrm{d} p = \left|\frac{\mathrm{d}f(p)}{\mathrm{d}p} \right|_{p'}^{-1}$, onde $p'$ cumpre $f(p') = 0$,

con $f(p_b) = E -  \sqrt{m^2_b + p^2_b}- \sqrt{m^2_c + p^2_b}$ respecto de $p = p_b$, e $p' = p^*$:

$$
\frac{\mathrm{d}f(p_b)}{\mathrm{d}p_b} = \frac{p_b}{\sqrt{m^2_b + p^2_b}} + \frac{p_b}{\sqrt{m^2_c + p^2_c}} = p_b \frac{E_b + E_c}{E_b E_c} = p_b \frac{E}{E_b E_c}
$$

A anchura de desintegración queda:

$$
\Gamma = \frac{1}{8 \pi^2 E} \frac{(p^*)^2}{4 E_b E_c} \frac{E_b E_c}{p^* E} \int |M_{fi}|^2 \, \mathrm{d}\Omega^* = 
\frac{1}{8\pi^2}\frac{p^*}{4 E^2} \int |M_{fi}|^2 \, \mathrm{d}\Omega^*
$$

A anchura de desintegración, neste caso, só depende de $p^*$, $m_a$ e do elemento de matriz, se poñemos $E = E_a = m_a$ no CM:

$$
\Gamma = \frac{p^*}{32 \pi^2 m^2_a} \int_{\Omega} |M_{fi}|^2 \mathrm{d}\Omega^*
$$

## 6. Sección eficaz dunha interacción de dous a dous corpos

Sexa a interacción entre dúas partículas incidentes, $a, b$, que dá lugar a dúas partículas finais, $c, d$.

| |
|:--:|
|<img src="./imgs/intro_drawing_int2.png" width = 400 align="center">|
| Esquema da interacción das partículas $a, b$ a $c, d$ no CM|

O sistema máis conveniente para calcular a sección eficaz é o CM. Neste sistema os momentos iniciais, e os finais, son opostos:

$$
{\bf p}^*_i = {\bf p}_a = - {\bf p}_b, \;\; {\bf p}^*_f = {\bf p}_c = - {\bf p}_d
$$

Nótese que as partículas están nun plano, e o único parámetro libre (ademais do azimut, que se integra) é o ángulo $\theta^*$ entre ${\bf p}^*_f$ e ${\bf p}^*_i$; este último define o eixe.

A sección eficaz obtense a partir da expresión xeral, onde $E = E_a + E_b$, $p_b = p_a = p^*_i$ e $p_d = p_c = p^*_f$.

O termo do espazo de fases calcúlase igual que na anchura de desintegración a dous corpos (sección 5), só que agora $E = E_a + E_b$:

$$
\frac{1}{(2\pi)^2}\frac{p^*_f}{4 E} \int  |M_{fi}|^2 \, \mathrm{d}\Omega^*
$$


O termo asociado ao fluxo:

$$
4 E_a E_b (\beta_b + \beta_a) = 4 E_a E_b \left(\frac{p_a}{E_a} + \frac{p_b}{E_b}\right) = 4 p^*_i (E_a + E_b) = 4 E \, p^*_i
$$

A sección eficaz queda:

$$
\sigma = \frac{1}{64 \pi^2 E^2} \frac{p^*_f}{p^*_i} \int_\Omega |M_{fi}|^2 \, \mathrm{d}\Omega^*
$$

Se substituímos $E^2 =  s$, onde $s = \left[(E_a, {\bf p}_a) + (E_b, {\bf p}_b)\right]^2 = [(E_a + E_b, {\bf 0})]^2$ é o cadrado do cuadrimomento total, é dicir, a enerxía no CM ao cadrado:

$$
\sigma = \frac{1}{64 \pi^2 s} \frac{p^*_f}{p^*_i} \int_\Omega |M_{fi}|^2 \, \mathrm{d}\Omega^*
$$

## 7. O invariante $t$


O invariante $t = (p_c - p_a)^2 = (p_d - p_b)^2$ é:

$$
t = (p_c - p_a)^2 = p^2_c + p^2_a - 2 p_c \cdot p_a = m^2_a + m^2_c + 2 |{\bf p}_a| |{\bf p}_c| \cos \theta_{ac} - 2 E_a E_c \\
t = (p_d - p_b)^2 = p^2_d + p^2_b - 2 p_d \cdot p_b = m^2_b + m^2_d + 2 |{\bf p}_b| |{\bf p}_d| \cos \theta_{bd} - 2 E_b E_d
$$

onde $\theta_{ac}, \theta_{bd}$ son os ángulos entre os momentos de $ac$ e $bd$, respectivamente.

No sistema de laboratorio, con $b$ en repouso, $p_b = (m_b, {\bf 0})$:

$$
 t = m^2_b + m^2_d - 2m_b E_d, \;\; E_d = \frac{m^2_b + m^2_d - t}{2m_b}
$$


No sistema CM (véxase a figura da interacción a dous corpos), $|{\bf p}_a| = p^*_i, \; |{\bf p}_c| = p^*_f$ e $\theta_{ac} = \theta^*$

$$
t = m^2_a + m^2_c + 2 p^*_i p^*_f \cos \theta^* - 2E^*_a E^*_c, \\
\cos \theta^* = \frac{t - m^2_a - m^2_c + 2E^*_aE^*_c}{2 p^*_i p^*_f}
$$




Como $t$ só depende de $\cos\theta^*$ a través de $\mathrm{d}t = 2\,p^*_i\,p^*_f \; \mathrm{d}\cos\theta^*$, e
$\mathrm{d}\Omega^* = \mathrm{d}\phi^* \, \mathrm{d}\cos\theta^*$, se $|M_{fi}|^2$ non depende do azimut a sección eficaz
diferencial en $t$ é:

$$
\boxed{
\frac{\mathrm{d}\sigma}{\mathrm{d}t} = \frac{|M_{fi}|^2}{64 \pi \, s \, p^{*2}_i}
}
$$

que é invariante Lorentz, e é a forma que se usa na práctica.
