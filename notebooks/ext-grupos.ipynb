## Sobre os grupos

Os grupos de simetría son unha das ferramentas matemáticas principais da Física de Partículas:

  * están relacionados coas **cantidades conservadas**, por exemplo coas cargas que definen unha partícula.
 
  * serven para **representar os estados das partículas**. Neste sentido usamos as bases das representacións irredutibles para representar os estados das partículas.

Imos lembrar o concepto de grupo e de representación, e despois revisaremos os grupos principais $U(1), SU(2)$ e $SU(3)$.

Os grupos máis comúns en física de partículas son:

 * O **grupo de Lorentz-Poincaré**, asociado ás rotacións, translacións e cambios de sistemas de referencia inerciais no espazo-tempo.
   
     As representacións máis comúns que atopamos son: **escalar, vectorial, espinorial**
     
     Para a discusión sobre o grupo e as representacións de Lorentz véxase a extensión [ext-trans-Lorentz].
     
 * Os **grupos unitarios** (cuxa representación son matrices unitarias) de dimensións pequenas:
   
   * **U(1)** rotación dunha fase dun número complexo, que asociamos coa conservación da carga eléctrica. 
      
   * **SU(2)** matrices unitarias de dimensión 2 con determinante positivo, que asociamos ao espín e a outras cantidades físicas como o isospín feble. A súa base fundamental é un doblete con posicións arriba e abaixo.
      
   * **SU(3)** matrices unitarias de dimensión 3 con determinante positivo, que asociamos á cor. A súa base fundamental é un triplete, que asociamos, no caso da cor, coas tres cores: vermello, verde e azul.

### Definición de grupo

Un grupo é unha colección de elementos, $\{g_i\}$, finita ou infinita, entre os que hai definida unha operación, a multiplicación ($\cdot$), e que cumpre as seguintes propiedades:

   * completitude: $g_i \cdot g_j = g_k$ (a multiplicación de dous elementos do grupo dá outro elemento do grupo)
   
   * asociatividade: $g_i \cdot (g_j \cdot g_k) = (g_i \cdot g_j) \cdot g_k$
   
   * existencia da identidade, $1$: $ g_i \cdot 1 = 1 \cdot g_i = g_i$
   
   * existencia da inversa, $g_i \cdot g^{-1}_i = 1$ para todo $g_i$ do grupo.
   
Dicimos que o grupo é abeliano se é conmutativo: $g_i \cdot g_j = g_j \cdot g_i$; se non, é non abeliano.



#### Clasificación de grupos

Os grupos principais son:

* **espazo-temporais**, o que inclúe o grupo de Lorentz

* dun **espazo interno**, chamado **isotópico**, que poden ser globais ou locais (isto é, que dependen do punto espazo-temporal).

Tamén se poden clasificar os grupos en:

* **continuos**, se o grupo depende dun parámetro $\theta$ continuo; poden ser compactos ou non dependendo de se o intervalo de valores que toma $\theta$ é aberto ou pechado. Por exemplo, o grupo de Lorentz, que depende do parámetro $\beta = v/c$, non é compacto porque non pode tomar o valor $\beta = 1$.

* **discretos**, onde hai un conxunto finito de elementos. Por exemplo, a inversión por paridade forma un grupo discreto dun elemento.

En Física de Partículas xogan un papel fundamental os **grupos de Lie**, que son aqueles cuxos elementos se relacionan de forma continua e diferenciable con respecto a un conxunto de parámetros $\theta$.


### Representacións

Chamamos **representación lineal $D(g)$** ao conxunto de matrices asociadas cos elementos $g$ dun grupo $G$ que cumpren **as mesmas regras de multiplicación** que os elementos de $G$.

Isto é:

$$
 g_i \cdot g_j = g_k \; \Rightarrow D(g_i) \, \cdot D(g_j) = D(g_k)
$$

A **dimensión da representación**, $n$, é a dimensión das matrices.

Os vectores de dimensión $n$ sobre os que actúan as matrices forman a **base da representación**.

Chamamos **representación fundamental** a aquela representación do grupo coa dimensión menor. 

#### Representación irredutible

Dicimos que unha representación $D(g)$ dun grupo $G$ é redutible se pode descompoñerse en representacións máis pequenas.

Sexa a representación $D(g)$ dun grupo, con matrices de dimensión $n\times n$; a representación é redutible se a matriz $D(g)$ se pode descompoñer en diagonal con matrices de menor dimensión, $D_1(g), D_2(g)$, da forma:

$$
D(g) = \begin{pmatrix} D_1(g) & 0 \\ 0 & D_2(g) \end{pmatrix}
$$

Tamén podemos dicir que a base da representación $D(g)$ se pode descompoñer en varias bases, de tal forma que os elementos de cada subbase forman o seu propio grupo. Isto é, os elementos da base de $D_1(g)$ non se mesturan cos elementos da base de $D_2(g)$.

Como dixemos antes, os estados das partículas represéntanse nas bases da representación irredutible.

### Exemplo: o espín

O conxunto de matrices complexas unitarias con determinante unidade (isto é, *propias*) forma o grupo $SU(2)$. 

A álxebra deste grupo é igual á do momento angular, e usamos este grupo para representar o espín dos fermións $s = 1/2$, que contén dous estados na súa base fundamental, o doblete:

$$
\begin{pmatrix} 1 \\ 0 \end{pmatrix} \equiv \, \uparrow, \;\; \begin{pmatrix} 0 \\ 1 \end{pmatrix} \equiv \, \downarrow.
$$


Os xeradores do grupo son os operadores proxección de espín nas coordenadas $x, y, z$:
$$
S^i = \frac{1}{2} \sigma^i
$$
onde $\sigma^i$ son as matrices de Pauli.

$$
\sigma^1 = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \;\;
\sigma^2 = \begin{pmatrix} 0 & -i \\ i & 0 \end{pmatrix}, \;\;
\sigma^3 = \begin{pmatrix} 1 & 0 \\ 0 & -1\end{pmatrix}.
$$


que cumpren a seguinte álxebra
$$
[S^i, S^j] = \epsilon_{ijk} S^k
$$
onde $\epsilon_{ijk}$ é o tensor de Levi-Civita. Esta é a álxebra do momento angular.


As matrices do grupo $SU(2)$ constrúense a partir dos xeradores mediante exponenciación e o seu desenvolvemento de Taylor. 
$$
U(\theta) = e^{i\vec{\theta} \cdot {\bf S}} = e^{i \theta_i S^i}
$$
onde $\theta_i$ son tres ángulos de rotación arredor dos eixes $x, y, z$. 

Por exemplo, a rotación dun ángulo $\theta$ arredor do eixe $y$ queda:
$$
U(\theta) = e ^{i \theta S^2} = \sum_n \frac{1}{n!}(i\theta S^2)^n = 1 + i \theta S^2 - \frac{1}{2} \theta^2  (S^2)^2 + \dots 
$$

dado que $(\sigma^2)^2 = I$ obtemos

$$
U(\theta) = \begin{pmatrix} \cos \theta/2 & - \sin \theta/2 \\ \sin \theta/2 & \cos \theta/2 \end{pmatrix}
$$




Lembremos que, dado que $U$ é unitaria, os xeradores, $S^i$, son hermíticos e, polo tanto, teñen asociados valores observables.

Os tres xeradores, $S^i$, non conmutan entre si. Pero o operador momento total ao cadrado
$$
{\bf S}^2 = (S^1)^2 + (S^2)^2 + (S^3)^2
$$
conmuta con $S^3$, isto é:
$$
[{\bf S}^2, S^3] = 0.
$$

De tal forma que damos os estados a partir dos números cuánticos de ambos os operadores, como $|j, m \rangle$, onde $j$ é o momento angular total e $m$ a terceira compoñente, así:

$$
{\bf S}^2 \, |j, m \rangle = j (j+1) \, | j, m \rangle, \;\;\; S^3 \, | j, m \rangle = m \, |j, m \rangle
$$

E facemos a asociación:

$$
\uparrow \equiv |1/2, 1/2 \rangle, \;\;\; \downarrow \equiv | 1/2, -1/2 \rangle
$$

O produto de dous espíns $s=1/2$, dous dobletes, que se denota por ${\bf 2} \otimes {\bf 2}$, dános as seguintes combinacións:
$$
\uparrow\uparrow, \, \uparrow\downarrow, \,
\downarrow\uparrow, \, \downarrow\downarrow
$$.

A representación da composición ${\bf 2} \otimes {\bf 2}$ ten dimensión 4. A súa representación irredutible está formada por un triplete e un singlete, que se denota como ${\bf 3} \oplus {\bf 1}$.

O triplete é:
$$
|1, 1 \rangle = \; \uparrow\uparrow, \;\;\;
|1, 0 \rangle = \frac{1}{\sqrt{2}} (\uparrow\downarrow + \downarrow\uparrow),  \;\;\; |1, -1 \rangle, = \; \downarrow\downarrow, 
$$
e o singlete:
$$
|0, 0 \rangle = \frac{1}{\sqrt{2}}(\uparrow\downarrow - \downarrow\uparrow)
$$

Que asociamos cos estados $|j, m\rangle$, onde $j = 1$ e $m = 1, 0, -1$, e $j = 0, m=0$ 

Para construír a representación irredutible usamos os operadores escada:

$$
S^\pm \equiv S^1 \pm i S^2 
$$

que na representación fundamental son simplemente:
$$
S^+ = \sigma^+ = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}, \;\;\;
S^- = \sigma^- = \begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}.
$$

O produto ${\bf 2} \otimes {\bf 2}$ constrúese a partir, por exemplo, do estado $\uparrow\uparrow$ e o operador $S^{-1}$

$$
|1, 1 \rangle = \uparrow \uparrow\\
|1, 0 \rangle \propto S^- \, \uparrow\uparrow = (S^-\uparrow)\uparrow + \uparrow (S^- \uparrow), \\
|1, -1 \rangle \propto S^- \left(\uparrow\downarrow + \downarrow\uparrow \right) = 2 \downarrow \downarrow
$$

Que, se normalizamos:
$$
|1, 1 \rangle =  \uparrow\uparrow, \;\; 
|1, 0 \rangle = \frac{1}{\sqrt{2}} (\uparrow\downarrow + \downarrow\uparrow), \;\;
|1, -1 \rangle = \downarrow\downarrow.
$$

e finalmente, o membro do singlete, $|0, 0 \rangle$, é o ortogonal a $|1, 0 \rangle$:

$$
|0, 0 \rangle = \frac{1}{\sqrt{2}}(\uparrow\downarrow - \downarrow\uparrow)
$$

Os membros do triplete son simétricos baixo o intercambio $\uparrow \leftrightarrow \downarrow$ e o singlete antisimétrico.



De forma xenérica, para o produto de dúas representacións $|j_1, m_1 \rangle |j_2, m_2 \rangle$, as representacións irredutibles veñen dadas por:

* $j = |j_1 - j_2|, \, |j_1 - j_2| + 1, \dots, \, j_1+j_2, \, j_1+j_2$

* $m = -j, -j+1, \dots, j-1, j$.

Para obter a base $|j, m \rangle$ usamos os operadores escada, $J^+, J^-$, que, lembremos, actúan da seguinte forma:
$$
J^+ \, |j, m \rangle = \sqrt{j (j+1) - m (m+1)} \, |j, m \rangle \\
J^- \, |j, m \rangle = \sqrt{j (j+1) - m (m-1)} \, |j, m \rangle 
$$
isto é, soben e baixan nos elementos da base.

Lembremos finalmente que os coeficientes de [Clebsch-Gordan](https://pdg.lbl.gov/2018/reviews/rpp2018-rev-clebsch-gordan-coefs.pdf) nos relacionan os elementos da base $|j, m \rangle$ cos iniciais $|j_1, m_1 \rangle |j_2, m_2 \rangle$.

Para construír as matrices unitarias dunha representación irredutible obtemos primeiro os seus xeradores a partir dos operadores escada.

Para $j=1$ os operadores escada, cuxos coeficientes poden obterse das regras de operación anteriores, son:

$$
J^+ = \begin{pmatrix} 0 & \sqrt{2} & 0 \\ 0 & 0 & \sqrt{2} \\ 0 & 0 & 0 \end{pmatrix}, \;\;\;
J^- = \begin{pmatrix} 0 & 0 & 0 \\ \sqrt{2} & 0 & 0\\ 0 & \sqrt{2} & 0\end{pmatrix}
$$

O xerador das rotacións, $J^2$, para $j=1$ arredor do eixe $y$, vén dado por:
$$
J^2 = - \frac{i}{2} (J^+ - J^-), \;\; J^2 = -\frac{i}{\sqrt{2}} \begin{pmatrix} 0 & 1 & 0 \\ -1 & 0 & 1 \\ 0 & -1 & 0\end{pmatrix}
$$

De tal forma que a matriz de rotación dun ángulo $\theta$ arredor do eixe $y$ para $j=1$ vén dada por:
$$
U(\theta) = e^{i \theta J^2} = \sum_n \frac{1}{n!} (i \theta J^2)^n =  \begin{pmatrix} 
\frac{1}{2} (1+\cos \theta) & \frac{1}{\sqrt{2}} \sin \theta &\frac{1}{2}(1-\cos \theta) \\
-\frac{1}{\sqrt{2}} \sin \theta & \cos \theta &  \frac{1}{\sqrt{2}}\sin \theta  \\
\frac{1}{2} (1-\cos \theta) & -\frac{1}{\sqrt{2}} \sin \theta &\frac{1}{2}(1+\cos \theta) \\
\end{pmatrix}
$$

Os valores $d^j_{m', m}(\theta)$ das configuracións de $j$ menores recóllense na táboa do PDG dos [Clebsch-Gordan](https://pdg.lbl.gov/2018/reviews/rpp2018-rev-clebsch-gordan-coefs.pdf); a partir deles podemos reconstruír as matrices de rotación para a representación $j$.


## O grupo $U(1)$

O grupo $U(1)$ fórmano as rotacións, $D(\theta)$, de 1 dimensión no espazo complexo.

Consideremos un número complexo, $x$, relacionado con outro número complexo, $x'$, simplemente por unha fase $x' = e^{i\theta} x$. A rotación entre eles é:

$$
D(\theta) = e^{i\theta}, \;\;\, x' = D(\theta) \, x 
$$

Estas rotacións deixan invariante a norma:
$$
|x'|^2 = |x|^2
$$

A multiplicación dos elementos do grupo é:

$$
D(\theta_1) \cdot D(\theta_2) = D(\theta_1 + \theta_2), \;\;\; D(\theta + 2 \pi) = D(\theta)
$$

Isto dinos que é un grupo abeliano: non importa a orde en que se realizan dúas rotacións, isto é, a orde de multiplicación.

A inversa é simplemente:

$$
D(\theta) = e^{i\theta}, \;\;\; D^{-1}(\theta) = D(-\theta) = e^{-i\theta}
$$

A representación fundamental ten dimensión 1. E o xerador das rotacións é tamén a unidade.

### O grupo $SO(2)$

O grupo $SO(2)$ está formado polas matrices das rotacións, $R(\theta)$, propias en dúas dimensións, $R(\theta)$.
Lembremos que as matrices propias teñen o determinante positivo.

Sexa $(x, y)$ un punto do espazo bidimensional; o efecto dunha rotación é:

$$
\begin{pmatrix} x' \\ y'\end{pmatrix} =
\begin{pmatrix} \cos \theta & \sin \theta \\ - \sin \theta & \cos \theta \end{pmatrix} \, 
\begin{pmatrix} x \\ y \end{pmatrix}
$$

Onde 
$$
R(\theta) = \begin{pmatrix} \cos \theta & \sin \theta \\ - \sin \theta & \cos \theta \end{pmatrix}, \;\;\; 
\mathrm{det}(D(\theta)) = 1
$$

O grupo SO(2) é continuo e compacto, con $\theta \in [0, 2 \pi]$.

As rotacións preservan o módulo do vector:

$$
x'^2 + y'^2 = x^2 + y^2
$$


As regras de multiplicación das matrices $R(\theta)$ son:

$$
R(\theta_1) \cdot R(\theta_2) = R(\theta_1 + \theta_2), \;\;\; R(\theta + 2 \pi) = R(\theta)
$$

Facer unha rotación de $\theta_2$ despois dunha rotación previa de $\theta_1$ é equivalente a rotar desde o principio con $\theta_1+\theta_2$
 
Trátase dun grupo abeliano, $D(\theta_1) \cdot D(\theta_2) = D(\theta_2) \cdot D(\theta_1)$

A inversa dunha rotación con $\theta$ é simplemente rotar con $-\theta$:
$$
R^{-1}(\theta) = R(-\theta)
$$

Na representación matricial $2\times2$, a inversa é a trasposta:

$$
R(\theta)  = \begin{pmatrix} \cos \theta & \sin \theta \\ - \sin \theta & \cos \theta \end{pmatrix}, \;\;\;
R^{-1}(\theta)  = R(-\theta) = R^T(\theta) = \begin{pmatrix} \cos \theta & -\sin \theta \\  \sin \theta & \cos \theta \end{pmatrix}
$$


O xerador das rotacións é:

$$
T = -i \begin{pmatrix} 0 & 1 \\ -1 & 0 \end{pmatrix}
$$

De tal forma que:
$$
R(\theta) = e ^{i \theta T} = \sum_n \frac{(i\theta T)^n}{n!} =
\begin{pmatrix} \cos \theta & \sin \theta \\ - \sin \theta & \cos \theta \end{pmatrix},
$$


As regras de multiplicación dos grupos $SO(2)$ e $U(1)$ son as mesmas:

$$
D(\theta_1) \cdot D(\theta_2) = D(\theta_1 + \theta_2), \;\;\; D(\theta + 2 \pi) = D(\theta)
$$

Dicimos que dous grupos son **isomorfos** se comparten as mesmas regras de multiplicación.

Neste caso dicimos que SO(2) e U(1) son isomorfos.

$$
SO(2) \sim U(1)
$$

As bases dun e doutro relaciónanse por:

$$
|\rho| e^{i\theta} \leftrightarrow (|\rho| \cos \theta, |\rho| \sin \theta)
$$


### A carga eléctrica

O grupo $U(1)$ utilizámolo en Física de Partículas para dar a invariancia gauge dos espinores de Dirac fronte ao cambio dunha fase global (o que asociaremos coa conservación de números cuánticos como o número leptónico) ou local (o que asociaremos coa conservación da carga eléctrica e que dará lugar á interacción dos fermións cargados co campo electromagnético)

O grupo $U(1)$ global respecto aos espinores de Dirac:
$$
\Psi(x) \to e^{i \theta} \Psi(x)
$$

E o grupo $U(1)$ local:
$$
\Psi(x) \to e^{i \theta(x)} \Psi(x)
$$
onde $\theta(x)$ depende de forma continua e *local* con respecto ao punto espazo-temporal $x$.




## O grupo $SU(2)$

O grupo $SU(2)$ fórmano as **matrices complexas unitarias $2 \times 2$ con determinante unidade**. A $S$ é por propias (con determinante unidade), a $U$ por unitarias e o $2$ corresponde á dimensión.

As matrices cumpren:
$$
U^\dagger U = I, \;\; \mathrm{det}(U) = 1
$$

Sexa un vector de dúas compoñentes complexas ${\bf x} = (x, y)$; as matrices transfórmano así:
$$
{\bf x}' = U \, {\bf x}
$$
e conservan o módulo do vector:
$$
|{\bf x}'|^2 = |{\bf x}|^2
$$

As matrices, $U(\theta)$, poden construírse a partir dos ángulos $\theta_i$ e 3 **xeradores** hermíticos $T_i$, con $i=1, 2, 3$:
$$
U({\bf \theta}) = e^{i \vec{\theta }\cdot {\bf T}} = e^{i \sum_i \theta_i T^i}
$$

Os xeradores $T^i$ son:
$$
T^i = \frac{1}{2} \sigma^i
$$
e $\sigma^i$ as matrices de Pauli. Cumpren a seguinte álxebra de conmutación:
$$
[T^i, T^j] = i \epsilon_{ijk} T^k,
$$
onde $\epsilon_{ijk}$ é o tensor de Levi-Civita.




Os elementos da base dámolos a partir dos números cuánticos dos operadores ${\bf T}^2, T^3$, dados por $|j, m \rangle$. Onde:

$$
{\bf T}^2 | j, m \rangle = j (j +1) | j, m \rangle, \;\;\; T^3 |j, m \rangle = m | j, m \rangle
$$

A representación fundamental, $j=1/2$, é:
$$
\begin{pmatrix} 1 \\ 0 \end{pmatrix} = |1/2, 1/2 \rangle, \;\; 
\begin{pmatrix} 0 \\ 1 \end{pmatrix} = |1/2, -1/2 \rangle, \;\; 
$$

A discusión sobre composición de estados e representacións irredutibles é idéntica á do espín (véxase arriba)

### O grupo $SO(3)$

O grupo $SO(3)$ fórmano as matrices ortogonais, $R$, de determinante positivo, que corresponden ás rotacións no espazo vectorial de tres dimensións. O $O$ do nome do grupo indica que son ortogonais.

$$
R^T \, R = I, \;\;\; \mathrm{det}(R) = 1
$$

As rotacións déixannos invariante a norma do vector ${\bf x} = (x, y, z)$.

$$
{\bf x}' = R \, {\bf x}, \;\; |{\bf x}'|^2 = |{\bf x}|^2
$$

As matrices, $R(\theta)$, quedan definidas con tres ángulos, $\theta_i$, con $i=1, 2, 3$, a través de tres xeradores hermíticos:
$$
R(\theta) = e^{i \vec{\theta} \cdot T} = e^{i \sum_i \theta_i T^i}
$$
onde:
$$
T^1 = -i \begin{pmatrix} 0 & 0 & 0 \\ 0 & 0 & 1 \\ 0 & -1 & 0\end{pmatrix}, \;\;
T^2 = -i \begin{pmatrix} 0 & 0 & -1 \\ 0 & 0 & 0 \\ 1 & 0 & 0\end{pmatrix}, \;\;
T^3 = -i \begin{pmatrix} 0 & 1 & 0 \\ -1 & 0 & 0 \\ 0 & 0 & 0\end{pmatrix}. 
$$

Que cumpren a seguinte álxebra de conmutación:
$$
[T^i, T^j] = i \, \epsilon_{ijk} \, T^k
$$
onde $\epsilon_{ijk}$ é o tensor de Levi-Civita.

Estas son novamente as regras de conmutación do momento angular e do grupo $SU(2)$.


Como comparten a mesma álxebra, dicimos que o grupo $SU(2)$ é isomorfo a $SO(3)$

$$
SU(2) \sim SO(3)
$$

A relación entre os elementos de cada grupo é:

$$
(x, y, z) \leftrightarrow \sigma \cdot x = \begin{pmatrix} z & x - i y \\ x + iy & z \end{pmatrix}
$$


### O isospín de sabor

O quark $u$ e o quark $d$ diferéncianse principalmente pola súa carga, dado que teñen masas moi similares. A forza forte, porén, non fai distinción entre eles: non distingue o sabor dos quarks.

Podemos definir entón un *isospín* de sabor e asociar cada un dos quarks cos elementos da base fundamental de $SU(2)$, (ou ${\bf 2})$. A forza forte debe ser invariante respecto dunha transformación unitaria entre ambos.

$$
u \equiv \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \;\;\; d \equiv \begin{pmatrix} 0 \\ 1 \end{pmatrix} 
$$

Igualmente podemos asociar tamén os antiquarks á representación fundamental ${\bf \bar{2}} $

De tal forma que os mesóns lixeiros podemos asocialos aos elementos das representacións irredutibles dun par quark-antiquark ${\bf 2} \otimes {\bf \bar{2}}$ e os barións lixeiros a tres quarks ${\bf 2} \otimes {\bf 2} \otimes {\bf 2}$

### O isospín feble

Para as interaccións febles definimos un *isospín feble* asociando o espinor levógiro do leptón cargado e o do neutrino do seu sabor cos elementos dun doblete de $SU(2)$, e de igual forma facemos co espinor levógiro dos quarks de abaixo e de arriba:

Por exemplo, para a primeira xeración de leptóns e quarks teriamos os dobletes
$$
L \equiv \begin{pmatrix} \nu_e|_L \\ e_L \end{pmatrix}, \;\; Q \equiv \begin{pmatrix} u_L \\ d_L \end{pmatrix},
$$


e esiximos que o lagrangiano feble sexa invariante baixo rotacións *locais* no espazo de *isospín feble*

$$
L' = e^{i \theta_i(x) T^i} L, \;\; Q' = e^{i \theta_i(x) T^i} Q
$$

onde $\theta_i(x)$ son funcións continuas e locais do punto espazo-temporal


## O grupo SU(3)

As **matrices unitarias $3 \times 3$ con determinante unidade** forman o grupo $SU(3)$.

As matrices, $U$, cumpren:
$$
U^\dagger U = I, \;\; \mathrm{det}(U) = 1
$$

Un vector complexo ${\bf x} = (x, y, z)$ mantén a norma baixo transformacións do grupo $SU(3)$.

$$
{\bf x}' = U \, {\bf x}, \;\; |{\bf x}'|^2 = |{\bf x}|^2
$$

O número de parámetros reais libres de SU(3) é 8. Hai $18 = 3 \times 3 \times 2$ parámetros reais en $U$, 9 ligaduras da condición de unitariedade e 1 por ser o determinante unidade.

As matrices de SU(3) poden construírse a partir de 8 xeradores hermíticos, $T_i$, e $\theta_i$ con $i = 1, \dots, 8$

$$
U(\theta) = e^{i \theta \cdot {\bf T}}
$$


As matrices $U(\theta)$ de $SU(3)$ constrúense con 8 ángulos, $\theta_i$, e 8 xeradores hermíticos, $T^i$,

$$
U(\theta) = e^{i \vec{\theta} \cdot {\bf T}} = e^{i \sum \theta_i T^i}
$$

O número de parámetros reais libres de SU(3) é 8. Hai $18 = 3 \times 3 \times 2$ parámetros reais en $U$, 9 ligaduras da condición de unitariedade e 1 por ser o determinante unidade.


Unha opción conveniente para os xeradores é construílos a partir das matrices $\lambda$ de Gell-Mann.

$$
T_i = \frac{1}{2} \lambda_i
$$

Onde:

$$
\lambda_1 = \begin{pmatrix} 0 & 1 & 0  \\  1 & 0 & 0 \\ 0 & 0 & 0 \end{pmatrix}, \;
\lambda_2 = \begin{pmatrix} 0 & -i & 0 \\  i & 0 & 0 \\ 0 & 0 & 0 \end{pmatrix}, \;
\lambda_3 = \begin{pmatrix} 1 & 0 & 0  \\ 0 & -1 & 0 \\ 0 & 0 & 0 \end{pmatrix}, \;
\lambda_4 = \begin{pmatrix} 0 & 0 & 1  \\ 0 & 0 & 0  \\ 1 & 0 & 0 \end{pmatrix}, \\
\lambda_5 = \begin{pmatrix} 0 & 0 & -i \\ 0 & 0 & 0  \\ i & 0 & 0 \end{pmatrix}, \;
\lambda_6 = \begin{pmatrix} 0 & 0 & 0  \\ 0 & 0 & 1  \\ 0 & 1 & 0 \end{pmatrix}, \;
\lambda_7 = \begin{pmatrix} 0 & 0 & 0  \\ 0 & 0 & -i \\ 0 & i & 0 \end{pmatrix}, \;
\lambda_8 = \frac{1}{\sqrt{3}}\begin{pmatrix} 1 & 0 & 0  \\ 0 & 1 & 0 \\ 0 & 0 & -2 \end{pmatrix}.
$$


Só tres dos xeradores conmutan entre eles, ${\bf T}^2, T_3, Y_8$.

Podemos escribir un elemento do grupo cos números cuánticos $|j, m, y \rangle$. Na representación fundamental:

$$
{\bf T}^2 = \sum_{i=1}^8 T^2_i = \frac{4}{3}    \begin{pmatrix} 1 & 0 & 0 \\ 0 &  1 & 0 \\ 0 & 0 & 1 \end{pmatrix}, \;
T_3 = \frac{1}{2} \lambda_3 = \frac{1}{2} \begin{pmatrix} 1 & 0 & 0 \\ 0 & -1 & 0 \\ 0 & 0 & 0 \end{pmatrix}, \;
\;\; Y = \frac{1}{\sqrt{3}} \lambda_8   = \frac{1}{3} \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & -2 \end{pmatrix}, 
$$

Os números cuánticos agora serán $m$, *terceira compoñente do isospín*, asociada a $T^3$, e $y$, a *hipercarga*, asociada a $Y_8$.

Os elementos da representación fundamental en función da terceira compoñente e da hipercarga quedan:

$$
\begin{pmatrix}  1 \\ 0 \\ 0 \end{pmatrix} = |1/2, 1/3 \rangle, \;\;
\begin{pmatrix} 0 \\ 1 \\ 0 \end{pmatrix}  = |-1/2, 1/3 \rangle, \;
\begin{pmatrix}  0 \\0 \\ 1 \end{pmatrix} = |0, -2/3 \rangle
$$


### Simetría $SU(3)$ de sabor

Aínda podemos considerar que os tres quarks lixeiros, de masas parecidas, se diferencian só polo seu sabor e a súa carga. 

Pero a forza forte é insensible ao sabor, polo que podemos asociar cada un dos quarks lixeiros, $u, d, s$, aos elementos dun triplete de $SU(3)$, de tal forma que a forza forte sexa invariante fronte a transformacións unitarias dos tres.

$$
u = \begin{pmatrix} 1 \\ 0 \\ 0 \end{pmatrix}, \;\; d = \begin{pmatrix} 0 \\ 1 \\ 0 \end{pmatrix}, \;\;
s = \begin{pmatrix} 0 \\ 0 \\ 1 \end{pmatrix}
$$

Desta forma, novamente, a combinación dun par quark-antiquark, ${\bf 3} \otimes {\bf \bar{3}}$, e de tres quarks, ${\bf 3} \otimes {\bf 3} \otimes {\bf 3}$, permítenos obter a composición dos mesóns e dos barións lixeiros.

### A cor

A forza forte establece que temos tres cores, vermello, verde e azul, $r, g, b$, que podemos asociar a un triplete de $SU(3)$. No Modelo Estándar esíxese que a forza forte sexa invariante baixo transformacións unitarias locais da cor.

As tres cores son os elementos do triplete:

$$
r = \begin{pmatrix} 1 \\ 0 \\ 0 \end{pmatrix}, \;\; g = \begin{pmatrix} 0 \\ 1 \\ 0 \end{pmatrix}, \;\;
b = \begin{pmatrix} 0 \\ 0 \\ 1 \end{pmatrix}
$$

Esta invariancia gauge local daranos lugar á cromodinámica cuántica (QCD).

Tamén nos permite dar a composición de sabor dos mesóns e barións. Esixiremos que os hadróns sexan *incoloros*, é dicir, que os seus números cuánticos de cor (terceira compoñente e hipercarga) sexan nulos.



A figura mostra os dous tripletes de cor, ${\bf 3}$ dos quarks e ${\bf \bar{3}}$ dos antiquarks, no plano $T_3, Y$. Nótese a oposición de cargas para os elementos da representación ${\bf \bar{3}}$.

Existen agora tres operadores escada:

$$
T_+ g = r, \; T_- r = g, \\ V_+ b = r, \; V_-r = b, \\ U_+b = g, \; U_-g = b
$$


A combinación das cores de quark e antiquark (mesón), ${\bf 3} \otimes {\bf \bar{3}}$, efectúase colocando o triplete ${\bf \bar{3}}$ sobre os vértices do ${\bf 3}$, o que dá lugar a un octete e un singlete na súa representación irredutible.
