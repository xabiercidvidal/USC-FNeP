**problema 2**

Considera un experimento de branco fixo $p+p$. Cal é a enerxía do feixe de protóns para que no CM teña a enerxía do LHC, $\sqrt{s} = 14$ TeV?

[MT1.P10]

*solución*

No sistema de laboratorio: $s = (E + m_p, {\bf p})$, polo que:

$$
(E+m_p)^2 - p^2 = s, \;\; 2E m_p + 2m^2_p = s \\
E = \frac{s- 2m^2_p}{2m_p}
$$

```python
ss   = (14 * 1e3) **2 ## GeV^2
ethr =  (ss - 2 * mp**2)/(2*mp)
print('Enerxía limiar {:e} TeV'.format(ethr/1e3))
```

**problema 3**

Sexan tres protóns de $p = 3$ GeV emitidos formando $120^o$ entre eles nun plano. Cal é a súa masa invariante? [AB1.P2]

*solución*

A masa invariante dun sistema de n partículas con cuadrimomentos $p_i$, con $i=1, \dots, n$, é:

$$
m^2 = \left( \sum_i p_i\right)^2
$$


Na suma dos cuadrimomentos o momento anúlase e a enerxía é:

$$
\sum_i p_i = (3E, 0)
$$

Logo a masa é $3E$

```python
pp = 3 # GeV
mp = 0.938 # GeV
ene = energy(mp, pp)
print('masa {:4.3f} GeV '.format(3 * ene))
```

**problema 4**

Considera a interacción $\gamma + p \to p + \pi^0$. Cal é a enerxía mínima dos fotóns para que teña lugar a interacción se estes golpean un branco formado por protóns en repouso? Que enerxía mínima deben ter os protóns dos raios cósmicos (isto é, ultrarrelativistas) para que a interacción teña lugar cos fotóns de 1 meV do fondo cósmico?


*solución*

Utilizaremos a variable de Mandelstam $s$, que é invariante Lorentz.

O caso no que as dúas partículas teñen menos enerxía no CM é

$$
s = (m_p + m_\pi, {\bf 0})^2
$$

No caso en que os protóns están en repouso, no sistema laboratorio, o invariante vale:

$$
s = ((E, E {\hat{k}}) + (m_p, {\bf 0}))^2 = (E + m_p, E \hat{k})^2 = E^2 + 2 E m_p + m^2_p - E^2 = 2 E m_p + m^2_p
$$

polo tanto:

$$
E = \frac{(m_p+m_\pi)^2 - m^2_p}{2 m_p}
$$

```python
mp  = 0.938 # GeV
mpi = 0.134 # GeV

E = ((mp + mpi)**2 - mp**2)/(2*mp)

print('E(gamma) lab {:4.3f} GeV'.format(E))
```

No caso dos protóns cósmicos, ultrarrelativistas, o seu cuadrimomento é $(E_p, E_p \hat{k})$, e golpean os fotóns de cuadrimomento $(E_\gamma, E_\gamma \hat{k}')$

$$
s = (E_p + E_\gamma, E_p {\hat k} + E_\gamma \hat{k}') = (E_p+E_\gamma)^2 -  (E_p {\hat k} + E_\gamma \hat{k}')^2 = \\
2 E_p E_\gamma -  2 E_p E_\gamma \cos \theta = 2 E_p E_\gamma (1 - \cos \theta)
$$

O valor mínimo de $E_p$ corresponde a $\cos \theta = -1$, ao choque frontal, logo:

$$
4 E_p E_\gamma = (m_p + m_\pi)^2 \to E_p = \frac{(m_p + m_\pi)^2}{4 E_\gamma}
$$

```python
mp     = 0.938 # GeV
mpi    = 0.134 # GeV
Egamma = 0.1e-3 * 1e-9 # meV en GeV

E = ((mp + mpi)**2)/(4*Egamma)

print('E(p) {:1.2e} GeV'.format(E))
```

**problema 5**

Calcula a luminosidade equivalente dun feixe de protóns de intensidade $10^{13}$ 1/s que golpea un tanque de 10 cm de hidróxeno líquido, $\rho = 60 \; \mathrm{kg/m}^3$. [AB1.6]

A taxa de interacción nun branco fixo é:
$$
R = \phi_a \sigma N_b
$$

onde $\phi_a$ é o fluxo; $\sigma$, a sección eficaz; e $N_b$, o número de partículas branco.

E en función da luminosidade $\mathcal{L}$ dun colisionador:
$$
R = \mathcal{L} \sigma
$$

Polo tanto, a luminosidade equivalente nun branco fixo é:

$$
\mathcal{L} = \phi_a N_b
$$


Neste caso, dada unha intensidade $I_a$, nunha sección $S$ temos un fluxo:

$$
\phi_a = I_a / S
$$

E nesa sección $S$, cunha lonxitude $L$ e densidade $\rho$ (en $\mathrm{kg/m}^3$), temos tantos nucleóns branco como:

$$
N_b = 10^{3} \rho S L N_A,
$$

onde $N_A = 6.022 \, 10^{23}$ é o número de Avogadro de átomos nun mol.

Logo a luminosidade equivalente é:
$$
\mathcal{L} = \phi_a N_b = I_a \rho 10^3 L N_A
$$

```python
Ia = 1e3 # s-1
L  = 0.1 # m
rho = 60e3 # g/m3
N_A = units.N_A

lumi = (Ia * rho * L * N_A)
print('Luminosidade {:1.2e} 1/(cm2 s)'.format(lumi * 1e4))

```

**problema 7**

A sección eficaz de produción do bosón de Higgs é 50  pbarn en colisións $p+p$ a $\sqrt{s} = $8 TeV, isto é, á enerxía no centro de masas da primeira toma de datos do LHC. Se a fracción de desintegración do $H \to \gamma + \gamma$ é de só $2.2 \times 10^{-3}$, cantos sucesos deste tipo se produciron nun ano de toma de datos no LHC? Canta luminosidade integrada se precisa para observar 100 sucesos deste tipo?

Datos: luminosidade do LHC $\mathcal{L} = 10^{34}$ $\mathrm{cm}^{-2} s^{-1}$; considera que 1 ano son $10^7$ s



*solución*

A taxa de produción vén dada por:

$$
\nu = \mathcal{L} \sigma \, \mathcal{B}
$$

En $T = 1e7$ producíronse

$$
N = \mathcal{L} T \, \sigma \, \mathcal{B}
$$


A luminosidade integrada $\mathcal{L}_T$ para producir $N$ sucesos é:

$$
\mathcal{L}_T = \frac{N}{\sigma \mathcal{B}}
$$


```python
sigma = 50e-12 * 1e-24 # cm2
B     = 2.2e-3
lumi  = 1e34  # 1/cm2 s
T     = 1e7

N     = lumi * T * sigma * B
print('Higgs -> gamma + gamma {:1.2e} producidos'.format(N))
```

```python
N = 100
LT = N/(sigma * B)
print('Luminosidade integrada {:1.2f} (1/pb) '.format(LT*1e-24*1e-12))
```

**problema 8**

A sección eficaz $e+e^+\to \tau + \tau^+$ é 1.7 nbarn a $\sqrt{s} = 91$ GeV. Se a luminosidade do colisionador é $10^{32}$ 1/
($\mathrm{cm}^2$ s), cal é a taxa de produción de $\tau+\tau^+$? Se a vida media do $\tau$ é 0.29 ps, que percorrido medio terá? No plano transverso da colisión, que relación garda o momento transverso do $\tau$ co do $\tau^+$?


*solución*

A taxa de produción é:

$$
N = \mathcal{L} \, \sigma
$$


```python
lumi = 1e32 # 1/cm2
sigma = 1.7e-9* 1e-24 # cm2

nu = lumi * sigma

print('taxa de produción {:4.3f} Hz'.format(nu))
```

A enerxía do tau é $45.5$ GeV, sendo a súa masa $m = 1.78$ GeV, con $\gamma = E/m$ e $\beta = p/E$

A distancia media percorrida é:

$$
d = \tau \beta \gamma  \, c
$$

```python
Etau = 45.5 # GeV
mtau = 1.78 # GeV

ptau = np.sqrt(Etau**2 - mtau**2)

gamma = Etau/mtau
beta  = ptau/Etau

tau = 0.29e-12 # s
c   = units.speed_of_light

dd  = tau * beta * gamma * c

print('momento do tau {:4.3f} GeV'.format(ptau))
print('beta tau {:4.3f} '.format(beta))
print('gamma {:4.3f}'.format(gamma))
print('distancia {:4.3f} mm'.format(dd *1e3))
```

No plano transverso a suma dos dous momentos anúlase, logo o momento transverso do $\tau$ é oposto ao do $\tau^+$
