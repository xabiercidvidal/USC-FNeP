### Exercicio  A.B 2.1)

Calcula a enerxía e o momento das partículas finais no CM na desintegración. Calcula tamén a $\beta$ dos leptóns cargados:

   * a) $\pi^+ \to \mu^+ + \nu_\mu$
   * b) $\pi^+ \to e^+ + \nu_e$.

Está resolto na clase (véxanse os apuntamentos de leptóns)

$$
E_l = \frac{m^2_\pi + m^2_l}{2 m_\pi}, \; \; p_l = \frac{m^2_\pi - m^2_l}{2 m_\pi}, \;\;
$$


```python
mpi = 139.51  # MeV
mmu = 105.66  # MeV
me  =   0.511 # MeV
ene_cm = lambda ml, mhad = mpi : (mhad**2 + ml**2)/(2*mhad)
p_cm   = lambda ml, mhad = mpi : (mhad**2 - ml**2)/(2*mhad)
a_R    = lambda ml, mhad = mpi : (1 - p_cm(ml)/(ene_cm(ml) + ml))/2

beta_mu, beta_e = p_cm(mmu)/ene_cm(mmu), p_cm(me)/ene_cm(me)
print(' beta mu    = {:7.6f}, beta       e = {:7.6f}    '.format(beta_mu, beta_e))
print(' enerxía mu = {:5.3f} MeV, enerxía e = {:5.3f} MeV'.format(ene_cm(mmu), ene_cm(me)))
print(' momento mu = {:5.3f} MeV,  momento e = {:5.3f} MeV'.format(p_cm(mmu), p_cm(me)))
```

### Exercicio 2.2)

Considera a desintegración $K^- \to \mu + \bar{\nu}_\mu$. Calcula a enerxía e o momento no CM do $\mu, \bar{\nu}_\mu$ e o momento máximo do $\mu$ se o $K^-$ ten un momento de 5 GeV.


A enerxía e o momento do leptón son:
$$
E_l = \frac{m^2_K + m^2_l}{2 m_K}, \; \; p_l = \frac{m^2_K - m^2_l}{2 m_K}, \;\;
$$

O momento máximo corresponderá ao momento do $\mu$ na dirección e sentido do $K$ incidente, que recibirá un *boost* de $\gamma = E_K/m_K, \; \beta = p_K/E_K$. Así:

$$
p'_\mu = \gamma \beta E^*_\mu + \gamma p^*_\mu
$$


```python
mkaon = 493.67 # MeV
e_mu, p_mu = ene_cm(mmu, mkaon), p_cm(mmu, mkaon)
e_nu, p_nu = p_mu, p_mu
print(' enerxía mu  = {:5.3f} MeV'.format(e_mu))
print(' momento mu = {:5.3f} MeV '.format(p_mu))

e_kaon = 5000 # MeV
p_kaon = momentum(mkaon, e_kaon)
gamma_kaon, beta_kaon = e_kaon/mkaon, p_kaon/e_kaon
p_max_mu = gamma_kaon * beta_kaon * e_mu + gamma_kaon * p_mu
print(' momento mu máximo no lab = {:5.3f} MeV '.format(p_max_mu))
p_min_nu = gamma_kaon * beta_kaon * e_nu - gamma_kaon * p_nu
print(' momento nu mínimo no lab = {:5.3f} MeV '.format(p_min_nu))
```

O colisionador PEP de SLAC produciu $e+e^+ \to \tau + \tau^+$ a unha enerxía $\sqrt{s} = 29$ GeV; calcula a distancia media que percorrían os $\tau$.

O cuadrimomento das partículas saíntes no CM é $p_c = (E_c, {\bf p}^*_f), \; p_d=(E_d, -{\bf p}^*_f)$, $s = (2E_\tau, {\bf 0})^2 = 29^2$ GeV$^2$. Isto é, $E_\tau = 29$ GeV


A distancia percorrida obtémola pola transformación de Lorentz de $(\tau, {\bf 0})$ na dirección $z$ con $\gamma = E/m$; temos $d = \gamma \beta \tau$.

```python
m_tau   = 1.77 # GeV
tau_tau = 0.29 * units.pico # s
sqrt_s  = 29.  # GeV
E_tau   = sqrt_s
p_tau   = momentum(m_tau, E_tau)
gamma   = gamma_lorentz(m_tau, E_tau)
beta    = beta_lorentz (p_tau, E_tau)
print(' tau gamma {:4.3f}, beta {:4.4f} '.format(gamma, beta))
print(' distancia {:4.3f} mm'.format(gamma * beta * tau_tau * units.c * units.kilo))
```

### Exercicio AB 7.5

Considera $\mu^+ \to e^+ + \nu_e + \bar{\nu}_\mu$ e $\tau^+ \to e^+ + \nu_e + \bar{\nu}_\tau$; a razón de desintegración da primeira é $\mathcal{B} = 100$ % e da segunda 17 %. Se a vida media do muón é $\tau_\mu = 2.2$ $\mu$s, calcula a vida media do tau $\tau_\tau$. 

Vimos na clase que a anchura de desintegración $l \to \nu_l + e + \bar{\nu}_e$, con $l = \mu, \tau$, vén dada por:

$$
\Gamma_l (l \to \nu_l + e + \bar{\nu}_e) =  \frac{\mathcal{B}(l \to \nu_l + e + \bar{\nu}_e)}{\tau_l} = \frac{G^2_F m^5_l}{192 \pi^3}
$$

Polo que a razón entre as vidas medias do $\tau$ e do $\mu$ vén dada por:

$$
\frac{\tau_\tau}{\tau_\mu} = \frac{m^5_\mu}{m_\tau^5} \frac{\mathcal{B}(\tau \to \nu_\tau + e + \bar{\nu}_e)}{\mathcal{B}(\mu \to \nu_\mu + e + \bar{\nu}_e)}
$$

```python
m_tau  = 1.77  # GeV
m_mu   = 0.106 # GeV
tau_mu = 2.2 * units.micro # s
B_tau_e = 0.17
B_mu_e  = 1.
tau_tau_exp = tau_mu * (m_mu/m_tau)**5  * B_tau_e
print(' vida media do tau {:e} s'.format(tau_tau_exp))
```

### Exercicio AB 7.7)

Cales son as diferenzas entre neutrinos e antineutrinos? Completa a interacción: $\nu_\mu + e \to \mu  + ?$. O Universo está cheo de neutrinos con temperatura $2$ K; se os neutrinos teñen unha masa de 50 meV, cal é a súa velocidade?


Os neutrinos teñen número leptónico total e de sabor positivo, e os antineutrinos negativo.

A interacción é $\nu_\mu + e \to \mu + \nu_e$

Da relación da enerxía cinética

$$
E_k = k_B T = m (\gamma - 1); \;\;\; \gamma = \frac{1}{\sqrt{1 - \beta^2}}
$$

Obtemos:

$$
\beta = \sqrt{1 - \frac{m^2}{(m + k_B T)^2}}
$$

```python
T   = 2 # K
kene = units.giga * units.k * T / GeV_to_Julio # J
mnu = 50 * units.milli
kappa = mnu/(mnu + kene)
beta = np.sqrt(1 - kappa*kappa)
gamma = 1./np.sqrt(1 - beta**2)
betap = np.sqrt(2 * kene/mnu)
print('enerxía cinética dos neutrinos {:e} eV '.format(kene))
print('masa dos neutrinos             {:e} eV '.format(mnu))
print('beta dos neutrinos             {:e}    '.format(beta))
print('gamma dos neutrinos            {:e}    '.format(gamma))
print('velocidade dos neutrinos       {:e} m/s  '.format(betap * units.c))
```

### Exercicio 7.16) 

O experimento GALLEX do Laboratorio Nacional do Gran Sasso (LNGS), en Italia, mediu o fluxo de $\nu_e$ do Sol a partir da reacción $\nu_e + ^{71}\mathrm{Ga} \to ^{71}\mathrm{Ge} + e$, que ten un limiar de enerxía $E = 233$ keV. O fluxo de neutrinos do Sol é $\phi = 6 \, 10^{14}$ m$^{-2}$s$^{-1}$. Consideremos que a sección eficaz é $\sigma = 10^{-48}$ m$^2$. Se consideramos unha eficiencia de selección do 40%, cantos núcleos de $^{71}$Ga son necesarios para ter unha interacción ao día? Se a abundancia natural do $^{71}$Ga é do 40%, canta masa de galio se necesita?

Sexa $n$ o número de moles necesarios para ter unha interacción, $N=1$, nos $t$ s do día:

$$
\frac{N}{t} = \phi \sigma \epsilon N_A \, n
$$

que convertemos en masa, se $\eta$ é a abundancia natural do $^{71}$Ga:

$$
M \eta = n A
$$

```python
phi     = 6e14  # /m2s
sigma   = 1e-48 # m2
epsilon = 0.4
N       = 1 
t       = 24 * 60 * 60 # s

print('1 día {:e} s'.format(t))

nmoles = N / ( t * epsilon * phi * sigma * units.N_A) 
eta     = 0.4
A       = 71e-3
M       = nmoles * A / eta
print('número de moles de 71Ga {:e} '.format(nmoles))
print('masa de 71Ga {:e} kg'.format(nmoles * A))
print('masa de Ga   {:e} kg'.format(M))
```
