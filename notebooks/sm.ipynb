## Introdución

O Modelo Estándar (SM) é un dos grandes logros da Física do século XX. Describe con gran precisión todos os resultados experimentais obtidos en Física de Partículas, coa excepción dos fenómenos de oscilación de neutrinos.

Non obstante, sabemos que o modelo é incompleto, xa que non explica certos aspectos fundamentais do Universo, como a asimetría materia–antimateria ou a natureza da materia escura.

O SM constrúese a partir de varios principios esenciais:

* O modelo debe **incorporar as partículas e interaccións coñecidas** e **respectar as leis de conservación** observadas experimentalmente. En particular, debe reproducir feitos empíricos como a violación de paridade nas correntes febles cargadas.

* Debe estar formulado mediante **un lagranxiano invariante de Lorentz e unitario**, en cumprimento dos principios da relatividade especial e da mecánica cuántica.

* As interaccións entre fermións e campos vectoriais xorden da **invariancia gauge local do lagranxiano**. Este principio, central na teoría cuántica de campos, garante a renormalizabilidade da teoría: que as magnitudes físicas observables sexan finitas e ben definidas.

* Finalmente, o mecanismo de Higgs proporciona unha forma coherente de xerar a masa dos bosóns $W^\pm$ e $Z$, así como dos fermións, sen romper a invariancia gauge fundamental do modelo.

### Fitos da creación do SM

* 1961–1967 — Formulación da teoría electrofeble por S. Glashow, A. Salam e S. Weinberg, unificando as interaccións electromagnética e feble.

* 1964 — Desenvolvemento do mecanismo de Higgs por F. Englert e R. Brout, e de forma independente por P. Higgs, introducindo o campo e o bosón asociados á ruptura espontánea de simetría gauge.

* 1973 — Descubrimento das correntes neutras no experimento Gargamelle (CERN), confirmando unha das predicións clave do modelo electrofeble.

* 1973 — G. ’t Hooft e M. Veltman demostran a renormalizabilidade das teorías gauge con rotura de simetría mediante o mecanismo de Higgs, consolidando a consistencia matemática do Modelo Estándar.

* 1983 — C. Rubbia e colaboradores (no experimento UA1, CERN) descobren os bosóns $W^\pm$ e $Z$, as partículas mediadoras da interacción feble.

* Década de 1990 — Os experimentos de precisión en LEP (CERN) confirman as predicións do Modelo Estándar, incluíndo a física do bosón $Z$ e a existencia de tres familias de neutrinos.

* 2012 — Descubrimento do bosón de Higgs polos experimentos ATLAS e CMS do LHC (CERN), completando o esquema do Modelo Estándar.

## Unificación electrofeble

### O descubrimento das correntes neutras

Nos anos 60 sospeitábase da posible existencia dun novo mediador feble neutro, o bosón vectorial $Z$. En particular, a sección eficaz do proceso $e^+ + e \to W^+ + W^-$ medraba infinitamente coa enerxía se no seu cálculo só intervía como mediador o fotón, pero se se consideraba a mediación dun $Z$, a sección eficaz moderábase e acadaba un valor máximo.

A predición máis xenuína da presenza dun $Z$ era a existencia de correntes febles neutras, e de forma limpa entre neutrinos. Os neutrinos poderían agora interactuar debilmente coa materia, mediante o $Z$, e, se tiñan enerxía suficiente, romper os nucleóns e producir hadróns sen dar lugar ao seu leptón cargado asociado:

$$
\nu_\mu + N \to \nu_\mu + \mathrm{hadróns}
$$

onde $N$ é un nucleón.

$$
\nu_\mu + N \to \mu + X
$$


* Nas correntes neutras, o neutrino escapaba sen ser detectado, pero a ruptura do nucleón deixaba trazas observables:

$$
\nu_\mu + N \to \nu_\mu + X
$$

onde $X$ representa os posibles hadróns.

### A interacción electromagnética e a invariancia gauge

Imos agora introducir brevemente o Modelo Estándar e dar unhas pautas de como xorden os bosóns vectoriais e as súas interaccións febles e cargadas.

A partir deste punto é necesario introducir os lagranxianos.



Os lagranxianos e as ecuacións do movemento dos **campos libres** son:

| | | | |
| :--: |:--: | :--: | :--: |
| ----- Tipo ----- | ----- Escalar complexo ------ | -------- Espinor -------- | -- Bosón vectorial ----- |
| | Klein-Gordon | Dirac | Maxwell |
|Ecuación | $(\partial^\mu\partial_\mu + m^2) \, \phi = 0$ | $(i\gamma^\mu \partial_\mu - m) \, \Psi = 0$ | $\partial_\mu F^{\mu\nu} = 0$ |
|Lagranxiano| $(\partial_\mu \phi)^* (\partial^\mu \phi) - m^2 \phi^* \phi$ | $i\bar{\Psi}\gamma^\mu \partial_\mu \Psi - m \bar{\Psi}\Psi$|  $-\frac{1}{4}F^{\mu\nu}F_{\mu\nu}$|

Onde:

$$
F_{\mu\nu} = \partial_\mu A_\nu - \partial_\nu A_\mu
$$

Pero a física *interesante* sucede na **interacción entre os campos**.

A interacción electromagnética xorde ao esixir que o lagranxiano de fermións sexa **invariante gauge local**.

Podemos verificar facilmente que o lagranxiano de Dirac é invariante baixo un cambio de fase global $\theta$
 no espinor $\Psi(x)$.

$$
\Psi'(x) \to e^{i q\, \theta} \,\, \Psi(x)
$$

onde $q$ é a carga eléctrica do fermión.

Para que o lagranxiano sexa invariante baixo unha transformación local, isto é, cun cambio de fase que depende localmente, $\theta(x)$,

$$
\Psi'(x) \to e^{i q\, \theta(x)} \, \Psi(x)
$$

debemos considerar conxuntamente a existencia do campo fermiónico coa do campo vectorial do fotón.

Para iso, debemos substituír a derivada ordinaria pola derivada covariante xeneralizada.

$$
D_\mu = \partial_\mu + i q A_\mu
$$


O lagranxiano conxunto do fermión cargado e do fotón é:

$$
\mathcal{L}  = i \bar{\Psi} \gamma^\mu D_\mu \Psi - m \bar{\Psi} \Psi  - \frac{1}{4} F_{\mu\nu}F^{\mu\nu}  \\
=  i \bar{\Psi} \gamma^\mu \partial_\mu \Psi  - q  \, \bar{\Psi} \gamma^\mu \Psi  A_\mu - m \bar{\Psi}\Psi- \frac{1}{4} F_{\mu\nu}F^{\mu\nu}
$$

Aquí, $q$ é a carga eléctrica do fermión.

Desta forma, a interacción co fotón xorde da simetría, e non se introduce “a man”, cumprindo o principio fundamental de invariancia gauge.

Usade que a parte do campo vectorial non se ve afectada por un cambio:

$$
A'_\mu(x) = A_\mu(x) - \partial_\mu \theta(x)
$$

No lagranxiano, o termo:

$$
 q \bar{\Psi} \gamma^\mu \Psi \, A_\mu
$$

corresponde á interacción entre un fermión cargado con carga $q$ e o fotón, e ten asociado o seguinte vértice dos diagramas de Feynman.

|:--:|
|<img src="./imgs/sm_emcur.png" width = 250 align="center">|

A corrente fermiónica $\bar{\Psi}\gamma^\mu \Psi$ acóplase cun fotón $A_\mu$ cunha intensidade dada pola súa carga $q = Qe$ (onde $e$ é a carga eléctrica e $Q$ a carga do fermión en unidades de $e$).


Dicimos que o lagranxiano ten unha simetría $U(1)$, ou é invariante, baixo un cambio de fase.


G. t'Hooft demostrou que os lagranxianos con invariancia gauge local son *renormalizables*, isto é, ao cuantificar os campos e calcular as seccións eficaces pódense obter resultados finitos.


### Teoría electrofeble

Respecto ás interaccións febles cargadas, mediadas polos $W^\pm$, sabemos que:

* Só interveñen os espinores de quiralidade esquerda.

* Sempre aparecen en pares. Na primeira familia, os pares son:


  $$
  \begin{pmatrix} \nu_e \\ e \end{pmatrix}, \;\;\;   \begin{pmatrix} u \\ d \end{pmatrix}
  $$

Para preservar os dous feitos anteriores, a **teoría electrofeble** constrúese a partir de:

* considerar que os **fermións non teñen masa** -polo momento-

* situar os espinores de esquerdas nun doblete dun espazo interno que chamaremos **isospín feble** de $SU(2)$ e os de dereitas en singletes:

$$
L  \equiv \begin{pmatrix} \Psi^{\nu_e}_L \\ \Psi^{e}_L \end{pmatrix}; \;\; \Psi^{e}_R
$$
$$
Q  \equiv \begin{pmatrix} \Psi^{u}_L \\ \Psi^{d}_L \end{pmatrix}; \;\; \Psi^{u}_R, \,\Psi^{d}_R
$$


Nótese que, por construción, o neutrino só ten compoñente de quiralidade esquerda. E tamén que os fermións deben carecer de masa para poder separar as dúas compoñentes de quiralidade en campos independentes.

Con esta configuración, o lagranxiano da primeira familia de leptóns pódese escribir como:

$$
i\bar{L} \gamma^\mu \partial_\mu L + i \bar{\Psi}^e_R \gamma^\mu \partial_\mu \Psi^e_R
$$

O primeiro termo é un produto escalar entre dous dobletes.

Lembremos que non hai termo de masas, xa que consideramos fermións sen masa.

Se desenvolvemos o primeiro termo, o asociado ao doblete, obtemos:

$$
i \bar{L} \gamma^\mu  \partial_\mu L =
i \begin{pmatrix} \bar{\Psi}^{\nu}_L, \bar{\Psi}^{e}_L \end{pmatrix} \gamma^\mu \partial_\mu \begin{pmatrix} \Psi^{\nu}_L \\ \Psi^{e}_L \end{pmatrix} = \\
i \bar{\Psi}^{\nu}_L \gamma^\mu  \partial_\mu \Psi^{\nu}_L + i \gamma^\mu \bar{\Psi}^{e}_L \gamma^\mu  \partial_\mu \Psi^{e}_L   \\
$$

Se engadimos agora de novo o segundo termo, o asociado ao singlete, obtemos:

$$
i \bar{\Psi}^{\nu}_L \gamma^\mu  \partial_\mu \Psi^{\nu}_L + i \gamma^\mu \bar{\Psi}^{e}_L \gamma^\mu  \partial_\mu \Psi^{e}_L + i \gamma^\mu \bar{\Psi}^{e}_R \gamma^\mu  \partial_\mu \Psi^{e}_R
$$

onde usamos que:

$$
\bar{\Psi}^{e} \gamma^\mu  \partial_\mu \Psi^e  = \bar{\Psi}^{e}_L \gamma^\mu  \partial_\mu \Psi^{e}_L + \bar{\Psi}^{e}_R \gamma^\mu  \partial_\mu \Psi^{e}_R
$$

desta forma, recuperamos a forma inicial do lagranxiano de Dirac para un neutrino e un electrón sen masa:

$$
i \bar{\Psi}^{\nu}_L \gamma^\mu  \partial_\mu \Psi^{\nu}_L + i \bar{\Psi}^{e} \gamma^\mu  \partial_\mu \Psi^e
$$


O lagranxiano da primeira familia, a parte do doblete,

$$
i\bar{L} \gamma^\mu \partial_\mu L + i \bar{\Psi}^e_R \gamma^\mu \partial_\mu \Psi^e_R
$$

é agora invariante baixo catro cambios de fase:

| | |
| :--: | :--: |
| ---- do grupo $U(1)$ ---- | ---- do grupo $SU(2)$ ---- |
|  $L' = e^{i\theta} \, L$ |   $L' = e^{i\theta_i \frac{\sigma^i}{2}} \, L$ |

Lembremos que o xerador das rotacións en $U(1)$ é a identidade, 1, e en $SU(2)$ os xeradores son as matrices de Pauli $\sigma^i/2$.

*cuestión*: verificade que o lagranxiano anterior é invariante

Respecto á rotación $U(\theta)$:

$$
U(\theta) = e^{i\theta_i \frac{\sigma^i}{2}}
$$

$$
L' = U(\theta) \, L, \;\; \bar{L}' = U^\dagger(\theta) L
$$

logo
$$
i \bar{L}' \gamma^\mu \partial_\mu L' = i \bar{L} U^\dagger \gamma^\mu U \partial_\mu L =
i \bar{L} U^\dagger U \gamma_\mu \partial_\mu L = i \bar{L} \gamma^\mu \partial_\mu L
$$

dado que:

$$
U^\dagger U = I
$$

Esixindo que a invariancia do lagranxiano sexa local, para cada unha das posibles fases aparece agora un campo vectorial e a interacción entre o fermión e os campos.

Omitimos aquí o desenvolvemento da teoría electrofeble, que contén detalles sutís moi importantes, entre eles como aparece o ángulo de Weinberg (véxase a extensión e [MT]), para indicar que os campos vectoriais físicos son:

$$
W_\mu^\pm, \, Z_\mu, \, A_\mu
$$

Os $W^\pm_\mu$ son bosóns vectoriais cargados, con carga eléctrica $\pm e$, e os campos $Z_\mu, A_\mu$ son bosóns vectoriais neutros. O campo $A_\mu$ corresponde ao fotón e $Z_\mu$ ao bosón vectorial neutro $Z$.

A teoría electrofeble ten só dous parámetros:

  * A **carga eléctrica**, $e$, -ou, se se prefire, a constante de estrutura fina, $\alpha$-.

  * O acoplamento feble $g_W$.

En vez de $g_W$, na literatura utilízase por razóns históricas **o ángulo de Weinberg**, $\theta_W$, cuxo valor se coñece con precisión:

$$
\sin^2 \theta_W = 0.23146 \pm 0.000012
$$


A relación das constantes de acoplamento con $e$ e $\theta_W$ é

| | | |
| :--: | :--: | :--: |
| electromagnética | ------  feble cargada -------- | --------- feble neutra -------------|
| $e$ |  $g_W = \frac{e}{\sin \theta_W}$ | $g_Z  = \frac{g_W}{\cos \theta_W} = \frac{e}{\sin \theta_W \cos \theta_W}$ |


No lagranxiano aparecen agora catro termos de interacción entre cada un dos campos bosónicos e os fermións, que son:

| | |
| :--: | :--: |
| corrente | ----------------------   forma ----------------------------- |
| electromagnética |  $Q e \, (\bar{\Psi} \gamma^\mu \Psi) \, A_\mu$ |
| feble cargada | $\frac{g_W}{\sqrt{2}} (\bar{L} \gamma^\mu \sigma^+ L) \, W^+_\mu + \frac{g_W}{\sqrt{2}} (\bar{L} \gamma^\mu \sigma^- L) \, W^-_\mu$ |
| feble neutra | $g_Z \left( c_L \, \bar{\Psi}_L \gamma^\mu \Psi_L + c_R \, \bar{\Psi}_R \gamma^\mu \Psi_R \right) \, Z_\mu $|

onde:

  * $Q$ é a carga do fermión en unidades da do electrón, isto é, $-1$ para o electrón.

  * $\sigma^\pm = (\sigma^1\pm i\sigma^2)/2$ son as matrices escaleira que conectan os elementos de arriba cos de abaixo do doblete.

  * $c_L, c_R$ son dous coeficientes que dependen da carga eléctrica do fermión e da súa posición no doblete.

Se desenvolvemos o termo de interacción feble cargada:

$$
\frac{g_W}{\sqrt{2}} (\bar{L} \gamma^\mu \sigma^+ L) \, W^+_\mu =
$$

para a primeira familia de leptóns, recuperamos a expresión habitual das interaccións febles cargadas:

$$
 \frac{g_W}{\sqrt{2}}
 \begin{pmatrix} \bar{\Psi}^{\nu}_{L}, \bar{\Psi}^e_L \end{pmatrix} \,
\begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}  \gamma^\mu
\begin{pmatrix} \Psi^\nu_L \\ \Psi^e_L \end{pmatrix} \, W^+_\mu=
\frac{g_W}{\sqrt{2}} \bar{\Psi}^\nu_L \, \gamma^\mu \, \Psi^e_L \, W^+_\mu
$$

Vemos que, por construción, a interacción feble cargada viola a paridade de forma máxima, xa que só interveñen os espinores de esquerdas.

Cómpre salientar que as interaccións co $Z$ están completamente determinadas polo modelo.

Os coeficientes $c_L, c_R$ son:

$$
c_L  = T^3 - Q \sin^2 \theta_W, \;\; c_R = -Q \sin^2 \theta_W
$$

Onde $Q$ é a carga en unidades da do electrón $e$, isto é, -1 para o electrón, e $T^3$ é a terceira compoñente de isospín feble. Isto é, para o neutrino $Q = 0$ e $T^3 = 1/2$, e para o electrón $Q = -1, \; T^3 = -1/2$.

A seguinte táboa mostra os coeficientes para a primeira familia de leptóns:

| | | |
| :-- | :-- | :-- |
| | -------- $c_R$ -------- | -------- $c_L$ --------  |
| $\nu_e$ | 0 | $\frac{1}{2}$|
| $e$ | $s^2_W$ | $-\frac{1}{2} + s^2_W$|

onde $s^2_W \equiv \sin^2 \theta_W$.

Podemos comprobar que nas correntes neutras só intervén novamente a compoñente de quiralidade esquerda do neutrino.

### Asimetría adiante-atrás

Na rexión onde o $Z$ domine ou sexa relevante aparecen efectos de violación de paridade que dependerán de $s^2_W$ no proceso $e+e^+ \to \mu +\mu^+$.

Consideramos o eixe de choque $e+e^+$ e tomamos como dirección cara adiante a que define o $e$. Definimos a asimetría adiante-atrás, $A_{FB}$, a partir do número de $\mu$ que saen cara adiante, $N_F$ (ou *forward*), e cara atrás, $N_B$ (ou *backwards*), como:

$$
A_{FB} = \frac{N_F - N_B}{N_F + N_B}
$$

Se se conserva a paridade, debe cumprirse que $A_{FB} = 0$

### Anchura de desintegración do $Z$

Na década dos 90, o colisionador $e+e^+$ LEP do CERN operou a $\sqrt{s} = m_Z = 91$ GeV producindo millóns de $Z$, o que permitiu estudar a súa física con gran precisión.

Un dos resultados máis importantes de LEP é a medida da anchura de desintegración do $Z$, que se obtén a partir da sección eficaz.

#### Sección eficaz

A sección eficaz $e+e^+ \to \mu + \mu^+$ en función do invariante $s$ na rexión dominada polo $Z$ vén dada por:
$$
\sigma (e+e^+\to Z \to \mu+\mu^+)= \frac{12 \pi s}{m^2_Z} \frac{\Gamma_{ee} \Gamma_{\mu\mu}}{(s-m^2_Z)^2 + m^2_Z\Gamma^2_Z}
$$

Onde $m_Z, \, \Gamma_Z$ son a masa e a anchura de desintegración do $Z$ respectivamente; e $\Gamma_{ee}, \Gamma_{\mu\mu}$ son as anchuras parciais de desintegración do $Z$ a $e+e^+$ e $\mu+\mu^+$ respectivamente -o Modelo Estándar predí que son iguais-.

A sección eficaz segue unha función de tipo Breit-Wigner onde:

  * existe un polo, onde a sección eficaz é máxima, en $\sqrt{s} = m_z$
  $$
  \sigma_{\max} = \frac{12\pi}{m^2_Z} \frac{\Gamma_{ee}\Gamma_{\mu\mu}}{\Gamma^2_Z}
  $$

  * e nos valores $\sqrt{s} = m_Z \pm \Gamma_Z/2$ a sección eficaz vale a metade do máximo:

  $$
  \sigma \left(\sqrt{s} = m_Z \pm \Gamma_Z/2 \right) = \frac{\sigma_{\max}}{2}
  $$

  ou o que é o mesmo, a anchura a media altura, FWHM, é a anchura de desintegración $\Gamma_Z$

*Cuestión* Calcula o máximo da sección eficaz para $e+e^+\to \mu+\mu^+$ e $e+e^+\to \mathrm{hadróns}$ e a curva da sección eficaz para hadróns no pico do $Z$.

```python
Gee = 0.0838 # GeV
Gmm = 0.0838 # GeV
Gnn = 0.1657 # GeV
Guu = 0.2894 # GeV
Gdd = 0.3713 # GeV
Ghad = 2 * Guu + 3 * Gdd
Gtot = Ghad + 3 * Gee + 3 * Gnn
mz  = 91.19  # GeV
Gz  =  2.45  # GeV
sigma_mumu = (12 * np.pi)/(mz**2) * (Gee * Gmm)/Gz**2
sigma_had  = (12 * np.pi)/(mz**2) * (Gee * Ghad)/Gz**2
sigma_tot  = (12 * np.pi)/(mz**2) * (Gee * Gz)/(Gz**2)

sigma_to_barn = fn.GEV2_TO_BARN   # (hbar c)^2 / barn, véxase fnyp/common.py
print('sigma max (ee->mumu) {:3.1f} nbarns '.format(sigma_to_barn * sigma_mumu / units.nano))
print('sigma max (ee->had)  {:3.1f} nbarns '.format(sigma_to_barn * sigma_had /units.nano))
print('sigma max (ee->Z)    {:3.1f} nbarns '.format(sigma_to_barn * sigma_tot /units.nano))
```

```python
# resonancia de Breit-Wigner do Z
def sigma(s, Gff = Gmm):
    return (12 * np.pi * s)/(mz**2) * (Gee*Gff/((s-mz**2)**2 + (mz*Gz)**2))

ss = np.linspace(mz-8, mz+8, 100)
fn.plot_xy(ss, lambda x: sigma_to_barn * sigma(x*x, Ghad) / units.nano,
           r'$\sqrt{s}$ (GeV)', r'$\sigma$ (nbarn)');
```

#### Anchura de desintegración parcial

Por outra banda, a anchura parcial de desintegración do $Z \to f + \bar{f}$ vén dada por:

$$
\Gamma(Z \to f + \bar{f}) = \frac{g^2_Z m_Z}{24 \pi} (c^2_L + c^2_R)
$$

Por exemplo, para $\nu_e$, como $c^{(\nu_e)}_L = 1/2, \; c^{(\nu_e)}_R = 0$:

$$
\Gamma (Z \to \nu_e + \bar{\nu}_e) = \frac{g^2_Z m_Z}{96 \pi} = 166 \; \mathrm{MeV}
$$



*Cuestión*: Calcula as anchuras de desintegración parciais e total do $Z$ e as súas fraccións de desintegración.

```python
GF  = units.value("Fermi coupling constant")
s2t = units.value("weak mixing angle")
MW  = 80.34 # GeV
MZ  = 91.19 # GeV
g2w = 8 * MW**2 * GF /np.sqrt(2)
g2z = g2w/(1-s2t)
gamma_nu = g2z*MZ/(96*np.pi)
print('gw    {:5.4f}'.format(np.sqrt(g2w)))
print('gz    {:5.4f}'.format(np.sqrt(g2z)))
print('Gamma nue {:5.4f} GeV'.format(gamma_nu))
```

```python
nc   = 3 # 3 cores para os quarks
pars = {'e':(-0.5+s2t, s2t), 'nue': (0.5, 0), 'u' : (0.5-2*s2t/3, -2*s2t/3), 'd': (-0.5+s2t/3, s2t/3)}
gamma = lambda cl, cr: g2z*MZ/(24*np.pi)*(cl**2 + cr**2)
gammas = {}
for key in pars.keys():
    cl, cr = pars[key]
    gammai = gamma(cl, cr)
    gammai = nc * gammai if key in ('u', 'd') else gammai
    comment = 'Gamma ' + key + ' {:5.4f} GeV'.format(gammai)
    print(comment)
    gammas[key] = gammai
```

A anchura total do $Z \to f + \bar{f}$ é a suma das parciais:

$$
\Gamma_Z = 3 \Gamma_{ll} +  \Gamma_{\mathrm{had}} + N_\nu \Gamma_{\nu\nu}
$$

onde $\Gamma_{ll} = \Gamma_{ee} = \Gamma_{\mu\mu} = \Gamma_{\tau\tau}$ é a anchura de desintegración parcial a leptóns cargados -que o modelo predí que é a mesma-, $\Gamma_{\mathrm{had}}$ é a anchura de desintegración a hadróns (a total a $q\bar{q}$), e finalmente $N_\nu$ é o número de neutrinos, e $\Gamma_{\nu\nu}$ a anchura de desintegración a un neutrino dun só sabor -o modelo tamén predí que é a mesma para cada sabor de neutrinos-.


```python
gamma_total = 3 * gammas['nue'] + 3 * gammas['e'] + 2 * gammas['u'] + 3 * gammas['d']
print('Gamma total {:5.4f} GeV '.format(gamma_total))
for key in gammas.keys():
    #n  = 2 if key == 'u' else 3
    br = 100 * gammas[key]/gamma_total
    print('BR ' + key + ' {:3.1f} %'.format(br))
```

## O bosón de Higgs

O modelo de Higgs dá resposta (quizais a única posible) aos seguintes problemas graves que presentaba a teoría electrofeble.

  * Os bosóns febles son masivos; porén, as teorías gauge esixen bosóns sen masa. Como dotar entón de masa os bosóns $W^\pm, Z$?

  * A teoría electrofeble separa fermións de dereitas e de esquerdas, en dobletes e singletes, e esixe que non teñan masa. Pero sabemos que os fermións teñen masa!


O mecanismo de Higgs permite dotar de masa fermións e bosóns preservando a invariancia gauge da teoría.

É moi difícil esquivar o lagranxiano do Modelo Estándar para falar do Higgs.

Imos dar aquí só unhas ideas que nos permitan entender o seu comportamento e descubrimento.


### Masa dos fermións

O termo de masa asociado aos fermións que aparece no lagranxiano é:

$$
m \bar{\Psi} \Psi
$$

A masa é, en principio, unha propiedade *intrínseca* da partícula, pero e se fose a consecuencia dunha interacción cun novo campo?

Supoñamos que os fermións se acoplan cun campo escalar $\phi(x)$, cada un deles cun acoplamento adimensional, $\lambda$, -que chamaremos de Yukawa-. O termo de interacción no lagranxiano será:

$$
\lambda \bar{\Psi} \Psi \phi
$$


Ou graficamente:

| |
|:--:|
|<img src="./imgs/sm_higgs_2.png" width = 150 align="center">|
| Acoplamento do campo de Higgs co fermión|


que corresponde a un vértice de tres ramas cun acoplamento de Yukawa $\lambda$.


Consideremos agora un campo escalar complexo *singular*, $\phi(x)$, cun valor mínimo non nulo, o que chamamos un **valor esperado no baleiro**, $v$:

$$
\phi(x) = \frac{1}{\sqrt{2}}\left(v + h(x)\right)
$$

(o factor $1/\sqrt{2}$ simplifica logo os cálculos, pero non é relevante).

A interacción que escribimos anteriormente entre os fermións e o campo $\phi$ queda:

$$
\lambda \bar{\Psi} \Psi \, \phi = \frac{\lambda v}{\sqrt{2}} \bar{\Psi} \Psi + \frac{\lambda}{\sqrt{2}} \bar{\Psi} \, \Psi h(x)
$$


O primeiro termo podemos asocialo á masa do fermión:

$$
m \bar{\Psi} \Psi = \frac{\lambda v}{\sqrt{2}} \bar{\Psi} \Psi \; \Rightarrow \; m= \frac{\lambda v}{\sqrt{2}}.
$$

Podemos dicir entón que a masa aparece como a fricción, a través do acoplamento de Yukawa, $\lambda$, co valor esperado do campo escalar (de Higgs) no baleiro, $v$.

O segundo termo $\bar{\Psi} \Psi h$ interpretámolo como a interacción do campo $h(x)$ cun fermión.


| |
|:--:|
|<img src="./imgs/sm_higgs_0.png" width = 250 align="center">|
| Masa dos fermións e a interacción co Higgs, $h(x)$|



### O campo do bosón de Higgs

Vimos que podemos dotar de masa os fermións cun campo escalar, $\phi(x)$, cuxo valor no baleiro, $v$, sexa non nulo; pero como obtemos ese valor $v$?

Consideremos un campo escalar complexo, que chamaremos campo de Higgs; o seu lagranxiano é:

$$
(D_\mu \phi)^\dagger (D^\mu \phi) -  m^2 \, (\phi^\dagger \phi),
$$

Por certo, nótese que o lagranxiano presenta unha simetría respecto a un cambio de fase $\phi'(x) = e^{i \theta} \, \phi(x)$


O segundo termo do lagranxiano, o de masas, asociámolo a un potencial.

$$
V(\phi) =  m^2 \, (\phi^\dagger \phi)
$$

que ten unha dependencia parabólica en termos de $(\phi^\dagger \phi)$.

Consideremos a extensión máis sinxela dese potencial, que sería cuadrática.

$$
V(\phi) = \frac{\mu^2}{2} (\phi^\dagger \phi) + \frac{\lambda}{4} (\phi^\dagger \phi)^2
$$

que depende dos parámetros $\mu, \lambda$.

Nótese que se $\mu = \sqrt{2} m$ e $\lambda = 0$, recuperamos o potencial inicial.

Os factores $1/2, 1/4$ serven para simplificar os cálculos, pero non son relevantes.


As seguintes figuras mostran o potencial para os casos $\mu^2>0, \mu^2<0$ con $\lambda > 0$.
Observa o valor $v$, que corresponde ao mínimo.

```python
# V(phi) = mu^2 phi^2 / 2 + lambda phi^4 / 4 ;  mu imaxinario  <=>  mu^2 < 0
higgs_potential = lambda phi, mu, xlambda : ((mu**2).real * phi**2)/2 + (xlambda/4) * (phi**2)**2
phis = np.linspace(-10, 10, 100)
mu, xlambda = 6, 1

fn.plot_xy(phis, higgs_potential(phis, mu, xlambda), r'$\phi$', r'$V(\phi)$',
           title=r'$\mu^2 > 0$', ax=plt.subplot(1, 2, 1))
fn.plot_xy(phis, higgs_potential(phis, mu*1j, xlambda), r'$\phi$', r'$V(\phi)$',
           title=r'$\mu^2 < 0$', ax=plt.subplot(1, 2, 2))
plt.tight_layout();
```

O valor esperado no baleiro, $v$, é o estado de mínima enerxía:

* No caso $\mu^2 >0$ o potencial ten un mínimo en $v = 0$.

* Mentres que o caso $\mu^2 <0$ ten dous mínimos en:

$$
v = \pm  \sqrt{\frac{-\mu^2}{\lambda}}
$$

Referímonos tamén a $v$ como *vev* (do inglés *vacuum expectation value*, valor esperado no baleiro).

Para desenvolver o campo arredor do vev:

$$
\phi(x) = v + h(x),
$$

debemos elixir un mínimo, isto é, rompemos a simetría inicial!

Chamamos **rotura espontánea de simetría** ao feito de que, aínda que un sistema físico -isto é, o seu lagranxiano- teña unha simetría dada, o baleiro -isto é, o valor mínimo do campo- a rompa.

No caso de que $\phi$ sexa complexo, obtemos a curva do potencial coñecida como sombreiro mexicano, onde un dos eixes é a compoñente real do campo e o outro a imaxinaria:

| |
| :--: |
|  <img src="./imgs/sm_mexican_hat.png" width = 300 align="center"> |
| potencial do campo complexo de Higgs $V(\phi)$ [Wikipedia]|

Logo a existencia dun campo escalar complexo, cun potencial tipo *sombreiro mexicano*, permítenos obter un valor esperado no baleiro, co que dotar de masa os fermións e os bosóns vectoriais.

Pero, porén, non temos un principio ou unha explicación de *por que este potencial*. *Outro gran misterio!*

O potencial de Higgs introduce dous parámetros libres (indeterminados), $v$ e $\lambda$ -ou, se se prefire, $v$ e $m_H$, o valor esperado no baleiro e a masa do Higgs-.

O mecanismo introduce ademais un parámetro adicional, o acoplamento de Yukawa $\lambda_f$, por cada fermión.

A cambio, tras a rotura espontánea de simetría, o campo de Higgs dá lugar:

 * ás **masas dos fermións** (aínda que non as predí, porque cada fermión ten asociada unha constante de Yukawa propia)

 * (aínda que non o vemos) á **masa dos bosóns vectoriais**, $m_W, m_Z$. Neste caso predí o cociente entre as masas $m_W$ e $m_Z$.

 * á súa propia masa, $m_H$ (aínda que non a predí)

O modelo dános un mecanismo para dotar de masa os fermións, pero non fai ningunha predición sobre elas.

Pero si predí:

 * que existen interaccións entre o Higgs, $h(x)$, e os bosóns febles, $W, Z$, que dan lugar ás **desintegracións do Higgs en bosóns vectoriais**

 * que as interaccións entre o Higgs, $h(x)$, e os fermións dan lugar a **desintegracións en pares fermión-antifermión**

Non só iso, senón que o Higgs se acopla tamén consigo mesmo en vértices de tres e catro ramas, e cos bosóns vectoriais febles en vértices de catro ramas. Isto é, xera novas interaccións consigo mesmo e cos bosóns vectoriais. Noutras palabras, o Higgs presenta unha física moi variada, na súa maior parte aínda inexplorada.

Por último, aínda que non o desenvolvemos, o Modelo Estándar predí a masa do $W^\pm$, $m_W$

$$
m_W = \frac{1}{2} g_W v
$$

Polo tanto, a partir da medida de $M_W$ e de $g_W$ podemos determinar $v$

$$
v = \frac{2m_W}{g_W} = \sqrt{\frac{1}{\sqrt{2} G_F}} = 246 \; \mathrm{GeV}
$$

isto é, $v$ vén dado pola constante de Fermi, $G_F$.

```python
GF = units.value("Fermi coupling constant") # 1/GeV^2
v  = np.sqrt(1/(np.sqrt(2)*GF))
print('valor esperado no baleiro {:4.1f} GeV'.format(v))
```

### Descubrimento do bosón de Higgs

O bosón de Higgs descubriuse en 2012 nos experimentos ATLAS e CMS do LHC do CERN.

Con anterioridade, os experimentos de LEP e os experimentos CDF e D0 de Fermilab restrinxiran o valor da súa masa, para o Higgs máis lixeiro, entre $115-150$ GeV.

Os experimentos e o acelerador LHC deseñáronse nos anos 80 e entraron en funcionamento a finais dos anos 2000.

No LHC colisionan $p+p$ a unha enerxía $\sqrt{s}= 7- 13$ TeV e cunha luminosidade $\mathcal{L}(t) = 10^{34} \; \mathrm{cm}^{-2}$. Hai $10^{11}$ protóns por paquete e unha frecuencia de colisión de 40 MHz (cada 25 ns).

O número de colisións é de $10^{9}$ por segundo e en cada cruzamento hai como máximo $35$ colisións, o que se denomina *pile-up* (amoreamento).

A sección eficaz $p+p$ é duns ~100 mbarns no LHC.

A sección eficaz medida de produción do $H$ para $\sqrt{s}=13$ TeV é de 54 pbarns.

As canles de desintegración onde o sinal se distingue máis facilmente do fondo son $H\to Z+Z^*$ e $H \to \gamma \gamma$.

Iso equivale a buscar unha desintegración relevante entre $10^{13}$.

## Conclusións

O SM:

  * O SM clasifica as partículas coñecidas e describe con precisión as interaccións electrofebles e fortes entre elas.

  * O SM verificouse en numerosos procesos físicos e experimentos.

  * Non obstante, sabemos que os neutrinos teñen masa, o que implica que debemos estender o SM.

Pero:

  * Non inclúe outra física coñecida, como a materia escura: que partículas a forman?

  * O modelo ten cinco parámetros fundamentais $e, g_W, g_S, v, \lambda$, dous deles corresponden ao Higgs. Experimentalmente, $\alpha, G_L, \alpha_S, M_Z, M_H$.

  * Ten 9 parámetros adicionais, que son os acoplamentos de Yukawa (as masas dos leptóns cargados e dos quarks); 4 máis para acomodar a matriz CKM dos quarks, e 4 máis para a PMNS dos leptóns (a que rexe as oscilacións de neutrinos).
