### Os quarks

Os quarks son os constituíntes dos hadróns.

Existen 6 quarks, agrupados en tres dobretes de isospín feble.

$$
\begin{pmatrix} u \\ d \end{pmatrix}, \begin{pmatrix} c \\ s \end{pmatrix}, 
\begin{pmatrix} t \\ b \end{pmatrix} \\
$$

A táboa de quarks

|     | xeración   |              | Q (e)  | 
|:--  | :--          | :--          | :--    |
|  u  (2 MeV) | c (1.2 GeV) | t (170 GeV)  | $\frac{2}{3}$   | 
|  d  (5 MeV)  | s (93 MeV)  | b (4.2 GeV)  |  $-\frac{1}{3}$ | 


Cada quark está nun estado dun triplete de cor ($r, g, b$, vermello, verde e azul):

$$
c_r = \begin{pmatrix} 1 \\ 0 \\ 0 \end{pmatrix}, \;\;
c_g = \begin{pmatrix} 0 \\ 1 \\ 0 \end{pmatrix}, \;\;
c_b = \begin{pmatrix} 0 \\ 0 \\ 1 \end{pmatrix}. 
$$

Polo tanto hai 18 quarks e 18 antiquarks.

Os estados de quark son, por helicidade $\pm$, e cor, $a = r, g, b$:

$$
u_{\pm}(p) \, c_a
$$

Pero cada hadrón está nun singlete de cor, o que limita as combinacións permitidas de quarks en barións e mesóns.

## Características das interaccións fortes

### Confinamento

Os quarks non se presentan illados, sempre aparecen en parellas nos mesóns ou en tríos nos barións. 

Sería doado identificar unha partícula cargada con carga fraccionaria $2/3, -1/3$ e iso non sucedeu. Os quarks non son libres.

Este fenómeno coñécese como **confinamento**.

Non existe unha explicación analítica, pero podemos entendelo como o feito de que, ao separar dous quarks, estes están ligados fortemente polos gluóns, que actuarían como unha "corda" estirada que, ao romper, crearía un quark-antiquark nos seus extremos de rotura (ver [MT10.4])

Matematicamente veremos que a condición de confinamento se traduce en que os quarks teñen que formar singletes de cor.

### Liberdade asintótica

Paradoxalmente, **as constantes de acoplamento non son constantes!**: dependen da escala de enerxía da interacción, aínda que a súa dependencia é suave, logarítmica co $q^2$ (co cadrado do momento transferido). A súa explicación escapa ao contido do curso (ver [MT10.5, AB5.8]) e está relacionada coa escala á que observamos a interacción, que converte a constante en "efectiva". Pero se medimos a constante a unha determinada escala podemos estimar canto vale a unha escala superior.

A teoría cuántica de campos conviviu durante décadas cun problema matemático grave: se considerásemos no propagador (por exemplo, do fotón) todos os posibles diagramas de Feynman de orde superior que poderían intervir, atopariamos que a suma de amplitudes non converxe, isto é, é infinita! 

G. 't Hooft demostrou nos anos 70 que unha **teoría con simetría gauge** pode reabsorber eses diagramas e proporcionar unha amplitude de transición finita. Pero, como consecuencia, a constante de acoplamento, $\alpha$, depende agora de $q^2$ de forma logarítmica. Este mecanismo denomínase **renormalización**.

Valores experimentais da constante de acoplamento forte $\alpha_s(q^2)$ en función de $q$

| | 
| :--: |
|  <img src="./imgs/hadrons_alphas_running.png" width = 350 align="center"> |
| valores experimentais de $\alpha_S(q^2)$ fronte a q [PDG]|

A constante de acoplamento forte é $\alpha_S \sim 1$ a baixas enerxías, pero ás enerxías do LHC $\alpha_S \sim 0.17$, o que permite realizar algúns cálculos perturbativos en QCD válidos para o LHC.

Canto maior é a enerxía, menor é a constante de acoplamento forte e, polo tanto, os quarks están máis libres dentro dos hadróns, aínda que nunca completamente libres; este fenómeno denomínase **liberdade asintótica**.

Pola contra, as constantes de acoplamento electromagnética, $\alpha$, e feble, $\alpha_W$, decrecen con $q^2$. 

*Nota*: Debe existir unha escala de enerxía $\Lambda$ onde as constantes, polo menos dúas, sexan iguais. Nas teorías supersimétricas, as tres constantes igualan a unha determinada escala.

## Estrutura dos nucleóns

A estrutura interna do protón (ou neutrón) estúdase mediante experimentos de dispersión:

|      |      |      |
| :--: | :--: | :--: |
| &emsp; electróns (QED)   &emsp;|  &emsp; &emsp; feble  &emsp; &emsp; |  &emsp; &emsp; forte  &emsp; &emsp;|
|  $e + p$ |  &emsp; $\nu + p$ | &emsp; $p + p$|

cada un é sensible a diversas forzas: (QED e feble), feble e forte, respectivamente.

Os procesos divídense en:

|     | |
| :--: | :--: |
| &emsp;&emsp;   elásticos &emsp;&emsp;  | &emsp;&emsp;   inelásticos &emsp;&emsp;   |
| $e + p \to e + p$ | $e + p \to e + X$ |
  
Onde $X$ indica a produción de diversos hadróns.

A dispersión inelástica tórnase dominante conforme a enerxía aumenta.

A unha determinada enerxía, os procesos inelásticos correspóndense con procesos elásticos da sonda cos partóns (quarks de valencia).

Conforme a enerxía da sonda $e, \nu, p$ aumenta, podemos explorar escalas distintas do protón.

  * O protón como partícula 'puntual' con espín $1/2$ e carga $+e$.
  
  * A distribución de carga, $\rho(r)$, e o momento magnético do protón.
  
  * A estrutura de quarks de valencia (uud), que se chamaron partóns.
  
  * Os quarks de valencia e o mar de gluóns e quarks.

A estrutura do protón descubriuse grazas ás probas $e+ p \to e + X$, principalmente en SLAC nos anos 60 [ver MT8, AB6.3]. A interacción é complexa porque é electrofeble.

Tamén podemos estudar a estrutura dos nucleóns a partir das dispersións con $\nu_\mu$, dado que *só interaccionan debilmente*, e é máis sinxelo. Estes experimentos realizáronse nos anos 80 [ver MT12.2-3].

Os cálculos da sección eficaz, usando as regras dos diagramas de Feynman, son relativamente sinxelos en teoría cuántica de campos:

No caso $\nu + q \to \mu + q'$:

$$
\langle |\mathcal{M}_{fi}|^2 \rangle = \frac{1}{2} \left(\frac{g^2_W}{m^2_W}\right)^2 s^2, \;\;\;
\frac{\mathrm{d}\sigma}{\mathrm{d}\Omega^*} = \frac{G^2_F}{4\pi^2} s, \;\;\;
\sigma = \frac{G^2_F}{\pi} s.
$$

E cos antineutrinos,  $\bar{\nu} + q \to \mu + q'$:

$$
\frac{\mathrm{d}\sigma}{\mathrm{d}\Omega^*} = \frac{1}{4}(1+\cos\theta^*)^2 \frac{G^2_F}{4\pi^2} s, \;\;\;
\sigma = \frac{G^2_F}{3 \pi} s
$$

É dicir, a razón entre ambas é simplemente:

$$
\frac{\sigma(\nu_\mu + N)}{\sigma(\bar{\nu}_\mu + N)} = 3
$$

Os experimentos de dispersión $e+p$ xa descubriran que a composición do protón é moi complexa. Está formado polos quarks de valencia ($uud$), pero tamén por gluóns e por pares quark-antiquark do mar.

Sexa $f_q$ a fracción de quarks no nucleón e $f_{\bar q}$ a de antiquarks. Podemos dar as seccións eficaces nun nucleón composto por $q$ e $\bar{q}$ por:

$$\sigma(\nu_\mu + N \to \mu + X) = \frac{G^2_F}{\pi} s \left(f_q + \frac{1}{3} f_{\bar{q}} \right), \\

Nótese que os neutrinos non interaccionan cos gluóns, só cos quarks e antiquarks!

Cos valores experimentais anteriores obtemos:
$$f_q \simeq 0.41, \;\; f_{\bar{q}} \simeq 0.1$$

Concluímos que **o protón está composto nun 41% de quarks e nun 10% de antiquarks; o resto son gluóns!**

O que está de acordo cos experimentos $e + p \to e + X$.

$$\frac{g_{\mu\nu}}{q^2}, \;\; \frac{g_{\mu\nu}}{q^2 - m_Z^2},$$

  * Na zona intermedia, os dous diagramas contribúen.
  

#### Sección eficaz $e+e^+\to f+\bar{f}$.

Consideremos o rango de $q^2$ de aniquilación no que domina o fotón (QED).

### Número de cores

A carga eléctrica dos quarks entra no vértice do elemento de matriz cun factor $Q^2_q e^2$.

A sección eficaz  $e+e^+ \to q+\bar{q}$ (hadróns) dependerá da enerxía dispoñible, $\sqrt{s}$, dos tipos de quarks (os distintos sabores) que se poidan crear con esa enerxía e da súa carga.

Pero cada quark preséntase ademais en tres cores, polo que cada cor de quark introduce un novo elemento de matriz, isto é, o elemento de matriz total está multiplicado por $N_c = 3$.

O factor que temos que engadir para un quark $q$ é:

$$
N_c Q^2_q
$$

A sección total é:

$$\sigma(e+e^+\to \mathrm{hadróns}) = \frac{4\pi \alpha^2}{s} \, N_c \, \sum_q Q^2_q,$$

onde $q = u, d, s, c, b$ dependendo da enerxía $\sqrt{s}$.

Cando $\sqrt{s}$ supera o limiar de produción dun quark, $q$, a sección eficaz $\sigma(e+e^+\to q+\bar{q})$ incrementarase en $N_c Q^2_q$

Podemos comparar esta sección eficaz coa de $\sigma(e+e^+ \to \mu+\mu^+)$, que son idénticas en valor agás o factor asociado á carga e ao número de cores dos quarks.

A razón $R_\mu$ entre a produción de hadróns e de pares $\mu, \mu^+$ dependerá de:

$$R_\mu = \frac{\sigma(e+e^+ \to q+\bar{q})}{\sigma(e+e^+ \to \mu+\mu^+)} = N_c \, \sum_q Q^2_q$$

  * a corrección adicional de QCD axústase aos datos. 
  
No capítulo do Modelo Estándar discutiremos a resonancia do $Z$ (á dereita nas figuras).

Nota adicional: o efecto da corrección de QCD débese a que os gluóns transmiten cor. O bo axuste demostra que efectivamente é así.

## Desintegracións febles de hadróns.

Sabemos que o bosón $W$ se acopla cun leptón de arriba e un de abaixo, pero sempre da mesma xeración.

$$
\begin{pmatrix}\nu_e \\ e \end{pmatrix}, \; \begin{pmatrix}\nu_\mu \\ \mu \end{pmatrix}, \;
\begin{pmatrix}\nu_\tau \\ \tau \end{pmatrix}
$$

Vimos (no tema de leptóns) que a constante de acoplamento é a mesma para as tres xeracións, $g_W$, o que chamamos **universalidade** leptónica.

| |
| :--: |
| <img src="./imgs/feynman_wleptons.png" width = 600 align="center">|



Nas desintegracións febles dos leptóns vimos que a constante de acoplamento de cada vértice era universal, $G^{(e)}_F = G^{(\mu)}_F = G^{(\tau)}_F$.

A partir das medidas das desintegracións $\beta$ obtemos a constante de acoplamento do vértice asociado á corrente dos quarks, $G^{(\beta)}_F$, que podemos comparar coa $G^{(\mu)}_F$ do vértice das desintegracións $\mu \to \nu_\mu + e + \bar{\nu}_e$:

| | 
| :--: |
|  <img src="./imgs/feynman_mu_beta_decay.png" width = 350 align="center"> |
| Diagramas de Feynman da desintegración $\beta$ e do $\mu$|

$$G^{(\beta)}_F = 1.1066 \pm 0.001 \, 10^{-5} \, \mathrm{GeV^{-2}}, \\
G^{(\mu)}_F = 1.1663787 \pm 0.0000006 \, 10^{-5} \, \mathrm{GeV^{-2}}.$$

que é un 5% máis pequena.

Cabibbo (1963) propuxo que as interaccións febles son universais para os autoestados febles $d', s'$, pero que experimentalmente observamos os estados de masa $d, s$ dentro dos hadróns. Entre ambos media unha matriz de mestura unitaria.

Se só consideramos dous quarks de abaixo, a matriz é simplemente:

$$\begin{pmatrix} d' \\ s' \end{pmatrix} = 
\begin{pmatrix} \cos \theta_c & \sin \theta_c \\ - \sin \theta_c & \cos \theta_ c \end{pmatrix}
\begin{pmatrix} d \\ s \end{pmatrix},$$
onde $\theta_C$ é o ángulo de Cabibbo.

De tal forma que os vértices das correntes cargadas entre quarks serán:

|  <img src="./imgs/feynman_cabibbo.png" width = 350 align="center"> |
| Vértices febles co ángulo de Cabibbo|


A constante $G_F^{(\beta)}$ estaría suprimida por un factor $\cos^2 \theta_C$.

Isto explica a diferenza entre as anchuras de desintegración de $\pi^- \to \mu + \bar{\nu}_\mu$ e $K^- \to \mu +  \bar{\nu}_\mu$.

| | 
| :--: |
|  <img src="./imgs/feynman_piK_cabibbo.png" width = 450 align="center"> |
| Vértices febles na desintegración $\pi^, K^-$|

A razón de desintegración entre ambos estará suprimida por un factor $\tan^2 \theta_c$.

O valor de $\theta_C \sim 13$º.

Agora sabemos experimentalmente que os quarks se acoplan co $W^\pm$ en pares de arriba e abaixo entre as distintas xeracións:

$$
\begin{pmatrix}u \\ d \end{pmatrix}, \; \begin{pmatrix} u \\ s \end{pmatrix}, \; \begin{pmatrix}u \\ b \end{pmatrix} \\
\begin{pmatrix}c \\ d \end{pmatrix}, \; \begin{pmatrix} c \\ s \end{pmatrix}, \; \begin{pmatrix}c \\ b \end{pmatrix} \\
\begin{pmatrix}t \\ d \end{pmatrix}, \; \begin{pmatrix} t \\ s \end{pmatrix}, \; \begin{pmatrix}t \\ b \end{pmatrix}
$$



A constante de acoplamento xa non é universal, senón que depende do par de arriba e abaixo que se acopla, $g_W V_{ij}$ (onde $i = \{u, c, t\}, j = \{d, s, b\}$).

| |
| :--: |
|<img src="./imgs/feynman_vertexW.png" width = 600 align="center">|
|Vértices do $W$ con parellas de quarks|


A extensión a 3 xeracións dá lugar á matriz $V_{CKM}$ de Cabibbo-Kobayashi-Maskawa (CKM).

$$\begin{pmatrix} d' \\ s' \\ b' \end{pmatrix} = 
\begin{pmatrix} V_{ud} & V_{us} & V_{ub} \\ 
                V_{cd} & V_{cs} & V_{cb} \\
                V_{td} & V_{ts} & V_{tb} \end{pmatrix}
\begin{pmatrix} d \\ s \\ b \end{pmatrix}$$

Onde $V$ é unha matriz unitaria, $V^\dagger V = I$, que queda definida por tres ángulos de rotación e unha fase complexa.

As correntes febles entre quarks $u \to d$ son:

$$
j^\mu_{u\to d} =  \bar{u}(p_d) \, \left[ \frac{g_w}{\sqrt{2}} V^*_{ud} \gamma^\mu \frac{1}{2}(I - \gamma^5) \right] u(p_u), 
$$

onde $p_u, p_b$ son os cuadrimomentos de $u, b$, respectivamente. 

A matriz CKM dá lugar a unha **fenomenoloxía moi rica entre hadróns**, onde existen **ligaduras** que deben cumprirse.

En particular, a matriz debe ser unitaria.

Por exemplo, as seguintes desintegracións permiten determinar $|V_{us}|, |V_{ub}|$

|  <img src="./imgs/feynman_K0_B0_decays.png" width = 450 align="center"> |
| Diagramas de Feynman $K^0 \to \pi^- + e^+ + \nu_e, \; B^0 \to \pi^- + e^+ + \nu_e$|


Os valores experimentais dos módulos dos elementos da matriz CKM:
  
$$\begin{pmatrix} |V_{ud}| \simeq 0.974& |V_{us}| \simeq 0.225 & |V_{ub}| \simeq 0.004\\ 
                |V_{cd}| \simeq 0.225& |V_{cs}| \simeq 0.973 & |V_{cb}| \simeq 0.041 \\
                |V_{td}| \simeq 0.009& |V_{ts}| \simeq 0.040 & |V_{tb}| \simeq 0.999 \end{pmatrix}$$

Observamos:
  
  * os valores diagonais son próximos a 1.
  
  * a mestura entre as xeracións 1-2 é máis intensa que entre as 1-3 e 2-3.


Se escribimos a matriz en termos dos tres ángulos de mestura $\theta_{12}, \theta_{13}, \theta_{23}$ e unha fase complexa $\delta$:

$$
V_{CKM} =  
\begin{pmatrix} 1 & 0 & 0 \\ 0 & c_{23} & s_{23} \\ 0 & -s_{23} & c_{23} \end{pmatrix}
\begin{pmatrix} c_{13} & 0 & s_{13}e^{-i\delta} \\ 0 & 1 & 0 \\  -s_{13} e^{i\delta}& 0 & c_{13} \end{pmatrix}
\begin{pmatrix} c_{12} & s_{12} & 0 \\ -s_{12} & c_{12} & 0 \\ 0 & 0 & 1 \end{pmatrix}
$$

onde $c_{ij} = \cos \theta_{ij}, \; s_{ij} = \sin \theta_{ij}$.

Os valores experimentais:

$$
s_{12} = 0.2265 \pm 0.0005, \; s_{13} = 0.0036\pm 0.0001, \; s_{23} = 0.04505 \pm 0.0008, \delta = 1.196\pm0.045
$$

Os ángulos de mestura son "pequenos". Existe unha **fase complexa** $\delta$, unha constante da Natureza, que dará lugar a unha fenomenoloxía moi rica de procesos de **violación de carga e paridade en hadróns** [ver MT14, AB8] que escapan ao nivel deste curso.

## Sobre a cromodinámica cuántica, QCD.

Vimos antes, ao medir $R_\mu$, que os quarks se presentaban en **tres cores**, que chamamos vermello, verde e azul.
En realidade son os nomes de cargas; non teñen nada que ver coas cores habituais.

A teoría detrás das interaccións fortes, a cromodinámica cuántica (Quantum Chromo Dynamics, QCD), desenvolveuse nos anos 70 por Gross, Politzer, Wilczek, Fritzsch, Gell-Mann e 't Hooft, entre outros.

En QCD cada **quark** ten unha posible **cor**: vermello, verde e azul, que podemos asignar como as compoñentes dun triplete.

$$c_r = \begin{pmatrix} 1 \\ 0 \\ 0 \end{pmatrix}, \;\;
c_g = \begin{pmatrix} 0 \\ 1 \\ 0 \end{pmatrix}, \;\;
c_b = \begin{pmatrix} 0 \\ 0 \\ 1 \end{pmatrix}. $$

De forma similar aos dobretes de espín:

$$\chi_\uparrow   = \begin{pmatrix} 1 \\ 0 \end{pmatrix}, \;\;
\chi_\downarrow = \begin{pmatrix} 0 \\ 1  \end{pmatrix}.$$

De forma matemática podemos entender o espinor de Dirac, $u(p)$, engadíndolle un triplete de cor, $c_a$: 

$$u(p) \to u(p) \, c_i, $$

con $a = r, g, b$.

O dobrete corresponde á representación irredutible ${\bf 2}$ do grupo SU(2), que tiña tres xeradores, $T_i = \frac{1}{3}\sigma_i$, dados polas matrices de Pauli, que obedecían as regras de conmutación do momento angular. 

En SU(2) só dous operadores conmutan, $T^2, T_3$, o espín total e a súa terceira compoñente. Os estados irredutibles etiquetabámolos con $T_3$. As combinacións de estados daban lugar a novos estados que se relacionaban cos orixinais a través dos coeficientes de Clebsch-Gordan, que á súa vez se obtiñan da aplicación dos operadores escaleira.

O triplete de cor é a representación irredutible ${\bf 3}$ do grupo SU(3), que ten 8 xeradores, $T_i = \frac{1}{2} \lambda_i$, dados polas matrices de Gell-Mann.

En SU(3) só tres operadores conmutan, $T^2, T_3, Y$, onde $T_3$ é a terceira compoñente de cor e $Y$ é a hipercarga de cor. Cada estado podemos etiquetalo agora con dous valores, $T_3, Y$.

Cando combinabamos dobretes de SU(2) obtiñamos un triplete e un singlete: ${\bf 2} \otimes {\bf 2} = {\bf 3}\oplus  {\bf 1}$

Na figura móstrase graficamente a combinación dun triplete cun antitriplete (${\bf 3} \otimes {\bf \bar{3}}$),  dous tripletes (${\bf 3} \otimes {\bf 3}$) e tres tripletes de cor (${\bf 3} \otimes {\bf 3} \otimes {\bf 3}$):

QCD esixe que os hadróns sexan **singletes de cor**, partículas **sen cor** ou *brancas*. 

As combinacións que dan singletes son:

$${\bf 3} \otimes {\bf 3} \otimes {\bf 3}   = {\bf 10} \oplus {\bf 8} \oplus {\bf 8} \oplus {\bf 1} \\
{\bf 3} \otimes {\bf \bar{3}}  = {\bf 8} \oplus {\bf 1}$$

Podemos formar unha partícula *sen cor*:

   * cun **trío de quarks**, que son os **barións**
   
   * cun par **quark-antiquark**, que son os **mesóns**.

O factor asociado aos vértices de cor nos diagramas de Feynman é:

$$-i \frac{g_S}{2} \lambda^a,$$
onde $g_S$ é a constante de acoplamento forte e $\lambda^a$ é a matriz de Gell-Mann asociada a cada un dos 8 gluóns, $a = 1, \dots, 8$.

O propagador dos gluóns é:

$$-i \frac{g_{\mu\nu}}{q^2} \delta^{ab}$$

onde $\delta^{ab}$ garante que un mesmo gluón media entre os dous vértices.


A corrente dos quarks que cambian de cor $i \to k$ mediada polo gluón $a$ queda:

$$j^\mu_q  = \bar{u}(p_c) \left[ c^\dagger_k \left( -i \frac{g_s}{2} \lambda^a \right)   \gamma^\mu  c_i \right] u(p_a),$$

onde $p_a$ é o cuadrimomento do quark entrante e $p_c$ o do saínte.

Nótese que na corrente só introducimos un factor de cor, que corresponde ao elemento $k, i$ da matriz $\lambda^a$:

$$c^\dagger_k \lambda^a c_i = \lambda^a_{ki}$$

### QCD e QED

QED (electrodinámica cuántica) e QCD (cromodinámica cuántica) son teorías similares: 

   * Son teorías con simetrías gauge locais que conservan paridade, carga e sabor.
   
   * Os mediadores son campos vectoriais sen masa.
   
Porén, son radicalmente diferentes:

   * O fotón non ten carga eléctrica e o seu alcance é infinito. Non hai interaccións entre fotóns. 
   
   * Os gluóns transfiren cor e por iso interaccionan entre eles, o que fai que o seu alcance sexa de fm e que só sexan libres as partículas singletes de cor.
   
   * O fotón provén do xerador do grupo U(1), que é a simetría gauge local do electromagnetismo.
   
   * Os gluóns proveñen dos xeradores do grupo SU(3), as matrices de Gell-Mann, que é a simetría gauge local de QCD. Existen tres cargas en QCD, ou cores, $r, g, b$, e as súas correspondentes anticargas $\bar{r}, \bar{g}, \bar{b}$.
   
Nota: podemos entender $r$ como unha carga e pensar en $r = +, \bar{r} = -1$, de forma similar á carga eléctrica.
