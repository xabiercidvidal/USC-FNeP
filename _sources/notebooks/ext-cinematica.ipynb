## 1. Unidades naturais

### 1.1. Todo en potencias do GeV

Con $\hbar = c = 1$ só queda unha unidade, a de enerxía. Cada magnitude é unha potencia do GeV:

| magnitude | dimensión en UN | pásase ao SI con |
|:--|:--:|:--|
| enerxía, momento, masa | GeV | — ($c$ e $c^2$ sobreentendense: GeV$/c$, GeV$/c^2$) |
| lonxitude | $\mathrm{GeV}^{-1}$ | $\hbar c = 0.1973$ GeV fm |
| tempo | $\mathrm{GeV}^{-1}$ | $\hbar = 6.582 \cdot 10^{-25}$ GeV s |
| sección eficaz | $\mathrm{GeV}^{-2}$ | $(\hbar c)^2 = 0.3894$ $\mathrm{GeV}^2$ mb |
| anchura, ritmo | GeV | $\Gamma = \hbar/\tau$ |

**A receita.** Unha magnitude que en UN vale $X$ $\mathrm{GeV}^n$ pásase ao SI multiplicando pola
combinación de $\hbar$ e $c$ que dá as dimensións correctas. Abonda con lembrar dúas cousas: $\hbar c$
converte enerxía en inverso de lonxitude, e $\hbar$ converte enerxía en inverso de tempo.

### 1.2. Exemplos do curso

* **Unha anchura e unha vida media.** $\Gamma_Z = 2.495$ GeV dá
  $\tau_Z = \hbar/\Gamma_Z = 6.58 \cdot 10^{-25} / 2.495 = 2.6 \cdot 10^{-25}$ s, e
  $c\tau_Z = \hbar c/\Gamma_Z = 0.197/2.495 = 0.079$ fm.
* **Unha sección eficaz.** $\sigma(e^+ + e \to \mu^+ + \mu) = 4\pi\alpha^2/3s$ a $\sqrt{s} = 10$ GeV vale
  $2.2 \cdot 10^{-6}$ $\mathrm{GeV}^{-2}$, é dicir, $2.2 \cdot 10^{-6} \times 0.389$ mb $= 0.87$ nb.
* **A lonxitude de desintegración.** En UN, $L = \gamma\beta\,\tau_0$: tempo e lonxitude teñen a
  mesma dimensión e non fai falta ningunha $c$. Para o muón,
  $\tau_0 = 2.197 \cdot 10^{-6}\ \mathrm{s} / 6.58 \cdot 10^{-25}\ \mathrm{GeV\,s} = 3.34 \cdot 10^{18}$ $\mathrm{GeV}^{-1}$, e
  $\tau_0 \times \hbar c = 3.34 \cdot 10^{18} \times 0.197$ fm $= 659$ m: é o $c\tau_0$ da introdución.
* **A constante de estrutura fina.** No SI, $\alpha = e^2/(4\pi\varepsilon_0\hbar c) \simeq 1/137$. Con
  $\hbar = c = 1$ e $\varepsilon_0 = 1$ (unidades de Heaviside-Lorentz), $\alpha = e^2/4\pi$ e
  $e = \sqrt{4\pi\alpha} = 0.303$. $\alpha$ é adimensional: vale o mesmo en calquera sistema.

*Cuestións:* canto vale en fm o raio do protón, $r_p = 4.1$ $\mathrm{GeV}^{-1}$? E en $\mathrm{GeV}^{-2}$ unha
sección eficaz de $1$ pb?

## 2. A transformación de Lorentz

### 2.1. Dedución

Dous sistemas inerciais, $\Sigma'$ movéndose con velocidade $\beta$ en $z$ respecto de $\Sigma$, cos
orixes coincidentes en $t = t' = 0$. Tres ingredientes:

1. **O intervalo é invariante.** Un escintileo na orixe cumpre $t^2 - {\bf r}^2 = 0$ nos dous
   sistemas; Einstein esixe máis: $t^2 - {\bf r}^2 = t'^2 - {\bf r}'^2$ para calquera suceso.
2. **Linearidade.** A transformación é linear (o espazo-tempo é homoxéneo), e por simetría as
   direccións transversas non cambian: $x' = x$, $y' = y$.
3. **A orixe de $\Sigma'$** ($z' = 0$) móvese en $\Sigma$ como $z = \beta t$.

Escribimos entón $t' = A\,t + B\,z$, $z' = C\,t + D\,z$. A condición 3 dá $C\,t + D\,\beta t = 0$, é
dicir, $C = -\beta D$. A condición 1, $t'^2 - z'^2 = t^2 - z^2$ para todo $t, z$, dá

$$
A^2 - C^2 = 1, \qquad D^2 - B^2 = 1, \qquad AB - CD = 0 .
$$

Con $C = -\beta D$, a terceira dá $B = -\beta D^2/A$, e a segunda,
$D^2\,(A^2 - \beta^2 D^2) = A^2$. Como a primeira é $A^2 - \beta^2 D^2 = 1$, queda $D^2 = A^2$: $D = A$
(con signo positivo, para que $\beta \to 0$ sexa a identidade). Entón $A^2(1 - \beta^2) = 1$:

$$
A = D = \frac{1}{\sqrt{1 - \beta^2}} \equiv \gamma, \qquad B = C = -\gamma\beta
$$

$$
\boxed{
t' = \gamma\,(t - \beta z), \qquad x' = x, \qquad y' = y, \qquad z' = \gamma\,(z - \beta t)
}
$$

A inversa obtense cambiando $\beta \to -\beta$ (a orixe de $\Sigma$ móvese con $-\beta$ en
$\Sigma'$). É a transformación da [introdución](./introduccion.ipynb).

### 2.2. Consecuencias

* **Dilatación do tempo.** Un reloxo en repouso en $\Sigma'$ ($z'$ fixo) marca un intervalo $\Delta\tau$. Coa
  inversa, $t = \gamma\,(t' + \beta z')$, en $\Sigma$ dura

  $$\Delta t = \gamma\,\Delta\tau \;\geq\; \Delta\tau .$$

  É a vida media no laboratorio, $\tau = \gamma\tau_0$.
* **Contracción das lonxitudes.** Unha regra en repouso en $\Sigma'$ mide $L_0 = \Delta z'$. En $\Sigma$ mídense
  os seus extremos á vez, $\Delta t = 0$, e $z' = \gamma\,(z - \beta t)$ dá $L_0 = \gamma L$:

  $$L = \frac{L_0}{\gamma} .$$

  Un volume contráese na dirección do movemento, $V = V_0/\gamma$: é o que obriga á
  normalización $2E$ da sección 6.3.
* **Composición de velocidades.** Un obxecto con velocidade $u'$ en $z$ en $\Sigma'$ ten en $\Sigma$

  $$u = \frac{\mathrm{d}z}{\mathrm{d}t} = \frac{u' + \beta}{1 + u'\beta}$$

  (dividindo $\mathrm{d}z = \gamma(\mathrm{d}z' + \beta\,\mathrm{d}t')$ entre $\mathrm{d}t = \gamma(\mathrm{d}t' + \beta\,\mathrm{d}z')$). Nunca supera
  $1$: con $u' = 1$, $u = 1$.

### 2.3. A rapidez

Como $\gamma^2 - (\gamma\beta)^2 = 1$, pódese pór $\gamma = \cosh\eta$, $\gamma\beta = \sinh\eta$, $\beta = \tanh\eta$: a
transformación é unha rotación **hiperbólica** no plano $(t, z)$, de «ángulo» $\eta$, a **rapidez**:

$$
\begin{pmatrix} t' \\ z' \end{pmatrix} =
\begin{pmatrix} \cosh\eta & -\sinh\eta \\ -\sinh\eta & \cosh\eta \end{pmatrix}
\begin{pmatrix} t \\ z \end{pmatrix}
$$

**As rapidezas súmanse.** Dous *boosts* seguidos en $z$ son o produto das dúas matrices, e polas
fórmulas de suma do coseno e o seno hiperbólicos, o resultado é outro *boost* de rapidez
$\eta_1 + \eta_2$. É a composición de velocidades escrita doutra forma:
$\tanh(\eta_1 + \eta_2) = (\beta_1 + \beta_2)/(1 + \beta_1\beta_2)$.

Para unha partícula, o *boost* desde o seu repouso ten $\cosh\eta = E/m$ e $\sinh\eta = \mathrm{p}/m$, logo

$$
\eta = \frac{1}{2}\ln\frac{E + \mathrm{p}}{E - \mathrm{p}} .
$$

Con $\mathrm{p}$ na dirección do *boost*, é a **rapidez** $y$ da sección 5.

## 3. Cuadrivectores

### 3.1. Métrica e índices

Un **cuadrivector contravariante** (índice arriba), $a^\mu = (a^0, {\bf a})$, é calquera cuaterna que se
transforma como $x^\mu = (t, {\bf x})$. Coa métrica de Minkowski báixase o índice e obtense o
**covariante**:

$$
a_\mu = g_{\mu\nu}\,a^\nu = (a^0, -{\bf a}), \qquad
a \cdot b = a^\mu b_\mu = g_{\mu\nu}\,a^\mu b^\nu = a^0 b^0 - {\bf a}\cdot{\bf b} .
$$

A norma, $a^2 = (a^0)^2 - {\bf a}^2$, pode ser positiva (tipo tempo), nula (tipo luz) ou negativa
(tipo espazo). O **cuadrigradiente**, $\partial_\mu = \partial/\partial x^\mu = (\partial_t, \nabla)$, leva o índice abaixo.

### 3.2. As transformacións de Lorentz como matrices

$x'^\mu = \Lambda^\mu{}_\nu\, x^\nu$, con $\mu$ o índice de fila e $\nu$ o de columna. O que **define** unha
transformación de Lorentz é que conserva o produto escalar: para todo $x, y$,
$x'^{T} g\, y' = x^T \Lambda^T g \Lambda\, y = x^T g\, y$, logo

$$
\boxed{
\Lambda^T g\, \Lambda = g
}
$$

Tomando determinantes, $\det\Lambda = \pm 1$. As **propias** ($\det\Lambda = +1$: rotacións e *boosts*)
conéctanse de forma continua coa identidade; a **paridade**, ${\bf x} \to -{\bf x}$, é impropia. Por seren
«pseudo-ortogonais» respecto de $g$, os covariantes transfórmanse coa inversa,
$a'_\mu = a_\nu\,(\Lambda^{-1})^\nu{}_\mu$, e por iso $a'^\mu b'_\mu = a^\mu b_\mu$.

Rotación arredor de $z$ e *boost* en $z$ (activos):

$$
\Lambda(\phi) =
\begin{pmatrix} 1 & 0 & 0 & 0 \\ 0 & \cos\phi & -\sin\phi & 0 \\ 0 & \sin\phi & \cos\phi & 0 \\ 0 & 0 & 0 & 1 \end{pmatrix},
\qquad
\Lambda(\eta) =
\begin{pmatrix} \cosh\eta & 0 & 0 & \sinh\eta \\ 0 & 1 & 0 & 0 \\ 0 & 0 & 1 & 0 \\ \sinh\eta & 0 & 0 & \cosh\eta \end{pmatrix}
$$

O formalismo completo (xeradores, representación dos espinores) está en
[ext-trans-Lorentz](./ext-trans-Lorentz.ipynb).

### 3.3. Cuadrivelocidade e cuadrimomento

O **tempo propio**, $\mathrm{d}\tau^2 = \mathrm{d}t^2 - \mathrm{d}{\bf x}^2 = \mathrm{d}t^2/\gamma^2$, é invariante. Dividindo o
cuadrivector $\mathrm{d}x^\mu$ por el obtéñense outros dous cuadrivectores:

$$
u^\mu = \frac{\mathrm{d}x^\mu}{\mathrm{d}\tau} = \gamma\,(1, {\bf v}), \qquad
p^\mu = m\,u^\mu = (\gamma m, \gamma m{\bf v}) = (E, {\bf p}), \qquad
p^2 = m^2 .
$$

(É a tarefa «Por que $(E, {\bf p})$ se transforma como $(t, {\bf r})$?» da introdución.) Nunha
reacción consérvase o cuadrimomento total, e a súa norma é o invariante $s$ (sección 4).

### 3.4. Unha medida invariante: $\mathrm{d}^3{\bf p}/2E$

$\mathrm{d}^3{\bf p}$ non é invariante (contráese na dirección do *boost*), pero $\mathrm{d}^3{\bf p}/2E$ si. Vese
partindo de algo que obviamente o é:

$$
\mathrm{d}^4p\;\delta(p^2 - m^2)\,\theta(p^0)
$$

($\mathrm{d}^4p$ é invariante porque $|\det\Lambda| = 1$; $p^2$ é invariante; e un *boost* non cambia o signo de
$p^0$). Integrando en $p^0$ con $\delta(p_0^2 - E^2) = \frac{1}{2E}\left[\delta(p^0 - E) + \delta(p^0 + E)\right]$, só
sobrevive $p^0 = E > 0$:

$$
\int \mathrm{d}p^0\; \delta(p^2 - m^2)\,\theta(p^0) = \frac{1}{2E}
\qquad \Rightarrow \qquad
\frac{\mathrm{d}^3{\bf p}}{2E} \;\text{ é invariante.}
$$

É a medida do espazo fásico relativista da sección 6.

## 4. Invariantes e centro de masas

### 4.1. O invariante $s$ e a velocidade do CM

En $a + b \to \dots$, con $b$ en repouso no laboratorio, o cuadrimomento total é
$P = (E_a + m_b, {\bf p}_a)$ e a súa norma,

$$
s = P^2 = m_a^2 + m_b^2 + 2\,m_b E_a ,
$$

vale o mesmo en calquera sistema. O CM é o sistema no que ${\bf P} = 0$: móvese respecto do
laboratorio coa velocidade do cuadrimomento total, e como $P^2 = s$ fai o papel dunha masa,

$$
\beta_{CM} = \frac{|{\bf p}_a|}{E_a + m_b}, \qquad \gamma_{CM} = \frac{E_a + m_b}{\sqrt{s}}, \qquad
\gamma_{CM}\beta_{CM} = \frac{|{\bf p}_a|}{\sqrt{s}} .
$$

En branco fixo e a alta enerxía, $\gamma_{CM} \simeq \sqrt{E_a/2m_b}$: o CM móvese moi axiña, e os
produtos saen concentrados **cara adiante**. Nun colisionador simétrico o CM é o laboratorio.

### 4.2. Do CM ao laboratorio

Unha partícula que no CM ten enerxía $E^*$, momento $p^*$ e ángulo $\theta^*$ respecto da dirección
do *boost* ten no laboratorio

$$
E = \gamma_{CM}\left(E^* + \beta_{CM}\,p^*\cos\theta^*\right), \qquad
p\cos\theta = \gamma_{CM}\left(p^*\cos\theta^* + \beta_{CM}\,E^*\right), \qquad
p\sin\theta = p^*\sin\theta^* ,
$$

e dividindo as dúas últimas,

$$
\tan\theta = \frac{\sin\theta^*}{\gamma_{CM}\left(\cos\theta^* + \beta_{CM}/\beta^*\right)}, \qquad \beta^* = \frac{p^*}{E^*} .
$$

Se a partícula é máis lenta no CM que o propio CM, $\beta^* < \beta_{CM}$, o denominador non se anula
nunca: **todas** saen cara adiante, dentro dun cono de ángulo máximo
$\sin\theta_{max} = \beta^*\gamma^*/(\beta_{CM}\gamma_{CM})$. É a xeometría dos experimentos de branco fixo e
*cara adiante* de perspectivas.

### 4.3. Cinemática a dous corpos

No CM de $a + b \to c + d$ os momentos iniciais son opostos, e os finais tamén:
${\bf p}^*_i = {\bf p}_a = -{\bf p}_b$, ${\bf p}^*_f = {\bf p}_c = -{\bf p}_d$. Como a enerxía total é $\sqrt{s}$:

$$
s = (p_a + p_b)^2 = (E^*_a + E^*_b)^2 .
$$

As enerxías e os momentos no CM quedan fixados por $s$ e as masas.

Podemos calcular $E^*_a, E^*_b$ a partir de $s$ e das masas $m_a, m_b$.

Se calculamos $E^{*2}_a$ tendo en conta que $E^*_a = (\sqrt{s} - E^*_b)$, e $p^*_a = p^*_b = p^*_i$, obtemos:

$$
\begin{gathered}
E^{*2}_a = m^2_a + p_i^{*2} = E^{*2}_b - 2 \sqrt{s}E^*_b + s = m^2_b + p_i^{*2} - 2 \sqrt{s}E^*_b + s \\
m^2_a = m^2_b + s - 2 \sqrt{s} E^*_b
\end{gathered}
$$


e como o tratamento para $a$ é idéntico ao de $b$, obtemos:

$$
E^*_a = \frac{s + m^2_a - m^2_b}{2 \sqrt{s}}, \;\; E^*_b = \frac{s + m^2_b - m^2_a}{2 \sqrt{s}}.
$$

O momento será:

$$
p^{*}_i = \sqrt{E^{*2}_a - m^2_a} = \sqrt{E^{*2}_b - m^2_b}, \qquad 2 p^{*2}_i = E^{*2}_a + E^{*2}_b - m^2_a - m^2_b
$$

polo tanto

$$
\begin{gathered}
p^{*2}_i = \frac{E^{*2}_a + E^{*2}_b - m^2_a - m^2_b}{2} = \\
\frac{\left[s + (m^2_a - m^2_b) \right]^2 + \left[s - (m^2_a - m^2_b)\right]^2 - 4 s (m^2_a + m^2_b)}{8 s} = \\
\frac{s^2 + (m^2_a - m^2_b)^2 - 2 s (m^2_a + m^2_b)}{4s}
\end{gathered}
$$

como:

$$
\begin{gathered}
m^2_a - m^2_b = (m_a + m_b) (m_a - m_b), \\ 2(m^2_a + m^2_b) = (m_a+m_b)^2 + (m_a-m_b)^2
\end{gathered}
$$

temos:

$$
\begin{gathered}
p^{*2}_i = \frac{s^2 + (m_a+m_b)^2(m_a-m_b)^2 - s \left[ (m_a+m_b)^2 + (m_a - m_b)^2 \right] }{4s} = \\
       = \frac{\left[s- (m_a+m_b)^2\right] \, \left[s - (m_a-m_b)^2\right]}{4s}
\end{gathered}
$$

polo que o momento inicial é:

$$
p^*_i = \frac{\sqrt{\left[s- (m_a+m_b)^2\right] \, \left[s - (m_a-m_b)^2\right]}}{2\sqrt{s}}
$$

Como a situación é idéntica para as partículas finais, temos:

$$
\begin{gathered}
E^*_c = \frac{s + m^2_c - m^2_d}{2 \sqrt{s}}, \;\; E^*_d = \frac{s + m^2_d - m^2_c}{2 \sqrt{s}} \\
p^*_f = \frac{\sqrt{\left[s- (m_c+m_d)^2\right] \, \left[s - (m_c-m_d)^2\right]}}{2 \sqrt{s}}
\end{gathered}
$$

No caso da desintegración a dous corpos, $a \to c + d$, a enerxía no CM é $ \sqrt{s} = m_a$.

Os momentos e enerxías finais son nese caso:

$$
\begin{gathered}
E^*_c = \frac{m^2_a + m^2_c - m^2_d}{2 m_a}, \;\; E^*_d = \frac{m^2_a + m^2_d - m^2_c}{2 m_a} \\
p^*_f = \frac{1}{2m_a}\sqrt{\left[m^2_a- (m_c+m_d)^2\right] \, \left[m^2_a - (m_c-m_d)^2\right]}
\end{gathered}
$$

### 4.4. As variables de Mandelstam

Cos catro cuadrimomentos de $a + b \to c + d$ fórmanse tres invariantes:

$$
s = (p_a + p_b)^2, \qquad t = (p_a - p_c)^2, \qquad u = (p_a - p_d)^2 .
$$

$s$ é a enerxía no CM ao cadrado e $t$ o cuadrimomento transferido de $a$ a $c$ (o $q^2$ do
propagador na canle $t$). Só dous son independentes:

$$
s + t + u = m_a^2 + m_b^2 + m_c^2 + m_d^2 ,
$$

porque, desenvolvendo, $s + t + u = 3m_a^2 + m_b^2 + m_c^2 + m_d^2 + 2\,p_a\cdot(p_b - p_c - p_d)$ e, por
conservación, $p_b - p_c - p_d = -p_a$.

**O invariante $t$.**


O invariante $t = (p_c - p_a)^2 = (p_d - p_b)^2$ é:

$$
\begin{gathered}
t = (p_c - p_a)^2 = p^2_c + p^2_a - 2 p_c \cdot p_a = m^2_a + m^2_c + 2 |{\bf p}_a| |{\bf p}_c| \cos \theta_{ac} - 2 E_a E_c \\
t = (p_d - p_b)^2 = p^2_d + p^2_b - 2 p_d \cdot p_b = m^2_b + m^2_d + 2 |{\bf p}_b| |{\bf p}_d| \cos \theta_{bd} - 2 E_b E_d
\end{gathered}
$$

onde $\theta_{ac}, \theta_{bd}$ son os ángulos entre os momentos de $ac$ e $bd$, respectivamente.

No sistema de laboratorio, con $b$ en repouso, $p_b = (m_b, {\bf 0})$:

$$
 t = m^2_b + m^2_d - 2m_b E_d, \;\; E_d = \frac{m^2_b + m^2_d - t}{2m_b}
$$


No sistema CM (véxase a figura da interacción a dous corpos), $|{\bf p}_a| = p^*_i, \; |{\bf p}_c| = p^*_f$ e $\theta_{ac} = \theta^*$

$$
\begin{gathered}
t = m^2_a + m^2_c + 2 p^*_i p^*_f \cos \theta^* - 2E^*_a E^*_c, \\
\cos \theta^* = \frac{t - m^2_a - m^2_c + 2E^*_aE^*_c}{2 p^*_i p^*_f}
\end{gathered}
$$

## 5. Cinemática en colisionadores

### 5.1. Momento transverso, rapidez e pseudorrapidez

Nun colisionador de hadróns chocan partóns con fraccións descoñecidas do momento do protón:
o sistema da colisión ten un *boost* descoñecido **ao longo do feixe** ($z$). Convén usar
variables que se transformen de forma sinxela baixo ese *boost*:

* o **momento transverso**, $p_T = \sqrt{p_x^2 + p_y^2}$, e o ángulo azimutal $\phi$: non cambian;
* a **rapidez**,

  $$y = \frac{1}{2}\ln\frac{E + p_z}{E - p_z}, \qquad E = m_T\cosh y, \quad p_z = m_T\sinh y, \quad m_T = \sqrt{m^2 + p_T^2} ,$$

  que baixo un *boost* en $z$ de rapidez $\eta$ se **despraza**, $y \to y + \eta$ (sección 2.3): as
  diferenzas de rapidez e a forma de $\mathrm{d}N/\mathrm{d}y$ non cambian.

Para medir $y$ fai falta a masa. Se $E \gg m$, $y$ só depende do ángulo polar $\theta$, e úsase a
**pseudorrapidez**:

$$
\eta_{ps} = -\ln\tan\frac{\theta}{2} .
$$

(Na literatura escríbese $\eta$; aquí poñémoslle subíndice para non confundila coa rapidez $\eta$ do
*boost* de perspectivas.) $\eta_{ps} = 0$ é perpendicular ao feixe; $|\eta_{ps}| = 2.5$, o límite dos
detectores de trazas de ATLAS e CMS, é $\theta = 9.4^\circ$; LHCb cobre $2 < \eta_{ps} < 5$, entre $15^\circ$ e
$0.8^\circ$.

### 5.2. Desintegración en voo

Unha partícula de masa $M$ que se move con $\gamma, \beta$ desintégrase en dúas. No seu repouso cada filla
ten $E^*$ e $p^*$ fixos (sección 4.3), e no laboratorio (sección 4.2)

$$
E = \gamma\left(E^* + \beta\,p^*\cos\theta^*\right) .
$$

Se a desintegración é isótropa no repouso (espín 0), $\cos\theta^*$ repártese uniforme entre $-1$ e $1$,
e a **enerxía no laboratorio tamén**: o espectro é **plano** entre
$\gamma(E^* - \beta p^*)$ e $\gamma(E^* + \beta p^*)$.

**Exemplo: $\pi^0 \to \gamma + \gamma$.** Con $E^* = p^* = m_\pi/2$, cada fotón ten unha enerxía uniforme entre
$E_\pi(1 - \beta)/2$ e $E_\pi(1 + \beta)/2$, e o ángulo entre os dous fotóns é como mínimo
$\sin(\theta_{min}/2) = m_\pi/E_\pi$. Un $\pi^0$ de $10$ GeV dá fotóns separados uns $27$ mrad: nun
calorímetro a un metro, $2.7$ cm, da orde do tamaño dunha cela. Por iso, a alta enerxía, un $\pi^0$
pode confundirse cun fotón.

A mesma conta, proxectada no plano transverso, dá o **pico xacobiano** da masa transversa
do $W$ (tarefa de perspectivas): unha distribución uniforme en $\cos\theta^*$ amoréase onde a
variable medida deixa de medrar.

## 6. Da regra de ouro a $\Gamma$ e $\sigma$

### 6.1. A regra de ouro

Un estado inicial $|i\rangle$ e unha perturbación $H_{int}$. A primeira orde, a amplitude de atopar o
sistema en $|f\rangle$ ao cabo dun tempo $t$ é

$$
c_f(t) = -i\,T_{fi}\int_0^t e^{i(E_f - E_i)t'}\,\mathrm{d}t', \qquad T_{fi} = \langle f|H_{int}|i\rangle ,
$$

e o seu módulo ao cadrado,

$$
|c_f(t)|^2 = |T_{fi}|^2\,\frac{\sin^2(\Delta t/2)}{(\Delta/2)^2}, \qquad \Delta = E_f - E_i ,
$$

é unha función cada vez máis picuda en $\Delta = 0$, de área $2\pi t$: para $t$ grande,
$|c_f|^2 \to 2\pi\,t\,\delta(E_f - E_i)\,|T_{fi}|^2$. A probabilidade **crece linealmente** co tempo: hai
un **ritmo** constante. Sumando sobre os estados finais, con $\mathrm{d}N = \rho(E_f)\,\mathrm{d}E_f$ estados en
$\mathrm{d}E_f$:

$$
\boxed{
\Gamma_{fi} = 2\pi\,|T_{fi}|^2\,\rho(E_i)
}
$$

É a regra de ouro de perspectivas. O traballo está en contar os estados, $\rho$.

### 6.2. Contar estados: de onde saen $\delta^4$ e os $(2\pi)$

Normalizamos cada partícula nunha caixa de volume $V = L^3$ con condicións periódicas: os momentos
permitidos son ${\bf p} = 2\pi{\bf n}/L$, e en $\mathrm{d}^3{\bf p}$ caben $V\,\mathrm{d}^3{\bf p}/(2\pi)^3$ estados. Con $V = 1$ (unha
partícula por unidade de volume), $\mathrm{d}^3{\bf p}/(2\pi)^3$.

Para $n$ partículas finais hai dúas restricións:

* **Momento.** O elemento de matriz leva a integral das ondas planas,
  $\int e^{i({\bf p}_a - \sum{\bf p}_k)\cdot{\bf x}}\,\mathrm{d}^3x = (2\pi)^3\,\delta^3({\bf p}_a - \sum{\bf p}_k)$ (con $V \to \infty$): o momento
  consérvase, e o último momento queda fixado polos demais.
* **Enerxía.** A regra de ouro impóna coa derivada $\rho = \mathrm{d}N/\mathrm{d}E$, que é o mesmo que integrar
  con $\delta(E_a - \sum E_k)$.

Xuntando, para unha desintegración:

$$
\Gamma = 2\pi\int |T_{fi}|^2\,\delta\Big(E_a - \sum E_k\Big)\,(2\pi)^3\delta^3\Big({\bf p}_a - \sum{\bf p}_k\Big)\prod_{k=1}^{n}\frac{\mathrm{d}^3{\bf p}_k}{(2\pi)^3}
= (2\pi)^4\int |T_{fi}|^2\,\delta^4\Big(p_a - \sum p_k\Big)\prod_{k=1}^{n}\frac{\mathrm{d}^3{\bf p}_k}{(2\pi)^3}
$$

O $(2\pi)^4$ é o $2\pi$ da regra de ouro polo $(2\pi)^3$ da conservación do momento; cada
partícula final achega un $1/(2\pi)^3$.

### 6.3. Normalización relativista e elemento de matriz

En mecánica cuántica normalizamos a función de ondas, $\psi$ nun cubo de lados $L$, con volume $V = L^3$, á unidade:

$$
\int_V \psi^* \psi \, \mathrm{d}^3x = 1
$$

É dicir, a partícula está contida en $V$.

O número de estados dispoñibles nese volume é:

$$
\frac{V \, \mathrm{d}^3{\bf p}}{(2\pi)^3},
$$

que vén de expresar $\frac{\mathrm{d}^3 {\bf x} \, \mathrm{d}^3 {\bf p}}{h^3}$ en unidades naturais, onde $h = 2\pi$.

O volume, porén, non é un invariante Lorentz.

Un cubo de lado $L$ visto desde un sistema no que se move con factor de Lorentz $\gamma$
contráese na dirección do movemento (sección 2.2):

$$
V' = \frac{L^3}{\gamma} = \frac{V}{\gamma},
$$

e a densidade de partículas, $1/V$, **aumenta** nun factor $\gamma$.

Se nese cubo estaba contida a partícula con enerxía $E_0 $, no sistema de referencia con factor $\gamma$ a súa enerxía é:

$$
E' = \gamma E_0
$$

Vemos que a densidade, $1/V$, e a enerxía, $E$, se transforman da mesma forma, cun factor $\gamma$.

En mecánica cuántica relativista, normalizamos a función de ondas, $\Psi$, nun volume $V$ a $2E$:

$$
\int_V \Psi^* \Psi \, \mathrm{d}^3 x = 2E,
$$

O factor $2$ é por conveniencia.

É dicir, $2E$ partículas por unidade de volume. Esta normalización si é invariante porque a densidade, $1/V$, e a enerxía, $E$, cambian co mesmo factor ao cambiar o sistema de referencia.

Como consecuencia da normalización relativista da función de ondas $\Psi$, o elemento de matriz relativista $M_{fi} = \langle \Psi_f| H_{int}| \Psi_i \rangle$ relaciónase co non relativista $T_{fi} = \langle \psi_f | H_{int}| \psi_i \rangle$ mediante un factor que provén da normalización.

$$
M_{fi} = \langle \Psi_f | H_{int}| \Psi_i \rangle = \left(\prod_i \sqrt{2E_i} \right) \langle  \psi_f | H_{int}| \psi_i \rangle =  \left(\prod_i \sqrt{2E_i} \right)  T_{fi}
$$
onde o índice $i$ percorre as $n$ partículas iniciais e finais.

As expresións da taxa de transición (a anchura) ou da sección eficaz no tratamento relativista incorporan polo tanto o seguinte factor:

$$
\frac{1}{\prod_i (2E_i)},
$$

debido á normalización relativista.

### 6.4. As fórmulas xerais: anchura e sección eficaz

A **regra de ouro de Fermi** dá a taxa de transición como o elemento de matriz ao cadrado pola
densidade de estados finais. Aplicada a unha partícula que se desintegra dá a **anchura**,
$\Gamma$; aplicada a dúas que chocan, dividida polo fluxo, a **sección eficaz**, $\sigma$.
Escribímolas primeiro na súa forma non relativista e despois na relativista, cos factores
$1/(2E)$ da normalización.

**Anchura de desintegración.**

A expresión non relativista (sección 6.2) da anchura de desintegración dunha partícula de enerxía $E_a$ a $m$ partículas finais é:


$$
\Gamma  = (2\pi)^4 \int |T_{fi}|^2 \delta^4 \left(\sum_k p_k - p_a \right) \, \prod_i \frac{\mathrm{d}^3{\bf p}_i}{(2\pi)^3}
$$

onde $p_a$ é o cuadrimomento da partícula inicial e $p_i$ (ou $p_k$) os das finais, $i = 1, \dots, m$. Lembremos que as deltas de Dirac impoñen a conservación da enerxía e do momento entre a partícula inicial e as finais.

E a relativista:

$$
\Gamma = \frac{(2\pi)^4}{2 E_a} \int |M_{fi}|^2 \delta^4 \left(\sum_k p_k - p_a \right) \, \prod_i \frac{\mathrm{d}^3{\bf p}_i}{(2\pi)^3 (2E_i)}
$$

**Sección eficaz da interacción de dúas partículas $a + b$.**

Sexa a interacción entre dúas partículas $a, b$ que dá lugar a $m$ partículas finais.

Sexa un fluxo de partículas $a$, cunha densidade de $N_a$ partículas nun volume $V$, que se move contra $N_b$ partículas branco nun volume $V$ que á súa vez se moven en sentido oposto con $v_b$.

Asignámoslles unha sección eficaz $\sigma$ de interacción ás partículas branco $b$. A frecuencia de interaccións corresponderá ao número de partículas $a$ que atravesan, por unidade de tempo, un cilindro de base $\sigma$ e altura $(v_a + v_b)\,\mathrm{d}t$ para cada partícula $b$.

| |
|:--|
|<img src="./imgs/fun_drawing_xsection.png" width = 400 align="center">|
| (a) unha partícula $a$ percorre en $\mathrm{d}t$ un cilindro de lonxitude $(v_a+v_b)\,\mathrm{d}t$ [MT3.4] |
| (b) visión transversal do cilindro, as partículas $b$ nel co disco da sección eficaz|

É dicir, o número de interaccións nun volume $V$ e nun tempo $\mathrm{d}t$ é:

$$
\mathrm{d}N = (v_a + v_b) \sigma \, n_a N_b \, \mathrm{d}t
$$

O factor $(v_a + v_b) n_a$ é o fluxo de partículas $a$, $\phi_a$, con velocidade $v_a$ que atravesan unha superficie perpendicular á súa velocidade, cando a partícula $b$ se move contra elas con velocidade $v_b$.

Logo, a frecuencia de interaccións, $R$, nun volume, V, onde hai $N_b$ partículas branco, ao que chega un fluxo $\phi_a$ de partículas $a$ é:

$$
R = \phi_a \, \sigma \, N_b
$$

En mecánica cuántica adóptase a normalización dunha partícula nunha unidade de volume, $V = 1, n_a = 1, n_b = 1$; nese caso:

$$
R = (v_a + v_b) \, \sigma, \;\;\; \sigma = \frac{R}{(v_a+v_b)}
$$

Con esa normalización a unha partícula por unidade de volume, a expresión non relativista para a sección eficaz de $a+b$ para producir $m$ partículas é:

$$
\sigma = \frac{(2\pi)^4}{|v_a + v_b|} \int |T_{fi}|^2 \delta^4 \left(\sum_k p_k - p_a - p_b \right) \, \prod_i \frac{\mathrm{d}^3{\bf p}_i}{(2\pi)^3},
$$

onde $v_a, v_b$ son os módulos das velocidades, considerando que as partículas colisionan nunha mesma dirección e teñen sentidos opostos. As deltas de Dirac impoñen a condición de conservación da enerxía e do momento entre os estados inicial e final.

A expresión relativista, co factor $1/(2E)$ de cada partícula, é:

$$
\sigma = \frac{(2\pi)^4}{(2 E_a) (2 E_b) (\beta_a + \beta_b)} \int |M_{fi}|^2 \delta^4 \left(\sum_k p_k - p_a -p_b\right) \, \prod_i \frac{\mathrm{d}^3{\bf p}_i}{(2\pi)^3 (2E_i)}
$$

Na versión relativista, cada un dos factores da sección eficaz é invariante Lorentz:

   * o elemento de matriz ao cadrado: $|M_{fi}|^2 = | \langle \Psi_f| H_{int} | \Psi_i\rangle |^2$

   * a densidade de estados: $\frac{\mathrm{d}^3 {\bf p}}{(2\pi)^3 (2E)}$ (sección 3.4)

   * o termo asociado ao fluxo: $4 E_a E_b (\beta_a + \beta_b)$ (para feixes colineais)

Nótese que, a partir da propiedade da función delta de Dirac, $\int_{E>0} \delta(E^2-{\bf p}^2 - m^2) \, \mathrm{d}E = \frac{1}{2E}$, podemos reescribir:

$$
\frac{\mathrm{d}^3{\bf p}}{(2\pi)^3 (2E)} = \int \delta(p^2-m^2) \, \theta(p^0) \frac{\mathrm{d}^4p}{(2\pi)^3}
$$

*Cuestión*: Demostrar que $4 E_a E_b (\beta_a + \beta_b) = 4 \sqrt{(p_a \cdot p_b)^2 - m^2_a m^2_b}$, que no CM vale $4 \, p^*_i \sqrt{s}$

### 6.5. Anchura de desintegración a dous corpos

Consideremos a desintegración dunha partícula $a$ en dúas, $b, c$, con masas $m_a, m_b, m_c$.

| |
|:--:|
|<img src="./imgs/intro_drawing_decay2.png" width = 400 align="center">|
| Esquema da desintegración da partícula $a$ en dúas, $b, c$, no CM|

O sistema máis conveniente para tratar o proceso é o CM, onde a partícula $a$ está en repouso e ten $E = m_a$.

As partículas $b, c$ saen con momentos opostos ${\bf p}^* = {\bf p}_b = - {\bf p}_c$.

O módulo do momento das partículas saíntes, $p^*$, é (sección 4.3):

$$
p^*= \frac{1}{2m_a} \sqrt{[m^2_a - (m_b+m_c)^2] \, [m^2_a - (m_b - m_c)^2] }
$$

A anchura de desintegración, $\Gamma$, é a taxa de transición da regra de ouro de Fermi.

Integramos usando $\delta({\bf p}_b + {\bf p}_c)$ en  $\mathrm{d}^3p_{c}$ directamente.

$$
\Gamma = \frac{1}{(2 E) (2\pi)^2} \int |M_{fi}|^2 \delta \left(E - E_b - E_c \right) \frac{p^2_b }{4 E_b E_c}\,  \mathrm{d}p_b \mathrm{d}\Omega^*,
$$

onde $E_b = \sqrt{m^2_b + p^2_b}, \; E_c = \sqrt{m^2_c + p^2_b}$ e $\mathrm{d}\Omega^*$ é o diferencial de ángulo sólido.

Se aplicamos a seguinte propiedade da delta de Dirac: $\int \delta(f(p))\,  \mathrm{d} p = \left|\frac{\mathrm{d}f(p)}{\mathrm{d}p} \right|_{p'}^{-1}$, onde $p'$ cumpre $f(p') = 0$,

con $f(p_b) = E -  \sqrt{m^2_b + p^2_b}- \sqrt{m^2_c + p^2_b}$ respecto de $p = p_b$, e $p' = p^*$:

$$
\frac{\mathrm{d}f(p_b)}{\mathrm{d}p_b} = \frac{p_b}{\sqrt{m^2_b + p^2_b}} + \frac{p_b}{\sqrt{m^2_c + p^2_b}} = p_b \frac{E_b + E_c}{E_b E_c} = p_b \frac{E}{E_b E_c}
$$

A anchura de desintegración queda:

$$
\Gamma = \frac{1}{8 \pi^2 E} \frac{(p^*)^2}{4 E_b E_c} \frac{E_b E_c}{p^* E} \int |M_{fi}|^2 \, \mathrm{d}\Omega^* =
\frac{1}{8\pi^2}\frac{p^*}{4 E^2} \int |M_{fi}|^2 \, \mathrm{d}\Omega^*
$$

A anchura de desintegración, neste caso, só depende de $p^*$, $m_a$ e do elemento de matriz, se poñemos $E = E_a = m_a$ no CM:

$$
\Gamma = \frac{p^*}{32 \pi^2 m^2_a} \int_{\Omega} |M_{fi}|^2 \mathrm{d}\Omega^*
$$

### 6.6. Sección eficaz dunha interacción de dous a dous corpos

Sexa a interacción entre dúas partículas incidentes, $a, b$, que dá lugar a dúas partículas finais, $c, d$.

| |
|:--:|
|<img src="./imgs/intro_drawing_int2.png" width = 400 align="center">|
| Esquema da interacción das partículas $a, b$ a $c, d$ no CM|

O sistema máis conveniente para calcular a sección eficaz é o CM. Neste sistema os momentos iniciais, e os finais, son opostos:

$$
{\bf p}^*_i = {\bf p}_a = - {\bf p}_b, \;\; {\bf p}^*_f = {\bf p}_c = - {\bf p}_d
$$

Nótese que as partículas están nun plano, e o único parámetro libre (ademais do azimut, que se integra) é o ángulo $\theta^*$ entre ${\bf p}^*_f$ e ${\bf p}^*_i$; este último define o eixe.

A sección eficaz obtense a partir da expresión xeral, onde $E = E_a + E_b$, $p_b = p_a = p^*_i$ e $p_d = p_c = p^*_f$.

O termo do espazo de fases calcúlase igual que na anchura de desintegración a dous corpos (sección 6.5), só que agora $E = E_a + E_b$:

$$
\frac{1}{(2\pi)^2}\frac{p^*_f}{4 E} \int  |M_{fi}|^2 \, \mathrm{d}\Omega^*
$$


O termo asociado ao fluxo:

$$
4 E_a E_b (\beta_b + \beta_a) = 4 E_a E_b \left(\frac{p_a}{E_a} + \frac{p_b}{E_b}\right) = 4 p^*_i (E_a + E_b) = 4 E \, p^*_i
$$

A sección eficaz queda:

$$
\sigma = \frac{1}{64 \pi^2 E^2} \frac{p^*_f}{p^*_i} \int_\Omega |M_{fi}|^2 \, \mathrm{d}\Omega^*
$$

Se substituímos $E^2 =  s$, onde $s = \left[(E_a, {\bf p}_a) + (E_b, {\bf p}_b)\right]^2 = [(E_a + E_b, {\bf 0})]^2$ é o cadrado do cuadrimomento total, é dicir, a enerxía no CM ao cadrado:

$$
\sigma = \frac{1}{64 \pi^2 s} \frac{p^*_f}{p^*_i} \int_\Omega |M_{fi}|^2 \, \mathrm{d}\Omega^*
$$

### 6.7. A sección eficaz diferencial en $t$

Co invariante $t$ da sección 4.4: como $t$ só depende de $\cos\theta^*$ a través de $\mathrm{d}t = 2\,p^*_i\,p^*_f \; \mathrm{d}\cos\theta^*$, e
$\mathrm{d}\Omega^* = \mathrm{d}\phi^* \, \mathrm{d}\cos\theta^*$, se $|M_{fi}|^2$ non depende do azimut a sección eficaz
diferencial en $t$ é:

$$
\boxed{
\frac{\mathrm{d}\sigma}{\mathrm{d}t} = \frac{|M_{fi}|^2}{64 \pi \, s \, p^{*2}_i}
}
$$

que é invariante Lorentz, e é a forma que se usa na práctica.

## 7. Tres corpos e a regra de Sargent

### 7.1. O espectro $\beta$

Na desintegración $\beta$ dun núcleo, $N \to N' + e + \bar\nu_e$, o núcleo fillo lévase o momento
que faga falta pero apenas enerxía (é miles de veces máis pesado). A enerxía dispoñible,
$E_0 = E_e + E_\nu$ (a enerxía **total** do electrón no punto final, $E_0 = Q + m_e$), repártese entre
o electrón e o neutrino, e a delta de momento cómese a integral do núcleo. Coa regra de ouro
(sección 6.1) e un elemento de matriz constante, $T$:

$$
\mathrm{d}\Gamma = 2\pi\,|T|^2\,\frac{\mathrm{d}N}{\mathrm{d}E_0}, \qquad
\mathrm{d}N = \frac{4\pi p_e^2\,\mathrm{d}p_e}{(2\pi)^3}\,\frac{4\pi p_\nu^2\,\mathrm{d}p_\nu}{(2\pi)^3} .
$$

A $p_e$ fixo, $E_\nu = p_\nu = E_0 - E_e$ e $\mathrm{d}p_\nu/\mathrm{d}E_0 = 1$, así que

$$
\boxed{
\frac{\mathrm{d}\Gamma}{\mathrm{d}p_e} = \frac{|T|^2}{2\pi^3}\; p_e^2\,(E_0 - E_e)^2
}
$$

É a **forma do espectro** $\beta$: nula en $p_e = 0$ e no punto final, cun máximo no medio. (Falta
a **función de Fermi**, que corrixe a atracción coulombiana entre o electrón e o núcleo.)

### 7.2. A regra de Sargent

A anchura total é a integral do espectro. Se $E_0 \gg m_e$, $E_e \simeq p_e$ e, con $x = E_e/E_0$,

$$
\Gamma = \frac{|T|^2}{2\pi^3}\int_0^{E_0} E^2(E_0 - E)^2\,\mathrm{d}E = \frac{|T|^2}{2\pi^3}\,E_0^5\int_0^1 x^2(1 - x)^2\,\mathrm{d}x
= \frac{|T|^2\,E_0^5}{60\,\pi^3} .
$$

$$
\boxed{\Gamma \propto E_0^5}
$$

É a **regra de Sargent**: a vida media dunha desintegración feble vai como a quinta potencia da
enerxía dispoñible. É o que explica, na tarefa da introdución, que o neutrón ($E_0 = 1.29$ MeV)
viva $10^8$ veces máis que o muón ($E_0 \simeq m_\mu$), coa mesma interacción.

A potencia cinco non é casual: na teoría de Fermi $|T|^2 \propto G_F^2$, con dimensións de $\mathrm{GeV}^{-4}$, e
$\Gamma$ ten dimensións de enerxía; a única escala é $E_0$. O cálculo completo da desintegración
do muón dá $\Gamma_\mu = G_F^2\,m_\mu^5/192\pi^3$. [MT11.5]

**O punto final e a masa do neutrino.** Se o neutrino ten masa, $p_\nu^2 = (E_0 - E_e)^2 - m_\nu^2$, e o
espectro cae a cero en $E_0 - m_\nu$ con outra forma: é o que mide KATRIN (tarefa da
introdución).

*Cuestións:* comproba $\int_0^1 x^2(1 - x)^2\,\mathrm{d}x = 1/30$. Con $\Gamma \propto E_0^5$, canto cambia a vida media
se a enerxía dispoñible se reduce á metade?
