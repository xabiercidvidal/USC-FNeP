## Introdución

Esta extensión é o **desenvolvemento matemático da quiralidade**, que o tema de [leptóns](./leptones.ipynb) presenta
na clase: a interacción feble só ve a man esquerda dos fermións. Parte do que se constrúe en
[ext-dirac](./ext-dirac.ipynb): a ecuación de Dirac, os espinores, as súas transformacións de Lorentz e as formas
bilineais. Ten nove seccións; cada unha responde a unha pregunta:

1. **A quiralidade.** *Que é?* Os autovalores de $\gamma^5$.
2. **Os proxectores e a paridade.** *Como se separa un espinor nas súas dúas quiralidades?* Con $P_L$ e $P_R$. A
   paridade intercámbiaas.
3. **A base quiral.** *En que base se ven as dúas partes por separado?* A representación de Weyl.
4. **Fermións sen masa: os espinores de Weyl.** *Que pasa se $m = 0$?* As dúas partes desacóplanse.
5. **Fermións con masa.** *Que fai a masa?* Acopla as dúas partes.
6. **Rotacións e *boosts* na base quiral.** *Como se transforma cada parte?* Por separado: a quiralidade é
   invariante Lorentz.
7. **Helicidade e quiralidade.** *Como se relacionan?* Coinciden sen masa; con masa, un estado de helicidade é unha
   mestura das dúas quiralidades.
8. **Helicidade e quiralidade baixo transformacións.** *Que cambia co observador e que ao propagarse?*
9. **As formas bilineais e a quiralidade.** *Que partes ve cada forma?* O escalar (a masa) une L con R; a
   corrente vectorial non as mestura, e a $V - A$ só ve a esquerda, en calquera sistema de referencia.

* $\gamma^5 \equiv i\gamma^0\gamma^1\gamma^2\gamma^3$, con $(\gamma^5)^2 = I$, $(\gamma^5)^\dagger = \gamma^5$ e $\{\gamma^5, \gamma^\mu\} = 0$ ([ext-dirac](./ext-dirac.ipynb), sección 1.7).

## 1. A quiralidade

### 1.1. Os autovalores de $\gamma^5$

$\gamma^5$ é hermítica e o seu cadrado é a identidade. Os seus autovalores son reais e o seu cadrado vale $1$: só poden
ser $\pm 1$. Como a súa traza é cero, hai **dous de cada**.

$$
\boxed{
\gamma^5\,\Psi_L = -\Psi_L, \qquad \gamma^5\,\Psi_R = +\Psi_R
}
$$

A **quiralidade** dun espinor é o seu autovalor de $\gamma^5$: $-1$, **levóxira** (L); $+1$, **dextróxira** (R). O
nome vén do grego *kheir*, man: a paridade converte unha na outra, como un espello converte a man
esquerda na dereita (sección 2.3).

### 1.2. En Pauli-Dirac

$\gamma^5 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix}$ intercambia os dous bloques do espinor, así que os seus autovectores son

$$
\Psi_L = \begin{pmatrix} \chi \\ -\chi \end{pmatrix}, \qquad \Psi_R = \begin{pmatrix} \chi \\ \chi \end{pmatrix},
$$

con calquera $\chi$ de dúas compoñentes. Un espinor calquera **non é autoestado** de $\gamma^5$: é unha mestura das
dúas quiralidades. Por exemplo, o espinor en repouso, $u(m) = \sqrt{2m}\,(\chi, 0)$, é metade e metade:

$$
\sqrt{2m}\begin{pmatrix} \chi \\ 0 \end{pmatrix} = \sqrt{\frac{m}{2}}\begin{pmatrix} \chi \\ \chi \end{pmatrix} + \sqrt{\frac{m}{2}}\begin{pmatrix} \chi \\ -\chi \end{pmatrix} .
$$

## 2. Os proxectores e a paridade

### 2.1. Os proxectores de quiralidade

$$
\boxed{
P_L \equiv \frac{1}{2}(I - \gamma^5), \qquad P_R \equiv \frac{1}{2}(I + \gamma^5)
}
$$

Con $(\gamma^5)^2 = I$:

* son **proxectores complementarios**: $P_L + P_R = I$, $P_{L,R}^2 = P_{L,R}$, $P_L P_R = 0$;
* **dan os autoestados**: $\gamma^5 P_L = \frac{1}{2}(\gamma^5 - I) = -P_L$ e $\gamma^5 P_R = +P_R$. Logo $P_L\Psi$ é levóxiro e $P_R\Psi$
  dextróxiro, sexa cal sexa $\Psi$.

Calquera espinor se separa, dunha soa maneira, nas súas dúas partes quirais:

$$
\Psi = \Psi_L + \Psi_R, \qquad \Psi_{L,R} = P_{L,R}\,\Psi .
$$

En Pauli-Dirac,

$$
P_L = \frac{1}{2}\begin{pmatrix} I & -I \\ -I & I \end{pmatrix}, \qquad
P_R = \frac{1}{2}\begin{pmatrix} I & I \\ I & I \end{pmatrix} .
$$

### 2.2. A paridade nos espinores

A paridade é a transformación de Lorentz impropia $\Lambda = \mathrm{diag}(1, -1, -1, -1)$. A condición de
[ext-dirac](./ext-dirac.ipynb) (sección 3.1), $S^{-1}\gamma^\mu S = \Lambda^\mu{}_\nu\gamma^\nu$, pide unha $S$ que deixe $\gamma^0$ igual e cambie o signo
das $\gamma^k$. $S = \gamma^0$ cúmpreo: $\gamma^0\gamma^0\gamma^0 = \gamma^0$ e $\gamma^0\gamma^k\gamma^0 = -\gamma^k$. Así,

$$
\boxed{
\Psi(t, {\bf x}) \;\to\; \gamma^0\,\Psi(t, -{\bf x}), \qquad \bar\Psi \;\to\; \bar\Psi\,\gamma^0
}
$$

**As formas bilineais baixo paridade.** Con $\gamma^0\gamma^5\gamma^0 = -\gamma^5$ (porque $\gamma^5$ anticonmuta con $\gamma^0$):

| forma | $\to$ | baixo paridade |
|:--|:--|:--|
| $\bar\Psi\Psi$ | $\bar\Psi\gamma^0\gamma^0\Psi = \bar\Psi\Psi$ | escalar |
| $\bar\Psi\gamma^5\Psi$ | $\bar\Psi\gamma^0\gamma^5\gamma^0\Psi = -\bar\Psi\gamma^5\Psi$ | pseudoescalar |
| $\bar\Psi\gamma^\mu\Psi$ | $(j^0, {\bf j}) \to (j^0, -{\bf j})$ | vector |
| $\bar\Psi\gamma^\mu\gamma^5\Psi$ | $(j^0_A, {\bf j}_A) \to (-j^0_A, +{\bf j}_A)$ | axial |

(para o axial, $\gamma^0\gamma^\mu\gamma^5\gamma^0 = (\gamma^0\gamma^\mu\gamma^0)(\gamma^0\gamma^5\gamma^0) = -\gamma^0\gamma^\mu\gamma^0\,\gamma^5$). Son as columnas de «baixo
paridade» de [perspectivas](./perspectivas.ipynb) e de [leptóns](./leptones.ipynb).

### 2.3. A paridade intercambia as quiralidades

Como $\gamma^0$ anticonmuta con $\gamma^5$, $\gamma^0 P_L = P_R\,\gamma^0$, e

$$
\gamma^0\,\Psi_L = \gamma^0 P_L\Psi = P_R\,(\gamma^0\Psi) :
$$

a parte esquerda de $\Psi$ convértese na parte dereita do espinor transformado, e ao revés:
$\Psi_L \leftrightarrow \Psi_R$.

* **A paridade non conserva a quiralidade: invértea.** Do transformado de $\Psi_L$ non queda nada levóxiro,
  $P_L(\gamma^0\Psi_L) = \gamma^0 P_R P_L\Psi = 0$.
* **Só conta $\gamma^0$.** O cambio ${\bf x} \to -{\bf x}$ actúa sobre o argumento, non sobre as compoñentes do
  espinor; e a fase global que pode acompañar a paridade, $\Psi \to \eta_P\,\gamma^0\Psi(t, -{\bf x})$ con $|\eta_P| = 1$,
  tampouco cambia o autovalor de $\gamma^5$.
* **Unha teoría que trate distinto as dúas quiralidades viola a paridade.** É o que fai a interacción
  feble: a imaxe especular dun $\nu_L$ sería un $\nu_R$, que non está no Modelo Estándar.

### 2.4. A quiralidade das antipartículas

Para os espinores $v$, os operadores dan o valor físico co signo cambiado ([ext-dirac](./ext-dirac.ipynb),
sección 2.6), e a quiralidade non é unha excepción: o antifermión dextróxiro é

$$
v_R = P_L\,v = \frac{1}{2}(I - \gamma^5)\,v .
$$

Por iso o mesmo $P_L$ do vértice $V - A$ selecciona os fermións levóxiros **e** os antifermións
dextróxiros, e a corrente do neutrino só contén $\nu_L$ e $\bar\nu_R$. [MT6.4]

## 3. A base quiral

### 3.1. A representación de Weyl

En Pauli-Dirac $\gamma^5$ non é diagonal. Cambiando de base coa matriz unitaria

$$
U = \frac{1}{\sqrt{2}}\begin{pmatrix} I & -I \\ I & I \end{pmatrix}, \qquad
\Psi_{W} = U\,\Psi_{PD}, \qquad \gamma^\mu_{W} = U\,\gamma^\mu_{PD}\,U^\dagger ,
$$

obtense a representación **de Weyl**, ou **quiral**:

$$
\boxed{
\gamma^0 = \begin{pmatrix} 0 & I \\ I & 0 \end{pmatrix}, \qquad
\gamma^k = \begin{pmatrix} 0 & \sigma_k \\ -\sigma_k & 0 \end{pmatrix}, \qquad
\gamma^5 = \begin{pmatrix} -I & 0 \\ 0 & I \end{pmatrix}
}
$$

e $P_L = \mathrm{diag}(I, 0)$, $P_R = \mathrm{diag}(0, I)$. Cumpre a mesma álxebra, $\{\gamma^\mu, \gamma^\nu\} = 2g^{\mu\nu}I$: a física non
cambia ([ext-dirac](./ext-dirac.ipynb), sección 1.5).

### 3.2. O espinor, partido en dous

Na base quiral, as dúas compoñentes de arriba son a parte L e as dúas de abaixo a parte R:

$$
\Psi = \begin{pmatrix} \psi_L \\ \psi_R \end{pmatrix}, \qquad
\psi_L = \frac{1}{\sqrt{2}}(\psi_a - \psi_b), \quad \psi_R = \frac{1}{\sqrt{2}}(\psi_a + \psi_b),
$$

con $\psi_a$, $\psi_b$ os bloques de arriba e de abaixo en Pauli-Dirac. Cada parte é un obxecto de **dúas compoñentes**.

* A **paridade**, $\gamma^0$, intercambia os dous bloques: $(\psi_L, \psi_R) \to (\psi_R, \psi_L)$.
* O **espín**, $\Sigma_k = \mathrm{diag}(\sigma_k, \sigma_k)$, é o mesmo nas dúas bases, e actúa igual sobre as dúas partes.
* O **adxunto**, $\bar\Psi = \Psi^\dagger\gamma^0 = (\psi_R^\dagger, \psi_L^\dagger)$, cruza os bloques.

## 4. Fermións sen masa: os espinores de Weyl

### 4.1. A ecuación de Dirac na base quiral

Con $\not{p} = \gamma^0 E - \gamma^k p^k = \begin{pmatrix} 0 & E - \vec\sigma\cdot{\bf p} \\ E + \vec\sigma\cdot{\bf p} & 0 \end{pmatrix}$, a ecuación $(\not{p} - m)\,u = 0$ queda

$$
\begin{pmatrix} -m & E - \vec\sigma\cdot{\bf p} \\ E + \vec\sigma\cdot{\bf p} & -m \end{pmatrix}
\begin{pmatrix} u_L \\ u_R \end{pmatrix} = 0
\qquad \Longleftrightarrow \qquad
\begin{cases}
(E + \vec\sigma\cdot{\bf p})\,u_L = m\,u_R \\
(E - \vec\sigma\cdot{\bf p})\,u_R = m\,u_L
\end{cases}
$$

**A masa é o único que acopla** $u_L$ con $u_R$.

### 4.2. Sen masa, as dúas partes sepáranse

$$
(E + \vec\sigma\cdot{\bf p})\,u_L = 0, \qquad (E - \vec\sigma\cdot{\bf p})\,u_R = 0 :
$$

son as **ecuacións de Weyl** (1929). Con $E = \mathrm{p}$ e ${\bf p} = \mathrm{p}\,\hat{\bf v}$:

$$
(\hat{\bf v}\cdot\vec\sigma)\,u_L = -u_L, \qquad (\hat{\bf v}\cdot\vec\sigma)\,u_R = +u_R .
$$

**Sen masa, a parte L ten helicidade negativa e a R positiva.** E son **dúas partículas independentes**, de dúas
compoñentes cada unha: pódese ter unha sen a outra. É o neutrino do Modelo Estándar, que só ten $u_L$
(e a súa antipartícula, $\bar\nu_R$).

***[+] Mirador.*** *Weyl propuxo a súa ecuación en 1929, e Pauli descartouna porque viola a paridade. Tras o
experimento de Wu (1957) resultou ser a do neutrino.*

## 5. Fermións con masa

### 5.1. As solucións

Con masa, as dúas ecuacións da sección 4.1 van xuntas: dada $u_L$, $u_R = (E + \vec\sigma\cdot{\bf p})\,u_L/m$. Con ${\bf p}$ en $z$ e
$\sigma_3\chi_\pm = \pm\chi_\pm$, normalizados a $u^\dagger u = 2E$:

$$
\boxed{
u_\pm = \begin{pmatrix} \sqrt{E \mp \mathrm{p}}\;\chi_\pm \\ \sqrt{E \pm \mathrm{p}}\;\chi_\pm \end{pmatrix}
}
$$

(con $\sqrt{E - \mathrm{p}}\sqrt{E + \mathrm{p}} = m$). **As dúas partes están sempre presentes.** En repouso, $\mathrm{p} = 0$, pesan o mesmo:
$u = \sqrt{m}\,(\chi, \chi)$. Ao aumentar a enerxía, en $u_+$ medra a parte R e en $u_-$ a L.

### 5.2. O termo de masa

Con $\bar\Psi = (\psi_R^\dagger, \psi_L^\dagger)$,

$$
\bar\Psi\Psi = \psi_R^\dagger\psi_L + \psi_L^\dagger\psi_R = \bar\Psi_R\Psi_L + \bar\Psi_L\Psi_R :
$$

o termo de masa do lagranxiano, $m\bar\Psi\Psi$, **acopla a esquerda coa dereita**. Un fermión con masa (de
Dirac) necesita as dúas partes.

### 5.3. A quiralidade non se conserva ao propagarse

O hamiltoniano de Dirac, $H = \vec\alpha\cdot{\bf p} + \beta m$, con $\alpha^k = \gamma^0\gamma^k$ e $\beta = \gamma^0$, na base quiral é

$$
H = \begin{pmatrix} -\vec\sigma\cdot{\bf p} & m \\ m & \vec\sigma\cdot{\bf p} \end{pmatrix} :
$$

o termo cinético é diagonal (non mestura) e a masa está **fóra da diagonal**. En xeral, $\gamma^5$ conmuta con
$\alpha^k$ (anticonmuta con cada factor) e anticonmuta con $\beta$:

$$
[\gamma^5, H] = 2m\,\gamma^5\gamma^0 \neq 0 .
$$

**Con masa, a quiralidade non se conserva**: un espinor que empeza sendo L desenvolve parte R ao evolucionar.
Sen masa, si se conserva.

## 6. Rotacións e *boosts* na base quiral

### 6.1. As dúas partes transfórmanse por separado

En [ext-dirac](./ext-dirac.ipynb) (seccións 3.2–3.6), os xeradores son o espín, $\frac{1}{2}\Sigma_k$, para as rotacións, e
$\frac{1}{2}\gamma^0\gamma^k$ para os *boosts*. Na base quiral, $\Sigma_k = \mathrm{diag}(\sigma_k, \sigma_k)$ e $\gamma^0\gamma^k = \mathrm{diag}(-\sigma_k, \sigma_k)$: **os dous son
diagonais por bloques**, e polo tanto tamén $S(\Lambda)$:

$$
\boxed{
S = \begin{pmatrix} S_L & 0 \\ 0 & S_R \end{pmatrix}, \qquad
\begin{aligned}
&\text{rotación } \theta\,\hat{\bf n}: && S_L = S_R = e^{-i\,\theta\,\hat{\bf n}\cdot\vec\sigma/2} \\
&\textit{boost } \eta\,\hat{\bf n}: && S_L = e^{-\eta\,\hat{\bf n}\cdot\vec\sigma/2}, \quad S_R = e^{+\eta\,\hat{\bf n}\cdot\vec\sigma/2}
\end{aligned}
}
$$

* **Ningunha transformación de Lorentz mestura L con R**: $\psi_L \to S_L\,\psi_L$ e $\psi_R \to S_R\,\psi_R$. Un espinor puramente L é
  puramente L para calquera observador. **A quiralidade é invariante Lorentz** (equivalente a $[\gamma^5, S] = 0$,
  ext-dirac, sección 3.2).
* As dúas partes son **dúas representacións distintas** do grupo de Lorentz: transfórmanse igual baixo rotacións,
  pero co *boost* oposto. Só a paridade as une (sección 2.3).
* **Rotacións:** a mesma matriz unitaria nas dúas partes: a proporción entre L e R non cambia.
* ***Boosts*:** non son unitarios e teñen expoñentes opostos: **reescalan** as dúas partes de forma distinta. A
  proporción cambia, pero o L segue sendo L.

### 6.2. Exemplo: o espinor en movemento, sen resolver a ecuación

O espinor en repouso con espín arriba é $u(m) = \sqrt{m}\,(\chi_+, \chi_+)$. Un *boost* de rapidez $\eta$ en $z$ actúa sobre
$\chi_+$, con $\sigma_3 = +1$:

$$
S\,u(m) = \sqrt{m}\begin{pmatrix} e^{-\eta/2}\,\chi_+ \\ e^{+\eta/2}\,\chi_+ \end{pmatrix}
= \begin{pmatrix} \sqrt{E - \mathrm{p}}\;\chi_+ \\ \sqrt{E + \mathrm{p}}\;\chi_+ \end{pmatrix} = u_+ ,
$$

porque $m\,e^{\mp\eta} = m(\cosh\eta \mp \sinh\eta) = E \mp \mathrm{p}$. É a solución da sección 5.1. O *boost* non cambia a dirección
de ningunha das dúas partes: só multiplica a L por $e^{-\eta/2}$ e a R por $e^{+\eta/2}$.

## 7. Helicidade e quiralidade

### 7.1. Un estado de helicidade, nas súas dúas quiralidades

Da sección 5.1, con $|u|^2 = 2E$, a fracción de cada quiralidade nun estado de helicidade é

| helicidade $\downarrow$ \ quiralidade $\rightarrow$ | R | L |
| :--: | :--: | :--: |
| $+$ | $\frac{E + \mathrm{p}}{2E} = \frac{1}{2}(1 + \beta)$ | $\frac{E - \mathrm{p}}{2E} = \frac{1}{2}(1 - \beta)$ |
| $-$ | $\frac{1}{2}(1 - \beta)$ | $\frac{1}{2}(1 + \beta)$ |

* **En repouso**, $\beta = 0$: metade e metade.
* **Sen masa**, $\beta = 1$: $u_+$ é puramente R e $u_-$ puramente L. **Helicidade e quiralidade coinciden.**
* **Con $E \gg m$**: a fracción «equivocada», $|P_L u_+|^2/|u_+|^2 = \frac{1}{2}(1 - \beta) \simeq m^2/4E^2$, é pequena. A
  propia compoñente (a súa raíz, a amplitude) vai como $m/2E$. A amplitude é a que entra en $M_{fi}$; a fracción,
  nas taxas.

### 7.2. En Pauli-Dirac

É o mesmo escrito na outra base. Con $u_+ = N(1, 0, \kappa, 0)$ e $u_- = N(0, 1, 0, -\kappa)$:

$$
P_R\,u_+ = \tfrac{1}{2}(1+\kappa)\,N\,(1, 0, 1, 0), \qquad P_L\,u_+ = \tfrac{1}{2}(1-\kappa)\,N\,(1, 0, -1, 0),
$$

$$
P_R\,u_- = \tfrac{1}{2}(1-\kappa)\,N\,(0, 1, 0, 1), \qquad P_L\,u_- = \tfrac{1}{2}(1+\kappa)\,N\,(0, 1, 0, -1) .
$$

As **amplitudes** $\frac{1}{2}(1 \pm \kappa)$ son as que usa o tema de leptóns; os seus cadrados, pola norma dos vectores,
$2N^2\cdot\frac{1}{4}(1 \pm \kappa)^2 = E \pm \mathrm{p}$, dan as fraccións da táboa de arriba. Na desintegración a dous corpos
$\pi \to l\,\nu$, con $E_l$ e $\mathrm{p}_l$ do CM ([ext-cinematica](./ext-cinematica.ipynb), sección 4.3), $\kappa = (m_\pi - m_l)/(m_\pi + m_l)$ e
$\frac{1}{2}(1 - \kappa) = m_l/(m_\pi + m_l)$: a supresión por helicidade de [leptóns](./leptones.ipynb).

## 8. Helicidade e quiralidade baixo transformacións

### 8.1. Cada unha ten unha virtude e un defecto

| | helicidade, $\vec\Sigma\cdot\hat{\bf p}$ | quiralidade, $\gamma^5$ |
| :-- | :-- | :-- |
| consérvase ao propagarse? (conmuta con $H$) | **si** | **non**, se hai masa (sección 5.3) |
| é a mesma para todo observador? (conmuta con $S(\Lambda)$) | **non**, se hai masa | **si** (sección 6.1) |
| tena definida unha partícula libre con masa? | **si** | **non**: leva as dúas partes |
| sen masa | consérvase, é invariante e coincide coa quiralidade | ídem |

**A helicidade depende do observador** se hai masa: un observador que adianta a partícula véa moverse cara
atrás, co espín igual, e a súa helicidade cambia de signo. Sen masa non se pode adiantar.

**A quiralidade non depende do observador**, pero unha partícula con masa non a ten definida. O que é igual
para todos os observadores é a **separación** do espinor nas súas dúas partes: a parte L para un é a parte L
para calquera outro. O que cambia é **canto pesa** cada parte.

### 8.2. Exemplo: adiantar a partícula

Tomemos $u_+$, con rapidez $\vartheta$ en $z$. As súas dúas partes están na proporción (sección 6.2)

$$
\frac{\text{amplitude R}}{\text{amplitude L}} = \sqrt{\frac{E + \mathrm{p}}{E - \mathrm{p}}} = e^{\vartheta} .
$$

Un *boost* de rapidez $\eta$ en $z$ multiplica a R por $e^{\eta/2}$ e a L por $e^{-\eta/2}$: a proporción pasa a $e^{\vartheta + \eta}$.

* **$\eta = -\vartheta$**, o sistema en repouso: metade e metade.
* **$\eta = -2\vartheta$**, o observador que a adianta: a proporción é $e^{-\vartheta}$, domina a parte L. A partícula
  móvese cara a $-z$ co espín cara a $+z$: helicidade negativa. É o mesmo estado, visto por outro observador.

**Ningunha parte cambiou de quiralidade**; cambiou o seu peso. Por iso a interacción feble se escribe coa
quiralidade: o $W$ acóplase á parte L, e todos os observadores están de acordo en cal é.

## 9. As formas bilineais e a quiralidade

### 9.1. Os adxuntos das partes quirais

Para separar as formas bilineais fan falta os adxuntos de $\Psi_L$ e $\Psi_R$. Como $\gamma^5$ é hermítica e
anticonmuta con $\gamma^0$, $P_L\gamma^0 = \gamma^0 P_R$:

$$
\bar\Psi_L \equiv (P_L\Psi)^\dagger\gamma^0 = \Psi^\dagger P_L\gamma^0 = \Psi^\dagger\gamma^0 P_R = \bar\Psi\,P_R, \qquad \bar\Psi_R = \bar\Psi\,P_L .
$$

**O adxunto cambia o proxector.** E como $\gamma^5$ anticonmuta con $\gamma^\mu$, ao pasar un proxector a través dunha $\gamma^\mu$
tamén cambia: $P_R\gamma^\mu = \gamma^\mu P_L$. Con estas dúas regras, e $P_L P_R = 0$, sae todo o demais.

### 9.2. O escalar e o pseudoescalar mesturan as quiralidades

Escribindo $\Psi = \Psi_L + \Psi_R$ e $\bar\Psi = \bar\Psi_L + \bar\Psi_R$, os termos coa mesma quiralidade anúlanse:

$$
\bar\Psi_L\Psi_L = \bar\Psi P_R\,P_L\Psi = 0, \qquad \bar\Psi_R\Psi_R = \bar\Psi P_L\,P_R\Psi = 0 ,
$$

e só sobreviven os cruzados:

$$
\boxed{
\bar\Psi\Psi = \bar\Psi_L\Psi_R + \bar\Psi_R\Psi_L
}
\qquad
\bar\Psi\gamma^5\Psi = \bar\Psi_L\Psi_R - \bar\Psi_R\Psi_L .
$$

**O escalar une a esquerda coa dereita.** É o termo de masa, $m\bar\Psi\Psi$ (sección 5.2).

### 9.3. O vector e o axial non as mesturan

Agora son os cruzados os que se anulan, porque a $\gamma^\mu$ cambia o proxector:

$$
\bar\Psi_L\gamma^\mu\Psi_R = \bar\Psi P_R\,\gamma^\mu P_R\Psi = \bar\Psi\gamma^\mu P_L P_R\Psi = 0 ,
$$

e o mesmo $\bar\Psi_R\gamma^\mu\Psi_L$. Sobreviven os da mesma quiralidade:

$$
\boxed{
\bar\Psi\gamma^\mu\Psi = \bar\Psi_L\gamma^\mu\Psi_L + \bar\Psi_R\gamma^\mu\Psi_R
}
\qquad
\bar\Psi\gamma^\mu\gamma^5\Psi = -\bar\Psi_L\gamma^\mu\Psi_L + \bar\Psi_R\gamma^\mu\Psi_R
$$

(no axial, $\gamma^5\Psi_L = -\Psi_L$ e $\gamma^5\Psi_R = +\Psi_R$). Restando:

$$
\boxed{
\bar\Psi\gamma^\mu\,\frac{1}{2}(1 - \gamma^5)\,\Psi = \frac{1}{2}\left(V^\mu - A^\mu\right) = \bar\Psi_L\gamma^\mu\Psi_L
}
$$

**A regra:** un número **par** de matrices $\gamma$ (S, P e tamén o tensor, $\sigma^{\mu\nu}$) une L con R; un número **impar**
(V, A) une L con L e R con R.

| forma | en partes quirais | |
| :-- | :-- | :-- |
| $\bar\Psi\Psi$ | $\bar\Psi_L\Psi_R + \bar\Psi_R\Psi_L$ | **mestura** L e R |
| $\bar\Psi\gamma^5\Psi$ | $\bar\Psi_L\Psi_R - \bar\Psi_R\Psi_L$ | mestura |
| $\bar\Psi\gamma^\mu\Psi$ | $\bar\Psi_L\gamma^\mu\Psi_L + \bar\Psi_R\gamma^\mu\Psi_R$ | **non mestura**; L e R por igual |
| $\bar\Psi\gamma^\mu\gamma^5\Psi$ | $-\bar\Psi_L\gamma^\mu\Psi_L + \bar\Psi_R\gamma^\mu\Psi_R$ | non mestura; L e R con signo oposto |
| $\frac{1}{2}(V - A)$ | $\bar\Psi_L\gamma^\mu\Psi_L$ | **só L** |

### 9.4. Na base quiral

Con $\Psi = (\psi_L, \psi_R)$ e $\bar\Psi = (\psi_R^\dagger, \psi_L^\dagger)$ (sección 3.2), todo é explícito. Con $\gamma^0\gamma^0 = I$ e
$\gamma^0\gamma^k = \mathrm{diag}(-\sigma_k, \sigma_k)$, e chamando $\sigma^\mu = (I, \vec\sigma)$ e $\bar\sigma^\mu = (I, -\vec\sigma)$:

$$
\bar\Psi\Psi = \psi_R^\dagger\,\psi_L + \psi_L^\dagger\,\psi_R, \qquad
\bar\Psi\gamma^\mu\Psi = \psi_L^\dagger\,\bar\sigma^\mu\,\psi_L + \psi_R^\dagger\,\sigma^\mu\,\psi_R .
$$

O escalar está **fóra da diagonal**: une os bloques de arriba e de abaixo. A corrente é **diagonal**: cada parte
coa súa, e cada unha coa súa propia matriz, $\bar\sigma^\mu$ ou $\sigma^\mu$ (por iso as dúas partes son representacións
distintas, sección 6.1). A corrente $V - A$ é $\psi_L^\dagger\bar\sigma^\mu\psi_L$: **só necesita o obxecto de dúas compoñentes** $\psi_L$.

### 9.5. Que significa

* **As interaccións vectoriais**, a electromagnética e a forte, acoplan a esquerda e a dereita **por igual
  e por separado**: non distinguen as mans, e conservan a paridade.
* **A corrente cargada da feble**, $V - A$, só acopla a esquerda: a parte dereita non participa.
* **Sen masa, a quiralidade consérvase no vértice**: un electrón que entra L sae L, sexa o vértice do fotón ou do
  $W$. Como sen masa a quiralidade é a helicidade, a alta enerxía **a helicidade consérvase nos vértices**
  vectoriais.
* **A masa é o único que une as dúas partes.** Un fermión con só a parte L, como o neutrino do Modelo
  Estándar, non pode ter un termo $m\bar\Psi\Psi$: non ten masa de Dirac.

***[+] Mirador.*** *No Modelo Estándar, $e_L$ e $e_R$ son obxectos distintos para a interacción feble: $e_L$ vai con
$\nu_L$ e $e_R$ vai só. Un termo $m\,\bar e_L e_R$ mesturaría dous obxectos que a feble trata de forma distinta, e non está permitido.
A masa do electrón dáa o seu acoplamento co campo de Higgs (tema do Modelo Estándar).*

### 9.6. É a mesma en todos os sistemas inerciais?

A corrente cargada une dous campos distintos, por exemplo o electrón e o seu neutrino:

$$
j^\mu = \bar e\,\gamma^\mu\,\frac{1}{2}(1 - \gamma^5)\,\nu = \bar e_L\,\gamma^\mu\,\nu_L .
$$

Cunha transformación de Lorentz, $\nu \to S\nu$, $\bar e \to \bar e\,S^{-1}$, e como $P_L$ conmuta con $S$:

$$
j'^\mu = \bar e\,S^{-1}\gamma^\mu P_L\,S\,\nu = \bar e\,(S^{-1}\gamma^\mu S)\,P_L\,\nu = \Lambda^\mu{}_\nu\;\bar e\,\gamma^\nu P_L\,\nu .
$$

* **A súa forma é a mesma en todos os sistemas inerciais**: segue sendo unha corrente entre partes L, $\bar e_L\gamma^\mu\nu_L$.
  Ningún observador ve intervir a parte R.
* **O seu valor non**: é un **cuadrivector**, e as súas compoñentes cambian como as de calquera cuadrivector. O que é
  igual para todos é a súa contracción con outro cuadrivector, $j^\mu W_\mu$ ou $j_{lep}\cdot j_{had}$: un escalar, e con el $M_{fi}$.
* **Coa paridade, en cambio, a forma cambia**: $\gamma^0$ leva $P_L$ a $P_R$ (sección 2.3), e a corrente convértese en
  $\bar e_R\gamma^\mu\nu_R$, que non aparece na teoría. É a violación da paridade.
